## 1. A real collimated laser beam

**Laser → collimated beam → screen.** Move the screen along the beam. It records a localized intensity spot, not the rapidly oscillating signed field.

A finite beam concentrates light in a transverse region and carries finite power. Its envelope may have smooth tails rather than a sharp edge. A quoted diameter needs an intensity convention. An aperture can impose an edge at its own plane.

A collimated beam has a small angular spread, so its spot changes slowly over a useful distance. It still generally diverges. Its wavefronts may be nearly flat across a limited region, and its amplitude may vary little across a small central patch. That patch can be described approximately by a plane wave, even though the whole beam cannot. Ask both **how large a region** and **over what distance** before calling a beam plane.

## 2. The ideal plane wave

For one optical frequency in a homogeneous medium, choose

$$U_{\mathrm{PW}}(\mathbf r)=A_0e^{i\mathbf k\cdot\mathbf r}.$$

Here $A_0$ is constant; take it real and positive for this lesson. The exponential changes spatial phase, not amplitude. Surfaces on which $\mathbf k\cdot\mathbf r$ is constant are planes. One wavevector specifies their orientation and phase spacing; its magnitude is $k=2\pi/\lambda$. The next lesson develops that geometry carefully. With our $e^{-i\omega t}$ convention, the crests travel along $\mathbf k$.

The model has the same nonzero amplitude arbitrarily far across each transverse plane. Consequently its time-averaged intensity is uniform, with no physical beam boundary. **A finite plotting window does not imply a finite physical beam.** Cropping the drawing changes neither the equation nor the light it describes.

Integrating this constant intensity over an unbounded transverse plane gives infinite power. No finite-power source prepares that globally uniform field. This spatial limitation is separate from monochromaticity: choosing one frequency is a temporal idealization, whereas removing the transverse envelope is a spatial one. A finite-width field can also be modeled monochromatically. We use the ideal plane wave because its simple phase structure is useful, not because an actual laser fills all space.

## 3. A finite optical beam

A more general scalar description separates amplitude and phase:

$$U(\mathbf r)=A(\mathbf r)e^{i\Phi(\mathbf r)}.$$

The envelope sets where the field is strong. Constant phase defines wavefronts wherever the field is nonzero. Changing beam width does not necessarily change its central direction.

As an example, prescribe a Gaussian-shaped cross-section at $z=0$:

$$A(x,0)=A_0\exp(-x^2/w_0^2).$$

The radius $w_0$ is where amplitude falls to $A_0/e$, so intensity falls to $1/e^2$ of its axial value. It is not a hard boundary. The expression specifies one transverse profile at one plane. It does not, by itself, say how that profile propagates. Simply carrying the same Gaussian unchanged along $z$ with a plane-wave carrier is generally not an exact Helmholtz solution.

A finite beam requires a spread of plane-wave directions. Narrower localization generally means greater angular spread and divergence. See [0.4: angular-spectrum propagation](0_4_Propagation_of_Scalar_Optical_Fields.ipynb).

## 5. What does the wavevector arrow tell us?

Both fields have the same central $+z$ arrow. It marks direction, not beam width or travel distance.

A Gaussian waist has a flat wavefront but nonuniform amplitude. Farther away, the wavefront curves. The arrow marks the central axis, not every local phase normal.

Image edges are viewing limits. Read each arrow as labeled: direction, wavevector or geometrical boundary.

## 6. Small executable calculations

### A. Uniform versus Gaussian-shaped intensity

**Predict:** at $x=w_0$, is the beam intensity $1/e$ or $1/e^2$ of its central value? **Calculate** the two profiles, then **change** only `w0` from 3 to 2 µm. **Interpret:** does the uniform profile acquire an edge? These calculation variables are independent of the explorer controls.

```python
x = np.linspace(-12.0, 12.0, 2401)  # µm
A0, w0 = 1.0, 3.0                 # relative field amplitude; µm
I_pw = np.full_like(x, abs(A0)**2)
I_beam = abs(A0)**2 * np.exp(-2*x**2/w0**2)
for position in (0.0, w0, 2*w0):
    print(f"x = {position:g} µm: plane = {np.interp(position, x, I_pw):.4f}, "
          f"beam = {np.interp(position, x, I_beam):.4f}")
```

:::{dropdown} Answer A

At the axis both intensities are 1. At one waist radius the Gaussian intensity is $e^{-2}\approx0.1353$, not $e^{-1}$. At twice the radius it is about 0.000335. Reducing $w_0$ narrows the beam profile but leaves the ideal plane-wave intensity unchanged everywhere. The envelope changes continuously; these radii are not edges.

:::

### B. Width and integrated intensity

**Predict:** what happens when you double the integration window for each profile? **Calculate** symmetric one-dimensional integrals. Then **change** the largest half-window and interpret whether each result approaches a limit. We hold the peak intensity fixed.

```python
integrate = getattr(np, 'trapezoid', None)  # NumPy 2 or NumPy 1 compatibility
if integrate is None:
    integrate = np.trapz
for half_window in (3.0, 6.0, 12.0, 24.0):  # µm
    grid = np.linspace(-half_window, half_window, 10001)
    uniform_integral = integrate(np.ones_like(grid), x=grid)
    gaussian_integral = integrate(np.exp(-2*grid**2/w0**2), x=grid)
    print(f"±{half_window:g} µm: uniform = {uniform_integral:.4f}, "
          f"Gaussian = {gaussian_integral:.4f} (relative intensity × µm)")
print("Gaussian cross-section limit:", w0*np.sqrt(np.pi/2))
```

These are line integrals, not total optical power: they have relative-intensity-times-length units. For a circular three-dimensional Gaussian beam, the waist-plane intensity depends on both transverse coordinates as $\exp[-2(x^2+y^2)/w_0^2]$. Its area integral is finite, $\pi w_0^2/2$ times peak intensity. Actual power in watts also requires calibrated intensity units. An ideal nonzero uniform plane wave instead has an unbounded area integral.

:::{dropdown} Answer B

The uniform line integral is twice the half-window and keeps growing. The Gaussian approaches $w_0\sqrt{\pi/2}$, about 3.760 for $w_0=3$ µm. Its weak distant tails add progressively less. The finite line integral illustrates localization, but only a two-dimensional transverse integral yields total power for a physical three-dimensional beam. A finite plotting interval cannot make the ideal plane wave's total power finite.

:::

## 7. When is the plane-wave approximation useful?

Use a plane wave locally where amplitude is nearly constant and wavefronts are nearly flat. Check both the region size and propagation distance. This approximation does not predict total beam power or a spot boundary.

Similarly, a grating may experience an approximately planar incident wavefront over the illuminated patch. An ideal infinite-periodic calculation represents each diffraction order by a plane-wave component. That component describes a direction and complex amplitude; it does not assert that a laboratory order is an infinitely wide beam. Finite illumination introduces envelopes and angular spread, which matter when the question concerns beam size or where light reaches a screen.

Plane-wave decomposition represents a field mathematically. It does not require the apparatus to prepare separate infinite beams.

**An ideal plane wave describes one propagation direction with uniform amplitude and unlimited transverse extent. A real optical beam has a finite spatial envelope. Plane waves remain useful because more complicated fields can be represented using plane-wave components.**

:::{dropdown} Optional historical context: why elementary waves?

- Huygens's *Treatise on Light* (1690) sought wave-based explanations for reflection and refraction; Chapter I explicitly starts from experimentally established optical behaviour [\[2\]](#ref-1-2-huygens). This predates lasers and is not a description of Gaussian-beam preparation.
- In modern scalar wave theory, plane waves are particularly simple elementary solutions in a homogeneous medium. Expanding a field in these solutions separates spatial structure into directions that can be propagated and recombined [\[1\]](#ref-1-2-gaussian), [course connections](#course-1-2-1).
- That mathematical convenience explains their continuing role; it does not make an infinite plane wave a laboratory object.

:::
