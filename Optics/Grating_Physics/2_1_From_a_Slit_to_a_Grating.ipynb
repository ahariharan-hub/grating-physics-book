## 1. Start with one isolated slit

**One isolated slit has a continuous range of candidate outgoing directions.** Here we ask only which plane-wave directions are possible in the surrounding medium. We do not yet calculate their amplitudes.

Use an $x$–$z$ cross-section of an opaque screen at $z=0$. The slit runs along $y$ and the incident wave, with wavevector $\mathbf k_i$, approaches from above, toward $-z$. The slit is uniform along $y$ and the incident wave has $k_{y,i}=0$, so the outgoing components also lie in the $x$–$z$ plane.

Measure the outgoing angle $\theta$ **from $-z$ toward $+x$**: zero points straight down. This angle reference differs from the $+z$ reference in notebook one; the Cartesian axes and the meaning of $\mathbf k$ stay the same. We write $\mathbf k(\theta)=(k_x,0,k_z)$ for the chosen outgoing candidate.
- **Bold $\mathbf k$ denotes a vector; plain $k$ denotes its positive magnitude.** At a fixed wavelength in the surrounding medium, each propagating candidate has the same magnitude $|\mathbf k(\theta)|=k$, with

$$k=\frac{2\pi}{\lambda},\qquad k_x=k\sin\theta,\qquad k_z=-k\cos\theta.$$

The outgoing angle $\theta$ can vary continuously between $-90^\circ$ and $+90^\circ$. The endpoints are grazing directions along the screen. In wavevector space, the tips fill a **continuous semicircle**, since $k_x^2+k_z^2=k^2$. This is the plane-wave direction constraint used in the angular-spectrum description. [8, §§3.5–3.6](#ref-8)

- **There is no grating period $d$ for one isolated slit.** Moving sideways does not bring us to an identical neighbouring opening.
- We therefore have no period-to-period phase agreement condition and no integer grating-order number $m$ to assign.
- We call these a continuum of candidate directions, rather than infinitely many grating orders.

**A possible direction does not guarantee nonzero intensity.** The propagating part of the slit field combines plane-wave components from this continuous set. Their strengths depend on the field across the opening. Contributions from different parts of the slit can reinforce or cancel, so the far-field intensity generally varies continuously with direction and can have dark zeros. [9, Figs. 4.4–4.5](#ref-9)

The semicircle answers **which wavevectors are geometrically possible**. It does not answer **how strongly the slit excites each one**. Its arrows do not represent equal intensity, and choosing a direction does not establish that its amplitude is nonzero.

<details><summary>Example: a possible direction with zero intensity</summary>

For this separate example, take a uniformly illuminated slit of width $a$ at **normal incidence**, using the standard scalar far-field model. Its dark directions satisfy

$$a\sin\theta=p\lambda,\qquad p=\pm1,\pm2,\ldots,$$

when those angles exist. Here $p$ labels **intensity minima**, not grating orders; $a$ is the width of this one opening, not a grating period. [9, Eq. (4.1)](#ref-9)

If $a=2\lambda$, the first zeros occur at $\theta=\pm30^\circ$. Both directions have real wavevectors on our semicircle, yet their far-field intensity is zero in this model because contributions across the slit cancel.

The width assumption belongs to this example. The direction explorer below does not calculate a slit field or its intensity pattern. We will develop the strengths later.

</details>

**Try the direction control below.** The orange arrow selects one candidate, while the green projection shows its $k_x$. The grey arrows are just a few samples of the continuum. The screen drawing shows normal incidence; the orange arrow is a candidate component, not the entire diffracted field.

The arrows select directions, not finite beam outlines. Each plane-wave component is laterally extended; the diffracted slit field combines a continuous range of them. Drawing only a few arrows neither confines the field to those lines nor replaces the aperture’s continuous spectrum. [8, §3.6](#ref-8)

### Small calculation 1: choose any outgoing direction

The interactive uses the same wavevector components as notebook one, with the angle measured from $-z$. Here, `np.sin` and `np.cos` turn a chosen angle into $k_x$ and $k_z$; `np.hypot` checks that the wavevector length stays $k$.

**Predict, then try:** replace $30^\circ$ with $14.5^\circ$ or $37.2^\circ$. Does the length change? Then keep the angle fixed and double the wavelength. There is no period or integer-order test in this calculation.

```python
import numpy as np

wavelength = 2.0  # micrometres; independent of the sliders
candidate_angle_deg = 30.0
if wavelength <= 0 or not -90 <= candidate_angle_deg <= 90:
    raise ValueError('Use a positive wavelength and an angle between -90 and 90 degrees.')
k = 2*np.pi / wavelength
angle = np.deg2rad(candidate_angle_deg)
kx = k*np.sin(angle)
kz = -k*np.cos(angle)
print(f'Candidate angle: {candidate_angle_deg:+.1f} degrees')
print(f'kx = {kx:+.3f}; kz = {kz:+.3f} rad/um')
print(f'Wavevector length = {np.hypot(kx, kz):.3f}; 2*pi/lambda = {k:.3f} rad/um')
```

<details><summary>Answer for the stated starting values and changes</summary>

At $\lambda=2\,\mu\mathrm m$ and $\theta=30^\circ$, $k_x\approx1.571$ and $k_z\approx-2.721\,\mathrm{rad}/\mu\mathrm m$. The length is $k=\pi\,\mathrm{rad}/\mu\mathrm m$.

Choosing $14.5^\circ$, $37.2^\circ$, or any other angle in the outgoing half-plane changes the components but preserves that length. Doubling the wavelength halves $k$, $k_x$ and $k_z$, while leaving the direction unchanged. This check says nothing about how strongly the slit excites the chosen component.

</details>

**Next, repeat the opening.** One isolated slit has no period. Placing identical openings a distance $d$ apart introduces a repeating structure. We will describe that structure and its illumination, then see the discrete directions it permits. [Notebook three](3_0_Why_Diffraction_Orders_Appear.ipynb) explains why the phase change across a period selects those directions.

**Preview: an order has amplitude and phase as well as direction**

Let the integer $m$ label a diffraction order. In the homogeneous output medium, a propagating monochromatic plane-wave order has the complex field

$$\boxed{U_m(\mathbf r)=C_m e^{i\mathbf k_m\cdot\mathbf r},
\qquad C_m=A_m e^{i\phi_m}.}$$

The wavevector $\mathbf k_m$ sets direction and spatial phase variation. The nonnegative magnitude $A_m=|C_m|$ and phase offset $\phi_m=\arg C_m$ together form the **complex diffraction amplitude** $C_m$; the phase is defined only when $C_m\ne0$. The real field is $u_m=\operatorname{Re}\{U_m e^{-i\omega t}\}$.

<details><summary>Fixed coefficient phases, spatially varying relative phase</summary>

A stationary deterministic grating illuminated by a fully coherent monochromatic plane wave gives definite complex coefficients within this scalar model. For two nonzero propagating orders, define relative phase as order $n$'s phase minus order $m$'s:

$$\boxed{\Delta\Phi_{mp}(\mathbf r)
=(\mathbf k_p-\mathbf k_m)\cdot\mathbf r+\phi_p-\phi_m.}$$

The coefficient difference $\phi_p-\phi_m$ is fixed for a common reference convention. The wavevector difference makes the **total** relative phase vary across space, even though it stays fixed in time at each point. This matters when orders overlap and interfere; [Notebook 7.1](7_1_Interference_and_Mutual_Coherence_of_Two_Plane_Waves.ipynb) develops the intensity calculation. Here the deterministic phase relationship is a model assumption, not an experimental measurement of temporal or spatial coherence.

</details>

This lesson introduces the orders; the following kinematics lessons establish their allowed wavevectors. They do **not** determine $C_m$: an allowed order can have zero amplitude. Nor is $|C_m|^2$ itself a diffraction efficiency $\eta_m$ (a power fraction); efficiencies require incident-power and normal-flux normalization, including propagation direction and medium.

<details><summary>Phase references and evanescent orders</summary>

Specify $\phi_m$ relative to a common coordinate origin and reference plane, taken here at $z=0$. Changing that reference changes the coefficient phase. An overall absolute phase is not independently observable; relative phases matter when order fields overlap.

For transmission toward $z<0$, propagating orders have $k_{z,m}<0$ (zero at grazing). The same exponential can represent evanescent orders using a complex normal component chosen to decay into $z<0$. These are not propagating beams.

</details>

## 2. Repeat the opening to make a grating

Now repeat the single opening: imagine an opaque sheet with many straight, equally spaced openings. A wave can pass through the openings; the strips between them block it in our simplest model. More generally, a grating can contain repeated regions with different transmission, or repeated reflecting structures. [1](#ref-1)

Two lengths describe our opening pattern:

- **Period $d$:** the distance between corresponding points in neighbouring openings. You can measure centre to centre or left edge to left edge, as in the diagram. Moving by $d$ reproduces the structure.
- **Opening width $a$:** the width of one transparent strip. It need not equal the period. The incident-wave explorer uses $a=d/2$, so half of each period is open. The small calculations also let you edit $a$.

The long parallel structures in ruled or etched gratings are called **grooves**. Our model uses long slits rather than a detailed groove shape. We still use "along the grooves" to identify their long direction. **One-dimensional grating** means that the pattern repeats in only one direction; the wave and physical structure still occupy three-dimensional space.

Here, all distances, including $a$, $d$, and the **wavelength in the medium** $\lambda$, are in micrometres ($\mu\mathrm m$). We use the same homogeneous, isotropic medium on both sides.

## 3. Name the directions

Walk across the openings along **$x$**: you cross the repeating pattern. Walk along an opening in **$y$**: nothing changes in this ideal model. The **$z$ axis** is perpendicular to the sheet, which lies at $z=0$.

- The incoming wave approaches the grating from $z>0$ with a **negative $z$ component** of its propagation direction. At normal incidence it points exactly along $-z$; a tilted wave also has an $x$ component.
- We keep $+z$ pointing upward in the side view, so the transmitted side is below the sheet, at $z<0$. The diagram shows a few periods of a pattern we idealize as indefinitely repeating.
- A dot in a circle means an axis points out of the page; a cross means it points into the page.

All examples in this notebook keep $k_y=0$: the incident and outgoing directions lie in the $x$-$z$ plane. The ideal pattern is uniform along $y$.

### Small calculation 2: describe one opening

Use a dimensionless **field-amplitude transmission** $M$: $M=1$ in an opening and $M=0$ in a blocked strip. This specifies our ideal on/off profile, not an intensity measurement. Centre an opening at $x=0$ and inspect one period, from $-d/2$ to $d/2$. [2, §2](#ref-2)

`np.arange(8)` makes the array $[0,1,\ldots,7]$. We use it to place eight sample points inside the period. `np.abs(x) < a/2` asks whether each point is inside the opening; `.astype(float)` turns the resulting true/false answers into 1/0.

- **This calculation starts with** $d=2\,\mu\mathrm m$ and $a=1\,\mu\mathrm m$, matching the opening profile of the incident-wave explorer introduced in Notebook 2.2.
- Wavelength and angle do not change the opening profile.
- **Predict, then try:** set `a = 0.5`.
- How many of the eight sample positions will have $M=1$?
- This width change is available in the code; the explorer keeps $a=d/2$.

```python
d, a = 2.0, 1.0  # period and opening width, in micrometres
x = -d/2 + (np.arange(8) + 0.5) * d/8
M = (np.abs(x) < a/2).astype(float)
print(' x (um)   M')
for position, transmission in zip(x, M):
    print(f'{position:7.3f}   {transmission:.0f}')
```

<details><summary>Answer for the stated starting values and changes</summary>

Initially four samples lie in the opening. Reducing $a$ to $0.5\,\mu\mathrm m$ leaves two. The period is still $2\,\mu\mathrm m$: narrowing an opening changes its width, not the spacing between openings. The samples avoid exact edges, where an ideal step needs an arbitrary endpoint convention.

</details>

## Checkpoint

1. Is the slit width the same quantity as the grating period?
2. Does a real candidate direction guarantee nonzero intensity?
3. Do two allowed order directions tell us their relative amplitude and phase?

<details><summary>Answers</summary>

Width describes one opening; period describes repetition. A real direction can have zero amplitude through cancellation. Direction information alone does not specify the complex coefficients $C_m$, so it cannot determine their relative amplitude or phase.

</details>
