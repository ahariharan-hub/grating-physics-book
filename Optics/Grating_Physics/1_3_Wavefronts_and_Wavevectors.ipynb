## 1. Wavefronts and the wavevector

Imagine many parallel copies of the 1D wave, all oscillating together. At a frozen time, joining points with the same phase gives a **wavefront**. In two dimensions these wavefronts can be straight lines; in three dimensions they can be planes. Successive **crest** lines or planes differ in phase by $2\pi$ and are separated by $\lambda$, measured perpendicular to them. A crest and the neighbouring trough are only $\lambda/2$ apart. Constant-phase surfaces also exist at all the intermediate phases.

A **plane wave** is this special model with parallel planar wavefronts. Here we also choose uniform amplitude across each wavefront and a sinusoidal oscillation with one frequency, $1/T$, called **monochromatic**. An ideal uniform plane wave extends indefinitely. The plots show finite windows into it; a real beam can approximate it over a region where its wavefronts are nearly flat and its amplitude nearly uniform [\[1\]](#ref-1-3-paschotta), [\[2\]](#ref-1-3-nde).

Wavelength tells us the spacing, but it cannot say whether the wavefronts are vertical, horizontal, or tilted. We need both **spacing and orientation**. A **vector** is a quantity with a magnitude and a direction, represented by an arrow. Let $\hat{\mathbf n}$ be a **unit vector**: a direction arrow of mathematical length 1, perpendicular to the wavefronts and pointing in their direction of travel.
- The hat indicates unit length. Define

$$\boxed{\mathbf k=\frac{2\pi}{\lambda}\hat{\mathbf n}=k\hat{\mathbf n}}.$$

This is the **wavevector** (bold $\mathbf k$). Its magnitude $|\mathbf k|=k$ describes spatial phase variation; its direction describes **phase propagation** for our $-\omega t$ convention. A frozen set of wavefronts has two opposite normals; watching the phase move tells us which one to use.

The wavevector's length **does not represent wave amplitude**. Its direction **does not represent electric-field polarization**, which describes the electric field's oscillation direction and is omitted from this scalar model. Nor does a phase-direction arrow always give the direction of energy flow: in anisotropic media, whose properties depend on direction, the two can differ [\[3\]](#ref-1-3-anisotropic).
- Our simple uniform, isotropic model has a single phase speed independent of direction.

**Predict:** if the whole wave pattern is rotated without changing crest spacing, what should happen to its wavevector? **Try:** sketch parallel crest lines and an arrow perpendicular to them; turn the sketch through a quarter-turn. **Explain:** which part of the definition changes?

:::{dropdown} Answer and explanation

The arrow rotates with the wavefront normal; its magnitude remains $2\pi/\lambda$. Orientation changes, but spacing does not. Keep following the same direction of phase travel when choosing between the two opposite normals.

:::

## 2. Components and the dot product

### Walk first, write the formula second

Freeze time again. How much phase do you accumulate over a walk of length $s$?

- **Along $\mathbf k$:** you cross the wavefronts directly, gaining phase $ks$.
- **Along a wavefront:** you stay at the same phase, gaining zero.
- **At angle $\alpha$ to $\mathbf k$:** only the part of the walk along $\mathbf k$ crosses the fronts. Elementary trigonometry gives this projected distance as $s\cos\alpha$, so the phase change is $ks\cos\alpha$. Walking partly against $\mathbf k$ gives a negative phase change.

Now describe a point in the $x$–$z$ plane by its **position vector** $\mathbf r=(x,z)$, the arrow from the chosen origin to that point. A displacement from one point to another is $\Delta\mathbf r=(\Delta x,\Delta z)$, where each $\Delta$ means final minus initial coordinate. Write the wavevector as $\mathbf k=(k_x,k_z)$.

The **dot product** is the following operation: multiply corresponding components and add. It gives one number, not another vector:

$$\boxed{\Delta\phi=\mathbf k\cdot\Delta\mathbf r
=k_x\Delta x+k_z\Delta z=ks\cos\alpha.}$$

The component formula and the projected-distance formula describe the same geometry. Each $k_x$ or $k_z$ is a signed phase change **per unit distance along that coordinate axis**. For example, moving only along $x$ gives $\Delta\phi=k_x\Delta x$; if $k_x=0$, such a move does not change the phase.
- Multiplying rad/length by length gives radians, as required.

For a numerical check, take $(k_x,k_z)=(3,4)$ rad per length unit. A displacement $(1,0)$ changes phase by 3 rad. A displacement $(4,-3)$ changes it by $3(4)+4(-3)=0$: this walk stays on one wavefront even though both coordinates change.

With phase zero at the origin at $t=0$, position and time now enter as

$$\phi(x,z,t)=k_x x+k_z z-\omega t
=\mathbf k\cdot\mathbf r-\omega t,\qquad
u(x,z,t)=A\cos\phi(x,z,t).$$

:::{dropdown} Optional enrichment: gradient notation and local wavevectors

(local-phase-gradient)=

If you later encounter calculus, the same component information is written

$$\nabla\phi=\left(\frac{\partial\phi}{\partial x},
\frac{\partial\phi}{\partial z}\right)=(k_x,k_z)=\mathbf k.$$

The symbol $\nabla\phi$, called the **gradient of phase**, collects the phase change per unit distance along each axis. Each partial derivative holds the other coordinate and time fixed. For this plane wave the rates are constant, so the finite-displacement calculation above already gives them. This is the gradient of **phase**, not of the field $u$: the field flattens at a crest even though spatial phase continues to change.

For a general spatial field $U(\mathbf r)=A(\mathbf r)e^{i\Phi(\mathbf r)}$, define the **local phase wavevector** wherever the field is nonzero and its phase is smooth:

$$\boxed{\mathbf k_{\rm local}(\mathbf r)=\nabla\Phi(\mathbf r)},\qquad
\Delta\Phi\simeq\mathbf k_{\rm local}(\mathbf r)\cdot\Delta\mathbf r$$

for a sufficiently small displacement. The common time term $-\omega t$ has no spatial gradient, so it does not change this result. Compare two fields from [0.2's amplitude and phase table](0_2_Solutions_of_the_Helmholtz_Equation.ipynb#amplitude-phase-comparison):

$$\begin{aligned}
\text{Plane:}\quad &\Phi=\mathbf k\cdot\mathbf r+\phi_0
&&\Longrightarrow\quad \nabla\Phi=\mathbf k,\\
\text{Spherical:}\quad &\Phi=kr
&&\Longrightarrow\quad \nabla\Phi=k\hat{\mathbf r}\quad(r>0).
\end{aligned}$$

The plane wave has the same phase-gradient arrow everywhere; the spherical wave has arrows pointing radially outward. Where the gradient is nonzero, it is normal to the constant-phase surfaces.

For an evanescent field $U=A_0e^{-\alpha z}e^{iqx}$, $\nabla\Phi=q\hat{\mathbf x}$. This real phase gradient does not contain the decay component $i\alpha$ of the full complex wavevector $(q,0,i\alpha)$. Nor must a general local phase gradient have magnitude $k$: here $q>k$.

The connection from phase normals to rays and optical path length belongs to a later geometrical-optics discussion, where the necessary approximation can be introduced. A phase-gradient arrow alone is not a general statement about electromagnetic energy flow. [TU Delft, geometrical-optics introduction](#ref-1-3-geometrical) [\[5\]](#ref-1-3-geometrical).

No calculus is needed for the remaining activities.

:::

### Calculate it: walk along or perpendicular to the wavevector

Here arrays list components in **$(x,z)$ order**. `np.deg2rad` converts the direction angle from degrees to radians before we use sine and cosine. The unit direction is $(\sin\theta,\cos\theta)$; $(\cos\theta,-\sin\theta)$ is perpendicular to it. Multiplying by `step` sets the walk's length.

`np.dot(k_vector, along)` multiplies matching components and adds them: $k_x\Delta x+k_z\Delta z$. The two printed phase changes are for walks **along the wavevector** and **perpendicular to it, along a wavefront**. Expect $\pi$ rad and zero for these settings.

**Explorer match:** wavelength 2, angle $45^\circ$, and $t/T=0$ in the 2D explorer. Starting at the origin, the first walk reaches a trough; the perpendicular walk stays on the crest.

**Predict and rerun:** halve `step` to `0.5`. How will each phase change respond? These are static walks through the frozen field; rerunning does not alter the explorer.

```python
wavelength = 2.0
theta_deg = 45.0  # measured from +z toward +x
step = 1.0
theta = np.deg2rad(theta_deg)
k = 2 * np.pi / wavelength
direction = np.array([np.sin(theta), np.cos(theta)])
k_vector = k * direction
along = step * direction
perpendicular = step * np.array([np.cos(theta), -np.sin(theta)])
delta_along = np.dot(k_vector, along)
delta_perpendicular = np.dot(k_vector, perpendicular)
print("kx, kz (rad/length):", np.round(k_vector, 3))
print("Along (dx, dz); phase change (rad):", np.round(along, 3), np.round(delta_along, 3))
print("Perpendicular (dx, dz); phase change (rad):", np.round(perpendicular, 3), np.round(delta_perpendicular, 3))
```

### Final check: what does the wavevector encode?

For each change below, predict the wavevector's **magnitude**, **direction**, and **components**. Hold the other properties fixed and follow the same choice of phase propagation direction.

1. The wavelength is halved.
2. The wavefronts are rotated without changing their spacing.
3. The amplitude is doubled without changing wavelength or direction.

**Try:** check the first two in the 2D or 3D explorer. The controls keep $A=1$, so use $u=A\cos\phi$ for the third. **Explain:** which changes the spacing or orientation of the wavefronts, and which changes only the field range?

:::{dropdown} Answers and explanations

| Change | Wavevector | Real field |
| --- | --- | --- |
| $\lambda\to\lambda/2$ | $\mathbf k\to2\mathbf k$: magnitude and every component double; direction stays fixed | Spatial phase changes twice as quickly per unit distance; amplitude is unchanged |
| Rotate wavefronts | $\mathbf k$ rotates with their normal; magnitude stays fixed, while components generally change | Spatial phase is redistributed across coordinate directions |
| $A\to2A$ | No change to magnitude, direction, or components | The field extrema double; its wavelength, direction, and phase stay fixed. [Chapter 6](6_0_Amplitude_and_Intensity_of_a_Scalar_Wave.ipynb) will relate this amplitude change to intensity |

At fixed $c$, halving wavelength also halves the physical period $T$. The normalized-playback convention does not remove this physical change. The wavevector describes **spatial phase geometry**. It does not specify the amplitude or the choice of initial phase.

:::
