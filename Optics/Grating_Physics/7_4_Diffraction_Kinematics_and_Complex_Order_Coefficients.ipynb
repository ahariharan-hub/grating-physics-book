## 1. Two different questions about diffraction

Periodicity fixes the permitted transverse wavevectors:

$$\mathbf k_{\parallel,m}=\mathbf k_{\parallel,i}+m\mathbf G,\qquad
\mathbf G=(2\pi/d)\hat{\mathbf x}.$$

The medium's dispersion relation and outgoing boundary condition supply the longitudinal component. A propagating transmitted order has negative real $k_{z,m}$; an evanescent order instead decays away from the structure. The integer label alone does not guarantee propagation.

The scattering problem supplies

$$C_m=|C_m|e^{i\phi_m}.$$

An allowed order can have zero coefficient: permission is not excitation. Two gratings with identical pitch, illumination and exterior media can share every allowed direction but have different coefficients. Changing material or opening fraction does not by itself change the periodicity rule, although changing wavelength, incidence or the exterior medium can change the directions.

## 2. What information does C contain?

**Connection to the solver notation:** $F_m(z)$ is the depth-dependent complex harmonic amplitude; $C_m(z_{\rm ref})=F_m(z_{\rm ref})$ specifies it at an output reference plane. They describe the same harmonic. Independent exponential evolution applies in the homogeneous outgoing region, while the patterned layer couples the functions $F_m(z)$.

$$\underbrace{C_m}_{\text{Complex coefficient}}=
\underbrace{|C_m|}_{\text{Magnitude}}\underbrace{e^{i\phi_m}}_{\text{Phase factor}}.$$

Choose the origin at the reference plane. For one outgoing homogeneous-medium order and its coherent reconstruction,

$$\underbrace{U_m(\mathbf r)}_{\text{One order field}}=
\underbrace{C_m}_{\text{Coefficient}}\underbrace{e^{i\mathbf k_m\cdot\mathbf r}}_{\text{Spatial evolution}},\qquad
\underbrace{U(\mathbf r)}_{\text{Total field}}=
\underbrace{\sum_m C_me^{i\mathbf k_m\cdot\mathbf r}}_{\text{Coherent order reconstruction}}.$$

More explicitly, when the reference plane is $z=z_{\rm ref}$,

$$U_m(x,y,z)=C_m(z_{\rm ref})e^{i[k_{x,m}x+k_{y,m}y+k_{z,m}(z-z_{\rm ref})]}.$$

At the exit reference plane, summing these orders gives the exit field used in Chapter 9:

$$\boxed{U_{\mathrm{exit}}(x,y)=\sum_m C_m(z_{\rm ref})e^{i(k_{x,m}x+k_{y,m}y)}.}$$

**Notation across the series:** the period $d$ here is $p$ in 9.1. Use $U_{\mathrm{exit}}$ at the exit plane and $U_{\mathrm{ref}}$ when the reference plane is moved. For a two-dimensional periodic grating, the same interpretation applies to the two-index coefficients $C_{mp}$.


Coefficient phase is specified at that plane and lateral origin; total phase also includes spatial evolution. Moving the reference plane changes the coefficient phase even when the physical field is unchanged. For a lossless propagating order its magnitude stays constant during homogeneous propagation. At zero magnitude, phase is undefined.

An isolated-order intensity measurement constrains amplitude under a known normalization. It does not reveal absolute phase. Interference supplies relative-phase information. Order intensities alone cannot reconstruct a coherent field.

## 3. Where do complex coefficients come from?

### A. Prescribed ideal mask

For a centered binary-amplitude opening with fraction $D_{\rm open}$, unit incident amplitude and phase zero at the chosen object plane $z=0$,

$$C_m(0)=D_{\rm open}\operatorname{sinc}(mD_{\rm open}),\qquad
\operatorname{sinc}(q)=\frac{\sin(\pi q)}{\pi q},\quad \operatorname{sinc}(0)=1.$$

This is a Fourier coefficient of prescribed transmission, not a material-scattering result. Changing opening fraction changes magnitudes, zeros and signs while keeping the pitch fixed. A negative real coefficient carries phase $\pi$ relative to a positive one; it does not represent negative intensity. Changing the lateral origin also changes coefficient phases, so comparisons require a common convention.

### B. Physical scalar grating

The next figure reuses the absorber from 0.5–0.7. Its complex coefficients come from the scalar scattering calculation, including reflection and fields inside the layer.

**What to compare:**

- **Magnitudes and phases:** an ideal mask and a physical grating can give different coefficients despite having the same opening fraction.
- **Zeros:** a coefficient that vanishes for the ideal half-open mask need not vanish for the physical grating.
- **Near field:** coefficient plots include evanescent harmonics; the collection panel includes only propagating orders. Phase is hidden below the stated magnitude threshold.

:::{dropdown} Phase reference used in the comparison

Both models are compared at the exit plane $z_e=-h$, with incident phase zero at $z=0$. The ideal mask is prescribed at $z=0$ and propagated order by order to $z_e$, giving $C_m^{\mathrm{ideal}}(z_e)=t_m e^{i\beta_mh}$. The plots show both models' raw exit coefficients, with no fitted or removed phase offset. The ideal coefficients at the exit are therefore generally complex.

:::

## 4. Complex coefficients versus diffraction efficiency

**Basis reminder:** selecting propagating orders for output is different from restricting the scattering solve to those labels. [0.5’s numerical example](0_5_From_Helmholtz_BVP_to_Diffracted_Field.ipynb#propagating-only-scattering-basis) shows why evanescent harmonics can change the propagating coefficients.


[6.2](6_2_One_Order_Across_a_Detector.ipynb) connected amplitude squared with relative scalar intensity. Power through a plane additionally depends on propagation direction. For unit-compatible scalar flux in the same lossless homogeneous incident and output medium,

$$\eta_m=\frac{|k_{z,m}|}{|k_{z,i}|}\left|\frac{C_m}{A_i}\right|^2.$$

Here $A_i$ and $C_m$ are incident and outgoing plane-wave amplitudes at stated reference planes; the incident wave is nongrazing and the transmitted order propagates. The normal-component factor accounts for oblique flow through the grating plane. Equal amplitudes at 0 and 60 degrees carry different normal power: the latter contributes half as much for normal incidence. This individual-order flux example is not a claim that all illustrated amplitudes form one energy-conserving scattering solution.

Do not apply this expression to evanescent orders or general polarization-coupled systems. Evanescent amplitude can be nonzero without outgoing normal power in the lossless exterior model. Energy balance and truncation convergence remain separate checks.

## 5. Why coefficient phase matters

For two coherent propagating orders sharing a reference convention,

$$I=|C_m|^2+|C_p|^2+2|C_m||C_p|
\cos(\Delta\mathbf k\cdot\mathbf r+\Delta\phi),$$

where $\Delta\mathbf k=\mathbf k_p-\mathbf k_m$ and $\Delta\phi=\phi_p-\phi_m$. Wavevector difference sets spatial frequency; coefficient-phase difference sets fringe position; magnitudes set average intensity and modulation. A common phase multiplying both coefficients changes no intensity. A relative phase does.

Coherence is assumed in this deterministic calculation. A fixed phase shift does not reduce mutual-coherence magnitude. Averaging fluctuating relative phases is a different operation, as 7.1 established. Knowing only squared magnitudes loses the phase information needed to reconstruct interference.

Return to the experiment with equal magnitudes and advance the relative phase by 180 degrees. Bright and dark positions exchange, but the individual intensities and nominal spacing remain unchanged. Now restore the phase and reduce one amplitude: minima become less deep. These two changes can both alter a detector picture, yet they have different physical causes. Keep that distinction when interpreting a pupil that changes the phases and amplitudes of different orders.

## 6. From diffraction coefficients to imaging

[5.2](5_2_Direction_Cosines_and_Spatial_Frequency_Coordinates.ipynb) locates components in spatial-frequency coordinates; [5.4](5_4_From_Diffraction_Coordinates_to_Pupils.ipynb) selects a collected subset. If an objective accepts $m=0,\pm1$, is that enough to predict an image? No: we still need the complex coefficients and the optical-system transformation.

A **schematic Series 2 preview**, retaining both transverse coordinates, is

$$U_{\mathrm{image}}(x,y)=\sum_{m\in\mathcal P}C_mH_m
 e^{i(k'_{x,m}x+k'_{y,m}y)}.$$

The set $\mathcal P$ records collection; $C_m$ describes grating output; $H_m$ describes system transfer. Image-space wavevectors require coordinate and magnification conventions. This component-wise expression assumes an appropriate imaging model; linearity alone does not establish a scalar shift-invariant transfer function. Normalized pupil coordinates remain angular normalizations unless a physical pupil mapping is specified.

## Final checkpoint

:::{dropdown} 1. Can an allowed order be absent?

Yes. Its coefficient can be zero even though periodicity permits its wavevector.

:::

:::{dropdown} 2. What distinguishes coefficient magnitude and phase?

Magnitude measures field amplitude; phase locates its phasor relative to the chosen reference. Zero magnitude has undefined phase.

:::

:::{dropdown} 3. Is coefficient phase the phase everywhere?

No. Spatial phase includes the wavevector term relative to the reference plane and lateral origin.

:::

:::{dropdown} 4. Do equal amplitudes imply equal efficiencies?

No. Normal flux includes the longitudinal-wavevector factor and incident normalization.

:::

:::{dropdown} 5. Does an integer order always propagate?

No. Dispersion can make its normal component imaginary; the outgoing branch then decays away from the interface.

:::

:::{dropdown} 6. Does propagation determine coefficients from a material grating?

No. Propagation evolves known fields; scattering determines them from illumination, material and boundaries.

:::

:::{dropdown} 7. Are C_m and H_m the same quantity?

No. The former describes grating output at a reference plane; the latter describes subsequent optical-system transfer under specified assumptions.

:::
