# 5.2 Direction Cosines and Spatial-Frequency Coordinates

Wavevector components, spatial frequencies and direction cosines express the same outgoing geometry on different scales. Comparing them shows which features come from the physical directions and which come from the chosen representation.

 For these direction constructions the incident and output media have the same index; refraction between unequal exterior media is not simulated.

The discussion builds on [5.1: 3D orders and conical diffraction](5_1_Three_Dimensional_Orders_and_Conical_Diffraction.ipynb). The grating is at $z=0$, periodic along $x$ with period $d$, and uniform along $y$. Transmission is into $z<0$ in the same homogeneous, isotropic, lossless medium. Lengths and in-medium wavelength $\lambda$ are in micrometres, with $k=2\pi/\lambda$. Recall $k_{x,m}=k_{x,i}+m2\pi/d$, $k_{y,m}=k_{y,i}$, and $k_{z,m}=-\sqrt{k^2-k_{x,m}^2-k_{y,i}^2}$ for real outgoing directions.

## 1. Harvey: make the directions easy to read

We already know the waves. Give each real wavevector normalized coordinates:
$$\hat{\mathbf k}_m=\frac{\mathbf k_m}{k}
=(\alpha_m,\beta_m,\gamma_m),\qquad
\alpha_m^2+\beta_m^2+\gamma_m^2=1.$$

These are **direction cosines**: components of a unit vector, or cosines of its angles with the coordinate axes. Here $\gamma_m<0$ for transmitted propagating waves. The incident fractions $a_i,b_i$ are now written $\alpha_i,\beta_i$.

With **Coordinates → Direction cosines**, the plot axes use unsubscripted $\alpha=k_x/k$, $\beta=k_y/k$ and $\gamma=k_z/k$ to label the coordinate system. An order point has coordinates $(\alpha_m,\beta_m,\gamma_m)$; the incident point uses subscript $i$.

Write the positive medium wavenumber explicitly as $k=nk_0$, where $k_0=2\pi/\lambda_0$ and $\lambda=\lambda_0/n$. The same component has three descriptions:
$$\mathbf f_m=\frac{\mathbf k_m}{2\pi},\qquad
(\alpha_m,\beta_m,\gamma_m)=\frac{\mathbf k_m}{nk_0}=\lambda\mathbf f_m.$$
Only real outgoing vectors define unit propagation directions; an evanescent candidate has a complex longitudinal component.


Project the outgoing hemisphere onto the $\alpha\beta$-plane:
$$\boxed{\alpha_m=\alpha_i+m\frac{\lambda}{d},\qquad
\beta_m=\beta_i,\qquad
\gamma_m=-\sqrt{1-\alpha_m^2-\beta_i^2}.}$$

The **Harvey direction-cosine diagram** has a simple structure:

Candidates form a straight row at $\beta=\beta_i$. Neighbouring orders have exactly equal horizontal spacing $\lambda/d$. The interior of the unit disk permits transmission; the boundary is grazing. Outside the disk, tangential coordinates remain meaningful, but there is no real point on the hemisphere.

Harvey's papers [\[1\]](#ref-5-2-harvey-vernold), [\[2\]](#ref-5-2-harvey-pfisterer) develop this representation. Our diagrams always use actual propagation vectors; the references explain the different signs in reflection observation diagrams.

**The same transverse rule with indices explicit:**
$$n_t\alpha_m=n_i\alpha_i+m\frac{\lambda_0}{d},\qquad n_t\beta_m=n_i\beta_i.$$
For the equal-medium explorer, $n_i=n_t=n$, giving $\alpha_m=\alpha_i+m\lambda_0/(nd)$ and $\beta_m=\beta_i$.


### 1.1. Spatial-frequency coordinates: Goodman's convention

[2.4](2_4_Order_Coordinates_and_Spacing.ipynb#one-order-two-triangles) expressed one wavevector in **cycles per unit length**. Extend that definition to all three components:
$$\boxed{\mathbf f_i=\frac{\mathbf k_i}{2\pi},\qquad
\mathbf f_m=\frac{\mathbf k_m}{2\pi}=(f_{x,m},f_{y,m},f_{z,m}).}$$
Here $f$ means spatial frequency, not oscillations per second. With lengths in micrometres, its units are cycles per micrometre.

This follows **Goodman's spatial-frequency convention** in *Introduction to Fourier Optics*: $f_x=\alpha/\lambda$ and $f_y=\beta/\lambda$ [\[3, §3.10.1, Eq. (3-62)\]](#ref-5-2-goodman-2005). We add the order subscript $m$ and extend the notation to the normal component $f_{z,m}$. Goodman develops propagation toward $+z$; our outgoing side remains $-z$, so real outgoing orders have $f_{z,m}\le0$.

At the grating surface, one order has the complex scalar field
$$U_m(x,y,0)=C_m\exp\!\left[i2\pi(f_{x,m}x+f_{y,m}y)\right].$$
Moving along $x$ samples phase cycles **across** the grooves; moving along $y$ samples phase cycles **along** them. For example, positive $f_{y,m}=0.2$ cycles/µm means a $1\,\mu\mathrm m$ step along $+y$ advances the phase by $0.2$ cycles. TU Delft uses the symbols $\xi,\eta$ for these transverse frequencies [\[4, Eqs. (371)–(376)\]](#ref-5-2-delft-diffraction).

**What does the grating do to those frequencies?** Divide Notebook [5.1](5_1_Three_Dimensional_Orders_and_Conical_Diffraction.ipynb)’s component equations by $2\pi$:
$$\boxed{f_{x,m}=f_{x,i}+\frac{m}{d},\qquad f_{y,m}=f_{y,i}.}$$
The orders form a row with horizontal step $1/d$ and a common along-groove frequency. Conical incidence shifts the row away from $f_y=0$; there is still only one order integer.

**Where can the real directions lie?** For propagating or grazing orders,
$$f_{x,m}^2+f_{y,m}^2+f_{z,m}^2=\frac{1}{\lambda^2},\qquad f_{z,m}\le0.$$
The outgoing hemisphere and its transverse disk have radius $1/\lambda$. Inside the disk an order propagates; on the boundary it grazes the surface. Outside, $f_{z,m}=k_{z,m}/(2\pi)$ is imaginary and describes normal decay. Such a candidate has a real transverse point but no real point on the hemisphere. The complex vector still obeys the algebraic sum of squares; its ordinary length is not $1/\lambda$.

The link to Harvey coordinates is
$$\boxed{(\alpha_m,\beta_m,\gamma_m)=\lambda(f_{x,m},f_{y,m},f_{z,m}).}$$
For real orders these are direction cosines; for evanescent candidates only the transverse coordinates remain real. The three scales describe the same order candidates:

| Coordinates | Units | Hemisphere/disk radius | Step between neighbouring orders along $x$ |
| --- | --- | --- | --- |
| Wavevector $\mathbf k_m$ | rad/µm | $2\pi/\lambda$ | $2\pi/d$ |
| Spatial frequency $\mathbf f_m$ (Goodman convention) | cycles/µm | $1/\lambda$ | $1/d$ |
| Normalized components $\mathbf k_m/k$ | dimensionless | $1$ | $\lambda/d$ |

Use **Coordinates** below to switch scales. The same controls, order labels, colours and selection apply to both panels. The axes rescale to fit; read their numbers and the radius/spacing readout when comparing units. These are coordinate representations, with no optical mapping or order strengths assigned.

**Then choose Spatial frequency (Goodman).** With **Conical: β = 0.4**, $\lambda=2\,\mu\mathrm m$ and $d=4\,\mu\mathrm m$: the disk radius is $0.5$ cycles/µm, the row sits at $f_y=0.2$ cycles/µm, and adjacent orders are separated by $0.25$ cycles/µm. Select $m=+1$ and compare the values in all three scales.

Keep Spatial frequency selected and move through the grazing and beyond presets. The row approaches the top of the disk, shortening the interval available to real orders. At a fixed row height its half-width is $\sqrt{1/\lambda^2-f_{y,i}^2}$. The selected point stays visible on the transverse plane after leaving the disk.

Change only $\lambda$: the radius $1/\lambda$ changes, but the step $1/d$ stays fixed. The incidence controls hold $\alpha_i,\beta_i$ fixed, so the row's frequency offset changes too: $f_{x,i}=\alpha_i/\lambda$, $f_{y,i}=\beta_i/\lambda$. Change only $d$ to change the frequency step. Return to **Direction cosines** for the comparison below.

The disk's cross-section at fixed $\beta_i$ has half-width $\rho=\sqrt{1-\beta_i^2}$. Increasing $|\beta_i|$ reduces the available interval in $\alpha$, explaining the cutoff from Notebook [5.1](5_1_Three_Dimensional_Orders_and_Conical_Diffraction.ipynb).

**A disk coordinate is not a location on a flat observation screen.** On a sphere of radius $R$, the projected coordinates are $(R\alpha,R\beta)$. On a plane $z=-L$, the intersection is instead
$$x_{\rm plane}=\frac{L\alpha}{-\gamma},\qquad
y_{\rm plane}=\frac{L\beta}{-\gamma}.$$
These are finite only for $\gamma<0$. Equal spacing in direction-cosine coordinates need not give equal angular or screen spacing. Near grazing, plane intersections can move far away.

### Small calculation 1: one order on three coordinate scales

Use the conical preset: $\lambda=2\,\mu\mathrm m$, $d=4\,\mu\mathrm m$, $(a_i,b_i)=(0.3,0.4)$, and $m=+1$. Build $\mathbf k_m$, divide by $2\pi$ to obtain $\mathbf f_m$, and divide by $k$ to obtain the normalized components.

The square root uses `+0j` so it can also return an imaginary normal component. The minus sign selects transmission or decay into $z<0$, as in section 3. `np.real_if_close` prints real arrays without an unnecessary `+0j`.

**Predict, then rerun:** change `b_i` to `0.6`, then `0.8`. Does changing coordinate scale change whether the order propagates? Restore `b_i = 0.4` and change only `wavelength` to `1.0`: which of the three order spacings changes? These settings are independent of the explorer.


```python
wavelength, d = 2.0, 4.0  # micrometres; wavelength in the medium
a_i, b_i, m = 0.3, 0.4, 1
if wavelength <= 0 or d <= 0 or a_i**2+b_i**2 >= 1 or m != int(m):
    raise ValueError("Use positive lengths, physical incidence and an integer m.")
m = int(m)
k = 2*np.pi/wavelength
k_i = k*np.array([a_i, b_i, -np.sqrt(1-a_i**2-b_i**2)])
k_m = k_i.astype(complex)
k_m[0] += m*2*np.pi/d
r = 1-(k_m[0].real/k)**2-(k_m[1].real/k)**2
if abs(r) <= 1e-12:
    r = 0.0
k_m[2] = -k*np.sqrt(r+0j)
f_m = k_m/(2*np.pi)
normalized = k_m/k
state = "grazing" if r == 0 else ("propagating" if r > 0 else "evanescent")
print(f"m = {m:+d}: {state}")
print("k_m (rad/um):", np.round(np.real_if_close(k_m), 6))
print("f_m (cycles/um):", np.round(np.real_if_close(f_m), 6))
print("k_m/k = wavelength*f_m:", np.round(np.real_if_close(normalized), 6))
print("Order steps: 2*pi/d, 1/d, wavelength/d =",
      np.round([2*np.pi/d, 1/d, wavelength/d], 6))
if r < 0:
    print("The imaginary normal component describes decay; no real hemisphere point.")
```

:::{dropdown} Answer

For the starting settings, $m=+1$ propagates:
$$\mathbf k_{+1}\approx(2.513274,\ 1.256637,\ -1.404963)\;\mathrm{rad}/\mu\mathrm m,$$
$$\mathbf f_{+1}\approx(0.400000,\ 0.200000,\ -0.223607)\;\text{cycles}/\mu\mathrm m,$$
$$(\alpha_{+1},\beta_{+1},\gamma_{+1})\approx(0.800000,\ 0.400000,\ -0.447214).$$
All three real vectors have the same direction. The radii are $\pi$ rad/µm, ${0.5}$ cycles/µm and ${1}$; the horizontal order steps are $\pi/2$ rad/µm, ${0.25}$ cycles/µm and ${0.5}$.

At $b_i=0.6$ the order is grazing; at $b_i=0.8$ it is evanescent in every coordinate scale. In the latter case $f_{z,+1}\approx-0.264575i\,\mu\mathrm m^{-1}$. This imaginary component controls decay; it is not a real frequency of phase oscillation along $z$.

With $b_i$ restored to ${0.4}$, changing only $\lambda$ from ${2}$ to $1\,\mu\mathrm m$ keeps $2\pi/d$ and $1/d$ unchanged, while the direction-cosine step halves from ${0.5}$ to ${0.25}$. The frequency disk radius doubles to ${1}$ cycle/µm. The incidence fractions stay fixed, so the incident spatial frequencies also change.

:::


### Small calculation 2: compare a sphere projection with a plane

Calculate both locations for the same orders. The plane is an observation surface; the disk is a coordinate representation.

**Predict and rerun:** set $b_i=0.6$. What happens to the plane intersection of $m=+1$?

```python
wavelength, d = 2.0, 4.0
a_i, b_i = 0.3, 0.4
R, L = 4.0, 4.0  # micrometres

print(" m   sphere projection (x,y)       plane intersection (x,y)")
for m in range(-2, 2):
    a = a_i+m*wavelength/d
    r = 1-a*a-b_i*b_i
    if r < -1e-12:
        print(f"{m:+2d}   no real outgoing ray")
    elif abs(r) <= 1e-12:
        print(f"{m:+2d}   ({R*a:+.3f}, {R*b_i:+.3f})       grazing: no finite intersection")
    else:
        minus_gamma = np.sqrt(r)
        print(f"{m:+2d}   ({R*a:+.3f}, {R*b_i:+.3f})       "
              f"({L*a/minus_gamma:+.3f}, {L*b_i/minus_gamma:+.3f})")
```

:::{dropdown} Answer

For $m=+1$ at $b_i=0.4$, $(\alpha,\beta,\gamma)=(0.8,0.4,-0.447214)$. Its sphere projection is $(3.2,1.6)\,\mu\mathrm m$; its intersection with $z=-4\,\mu\mathrm m$ is approximately $(7.155,3.578)\,\mu\mathrm m$.

At $b_i=0.6$, that order is grazing. Its sphere projection remains finite, but it never intersects the plane $z=-4$.

:::
