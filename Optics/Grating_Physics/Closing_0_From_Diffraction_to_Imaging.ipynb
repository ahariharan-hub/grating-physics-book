## 1. From illumination to diffracted fields

**What determines the optical field leaving a grating?**

In [Chapter 0](0_0_Foundations_of_Scalar_Optics.ipynb), a complex scalar function described monochromatic light. Helmholtz theory constrained possible fields; illumination, material properties and boundary conditions selected the scattering problem. Periodicity then allowed an order representation of its solution.

Keep two tasks separate:

- **Known boundary field → propagation:** extend a specified field through a homogeneous region, with the appropriate outgoing condition.
- **Illumination + material structure → scattering calculation:** determine the reflected and transmitted fields from a boundary-value problem.

An ideal binary mask prescribes transmission as zero or one. It is useful, but does not solve the scattering of a physical absorbing or dielectric grating. [Appendix 1](Appendix_1_Fourier_Modal_Method_for_Scalar_Grating_Scattering.ipynb) explains the scalar material solver; [Appendix 2](Appendix_2_Numerical_Diffraction_of_Finite_Grating_Illumination.ipynb) propagates a prescribed finite-mask field and discusses sampling, padding and order-isolation limits. Neither approach removes the need to check numerical convergence for the quantity being interpreted.

## 2. Diffraction orders and spatial frequencies

**What does an order represent mathematically?**

Periodicity constrains the transverse phase change:

$$k_{x,m}=k_{x,i}+m\frac{2\pi}{d}.$$

Divide by $2\pi$ to recover [5.2's Goodman spatial-frequency convention](5_2_Direction_Cosines_and_Spatial_Frequency_Coordinates.ipynb):

$$\boxed{f_{x,m}=f_{x,i}+\frac{m}{d},\qquad f_{y,m}=f_{y,i}.}$$

Frequency here means cycles per unit length, not cycles per second. At the reference plane, an ideal order is a complex exponential:

$$U_m(x,y,0)=C_m\exp[i2\pi(f_{x,m}x+f_{y,m}y)].$$

An order is therefore more than a ray: it has spatial phase and a complex coefficient. [7.4](7_4_Diffraction_Kinematics_and_Complex_Order_Coefficients.ipynb) distinguishes wavevector kinematics from scattering dynamics: periodicity constrains directions; material, geometry and illumination determine coefficient magnitudes and phases at a specified reference plane. Wavevector geometry has become a spatial-frequency representation. Wavevector coordinates, spatial frequencies and Harvey direction cosines describe the same direction on different scales. Projection onto a sphere and intersection with a flat observation plane are different geometrical mappings, not additional optical fields.

Outside the propagation disk, an integer candidate is evanescent rather than a real outgoing direction. Its normal decay belongs to near-field physics; a conventional far-field objective does not collect it as a propagating order.

## 3. From diffraction orders to optical-system collection

**Which orders can an optical system accept?**

As in [5.4](5_4_From_Diffraction_Coordinates_to_Pupils.ipynb), let $n$ be the local refractive index and $\alpha_m=k_{x,m}/k$, $\beta_m=k_{y,m}/k$. A propagating order is collected when

$$\boxed{(n\alpha_m)^2+(n\beta_m)^2\leq\mathrm{NA}_{\mathrm{obj}}^2.}$$

Propagation and collection have different boundaries. The propagation disk has radius one in direction cosines; the objective generally accepts a smaller disk. An order can propagate yet miss the objective.

Direction cosines $(\alpha,\beta)$ become NA coordinates $(n\alpha,n\beta)$. Divide by objective NA to obtain normalized angular coordinates:

$$\boxed{(\sigma_{x,m},\sigma_{y,m})=\frac{n}{\mathrm{NA}_{\mathrm{obj}}}(\alpha_m,\beta_m).}$$

The collection edge now has radius one. These dimensionless angular coordinates do not automatically specify millimetres in a pupil. For the specified ideal sine mapping with aligned axes,

$$\left(\frac{x_p}{R},\frac{y_p}{R}\right)=(\sigma_x,\sigma_y).$$

A physical radius $R$ and the optical mapping must still be supplied. A bare detector plane follows a different projection.

### Revisit the generic collection example

Use vacuum wavelength 1 um, grating pitch 4 um and collection NA 0.35 in air. At normal incidence, NA coordinates are $(0.25m,0)$ and $\boldsymbol\sigma_m=(0.714286m,0)$. Orders $0,\pm1$ are collected; $\pm2$ propagate outside the aperture. This is frequency selection, not yet the complete complex image transformation. The lithography application remains optional in [5.4](5_4_From_Diffraction_Coordinates_to_Pupils.ipynb).


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Generic grating and objective, using 5.4's collection rule; no imaging calculation.
orders = np.arange(-2, 3)
lambda0_um, pitch_um = 1., 4.
na_obj = .35
fx = orders / pitch_um
na_x = lambda0_um * fx  # air, normal incidence
sigma_x = na_x / na_obj
collected = np.abs(na_x) <= na_obj
assert orders[collected].tolist() == [-1, 0, 1]
assert np.all(np.abs(na_x) < 1)  # all five illustrated orders propagate

fig, axes = plt.subplots(1, 4, figsize=(13, 3.6), layout='constrained')
blue, orange = '#2866a3', '#b85c17'
a = axes[0]
a.plot([-1, 1], [0, 0], color='0.3', lw=3)
for x in np.linspace(-1, 1, 11):
    a.plot([x, x], [-.04, .04], color='0.5')
a.annotate('Incident', xy=(0, .05), xytext=(0, .8), ha='center',
           arrowprops=dict(arrowstyle='->', color='#17806b'))
for m, alpha, accept in zip(orders, na_x, collected):
    end = np.array([alpha, -np.sqrt(1-alpha**2)])
    a.annotate('', xy=end, xytext=(0, 0),
               arrowprops=dict(arrowstyle='->', color=blue if accept else orange))
    a.text(end[0], end[1]-.12, f'{m:+d}', ha='center', fontsize=8)
a.set(title='1. Illumination and grating', xlim=(-1.1, 1.1), ylim=(-1.25, 1.05))
a.text(0, -1.2, 'Direction arrows, not beam boundaries', ha='center', fontsize=8)
a.set_aspect('equal'); a.axis('off')

a = axes[1]
a.axhline(0, color='0.7', lw=.8)
for m, x, accept in zip(orders, fx, collected):
    a.plot(x, 0, 'o', color=blue if accept else orange)
    a.text(x, .10, f'{m:+d}', ha='center', fontsize=8)
a.set(title='2. Spatial frequencies (5.2)', xlabel='f_x (cycles / micrometre)',
      ylabel='f_y (cycles / micrometre)', xlim=(-.6,.6), ylim=(-.6,.6))
a.text(0, -.45, 'Positions label components, not a screen', ha='center', fontsize=8)

a = axes[2]
a.add_patch(Circle((0,0), 1, fill=False, color='0.35'))
for m, x, accept in zip(orders, sigma_x, collected):
    a.plot(x, 0, 'o', color=blue if accept else orange)
    a.text(x, .13, f'{m:+d}', ha='center', fontsize=8)
a.set(title='3. Collection coordinates (5.4)', xlabel=r'$\sigma_x$', ylabel=r'$\sigma_y$',
      xlim=(-1.7,1.7), ylim=(-1.25,1.25), aspect='equal')
a.text(0,-1.17,'Angular diagram; physical pupil needs mapping',ha='center',fontsize=7)

a = axes[3]; a.axis('off')
a.text(.5,.9,'4. Coherent observation',ha='center',weight='bold')
a.text(.5,.69,'Known fields: add U, then take |U|^2',ha='center',fontsize=9)
a.annotate('',xy=(.5,.43),xytext=(.5,.59),arrowprops=dict(arrowstyle='->'))
a.text(.5,.31,'5. Imaging system -> image field ?',ha='center',weight='bold',fontsize=9)
a.text(.5,.1,'Complex transfer and reconstruction\nremain to be specified in Series 2.',ha='center',fontsize=9)
fig.suptitle('Collected: blue 0, +/-1. Excluded: orange +/-2 (still propagating). No image is calculated.',fontsize=11)
plt.show()
```

### The two-dimensional transverse picture

The normal-incidence example lies along $f_y=0$, but collection depends on **both** transverse coordinates. Retain the same grating and objective, and tilt illumination along the grooves so that $n\beta_i=0.27$ while $n\alpha_i=0$. The order row shifts to nonzero $f_y$ without changing its horizontal spacing. Predict whether the same orders still fit inside the circular collection boundary.

In the diagrams below, the circles represent the objective acceptance in two coordinate scales, not physical detector or pupil boundaries. The tilted case collects only order zero: the first orders still propagate, but their combined transverse coordinates exceed the collection radius.

- A grating periodic only along $x$ still has one order index $m$; the field varies across the two-dimensional $(x,y)$ plane and propagates in three dimensions.
- [5.3](5_3_Two_Dimensional_Gratings_and_Diffraction_Order_Lattices.ipynb) established that a structure periodic along **both** surface directions has two integer labels and a two-dimensional order lattice. The row drawn here retains the one-period example. 

```python
# Same generic parameters; compare normal and along-groove tilted illumination.
fig_2d, ax_2d = plt.subplots(2, 2, figsize=(9, 6.8), layout='constrained')
frequency_radius = na_obj / lambda0_um
for row, na_y_inc in enumerate((0., 0.27)):
    fy_row = np.full_like(fx, na_y_inc / lambda0_um)
    sigma_y_row = np.full_like(sigma_x, na_y_inc / na_obj)
    accepted_row = np.hypot(sigma_x, sigma_y_row) <= 1
    propagating_row = np.hypot(na_x, na_y_inc) < 1  # outgoing medium is air
    assert propagating_row.all()
    expected = [-1, 0, 1] if row == 0 else [0]
    assert orders[accepted_row].tolist() == expected
    for col, (xx, yy, radius) in enumerate((
            (fx, fy_row, frequency_radius), (sigma_x, sigma_y_row, 1.))):
        ax = ax_2d[row, col]
        ax.add_patch(Circle((0, 0), radius, fill=False, color='0.35'))
        ax.axhline(0, color='0.85', lw=.7)
        ax.axvline(0, color='0.85', lw=.7)
        ax.plot(xx, yy, ':', color='0.6', lw=1)
        for m, x, y, accepted in zip(orders, xx, yy, accepted_row):
            ax.plot(x, y, 'o', color=blue if accepted else orange)
            ax.annotate(f'{m:+d}', (x, y), xytext=(0, 8),
                        textcoords='offset points', ha='center', fontsize=9)
        bound = .6 if col == 0 else 1.65
        ax.set(xlim=(-bound, bound), ylim=(-bound, bound), aspect='equal')
        if col == 0:
            ax.set(xlabel='f_x (cycles / micrometre)',
                   ylabel='f_y (cycles / micrometre)')
        else:
            ax.set(xlabel=r'$\sigma_x$', ylabel=r'$\sigma_y$')
        condition = 'Normal incidence' if row == 0 else 'Along-groove tilt'
        scale = 'Spatial frequencies' if col == 0 else 'Collection coordinates'
        ax.set_title(f'{condition}: {scale}', fontsize=10)
fig_2d.suptitle('Two transverse coordinates, one order index. Blue: collected; orange: excluded.\n'
               'Circles show objective acceptance, not the larger propagation boundary.', fontsize=11)
plt.show()
```

## 4. From collected orders to coherent interference

**What happens when selected fields overlap?**

Chapter 7 added fields before calculating relative intensity:

$$U(x,y,z)=\sum_m U_m(x,y,z),\qquad \boxed{I(x,y,z)=\left|\sum_m U_m(x,y,z)\right|^2.}$$

For two stable coherent components,

$$I=I_m+I_p+2\operatorname{Re}(U_m^*U_p).$$

Mutual coherence describes averaged field correlation; spatial overlap describes where both fields are appreciable. Fringe geometry follows spatial relative phase, while detector exposure determines the averaging. Nonzero orders from one deterministic coherent input have stable relative phase in the ideal linear stationary scattering model. Their intensities alone cannot reconstruct the cross term.

For two distinct, same-wavelength, ideal unredirected orders from one periodic grating, on a common plane parallel to it,

$$\Lambda_x=\frac{2\pi}{|k_{x,p}-k_{x,m}|}=\frac{d}{|p-m|}.$$

This is the transverse plane-wave spacing, not a general formula for redirected or curved wavefronts. [7.3](7_3_Finite_Diffraction_Beams_and_Experimental_Recombination.ipynb) assigns finite Gaussian envelopes: their centers can separate, and curvature changes the local relative-phase gradient. Their shapes are prescribed, not generated by the grating calculation. Mirror-assisted recombination is an optional experimental extension, not an unfinished requirement of this series.

These are the physical and geometrical ingredients of coherent imaging. Their mathematical relationship is the next subject. Write an optical system abstractly as

$$\boxed{U_{\mathrm{image}}=\mathcal H\{U_{\mathrm{object}}\}.}$$

The brief coherent LSI model in 5.4 states when spatial frequencies receive independent complex factors. How can that model be derived, and how can its validity region be established? How does a pupil alter complex amplitude and phase, and how are the transformed components reconstructed?

For an appropriately modeled coherent system, a **schematic preview** is

$$\boxed{U_{\mathrm{image}}(x,y)=\sum_m C_m H_m e^{i(k'_{x,m}x+k'_{y,m}y)}.}$$

Here $C_m$ is the complex input order coefficient, $H_m$ a complex system transfer factor, and $k'_{x,m}$ the image-space transverse wavevector after coordinate mapping. This expression requires specified magnification, pupil mapping, reference phases and system properties. Linearity alone does not guarantee a scalar shift-invariant transfer function or prevent mixing between frequency components. Series 2 will establish when this representation applies.

## 6. What have our models actually predicted?

| Model | Established result | Principal limitation |
| --- | --- | --- |
| Ideal order geometry | Allowed directions | No efficiencies |
| Harvey/spatial-frequency coordinates | Component locations | No field transfer |
| NA and normalized angular collection | Accepted directions | No complex transmission |
| Ideal two-wave interference | Fringe geometry | Infinite plane-wave extent |
| Prescribed Gaussian beams | Finite overlap and paraxial spreading | No grating-scattering prediction |
| Thin-mask FFT/ASM | Propagation of a prescribed boundary field | No rigorous material scattering |
| Scalar Helmholtz BVP | Scattering for a specified scalar material model | No general vector electromagnetic coupling |

These models answer complementary questions. Choose one by identifying its input and the quantity you need to predict. Increasing numerical resolution cannot supply physics omitted from a model; adding material detail does not automatically specify an imaging system.

## Final learning checkpoint

Predict each answer before expanding it.

:::{dropdown} 1. Why does a periodic grating have discrete order labels?

The outgoing translation factor must match the incident one. Their transverse phase steps differ by an integer number of complete cycles per period.

:::

:::{dropdown} 2. How do wavevector components relate to spatial frequencies?

$f_x=k_x/(2\pi)$ and $f_y=k_y/(2\pi)$: radians per length become cycles per length.

:::

:::{dropdown} 3. Can two propagating orders have different collection outcomes?

Yes. One can lie inside the objective collection disk and the other outside it; propagation permits a wider set of directions.

:::

:::{dropdown} 4. What does a normalized pupil coordinate represent?

Under a stated pupil mapping it is position divided by pupil radius. Normalized angular coordinates first describe NA-normalized directions, with collection edge one.

:::

:::{dropdown} 5. Why does sigma position not automatically give physical pupil position?

An optical mapping and a physical scale are also needed. Equality with $(x_p/R,y_p/R)$ assumes the specified sine mapping and axis alignment.

:::

:::{dropdown} 6. Why add coherent fields before calculating intensity?

Squaring their sum retains relative-phase cross terms. Adding individual intensities discards them.

:::

:::{dropdown} 7. What distinguishes mutual coherence from spatial overlap?

Coherence is averaged field correlation; overlap is shared spatial support or appreciable magnitude. One does not establish the other.

:::

:::{dropdown} 8. What does the prescribed Gaussian-order model omit?

It does not calculate the grating-generated output profiles or efficiencies. It assigns envelopes along order directions.

:::

:::{dropdown} 9. What else is needed to predict an image from collected orders?

Complex order coefficients, system amplitude and phase transfer, magnification, coordinate and reference-plane conventions, and a valid reconstruction model.

:::

:::{dropdown} 10. Why is linear-systems theory the natural next subject?

It connects field superposition and spatial-frequency representations to a specified optical transformation, clarifying when component-wise transfer and reconstruction are valid.

:::
