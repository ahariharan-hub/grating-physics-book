**What does a wavevector tell us about an optical wave, and what does it leave unspecified?**

## Motivation and prerequisites

A grating sends light in different directions. Before describing those directions, we need a language for phase, wavelength and wavefront orientation. This chapter builds that language from a travelling oscillation to a three-dimensional field.

Review [0.1: scalar fields](0_1_Scalar_Fields_and_Helmholtz.ipynb) for the meaning of a complex field, or begin with basic trigonometry and follow the visual examples. Vector components are introduced as they become useful.

## The central idea

A wavefront joins points of equal phase. The wavevector points normal to the wavefronts, toward increasing spatial phase; its magnitude gives the number of radians of phase change per unit distance. With the course convention $e^{-i\omega t}$, an ideal travelling plane wave has spatial field

$$U(\mathbf r)=C e^{i\mathbf k\cdot\mathbf r}.$$

Rotating the wavevector rotates the wavefronts. It does not give the wave a finite width. An ideal plane wave extends indefinitely across its transverse directions, whereas a real beam has an envelope. A direction arrow therefore represents direction, not a beam boundary or the extent of a field.
