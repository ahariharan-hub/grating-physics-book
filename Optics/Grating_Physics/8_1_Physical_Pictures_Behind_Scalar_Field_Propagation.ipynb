# 8.1 Physical Pictures Behind Scalar Field Propagation

An opening produces bright and dark fringes without supplying instructions about where they should go. How can we predict the field beyond it?

Follow three ideas: **Helmholtz constrains the field; Green gives a known response; Kirchhoff adds boundary contributions.** The historical journey from Huygens and Fresnel to Kirchhoff is a useful guide, but we need not memorize the guest list before letting the light move. [Historical context](Appendix_3_Historical_Context_and_Original_Sources.ipynb).

In the first two plots, physical source $P_0$ is outside a closed mathematical surface $S$. Point $P$ is inside; $Q$ runs over its boundary. These are cross-sections of three-dimensional geometry. Normals point **into the enclosed region**.

## 1. What does wavelength change?

Shorten **Wavelength**. More oscillations fit into the same space; the wavefronts move closer together. That is the physical meaning to recognize in

$$U=\frac{e^{ikr_0}}{r_0},\qquad k=\frac{2\pi}{\lambda},\qquad
\nabla^2U+k^2U=0.$$

**The blue fronts show physical illumination from $P_0$.** Here $r_0$ is distance from that source. The small inset shows the local balance between the field and its **full three-dimensional curvature**. A second derivative along one ray would miss spherical spreading. **P transverse** selects a location to inspect.

The source amplitude here is chosen independently; the Green function below uses unit-source normalization.

Helmholtz tells us which fields are possible. Illumination and boundary data select the particular field.

## 2. What changes between two points?

Move **Distance**, then watch the connecting segment and the complex arrow. Longer separation weakens the response and changes its phase:

$$G(Q,P)=\frac{e^{ikR}}{4\pi R},\qquad R=|Q-P|.$$

The factor $1/(4\pi R)$ sets the arrow's length; $kR$ sets its angle. Its horizontal and vertical components are the real and imaginary parts. The amplitude scale stays fixed, so a weaker response really looks shorter. **Q angle** selects a different patch of $S$.

**The green fronts show a mathematical probe centred at $P$.** The observation point has not become a physical emitter. We place the probe here because Green's identity must pick out $U(P)$. Away from its singularity at $P$, it obeys the same Helmholtz equation as $U$. It is the known response we can use to reconstruct an unknown field. [1](#ref-1)

Because this free-space response depends only on separation, $G(Q,P)=G(P,Q)$. The $P$-centred probe and the $Q\to P$ path in the aperture picture therefore use the same scalar response.

## 3. How can the boundary reconstruct the field?

Each boundary patch supplies two clues: its complex field value $U$ and its normal change $h=\partial_nU$. Green's **function** supplies the response; Green's **identity** turns the volume calculation into boundary terms. With the probe singularity accounted for, Kirchhoff's theorem gives

$$U(P)=\int_S\left[U(Q)\partial_{n_Q}G(Q,P)-G(Q,P)h(Q)\right]dA_Q.$$

Read it as: **apply two complex weights at each patch, then add over the boundary**. Both normal derivatives act at $Q$. The theorem is exact when the data belong to the same solution. The derivation and [surface-transformation explorer](supplement/8_1_Boundary_Mathematics_and_History.ipynb) are available in the companion.

For an aperture, choose $S=A\cup B\cup C$: opening $A$, the boundary behind opaque screen region $B$, and distant closure $C$. The outgoing condition removes the combined contribution on $C$, while $P$ stays fixed. The actual field and normal slope on $A+B$ are still needed.

![Aperture geometry: incident illumination, opening A, screen B, boundary point Q, observation point P, and the vectors connecting them.](supplement/figures/8_1_boundary_data.svg)

**Read this view:** $O$ is the coordinate origin; $\mathbf r'$ and $\mathbf r$ locate $Q$ and $P$. Illumination from $P_0$ reaches $Q$; the orange segment connects it to $P$. Its length $R=|\mathbf r-\mathbf r'|$ is the separation used in $G(Q,P)$. This shows how to read the boundary integral; the earlier $P$-centred probe explains its derivation. This original schematic is a cross-section of our three-dimensional model, with $s=-z$.

**Here Kirchhoff makes the approximation:**

| Boundary region | Prescribed field $U$ | Prescribed normal slope $h=\partial_nU$ |
| --- | --- | --- |
| Opening $A$ | $U_i$ | $h_i=\partial_nU_i$ |
| Opaque region $B$ | $0$ | $0$ |

Subscript $i$ denotes the incident field; on this plane, $n$ points toward positive $s$. Both columns are assumed boundary data.

Change **Opening** while holding **Boundary eta** fixed, then **Reveal assumptions**. Watch $Q$ switch regions. **Distance** reshapes the two Green-function weights plotted across the plane; the curves show weight magnitudes. Dividing the derivative curve by $k$ gives comparable units; the integral uses the full complex weights.

The integral theorem and the screen prescription have different status. Zero field does not automatically mean zero normal slope; the two guessed data are generally incompatible as exact boundary data. The unobstructed source field is therefore an incident field, not the exact field transmitted by the screen. [5](#ref-5), [13](#ref-13)
