# 8.1 Physical Pictures Behind Scalar Field Propagation

If we know the field around the boundary of a region, can we find it at a point inside? We will connect that question to aperture propagation. The wave equation gives the local rule; boundary values provide the information needed for the reconstruction.

Keep three roles in view: **Helmholtz constrains the field; Green gives a known response; Kirchhoff adds boundary contributions.** We will follow each role in the pictures before assembling the integral. The historical development from Huygens and Fresnel to Kirchhoff gives further context. [Historical context](Appendix_3_Historical_Context_and_Original_Sources.ipynb).

## 1. What does wavelength change?

Shorten **Wavelength**. More oscillations fit into the same space; the wavefronts move closer together. That is the physical meaning to recognize in

$$U=\frac{e^{ikr_0}}{r_0},\qquad k=\frac{2\pi}{\lambda},\qquad
\nabla^2U+k^2U=0.$$

**The blue fronts show physical illumination from $P_0$.** Here $r_0$ is distance from that source. The small inset shows the local balance between the field and its **full three-dimensional curvature**. A second derivative along one ray would miss spherical spreading. **P transverse** selects a location to inspect.

The source amplitude here is chosen independently; the Green function below uses unit-source normalization.

Helmholtz tells us which fields are possible. Illumination and boundary data select the particular field.

## 2. What changes between two points?

Move **Distance**, then watch the connecting segment and the complex arrow. Longer separation weakens the response and changes its phase:

$$G(Q,P)=\frac{e^{ikR}}{4\pi R},\qquad R=|Q-P|.$$

The factor $1/(4\pi R)$ sets the arrow's length; $kR$ sets its angle. Its horizontal and vertical components are the real and imaginary parts. The amplitude scale stays fixed, so a weaker response really looks shorter. **Q angle** selects a different patch of $S$.

**The green fronts show a mathematical probe centred at $P$.** The observation point has not become a physical emitter. We place the probe here because Green's identity must pick out $U(P)$. Away from its singularity at $P$, it obeys the same Helmholtz equation as $U$. It is the known response we can use to reconstruct an unknown field. [\[1\]](#ref-8-1-handbook)

Because this free-space response depends only on separation, $G(Q,P)=G(P,Q)$. The $P$-centred probe and the $Q\to P$ path in the aperture picture therefore use the same scalar response.

## 3. How can the boundary reconstruct the field?

A normal derivative measures how the field changes perpendicular to the boundary. For a complex field, that change can involve phase as well as amplitude. We need this information together with the field value itself.

Each boundary patch supplies two clues: its complex field value $U$ and its normal change $h=\partial_nU$. Green's **function** supplies the response; Green's **identity** turns the volume calculation into boundary terms. With the probe singularity accounted for, Kirchhoff's theorem gives

$$U(P)=\int_S\left[U(Q)\partial_{n_Q}G(Q,P)-G(Q,P)h(Q)\right]dA_Q.$$

Read it as: **apply two complex weights at each patch, then add over the boundary**. Both normal derivatives act at $Q$. The theorem is exact when the data belong to the same solution. The derivation below explains the shared-face cancellation and small-sphere limit.

For an aperture, choose $S=A\cup B\cup C$: opening $A$, the boundary behind opaque screen region $B$, and distant closure $C$. The outgoing condition removes the combined contribution on $C$, while $P$ stays fixed. The actual field and normal slope on $A+B$ are still needed.

![Aperture geometry: incident illumination, opening A, screen B, boundary point Q, observation point P, and the vectors connecting them.](assets/figures/8_1_boundary_data.svg)

**Read this view:** $O$ is the coordinate origin; $\mathbf r'$ and $\mathbf r$ locate $Q$ and $P$. Illumination from $P_0$ reaches $Q$; the orange segment connects it to $P$. Its length $R=|\mathbf r-\mathbf r'|$ is the separation used in $G(Q,P)$. This shows how to read the boundary integral; the earlier $P$-centred probe explains its derivation. This original schematic is a cross-section of our three-dimensional model, with $s=-z$.

**Here Kirchhoff makes the approximation:**

| Boundary region | Prescribed field $U$ | Prescribed normal slope $h=\partial_nU$ |
| --- | --- | --- |
| Opening $A$ | $U_i$ | $h_i=\partial_nU_i$ |
| Opaque region $B$ | $0$ | $0$ |

Subscript $i$ denotes the incident field; on this plane, $n$ points toward positive $s$. Both columns are assumed boundary data.

Change **Opening** while holding **Boundary eta** fixed, then **Reveal assumptions**. Watch $Q$ switch regions. **Distance** reshapes the two Green-function weights plotted across the plane; the curves show weight magnitudes. Dividing the derivative curve by $k$ gives comparable units; the integral uses the full complex weights.

The integral theorem and the screen prescription have different status. Zero field does not automatically mean zero normal slope; the two guessed data are generally incompatible as exact boundary data. The unobstructed source field is therefore an incident field, not the exact field transmitted by the screen. [\[2\]](#ref-8-1-buchwald), [\[3\]](#ref-8-1-cooper)


:::{dropdown} Derivation: recover the field from its boundary

(boundary-representation-derivation)=

The normal points into the source-free propagation region.

### Shared-face cancellation

Remove a small sphere $S'$ around $P$ and call the remaining volume $V$. Neither field has a source in this punctured region. Multiplying their equations crosswise and subtracting cancels the matching $k^2UG$ terms:

$$U\nabla^2G-G\nabla^2U=0\quad\text{in }V.$$

Green's identity, an integration-by-parts statement, gives our inward-normal form:

$$\int_V(U\nabla^2G-G\nabla^2U)\,dV
=-\int_{S\cup S'}(U\partial_nG-G\partial_nU)\,dA=0.$$

Each internal face is counted twice with opposite normals. Those contributions cancel, leaving the exterior surface **and the hole**. This is mathematical bookkeeping, not destructive interference of light. We have not discarded the singularity; we have given it its own boundary. [\[1, sections 18.2-18.3\]](#ref-8-1-handbook), [\[3, pp. 3-6\]](#ref-8-1-cooper)

### The small-sphere limit

On $S'$, the normal into $V$ points **away from $P$**. As its radius $\epsilon$ shrinks,

$$\operatorname{area}(S')=4\pi\epsilon^2,\qquad
\partial_nG\sim-\frac{1}{4\pi\epsilon^2},$$

$$\int_{S'}U\partial_nG\,dA\longrightarrow-U(P),\qquad
-\int_{S'}G\partial_nU\,dA\longrightarrow0.$$

The sphere is shrinking, but its contribution is not disappearing. The derivative grows as the sphere's area shrinks. When the boundary data come from the same solution, **their complete sum already equals $-U(P)$ at every finite radius for which this source-free construction applies.** Shrinking explains the separate limits; it does not repair an initially incorrect total.

Thus the outer boundary reconstructs the hidden field:

$$\boxed{U(P)=\int_S\left[U(Q)\partial_{n_Q}G(Q,P)
-G(Q,P)\partial_{n_Q}U(Q)\right]dA_Q.}$$

This is Kirchhoff's integral theorem for the free-space Green function. The two data are the complex field value and its spatial normal slope, $h=\partial_nU$. A slope may describe phase change even when field magnitude is constant. The field values and slopes must come from the same solution of the wave equation.

### Why the distant closure disappears

The outgoing condition has the form $R_C(\partial_{R_C}U-ikU)\to0$, with the same condition for $G$ and both fields spreading as $1/R_C$. Their matching leading derivative terms cancel in $U\partial_nG-G\partial_nU$. The remaining combined surface integral tends to zero as the closure expands. This limiting argument introduces no paraxial or far-field approximation at the fixed observation point.
