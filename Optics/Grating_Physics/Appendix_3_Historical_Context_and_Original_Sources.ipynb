## Historical context: follow the questions

Light passes through an opening and paints bright and dark fringes on a screen. The aperture has supplied no instructions about where those fringes should go. Yet, knowing the complex field in its plane lets us predict the pattern beyond it. **How does that work?**

The history is useful because each development answers part of this question. There are quite a few surnames involved. Fortunately, the light does not require us to memorize them before proceeding.

### Why does a shadow have fringes?

A straight-ray picture tells us where the geometrical shadow should fall. It leaves the fringes asking for an explanation. Huygens offers advancing wavelets; Young and Fresnel teach us to add contributions with their relative timing intact. Arriving in step, they reinforce; arriving in opposition, they cancel. Fraunhofer examines the angular patterns produced by slits and gratings. The question becomes: **what arrives at each observation point, and how do those arrivals add?** [2](#ref-2)-[4](#ref-4), [11](#ref-11)

### Can we calculate the contribution rather than guess it?

Green and Helmholtz supply tools for connecting a wave equation to boundary data; Stokes connects the secondary-wave picture to wave dynamics. Kirchhoff puts a boundary representation to work on optical diffraction. Now we have an integral, but we must still decide what field the aperture actually supplies.

That last step attracts some awkward questions. Poincare exposes a conflict in the assumed boundary data; Rayleigh and Sommerfeld give us alternatives that avoid prescribing both data independently. **A sound theorem does not automatically make our model of the screen sound.** Even a beautifully behaved integral deserves that question. [5, sections 3-6](#ref-5)

The main [8.1 lesson](8_1_Physical_Pictures_Behind_Scalar_Field_Propagation.ipynb) follows that boundary thread: **Helmholtz gives the local rule; Green supplies a known response; Green's identity moves the accounting to the boundary; Kirchhoff reconstructs the field.**

### Could we build the same field from directions instead?

Whittaker and Weyl open the plane-wave viewpoint. Booker and Clemmow connect angular spectra with aperture fields; Duffieux, Ratcliffe, Sherman and Goodman develop, interpret or synthesize the Fourier approach. The physical field is the same; the contributions are organized differently.

This gives us a powerful check: with matching boundary data and outgoing/decaying conditions, the full boundary and plane-wave formulations must agree. **Changing mathematical viewpoint should not change the experiment we predict.** [7](#ref-7), [8](#ref-8), [12](#ref-12)-[15](#ref-15), [18](#ref-18)

The [8.2 lesson](8_2_Rayleigh_Sommerfeld_Propagation.ipynb) develops the connection between boundary and plane-wave descriptions.

### What changes when a lens joins the experiment?

Abbe asks which diffracted components the pupil admits and how their recombination forms an image. Debye assembles converging directions near a focus; Richards and Wolf include the vector field and polarization at high aperture. Collins connects paraxial diffraction kernels to ray matrices. Hopkins asks how image intensity changes when illumination is only partially coherent.

These are extensions of the same field-building problem. Emil Wolf appears in several threads because focusing, boundary consistency and coherence intersect; they are not separate varieties of interference. [6](#ref-6), [9](#ref-9), [10](#ref-10), [16](#ref-16), [17](#ref-17), [19](#ref-19)

### A small map, not a roll call

| Thread | The question to carry forward | Foundational anchors |
| --- | --- | --- |
| Physical interference | How do contributions reinforce or cancel? | Huygens, Young, Fresnel; Fraunhofer's angular patterns |
| Boundary representations | Which surface data determine the field? | Green, Helmholtz, Kirchhoff, Rayleigh, Sommerfeld |
| Plane-wave/Fourier representations | Can we reconstruct the field from directional components? | Whittaker, Weyl, Booker-Clemmow, Sherman |
| Imaging and focusing | How do pupils, lenses and illumination change the result? | Abbe, Debye, Richards-Wolf, Hopkins, Collins |

The threads overlap; this is a learning map, not a strict historical succession or an approximation hierarchy. [The optional date index](#history-date-index) keeps the fuller cast and their contributions in one place. Chapter 18 of *Handbook of Optical Systems*, Vol. 2 guides the physical organization; original publications and Buchwald-Yeang support the history. [1, section 18.1](#ref-1), [5](#ref-5)

## Names, dates and contributions

These grouped entries collect the developments behind the historical overview. Dates identify selected works, not exclusive priority claims; related contributions are grouped rather than arranged as a single chain.

| Thread / development | Names and dates | Contribution |
| --- | --- | --- |
| Interference: wavefronts | Huygens: 1678; published 1690 [2](#ref-2) | Construct a new wavefront as the envelope of secondary wavelets. |
| Interference: relative phase | Young: 1801 lecture / 1802 publication; Fresnel: 1818 memoir [11](#ref-11), [3](#ref-3) | Coherent addition explains interference and diffraction fringes. |
| Interference: angular patterns | Fraunhofer: early 1820s / 1822 publication [4](#ref-4) | Measure diffraction directions from slits and gratings. |
| Boundary mathematics | Green: 1828; Stokes: 1856; Helmholtz: 1859 [5](#ref-5) | Integral identities; wave-equation treatment of secondary waves; acoustic wave-integral representations. |
| Boundary formulation and scrutiny | Kirchhoff: 1882-1883; Poincare: 1889, 1892 [5](#ref-5) | Optical boundary integral and aperture assumptions; demonstration of their incompatibility. |
| Boundary alternatives | Sommerfeld: 1896; Rayleigh: 1897; Sommerfeld lectures: 1934, volume: 1954 [5](#ref-5) | The 1896 rigorous half-plane solution is distinct from the plane-boundary alternatives associated with Rayleigh and Sommerfeld's later lectures. |
| Edge-integral representation | Maggi: 1888; Rubinowicz: 1917 [5](#ref-5) | Transform the surface integral and express diffraction through geometrical and rim contributions. |
| Consistency and reinterpretation | Kottler: 1923; Born: 1933; Marchand-Wolf: 1964, 1966 [5](#ref-5) | Field-jump formulations; an iterative-approximation interpretation; comparison and modified boundary formulations, respectively. |
| Early plane-wave representations | Whittaker: 1902-1903; Weyl: 1919 [12](#ref-12), [7](#ref-7) | Wave-equation superpositions; spherical-response/plane-wave connection. |
| Fourier/aperture descriptions | Duffieux: 1946 book; Booker-Clemmow: 1950, received 1949 [13](#ref-13), [8](#ref-8) | Fourier methods in optics; relations between aperture fields, angular spectra and directional patterns. |
| Fourier interpretation and synthesis | Ratcliffe: 1956; Sherman: 1967; Goodman: 1968 book [14](#ref-14), [15](#ref-15), [18](#ref-18) | Physical Fourier interpretation; convolution/transform formulation; optical linear-systems synthesis, respectively. |
| Imaging through a pupil | Abbe: 1873 [9](#ref-9) | Selected diffraction components recombine to form an image. |
| Focused fields | Debye: 1909; Richards-Wolf: 1959 [6](#ref-6), [16](#ref-16) | Near-focus angular-cone description; vector/high-aperture focal fields. |
| Partially coherent imaging | Hopkins: 1953 [10](#ref-10) | Predict image intensity when illumination is partially coherent. |
| Electromagnetic-optics synthesis | Born-Wolf: 1959 book [17](#ref-17) | Consolidate propagation, interference, diffraction and imaging. |
| Lens-system propagation | Collins: 1970 [19](#ref-19) | Write paraxial diffraction kernels using ray-transfer matrices. |

Vector focusing and partial coherence are included as historical connections; the main Chapter 8 lessons focus on coherent scalar propagation.
