# 5.4 From Wavevectors to Lithography Sigma Coordinates

Chapter 5.3 gave us the outgoing diffraction directions. Here we ask which of those directions a collecting lens can accept. A pupil-plane diagram displays that acceptance and locates the orders relative to it. We will first learn to read the diagram, then assign physical distances to its points. All directions and NA values in this lesson refer to the **object (mask) side**.

## What is a pupil plane?

Light passes through a finite opening in the collecting optics. The opening that limits the admitted bundle is called the **aperture stop**. Seen through the lenses, that opening has an optical image called a **pupil**. The pupil can coincide with the opening or lie in another plane. The **pupil plane** is the plane of that pupil. [\[5\]](#ref-5-4-pupils)

For the **pupil view assumed in this lesson**, the optics associate each diffraction direction with a position. Picture an ideal collecting objective, meaning a lens or lens group. A plane wave arriving along its axis focuses at the centre of its back focal plane; a tilted plane wave focuses away from the centre. This is a concrete example of how optics can turn a direction into a position. [\[3\]](#ref-5-4-tran)

In this view, the **centre** labels the axial direction. Positions **farther from the centre** label larger angles. The **circular rim** marks the largest accepted angle in our aligned circular model. The plotted points locate angular components; their marker sizes and colours do not give a calculated pupil intensity.

A screen placed at a chosen distance from the grating locates ray–plane intersections, as in [5.1](5_1_Three_Dimensional_Orders_and_Conical_Diffraction.ipynb). The pupil view here instead uses an angle-to-position relation supplied by the optics. We must state that relation before putting millimetres on its axes.

With this picture in mind, we follow one lithography example through three representations:

$$\underbrace{(k_x,k_y)}_{\text{wavevector}}
\xrightarrow{\ \div(k_0\mathrm{NA})\ }
\underbrace{(\sigma_x,\sigma_y)}_{\text{normalized angular coordinates}}
\xleftrightarrow{\ \text{declared sine mapping}\ }
\underbrace{(x_p/R,y_p/R)}_{\text{normalized physical pupil position}}.$$

The first step defines coordinates. The second needs an optical mapping for the specified pupil or relay. Read [5.2](5_2_Direction_Cosines_and_Spatial_Frequency_Coordinates.ipynb) and [5.3](5_3_Two_Dimensional_Gratings_and_Diffraction_Order_Lattices.ipynb) first.

## 1. Normalize the diffraction wavevectors

The **numerical aperture (NA)** describes how wide a cone of directions the objective accepts, including the refractive index on the stated side. We use it to scale the angular plot so that the accepted cone ends at a circle of radius one.

Use vacuum wavelength $\lambda_0$, $k_0=2\pi/\lambda_0$, and the objective NA referred to the same side as the wavevectors and grating pitches. In a medium of index $n$, $k_x=k_0n\alpha$ and $k_y=k_0n\beta$. Define

$$\boxed{\boldsymbol\sigma\equiv(\sigma_x,\sigma_y)_{\rm angular}
=\frac{(k_x,k_y)}{k_0\mathrm{NA}}
=\frac{n}{\mathrm{NA}}(\alpha,\beta)
=\frac{\lambda_0}{\mathrm{NA}}(f_x,f_y).}$$

All entries are dimensionless. We have rescaled the direction and frequency information from 5.2. This step assumes neither an angle-to-pupil-position law nor an aberration-corrected optical system. Lithography calls the resulting angular plot **sigma space**; [Mack, problem 3](#ref-5-4-mack) [\[1\]](#ref-5-4-mack) constructs it by scaling spatial frequencies by $\lambda_0/\mathrm{NA}$.

For an aligned circular objective, a propagating order directed toward it is collected when $|\boldsymbol\sigma|\le1$. Its angular acceptance therefore becomes a unit circle. With illumination and grating fixed, changing coordinates or the collection NA does not change the order's wavevector or propagation state.

## 2. Supply the optical mapping to a physical pupil

To turn the direction plot into distances, we need a definite angle-to-position law. We use an **ideal sine mapping**: distance from the pupil centre grows with the sine of the angle from the optical axis.

Let $r_p$ be radial position in a specified physical pupil, of radius $R$. For the **ideal sine mapping assumed here**, the optical system supplies

$$r_p=C\sin\theta,\qquad R=C\sin\theta_{\max},$$

where $C$ is the system's mapping scale. The angle $\theta$ is measured from the optical axis, and $\theta_{\max}$ is the largest accepted angle. That maximum angle maps to the pupil rim. Since $\mathrm{NA}=n\sin\theta_{\max}$,

$$\frac{r_p}{R}=\frac{\sin\theta}{\sin\theta_{\max}}
=\frac{n\sin\theta}{\mathrm{NA}}.$$

With aligned transverse axes,

$$\boxed{\left(\frac{x_p}{R},\frac{y_p}{R}\right)
=\frac{(k_x,k_y)}{k_0\mathrm{NA}}
=\boldsymbol\sigma.}$$

**This equality is the bridge from angular sigma space to normalized physical pupil position.** NA alone does not give $R$ in millimetres; that scale belongs to the chosen pupil or relay.

The sine law is an assumption about this selected pupil view. Its validity and scale depend on the optics; it is not a rule for every internal aperture. [Burge, Zhao and Lu, §§1–2](#ref-5-4-burge) [\[2\]](#ref-5-4-burge) discuss the optical conditions behind the sine relation. [Tran and Oldenbourg](#ref-5-4-tran) [\[3\]](#ref-5-4-tran) show how the direction-to-position relation can be calibrated experimentally. A bare downstream plane instead follows a tangent projection; the sine mapping comes from the optics.

## 3. Put source points and grating orders on the same axes

Keep every pitch, wavevector and NA **mask-side** in this example. The grating is in air, and the outgoing direction is toward $-z$.

| Quantity | Value |
| --- | --- |
| Vacuum wavelength | $\lambda_0=193$ nm |
| Mask pitches | $d_x=800$ nm, $d_y=1000$ nm |
| Mask-side projection NA | $0.3375$ |
| Chosen physical pupil radius | $R=3$ mm |
| One incident source direction | $\boldsymbol\sigma_s=(0.3,0)$ |

Each source point supplies an incident direction. A rectangular grating adds its reciprocal-lattice shifts, so

$$\boxed{\boldsymbol\sigma_{s,mp}
=\underbrace{\boldsymbol\sigma_s}_{\text{illumination direction}}
+\underbrace{\left(\frac{m\lambda_0}{d_x\mathrm{NA}},
\frac{p\lambda_0}{d_y\mathrm{NA}}\right)}_{\text{grating contribution}}.}$$

Here $m,p$ label diffraction orders. Their sigma-space spacings are approximately $0.715$ and $0.572$. The linked plots track the same nine candidate orders through wavevector, angular-sigma and physical-pupil coordinates. Move the source or change a pitch to translate or spread the order pattern. Change $R$ to rescale only the physical pupil positions. The physical panel uses the sine mapping from Section 2; wavelength, index and projection NA stay fixed.

## 4. Distinguish a source position from source size

The pair $(\sigma_x,\sigma_y)$ locates one angular component. For a centred circular source, the separate scalar

$$\sigma=\frac{\mathrm{NA}_{\mathrm{illum}}}{\mathrm{NA}_{\mathrm{obj}}}$$

gives the radius of the region occupied by illumination in sigma space. Both NAs must refer to the same optical side. See [Kirchauer, Eq. 2.4](#ref-5-4-kirchauer-coherence) [\[4\]](#ref-5-4-kirchauer-coherence). A dipole or annulus needs a shape or collection of source points, not just one radius. Each source point translates its own order pattern; the objective clips the translated orders at the unit circle.

This lesson locates and selects orders. Amplitudes, coherence and image formation follow later. [Appendix 5: source identity and image formation](Appendix_5_Formulae_Sheet.ipynb#formula-companion-pupils-and-imaging) states how coherent fields and mutually incoherent source contributions are combined.
