# 3.1 Phase change and the order rule

Compare two positions separated by one grating period. How much phase does the incoming wave gain between them? An outgoing order must gain the same phase, apart from whole extra cycles. Counting those extra cycles gives the integer order label.

:::{note} Surrounding medium

 The incident and output media have the same refractive index in these examples. Refraction between unequal exterior media is outside this model.
:::

[Chapter 2](2_0_Plane_Wave_Meets_a_Grating.ipynb) introduced the allowed directions. Here we compare the incident and outgoing phase changes across one period and derive $k_{x,m}=k_{x,i}+m2\pi/d$.

## 1. Recall the phase at the surface

The grating lies at $z=0$, and its pattern repeats along $x$ with period $d$.

From notebook two, the incident phase at the surface is

$$\phi_i(x,0,t)=k_{x,i}x-\omega t.$$

Keep $\mathbf k_i=(k_{x,i},0,k_{z,i})$ and $k=2\pi/\lambda$. The signed component $k_{x,i}$ measures phase change along $x$, in radians per micrometre when $x$ is in micrometres.

The plane-wave phase and the choice of time convention are reviewed in [\[1, §3.3, Eq. (8)\]](#ref-3-1-delft-wavefield).


## 2. Compare two positions at the same time

Choose A at position $x$ and B at $x+d$. Both are on the grating surface. Compare their phases at the **same instant**:

$$\phi_{i,A}=k_{x,i}x-\omega t,$$
$$\phi_{i,B}=k_{x,i}(x+d)-\omega t.$$

Define the phase difference as the phase at B minus the phase at A. The common time term cancels, giving

$$\boxed{\Delta\phi_i=\phi_{i,B}-\phi_{i,A}=k_{x,i}d.}$$

The meaning is direct: **phase change per unit distance × one period = phase change across that period**.

## 3. Explore the phase change here

Use the same incident-wave explorer as notebook two. It starts here with $\lambda=d=2\,\mu\mathrm m$ and $\theta_i=30^\circ$. The wavelength is measured in the medium; the incidence angle is measured from $-z$ toward $+x$.

$$k_{x,i}=\frac{2\pi}{\lambda}\sin\theta_i
=\frac{\pi}{2}\,\mathrm{rad}/\mu\mathrm m,
\qquad
\Delta\phi_i=k_{x,i}d=\pi\,\mathrm{rad}.$$

The **left plot** shows the incident wave above the grating at $t=0$. The **right plot** shows its phase along the surface: blue bands mark the openings, and the dots mark their centres.

Choose the centre at $x=0$ as A and the next one at $x=d$ as B. Their horizontal separation is $d$; their vertical separation in the phase plot is $\Delta\phi_i=k_{x,i}d$.

Adjust the **wavelength**, **period**, or **incidence angle** and compare those two centres. The readout gives the phase step directly. The phase plot keeps the full signed phase, including values beyond one cycle.

For this step, the result to keep is **the spatial phase difference of one plane wave across one grating period: $\Delta\phi_i=k_{x,i}d$**.

### Small calculation 1: phase across one period

Use section 3's starting settings: $\lambda=d=2\,\mu\mathrm m$ and $\theta_i=30^\circ$. Sample A at $x=0$ and B at $x=d$, at the same instant and at $z=0$.

`np.deg2rad` converts degrees to radians before `np.sin` is applied. `np.diff` subtracts neighbouring array entries, giving the B-minus-A phase change. Dividing by $2\pi$ expresses that change in full cycles.

**Predict, then try:** double `d` to `4.0`. Restore `d = 2.0`, then set `theta_i_deg = 0.0`. Compare the printed changes with the incident-phase explorer. This cell has its own editable settings; it does not move the sliders.


```python
wavelength, d = 2.0, 2.0  # micrometres
theta_i_deg = 30.0
k = 2*np.pi / wavelength
k_x_i = k * np.sin(np.deg2rad(theta_i_deg))
x = np.array([0.0, d])  # A, B
phase = k_x_i * x
delta_phi_i = np.diff(phase)[0]
print(' x (um)   incident phase/pi')
for position, phi in zip(x, phase):
    print(f'{position:7.2f}   {phi/np.pi:17.3f}')
print(f'B minus A: {delta_phi_i/np.pi:+.3f} pi rad')
print(f'Cycles across d: {delta_phi_i/(2*np.pi):+.3f}')
```

:::{dropdown} Answer for the stated starting values and changes

At the starting settings, $k_{x,i}=\pi/2\,\mathrm{rad}/\mu\mathrm m$. The phase at A is zero and at B it is $\pi$: a change of **half a cycle**.

Doubling $d$ gives $\Delta\phi_i=2\pi$, one full cycle. Restoring $d=2\,\mu\mathrm m$ and choosing normal incidence gives $k_{x,i}=0$ and no spatial phase change across the period. These are phase changes of the incident wave; they are not yet order labels.

:::


## 4. Connect the phase change to diffraction orders

Recall the [single-slit view at the start of notebook two](2_1_From_a_Slit_to_a_Grating.ipynb#single-slit-kinematics): one isolated opening has a continuous range of candidate outgoing directions.

**The relation $\Delta\phi_i=k_{x,i}d$ alone does not select discrete orders.** Repeating the opening introduces an additional condition: the outgoing field must fit the repeating grating and its illumination.

Assume an ideal linear grating that repeats indefinitely with period $d$.

**First, keep track of the incident phase change.** Keep the incident subscript $i$:

$$\Delta\phi_i=k_{x,i}d.$$

Move the whole description one period along $x$. The grating geometry looks identical, while the incident wave acquires this phase shift. Since the grating is unchanged and its response is linear, the outgoing field must acquire the same phase shift, up to complete cycles. This is the phase-repeat property called *pseudo-periodicity* in [\[2, §2.3, Eq. (2.46)\]](#ref-3-1-maystre).

**Now describe a candidate outgoing plane wave** with vector $\mathbf k_{\mathrm{out}}=(k_{\mathrm{out},x},0,k_{\mathrm{out},z})$. Its phase change obeys the same relation:

$$\Delta\phi_{\mathrm{out}}=k_{\mathrm{out},x}d.$$

To fit the repeating response, its phase change must agree with the incident phase change up to an integer number of full cycles. One full cycle is $2\pi$ radians, so

$$\boxed{\Delta\phi_{\mathrm{out}}-\Delta\phi_i=2\pi m},
\qquad m=0,\pm1,\pm2,\ldots$$

The repeat condition is the new physical constraint. Substituting the two expressions above gives

$$\left(k_{\mathrm{out},x}-k_{x,i}\right)d=2\pi m.$$

When the candidate satisfies this condition for an integer $m$, it is diffraction order $m$: write $\mathbf k_{\mathrm{out}}=\mathbf k_m$ and $k_{\mathrm{out},x}=k_{x,m}$. Then

$$\boxed{k_{x,m}=k_{x,i}+m\frac{2\pi}{d}.}$$

**The integer $m$ is the diffraction-order number.** It counts the extra complete spatial phase cycles per period relative to the incident wave. Once the illumination and period are fixed, the candidate horizontal wavevector components come in steps of $2\pi/d$. [\[2, §2.3, Eqs. (2.48)–(2.50)\]](#ref-3-1-maystre)

The integer rule can be checked against measured directions. Measure the incident angle
and the centres of resolved outgoing order peaks, using the angle convention
of Notebook 2. With known $d$ and $\lambda$, the preceding equation predicts

$$\frac{d}{\lambda}(\sin\theta_m-\sin\theta_i)=m.$$

For a finite grating, compare the peak centres with these integer values
within the angular measurement uncertainty. The phase-cycle diagrams explain
this observable angular pattern; a camera does not directly count the
unwrapped field-phase cycles across one grating period.

**A simpler case: normal incidence.** Here $k_{x,i}=0$, so

$$\Delta\phi_m=k_{x,m}d=2\pi m.$$

An outgoing component returns to the same phase after one grating period: its phase has changed by a whole number of cycles. This connects to the periodic-wave picture in the EMPossible explanation [\[3\]](#ref-3-1-em-discrete).

For tilted incidence, use the **difference** between outgoing and incident phase changes: $(k_{x,m}-k_{x,i})d=2\pi m$.

This selects candidate orders. Notebook [3.2](3_2_Selected_Orders_and_Phase_Continuity.ipynb) shows the candidates with real outgoing directions so that we can draw their plane waves.
