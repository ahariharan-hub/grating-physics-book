# 6. Amplitude and intensity of a scalar wave

The earlier chapters described fields and their directions. To connect those pictures to measurements, we now distinguish an instantaneous field snapshot from a detector's time-averaged signal.

 Relative $|U|^2$ uses a fixed-medium reference. It does not compare absolute irradiance across media. For electric-field amplitudes, irradiance also depends on refractive index, and normal power depends on direction. These plots do not apply that cross-medium calibration.

## Motivation and prerequisites

Review complex amplitude and the time convention in [0.1](0_1_Scalar_Fields_and_Helmholtz.ipynb), and the travelling-wave picture in [1.1](1_1_Oscillation_and_Wavelength.ipynb). No diffraction solver is needed.

## The central idea

An ordinary detector averages over many optical cycles. For a monochromatic travelling wave in a fixed medium and geometry, intensity scales with squared field amplitude. In relative units,

$$I_{\mathrm{rel}}=\frac{|U|^2}{A_{\mathrm{ref}}^2}.$$

A single ideal plane wave has uniform amplitude: the alternating bands of a signed field snapshot do not become bright and dark intensity fringes. This distinction prepares the next chapter, where adding a second field introduces a phase-dependent cross term.

The examples use a scalar, fixed-medium intensity normalization. Converting a signal to absolute irradiance or camera counts requires additional physical and detector calibration; squared amplitude is not universally a normal power-flux measurement.
