# 7.2 Interference of diffraction orders from a grating

**How does grating pitch determine the fringe spacing of two coherent diffraction orders?**

## Motivation and prerequisites

An order arrow indicates propagation direction, not a beam boundary. Two ideal plane-wave components extend throughout the output region and can interfere without mirrors. Finite beams instead require appreciable overlap, as [Notebook 7.1](7_1_Interference_and_Mutual_Coherence_of_Two_Plane_Waves.ipynb) demonstrated.

## Physical assumptions

The grating lies at $z=0$, repeats indefinitely along $x$ and is uniform along $y$. Coherent monochromatic illumination arrives from $z>0$; transmitted orders propagate into $z<0$. Wavelength and distances refer to the homogeneous medium, in micrometres. We retain $e^{-i\omega t}$, common polarization and relative scalar intensity $I_{\mathrm{rel}}=|U|^2$ (reference amplitude one); below we abbreviate it as $I$.

We select two strictly propagating components and prescribe their complex coefficients at $z=0$. This does not imply that other physical orders vanish. No scattering efficiencies, finite-beam propagation or calibrated camera response are calculated.

## 1. Two orders, one optical field

Start with $\lambda=2\,\mu\mathrm m$, pitch $d=2\,\mu\mathrm m$, incidence $30^\circ$, detector $z_D=-3\,\mu\mathrm m$ and orders $(m,p)=(-1,0)$. Incidence is measured from $-z$ toward $+x$. From Notebook 2.3,

$$k=nk_0=\frac{2\pi n}{\lambda_0}=\frac{2\pi}{\lambda},\qquad k_{x,i}=k\sin\theta_i,\qquad
k_{x,j}=k_{x,i}+j\frac{2\pi}{d},\qquad
k_{z,j}=-\sqrt{k^2-k_{x,j}^2}.$$

A positive radicand selects propagation into $z<0$; grazing and evanescent candidates are excluded.

Initially, $k_{x,-1}=-\pi/2$ and $k_{x,0}=+\pi/2$ rad/µm, with equal $k_z=-\sqrt3\pi/2$ rad/µm. Both fields occupy the same output region.

**Predict:** each order alone has uniform intensity. Will their sum? Inspect the overlapping wavefronts and detector fringes below. A dark stripe is field cancellation, not a missing strip in either order.

### Explorer extension: conical diffraction interference

The **incident azimuth** control rotates the incident transverse direction from $+x$ toward $+y$. The signed polar angle retains the original convention: a negative value reverses both transverse components. Every outgoing order obeys

$$k_{x,m}=k\sin\theta_i\cos\phi_i+m\frac{2\pi}{d},\qquad k_{y,m}=k\sin\theta_i\sin\phi_i,\qquad k_{z,m}=-\sqrt{k^2-k_{x,m}^2-k_{y,m}^2}.$$

Only strictly propagating transmitted orders are offered. The grating and detector are square patches spanning $-8$ to $8$ micrometres in both $x$ and $y$.

The orders share the same $k_y$, so $\Delta k_y=0$: their common $y$ phase cancels in intensity. Wavefronts and propagation directions extend in three dimensions, while the XY detector fringes remain uniform along $y$. Their spacing is $\Lambda_x=d/|p-m|$. Moving the detector shifts the fringes whenever $\Delta k_z\ne0$.

Try **azimuth = 30 degrees** with the default polar angle and orders. Both orders have nonzero, equal $k_y$; their $k_z$ values differ, so changing $z_d$ moves the fringes. Azimuth zero recovers the original in-plane example.

### Small calculation 1: calculate the two wavevectors

**Predict:** are both longitudinal components equal and negative? Calculate, then change incidence to $15^\circ$. Does the transverse separation stay fixed while longitudinal symmetry breaks? Keep both orders propagating.


```python
wavelength_um, pitch_um, incidence_deg = 2.0, 2.0, 30.0
orders = np.array([-1, 0])
k = 2*np.pi/wavelength_um
kx = k*np.sin(np.deg2rad(incidence_deg)) + orders*2*np.pi/pitch_um
radicand = k*k-kx*kx
assert np.all(radicand > 1e-12*k*k), 'Choose two strictly propagating orders.'
kz = -np.sqrt(radicand)
vectors = np.column_stack((kx, np.zeros(2), kz))
print('Rows: orders', orders, '; columns: kx, ky, kz (rad/µm)')
print(np.round(vectors, 6))
print('Magnitudes:', np.linalg.norm(vectors, axis=1))
```

<details><summary>Answer 1</summary>

The starting rows are approximately $(-1.570796,0,-2.720699)$ and $(+1.570796,0,-2.720699)$ rad/µm, each of magnitude $\pi$. At $15^\circ$, both remain propagating but their longitudinal components differ. The transverse difference remains $+2\pi/d$, because the common incident transverse wavevector cancels. The negative longitudinal signs select transmission toward $z<0$.

</details>

## 2. From grating pitch to fringe spacing

Write the selected fields as

$$U_j(x,z)=C_j e^{i(k_{x,j}x+k_{z,j}z)},\qquad U=U_m+U_p.$$

The complex coefficients specify reference amplitudes and phases. Let $\Delta\phi=\arg C_p-\arg C_m$. Field addition gives

$$I=|C_m|^2+|C_p|^2+2|C_m||C_p|
\cos\left[(k_{x,p}-k_{x,m})x+(k_{z,p}-k_{z,m})z+\Delta\phi\right].$$

We use second-minus-first, as in 7.1: with fields ordered $(m,p)$, $\Delta\mathbf k=\mathbf k_p-\mathbf k_m$ and $\Delta\phi=\arg C_p-\arg C_m$. The phase slider changes a prescribed coefficient, not source coherence.

The common incident transverse wavevector cancels:

$$k_{x,p}-k_{x,m}=(p-m)\frac{2\pi}{d}.$$

In wavevector space, dashed arrows resolve each full vector into signed components. The purple construction runs from $\mathbf k_m$ to $\mathbf k_p$; it represents their difference, not another order. With $G=2\pi/d$, its transverse component is $(p-m)G$.

$$\Delta\Phi(x,z)=\Delta k_x x+\Delta k_z z+\Delta\phi.$$

Thus $\Delta k_x$ sets transverse phase variation; $\Delta k_z$ sets the phase change with detector position. One fringe period advances phase by $2\pi$:

$$\boxed{\Lambda_x=\frac{2\pi}{|k_{x,p}-k_{x,m}|}=\frac{d}{|p-m|}.}$$

Adjacent orders give period $d$; orders two indices apart give $d/2$. This spacing is along physical $x$ on a constant-$z$ detector; magnification or detector tilt changes the coordinate mapping.

For a fixed available pair and pitch, wavelength and incidence change directions, not transverse spacing. If an order stops propagating, the explorer selects a valid pair where possible: check the order labels after changing geometry.

### Small calculation 2: pitch is not always fringe period

Can different pitches produce the same fringe spacing? Calculate the examples, then change pitch or order separation. Why must a pitch measurement identify the interfering pair?


```python
for d_um, m, p in [(2.0, -1, 0), (4.0, -1, 1), (4.0, 0, 1)]:
    period_um = d_um/abs(m-p)
    print(f'd = {d_um:g} µm; pair ({m:+d}, {p:+d}); fringe period = {period_um:g} µm')
```

<details><summary>Answer 2</summary>

The periods are 2, 2 and 4 µm. A 4 µm grating can produce 2 µm fringes from orders $-1$ and $+1$; adjacent orders from the same grating give 4 µm. The selected pairs are available at the lesson’s 2 µm wavelength and normal incidence for the 4 µm cases. Spacing alone does not identify the grating period without the order separation.

</details>

## 3. Move the detector and change phase

**Reset**, then move only the detector. The starting orders have equal $k_z$, so both acquire the same longitudinal phase: their relative phase and detector pattern stay unchanged.

Now set $d=4$ µm, normal incidence and pair $(0,+1)$. Unequal $k_z$ values make the fringes translate as the detector moves, while spacing stays 4 µm. Tracking a fixed fringe phase gives

$$\Delta x=-\frac{k_{z,p}-k_{z,m}}{k_{x,p}-k_{x,m}}\Delta z_D.$$

A coefficient-phase change similarly gives $\Delta x=-\Delta(\Delta\phi)/(k_{x,p}-k_{x,m})$. Either operation can darken one pixel without reducing pattern visibility.

Compare one control at a time: double pitch for adjacent orders, change wavelength with the pair retained, then vary coefficient phase. Which changes direction, spacing or position? With fewer than two propagating orders, the display clears; increase pitch or reduce wavelength.

### Small calculation 3: add fields and measure the period

Restore calculation 1’s starting values. Predict 2 µm fringes, then compare the complex sum with the expanded intensity. The FFT measures the computed profile’s repetition, not grating coefficients. Change `coefficient_phase_deg` to 90, then `amplitude_ratio` to 0.5. Should either move the frequency peak?


```python
x_um = np.linspace(-8.0, 8.0, 4096, endpoint=False)
z_detector_um = -3.0
amplitude_ratio, coefficient_phase_deg = 1.0, 0.0
delta_phi = np.deg2rad(coefficient_phase_deg)
U_m = np.exp(1j*(vectors[0, 0]*x_um + vectors[0, 2]*z_detector_um))
U_p = amplitude_ratio*np.exp(1j*(vectors[1, 0]*x_um + vectors[1, 2]*z_detector_um + delta_phi))
intensity = abs(U_m+U_p)**2
delta_k = vectors[1]-vectors[0]
expanded = 1 + amplitude_ratio**2 + 2*amplitude_ratio*np.cos(
    delta_k[0]*x_um + delta_k[2]*z_detector_um + delta_phi)
assert np.allclose(intensity, expanded)
frequency = np.fft.rfftfreq(x_um.size, d=x_um[1]-x_um[0])  # cycles/µm
spectrum = abs(np.fft.rfft(intensity-intensity.mean()))
if spectrum[1:].max() > 1e-10:
    peak = 1 + np.argmax(spectrum[1:])
    print('Numerical period (µm):', 1/frequency[peak])
else:
    print('No modulation: a fringe period cannot be measured from this profile.')
print('Analytical period (µm):', 2*np.pi/abs(delta_k[0]))
print('Largest intensity-formula mismatch:', np.max(abs(intensity-expanded)))
```

## 4. Contrast and mutual coherence

A stable coherent input and deterministic linear grating produce mutually coherent nonzero orders in this model. Their prescribed coefficient phases are fixed; the phase slider compares different prescriptions, not random source phases.

For resolved, fully coherent fringes, Notebook 7.1 gives

$$V=\frac{2|C_m||C_p|}{|C_m|^2+|C_p|^2}$$

Equal amplitudes give $V=1$; amplitude ratio 0.5 gives $V=0.8$. Try both. A zero coefficient removes modulation and makes normalized pair coherence undefined; it does not demonstrate source decoherence.

In a more general averaged measurement, coherence magnitude multiplies the intensity-balance factor. Unresolved pixels can also reduce recorded contrast. Neither effect is varied here: the detector shows ideal resolved scalar intensity, without pixel integration, source averaging or noise.

### What index changes in the fringe law

At fixed source frequency, refractive index changes $k=nk_0$ and the order angles. Throughout this lesson, $n$ denotes refractive index; $m$ and $p$ label the selected diffraction orders. For distinct unredirected orders of the same grating on a parallel observation plane,

$$\Delta k_x=(p-m)\frac{2\pi}{d},\qquad \Lambda_x=\frac{d}{|p-m|}.$$

The nominal transverse spacing is unchanged by index while both orders propagate. For arbitrary prescribed beam angles, the spacing instead uses the medium wavelength $\lambda_0/n$ and those angles. Fixed grating orders and fixed beam angles are different comparisons.
