# 2.2 Illumination and the surface field

The field illuminating a grating and the field immediately after an ideal thin mask are different stages of the calculation. Specifying the illumination first makes clear what the mask model prescribes and what must subsequently be propagated.

 For these direction constructions the incident and output media have the same index; refraction between unequal exterior media is not simulated.

The discussion builds on [2.1: slit and grating geometry](2_1_From_a_Slit_to_a_Grating.ipynb). The grating lies at $z=0$, repeats along $x$ with period $d$, and is uniform along $y$. Light approaches from $z>0$ and transmits into $z<0$.

We use $k=2\pi/\lambda$, $e^{-i\omega t}$, and angles measured from $-z$ toward $+x$. The binary amplitude mask $M$ is 1 in an opening and 0 in a blocked strip.

## 1. Illuminate the surface: straight on, then tilted

At **normal incidence**, the wave travels along $-z$. Its wavefronts are parallel to the grating: every point on the surface is at the same phase at a given time. This does not mean the field is constant in time; all surface points oscillate together.

Tilt the wave in the $x$-$z$ plane. Different grating positions now generally have different incident phases.

Measure $\theta_i$ **from $-z$ toward $+x$**. Positive angles lean toward $+x$; zero is downward normal incidence. [Chapter 1](1_0_Plane_Waves_and_Wavevectors.ipynb) used $+z$ as its angle reference. The Cartesian axes and the meaning of $\mathbf k$ are unchanged. $\mathbf k_i=(k_{x,i},k_{y,i},k_{z,i})$ labels the incident wavevector; $k=|\mathbf k_i|$ is its magnitude. Use $-90^\circ<\theta_i<90^\circ$ for illumination arriving from above.

The components are

$$k=\frac{2\pi}{\lambda},\qquad k_{x,i}=k\sin\theta_i,\qquad k_{y,i}=0,\qquad k_{z,i}=-k\cos\theta_i.$$

Use the same complex scalar plane wave as notebook one, now with the incident components above:

$$U_i(x,z)=A e^{i(k_{x,i}x+k_{z,i}z)}.$$

Restoring the time factor gives the signed field shown in the plots:

$$u_i(x,z,t)=\operatorname{Re}\{U_i(x,z)e^{-i\omega t}\}
=A\cos(k_{x,i}x+k_{z,i}z-\omega t).$$

Here $A$ is the scalar amplitude and $\omega$ is the angular frequency. We take the phase to be zero at the origin when $t=0$.

**For a vector electric field, put the direction into the amplitude.** For a linearly polarized plane wave, replace the scalar amplitude $A$ with the vector amplitude
$$\mathbf A_i=A\,\hat{\mathbf e}_i.$$
The electric field is then
$$\boxed{\mathbf E_i(x,z,t)=\mathbf A_i\cos(k_{x,i}x+k_{z,i}z-\omega t).}$$
Here $A$ gives the amplitude's magnitude, and the fixed unit vector $\hat{\mathbf e}_i$ gives the electric-field oscillation direction. The cosine remains a **scalar**: it determines the field's signed value at each position and time. In our homogeneous, isotropic medium, the amplitude must be transverse: $\mathbf k_i\cdot\mathbf A_i=0$. See [\[1, §3.3, Eqs. (8)–(11)\]](#ref-2-2-delft-wavefield) for the vector plane wave and scalar-component description.

For example, $\mathbf A_i=(0,A,0)$ gives $\mathbf E_i=(0,u_i,0)$, a **$y$-polarized** wave. The notebook plots $u_i$ without choosing a polarization direction. The condition $k_{y,i}=0$ specifies no phase variation along $y$, independently of that choice.

Circular and elliptical polarization use a complex vector amplitude to encode relative phases between electric-field components; those cases can be treated later.

At normal incidence the scalar expression becomes $u_i=A\cos(-kz-\omega t)$. Following a fixed phase as time increases gives $z(t)=z(0)-(\omega/k)t$: the wave moves toward smaller $z$. The time factor remains $e^{-i\omega t}$; the negative $k_{z,i}$ sets the downward direction.

At the surface, **set $z=0$**. Its phase is $\phi_i(x,0,t)=k_{x,i}x-\omega t$. Thus moving along the grating by one period changes the incident phase by

$$\Delta\phi_i=k_{x,i}d=\frac{2\pi d}{\lambda}\sin\theta_i.$$

For a fixed $\mathbf k_i$, this incident phase step is the same for either transverse polarization. It follows from the plane-wave phase, without specifying the grating's polarization-dependent response.

- The incident amplitude $A$ is the same at neighbouring openings, but their phases are **generally different** when the wave is tilted.
- A repeating structure does not imply an identically repeating instantaneous field.
- A phase offset that is a whole multiple of $2\pi$ is a special case: it returns to the same point in the cycle. [\[2, §2.3\]](#ref-2-2-maystre)

At normal incidence $k_{x,i}=0$, so the phase difference remains zero for every period. At fixed nonzero angle, doubling $d$ doubles $\Delta\phi_i$: the opening centres are farther apart along the phase slope. For $d=\lambda=2\,\mu\mathrm m$ and $\theta_i=30^\circ$, the phase step is $\pi$ radians. At this snapshot, neighbouring opening centres alternate between $u_i=+1$ and $u_i=-1$, although the amplitude of the incident wave is 1 everywhere.

:::

### Small calculation 1: compare neighbouring openings

Opening centres are at $x=jd$, where the integer $j$ labels an opening, **not a diffraction order**. We evaluate the time-independent complex incident field $U_i(x,0)=A e^{ik_{x,i}x}$. Its magnitude is the wave amplitude; its real part is the signed field at $t=0$.

`np.deg2rad` converts degrees to radians for `np.sin`. As in notebook one, `1j` is the imaginary unit, `np.exp(1j*phase)` gives a complex phase factor of magnitude one, `abs(...)` gives its magnitude, and `.real` gives its real part. `zip` lets us print corresponding array entries together.

**This calculation starts with these incident-wave settings:** $d=2$, $\lambda=2$ (both in $\mu\mathrm m$), $\theta_i=30^\circ$, $t=0$. **Predict, then try:** change `theta_deg` to `0`. Which output columns become constant?

```python
wavelength, d, A = 2.0, 2.0, 1.0  # lengths in micrometres
theta_deg = 30.0
k = 2*np.pi / wavelength
k_x_i = k * np.sin(np.deg2rad(theta_deg))
x = np.arange(-2, 3) * d
phase = k_x_i * x
incident = A * np.exp(1j * phase)
print(' x (um)   phase/pi   amplitude   u at t=0')
for position, phi, U_value in zip(x, phase, incident):
    print(f'{position:7.1f}   {phi/np.pi:8.2f}   {abs(U_value):9.2f}   {U_value.real:8.2f}')
```

:::{dropdown} Answer for the stated starting values and changes

At $30^\circ$, the printed phases advance by $\pi$ between openings. The amplitude column stays at 1 while the real field alternates in sign. At $0^\circ$, all phases become zero and all real fields become 1 at this snapshot; the amplitude column was already constant. At another time they would still oscillate together.

:::

## 2. What does the grating do right at the surface?

- For a first model, imagine the sheet as an **ideal thin mask**: at each position, an opening passes the incident field unchanged and a blocked strip sets it to zero.
- We prescribe this local action and neglect propagation through the thickness, reflections, and multiple scattering within the material. It is a scalar approximation, not a complete electromagnetic solution; it does not calculate polarization changes or polarization-dependent order strengths.
- A real opaque mask need not reproduce this boundary prescription exactly. Small openings or substantial material structure can require a more complete model.

Write the dimensionless complex amplitude transmission as $t(x)$; for our binary amplitude mask, $t(x)=M(x)$ is 1 in openings and 0 elsewhere. Within this prescribed thin-element model,

$$\boxed{U_{\mathrm{exit}}(x)=t(x)\,U_i(x,0)},\qquad t(x+d)=t(x).$$

This equality defines the transmitted **boundary trace**, including amplitude and phase, within the thin-mask model. It is not an exact electromagnetic interface solution.

The exit plane here is $z_e=0^-$, so $U_{\mathrm{exit}}(x)=U(x,0^-)$. Use $t_m$ for Fourier coefficients of $t(x)$ and $C_m$ for Fourier coefficients of the resulting field.

**Notation:** use $U_{\mathrm{exit}}$ at the exit plane, $U_{\mathrm{ref}}$ at a general reference plane, and $C_m$ (or $C_{mp}$) for field coefficients. [Shared convention](0_0_Foundations_of_Scalar_Optics.ipynb#exit-field-notation).


Here $0^-$ means **just on the transmission side**. Uppercase $U$ denotes the time-independent complex scalar field; recover the real field with $u_{\mathrm{exit}}=\operatorname{Re}\{U_{\mathrm{exit}}e^{-i\omega t}\}$. An opening preserves the incoming phase in this mask model. In a blocked region, $U_{\mathrm{exit}}=0$ and its phase is undefined. The binary transmission model is also the starting example in [\[3, §2\]](#ref-2-2-phasor).

This product specifies a field **at the surface**. It does not calculate the field farther away: propagating that surface field is another step. In particular, it does not say that the space beyond every blocked strip remains dark.

**The prescribed field is deterministic.** Writing $A_i=A$ for the incident amplitude used above,

$$U_i(x,0)=A_i e^{ik_{x,i}x},\qquad
U_{\mathrm{exit}}(x)=t(x)A_i e^{ik_{x,i}x}.$$

- For a stationary deterministic mask and a fully coherent monochromatic input, the nonzero diffraction-order coefficients have fixed relative phases at a common reference.
- Their **total relative phase can still vary across space** because their wavevectors differ; it stays fixed in time at each observation point.
- No random field fluctuations are introduced by this multiplication.

### Small calculation 2: apply the mask

To reuse the opening at $x=0$ across the whole grating, `(x + d/2) % d - d/2` finds each point's position within one centred period. `%` is the remainder operation. We then multiply the complex incident field by the real 0/1 transmission, point by point. The output shows the real and imaginary parts of the transmitted complex field; at $t=0$ its real part is the signed field.

**This calculation starts with these incident-wave settings:** $d=\lambda=2\,\mu\mathrm m$, $W_{\mathrm{open}}=d/2$, $\theta_i=30^\circ$, $t=0$. The explorer displays the incident wave; this code adds the thin-mask operation. **Predict, then try:** change `theta_deg` to `0`. Will blocked points start transmitting, or will only the values inside openings change?

```python
wavelength, d, W_open, A = 2.0, 2.0, 1.0, 1.0  # lengths in micrometres
theta_deg = 30.0
x = np.array([0.0, 0.25, 0.75, 1.0, 2.0, 2.25])
k_x_i = (2*np.pi / wavelength) * np.sin(np.deg2rad(theta_deg))
local_x = (x + d/2) % d - d/2
M = (np.abs(local_x) < W_open/2).astype(float)
incident = A * np.exp(1j * k_x_i * x)
after = M * incident
print(' x (um)   M    Re(incident)    Re(after)    Im(after)')
for position, mask, incoming, outgoing in zip(x, M, incident, after):
    print(f'{position:7.2f}   {mask:.0f}    {incoming.real:12.3f}    {outgoing.real:9.3f}    {outgoing.imag:9.3f}')
```

:::{dropdown} Answer for the stated starting values and changes

With the code's starting mask and $A=1$, the blocked samples ($x=0.75$ and $1.00\,\mu\mathrm m$) remain zero when you change the incidence angle. At normal incidence every opening receives $U_i=1$, so $U_{\mathrm{exit}}$ is 1 inside an opening and 0 outside. At $30^\circ$, the opening centres $x=0$ and $2\,\mu\mathrm m$ receive opposite real values at $t=0$. Multiplication by the repeating mask preserves that phase offset; it does not make the tilted field identical from period to period.

:::

**Next step: identify the component waves**

Incident field → prescribed thin-mask field → Fourier components → diffraction orders.

The periodic transmission supplies discrete spatial harmonics; the incident phase tilt shifts their transverse wavevectors to the allowed order values developed in [2.3](2_3_Allowed_Diffraction_Orders.ipynb). The tilted boundary field itself need not repeat identically from period to period. Fourier decomposition extracts the coefficients implied by this mask model; their numerical evaluation is reserved for a later lesson. [0.3](0_3_Boundary_Conditions_and_Diffraction.ipynb#grating-field-flow) shows the two routes to an output field and the coefficient formula.

A more detailed **scalar Helmholtz boundary-value model** would use the refractive-index geometry, incident field, interface conditions and outgoing condition to determine the complex field. Its computed output trace can differ from the thin-mask trace, giving different coefficients. Fourier projection on a plane in the homogeneous output region would then extract its outgoing order amplitudes. **The solver determines the field; decomposition identifies its harmonics.** A scalar solver can describe multidimensional geometry while solving for one scalar field; it is not automatically a vector Maxwell solution.

:::{dropdown} Keep transmission coefficients and field amplitudes distinct

If $t_m$ denotes a dimensionless Fourier coefficient of $t(x)$, the corresponding field coefficient at the reference plane $z=0^-$ is $C_m=A_i t_m$, using the same origin and phase convention. $C_m$ carries the incident field's amplitude scale; $t_m$ does not. Coefficients referred to another plane must include propagation from the chosen reference.

:::
