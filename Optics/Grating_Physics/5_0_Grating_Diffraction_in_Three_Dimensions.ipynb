# 5. Grating diffraction in three dimensions

[Download numerical explorers or standard Python scripts](Downloads.md).

**How do periodic structure, coordinate representation and optical collection fit together?**

First establish outgoing waves, then choose coordinates to compare them, then extend the structure to two periods. Only after those steps do we introduce a collecting optical system.

## Prerequisites and conventions

Review [1.4: plane waves in three dimensions](1_4_Plane_Waves_in_Three_Dimensions.ipynb), [2.3: allowed orders](2_3_Allowed_Diffraction_Orders.ipynb) and [Chapter 4: evanescent orders](4_0_Evanescent_Diffraction_Orders.ipynb). Detailed coefficient calculations in [7.4](7_4_Diffraction_Kinematics_and_Complex_Order_Coefficients.ipynb) are a forward extension, not a prerequisite for this geometry chapter.

Retain $Ue^{-i\omega t}$, incidence from $z>0$ and transmission toward $-z$. Lessons 5.1 and 5.2 preserve their controls for wavelength **in the common exterior medium**, $\lambda=\lambda_0/n$, with $k=nk_0=2\pi/\lambda$. Lessons 5.3 and 5.4 use explicit vacuum wavelength $\lambda_0$ and lossless exterior indices; their examples state which medium defines each coordinate. See the [course medium convention](0_0_Foundations_of_Scalar_Optics.ipynb#medium-and-wavelength).

## Before you begin

Does giving illumination an along-groove component create a second period in the grating? Predict which tangential component is conserved before opening 5.1. Allowed directions, actual excitation and optical-system transfer answer different questions.
