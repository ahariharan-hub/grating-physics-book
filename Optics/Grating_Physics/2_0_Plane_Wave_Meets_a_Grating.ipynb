# 2. A plane wave meets a grating

A single opening spreads light over directions; repeating an opening introduces an additional spatial pattern. This chapter connects that periodicity to discrete diffraction orders and gives you the coordinate tools needed to locate them.

:::{note} Surrounding medium

 The incident and output media have the same refractive index in these examples. Refraction between unequal exterior media is outside this model.
:::

## Motivation and prerequisites

Review [Chapter 1](1_0_Plane_Waves_and_Wavevectors.ipynb), especially wavevector components and the distinction between plane waves and finite beams. You will use wavelength, grating period and simple trigonometric relations.

## The central idea

The incoming wave sets a phase pattern across the grating. The repeating structure allows only certain phase changes from one period to the next. These fix the outgoing wavevector component along the surface. The wave equation supplies the component perpendicular to it and tells us whether the order can propagate.

Keep three questions separate: which order labels are permitted by periodicity, which correspond to propagating directions, and how much light occupies each order. The first two are geometric constraints. They do not determine diffraction efficiencies. The surface-field and mask examples in this chapter illustrate prescribed models; a physical material grating requires a boundary-value calculation to determine its coefficients.
