**What happens to an order when its transverse wavevector is too large for a travelling wave?**

 For these direction constructions the incident and output media have the same index; refraction between unequal exterior media is not simulated.

## Motivation and prerequisites

An integer order can satisfy the periodic phase rule without corresponding to a propagating angle. Rather than discard it, we can complete its wavevector with a complex normal component and examine the resulting near field.

Review [Chapter 3](3_0_Why_Diffraction_Orders_Appear.ipynb) and the propagating-order condition in [2.3](2_3_Allowed_Diffraction_Orders.ipynb). Complex exponentials will connect an imaginary wavevector component to exponential decay.

## The central idea

In a homogeneous output region, the fixed wavenumber relates the transverse and normal components. Beyond the propagating threshold, the outgoing physical branch decays away from the interface. Such an evanescent field can vary along the surface while decaying in the normal direction.

Decay with distance does not by itself imply absorption. In the lossless scalar examples, squared field amplitude and normal energy flux are different quantities. We study individual prescribed orders to understand this distinction; their excitation strengths and coupling to another nearby structure require a separate boundary-value problem.
