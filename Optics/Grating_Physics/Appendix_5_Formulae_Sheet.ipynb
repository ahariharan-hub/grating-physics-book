**Propagation signs.** Most grating lessons transmit toward $-z$.
Define downstream distance $s=z_e-z\geq0$ and
$\beta=\sqrt{k^2-|\mathbf k_\perp|^2}$ with nonnegative real and imaginary parts.
Then $k_z=-\beta$ and the outgoing factor is $e^{i\beta s}$.
The introductory $+z$ half-space instead uses $e^{i\beta z}$.
This sheet states the local choice when needed.

**Fourier convention.** Unless a subscript $f$ is shown, spectra use radians per length:
$$
\widetilde U(k_x)=\int_{\mathbb R}U(x)e^{-ik_xx}\,dx,
\qquad
U(x)=\frac1{2\pi}\int_{\mathbb R}\widetilde U(k_x)e^{ik_xx}\,dk_x.
$$
In cycles per length,
$\widetilde U_f(f_x)=\int U(x)e^{-i2\pi f_xx}\,dx=\widetilde U(2\pi f_x)$.
In two transverse dimensions use a double integral and inverse factor $(2\pi)^{-2}$
for the radians convention. The corresponding cycles convention has no inverse prefactor.
A one-dimensional FT value has units of field times length; its frequency axis has
reciprocal-length units. Infinite periodic-field spectra are delta distributions.

**Model limits.** Unless a material solve is explicitly named, the propagation
chains assume a homogeneous, isotropic, lossless exterior. A thin-mask product
prescribes exit data; it does not solve interfaces. Physical indices inside
a scattering structure may be complex. Gaussian propagation formulas are paraxial.

## 0 — Foundations


(formula-0-0)=
### [0.0 — Foundations: the whole calculation](0_0_Foundations_of_Scalar_Optics.ipynb)

First choose how the exit field will be obtained; then propagate that complex field.

$$
\begin{gathered}
\underbrace{\{\text{illumination, structure, model}\}}_{\text{physical inputs}}
\xrightarrow{\text{prescribe or solve}}
U_{\mathrm{exit}}\\[0.7em]
\xrightarrow{\text{propagate}}
U_{\mathrm{ref}}
\xrightarrow{\text{square magnitude}}
I_{\mathrm{rel}}=\frac{|U_{\mathrm{ref}}|^2}{A_{\mathrm{ref}}^2}.
\end{gathered}
$$

A propagation method does not determine the field transmitted by a material structure. [Shared notation](0_0_Foundations_of_Scalar_Optics.ipynb#exit-field-notation).


(formula-0-1)=
### [0.1 — Scalar fields and Helmholtz](0_1_Scalar_Fields_and_Helmholtz.ipynb)

In a homogeneous, nondispersive, lossless medium, select one temporal frequency.

$$
\nabla^2u-\frac{1}{v^2}\partial_t^2u=0
\xrightarrow{u=\operatorname{Re}\{Ue^{-i\omega t}\}}
(\nabla^2+k^2)U=0,
\qquad k=\frac{\omega}{v}=\frac{2\pi}{\lambda}.
$$

Helmholtz constrains possible fields. Boundary data and illumination select a particular one.


(formula-0-2)=
### [0.2 — Families of allowed fields](0_2_Solutions_of_the_Helmholtz_Equation.ipynb)

These are alternative solution families, rather than stages that turn one field into another.

$$
\begin{aligned}
(\nabla^2+k^2)U=0
&\xrightarrow{\text{choose a geometry}}
\begin{cases}
Ae^{i\mathbf k\cdot\mathbf r},&\mathbf k\cdot\mathbf k=k^2,\\
A_s e^{ikr}/r,&r>0,\\
A J_0(k_\perp\rho)e^{ik_z z},&k_\perp^2+k_z^2=k^2;
\end{cases}\\
U=e^{ikz}\psi
&\xrightarrow{\text{paraxial approximation}}
2ik\partial_z\psi+\nabla_\perp^2\psi\simeq0
\xrightarrow{\text{Gaussian solution}}\psi_{\mathrm G}.
\end{aligned}
$$

The notebook also treats standing, cylindrical and spherical-harmonic fields. A standard Gaussian beam solves the paraxial model; that is a different status from an exact Helmholtz solution.


(formula-0-3)=
### [0.3 — Boundary conditions select a field](0_3_Boundary_Conditions_and_Diffraction.ipynb)

For the introductory one-dimensional half-space $z>0$, impose a boundary value and exclude a wave arriving from $+\infty$.

$$
U(z)=B_+e^{ikz}+B_-e^{-ikz}
\xrightarrow{\substack{U(0)=U_0\\B_-=0}}
U(z)=U_0e^{ikz}.
$$

For an aperture, prescribing a transmitted field is a model choice. Solving material interfaces is another route; the two need not give the same field.


(formula-0-4)=
### [0.4 — Propagate a known field](0_4_Propagation_of_Scalar_Optical_Fields.ipynb)

In the outgoing positive-z example, Fourier components propagate independently through the homogeneous region.

$$
U_{\mathrm{exit}}
\xrightarrow{\mathcal F_\perp}\widetilde U_{\mathrm{exit}}
\xrightarrow{\times\,e^{i\beta z}}\widetilde U_z
\xrightarrow{\mathcal F_\perp^{-1}}U_z,
\qquad \beta=\sqrt{k^2-|\mathbf k_\perp|^2}.
$$

Choose $\operatorname{Re}\beta\geq0$ and $\operatorname{Im}\beta\geq0$ so propagating components travel outward and evanescent components decay.


(formula-0-5)=
### [0.5 — Solve the scalar grating scattering problem](0_5_From_Helmholtz_BVP_to_Diffracted_Field.ipynb)

Here the exit field is an output of a physical boundary-value calculation.

$$
\begin{gathered}
\{n(x,z),\,U_i,\,\text{interfaces, outgoing conditions}\}\\[0.7em]
\xrightarrow{\text{scalar FMM solve}}
U(x,z)
\xrightarrow{\text{evaluate at }z_e}
U_{\mathrm{exit}}(x)=U(x,z_e).
\end{gathered}
$$

Fourier harmonics are coupled inside a patterned layer. Fourier expansion alone is not the scattering solve. The solver retains its scalar assumptions.


(formula-0-6)=
### [0.6 — Field coefficients, reconstruction and power](0_6_From_Complex_Fields_to_Diffraction_Orders.ipynb)

At a fixed outgoing reference plane, projection and synthesis describe the same field.

$$
\begin{aligned}
U_{\mathrm{ref}}(x)
&\overset{\text{project periodic factor}}{\underset{\text{synthesis}}{\hspace{3em}\rightleftharpoons\hspace{3em}}}
\{C_m(z_{\mathrm{ref}}),\,k_{x,m}=k_{x,i}+mG\},\\
C_m(z_e)&\xrightarrow{\text{distance }s\text{ toward }-z}
C_m(z_e-s)=C_m(z_e)e^{i\beta_m s}.
\end{aligned}
$$

For the scalar flux convention used here, $\eta_m=\operatorname{Re}\beta_{m,\mathrm{out}}\,|C_m/A_i|^2/\beta_i$, with a propagating incident wave and consistent field normalization. Squared coefficients alone are not generally diffraction efficiencies.


(formula-0-7)=
### [0.7 - Ideal mask versus physical absorbing grating](0_7_Ideal_Binary_Masks_vs_Physical_Gratings.ipynb)

Prescribe the ideal binary mask at $z_{\mathrm{obj}}=0$, then compare both predictions at $z_{\mathrm{ref}}=-h-\ell$ with the same incident phase.

$$
\begin{aligned}
\{t,U_i(x,0)\}&\xrightarrow{\text{thin-mask rule at }0}
U_{\mathrm{thin}}(x,0^-)
\xrightarrow{\text{propagate }h+\ell}
U_{\mathrm{thin}}(x,z_{\mathrm{ref}}),\\[0.5em]
\{n(x,z),U_i,\text{boundary data}\}
&\xrightarrow{\text{scalar scattering solve}}
U_{\mathrm{exit}}^{\mathrm{phys}}(x,-h^-)
\xrightarrow{\text{propagate }\ell}
U_{\mathrm{phys}}(x,z_{\mathrm{ref}}).
\end{aligned}
$$

For unit normal incidence, $C_m^{\mathrm{thin}}(z_{\mathrm{ref}})=t_m e^{i\beta_m(h+\ell)}$. The common-plane physical coefficients instead follow from the material solve. Prescribing a binary field directly at $z=-h$ is a different approximation.

(chapter-1)=
## 1 — Plane waves


(formula-1-0)=
### [1.0 — Plane waves: wavelength, direction and phase](1_0_Plane_Waves_and_Wavevectors.ipynb)

Wavelength sets the wavevector length; direction sets its components.

$$
\begin{gathered}
\{\lambda,\widehat{\mathbf s}\}
\xrightarrow{\text{definition}}\mathbf k=\frac{2\pi}{\lambda}\widehat{\mathbf s}\\[0.7em]
\xrightarrow{\text{spatial phase}}U=Ae^{i(\mathbf k\cdot\mathbf r+\phi_0)}
\xrightarrow{\text{real field}}u=A\cos(\mathbf k\cdot\mathbf r+\phi_0-\omega t).
\end{gathered}
$$

Amplitude is an additional input.


(formula-1-1)=
### [1.1 — Oscillation and wavelength](1_1_Oscillation_and_Wavelength.ipynb)

A displacement of one wavelength along propagation advances spatial phase by one full cycle.

$$
\Delta \ell
\xrightarrow{\times\,k}\Delta\phi=k\Delta\ell
=2\pi\frac{\Delta\ell}{\lambda},
\qquad
\Delta\ell=\lambda\ \Longrightarrow\ \Delta\phi=2\pi.
$$

At a fixed point, the temporal cycle is $T=2\pi/\omega$. Spatial frequency and temporal frequency use different independent coordinates.


(formula-1-2)=
### [1.2 — Plane waves versus finite beams](1_2_Ideal_Plane_Waves_versus_Finite_Optical_Beams.ipynb)

Compare their transverse profiles at a stated plane, chosen here as a Gaussian waist.

$$
\begin{aligned}
U_{\mathrm{plane}}(\rho)&=A
\xrightarrow{|\cdot|^2}I_{\mathrm{plane}}=|A|^2,\\
U_{\mathrm G}(\rho)&=Ae^{-\rho^2/w_0^2}
\xrightarrow{|\cdot|^2}I_{\mathrm G}=|A|^2e^{-2\rho^2/w_0^2}.
\end{aligned}
$$

A finite beam has an envelope and an angular spread. Its standard Gaussian propagation law is paraxial.


(formula-1-3)=
### [1.3 — Wavefronts and wavevectors](1_3_Wavefronts_and_Wavevectors.ipynb)

A plane-wave wavefront is a surface of constant spatial phase.

$$
\Phi(\mathbf r)=\mathbf k\cdot\mathbf r+\phi_0
\xrightarrow{\text{displace by }\Delta\mathbf r}
\Delta\Phi=\mathbf k\cdot\Delta\mathbf r,
\qquad \nabla\Phi=\mathbf k.
$$

Displacement within a wavefront has $\mathbf k\cdot\Delta\mathbf r=0$; the wavevector is normal to that wavefront.


(formula-1-4)=
### [1.4 — Plane waves in three dimensions](1_4_Plane_Waves_in_Three_Dimensions.ipynb)

Here the spherical polar angle $\theta$ is measured from $+z$.

$$
\{k,\theta,\varphi\}
\xrightarrow{\text{coordinate conversion}}
\mathbf k=k(\sin\theta\cos\varphi,\sin\theta\sin\varphi,\cos\theta)
\xrightarrow{\mathbf k\cdot\mathbf r}
k_xx+k_yy+k_zz.
$$

The later transmitted-grating examples use an axis toward $-z$; keep the stated angle reference.


(chapter-2)=
## 2 — Illumination and grating orders


(formula-2-0)=
### [2.0 — A plane wave meets a grating](2_0_Plane_Wave_Meets_a_Grating.ipynb)

Periodicity fixes candidate tangential wavevectors; the surrounding medium fixes which can propagate.

$$
\{d,k_{x,i}\}
\xrightarrow{G=2\pi/d}k_{x,m}=k_{x,i}+mG
\xrightarrow{\text{dispersion, outgoing branch}}
k_{z,m}=-\sqrt{k_{\mathrm{out}}^2-k_{x,m}^2}.
$$

This locates orders. It does not determine their complex coefficients.


(formula-2-1)=
### [2.1 — From one slit to an infinite grating](2_1_From_a_Slit_to_a_Grating.ipynb)

Changing the physical aperture changes its spectrum.

$$
\begin{aligned}
a(x)&\xrightarrow{\text{FT}}\widetilde a(k_x)
\quad\text{(continuous slit spectrum)},\\
t(x)=\sum_{\ell\in\mathbb Z}a(x-\ell d)
&\xrightarrow{\text{FT}}
2\pi\sum_m t_m\delta(k_x-mG),
\qquad t_m=\frac{\widetilde a(mG)}{d}.
\end{aligned}
$$

The second row is an infinite periodic array. A finite array has broadened peaks rather than exact delta lines.


(formula-2-2)=
### [2.2 — Illumination and the surface field](2_2_Illumination_and_the_Surface_Field.ipynb)

Under the scalar thin-mask prescription, multiply the incident complex field by transmission.

$$
\begin{gathered}
\underbrace{t(x)}_{\text{periodic mask}}\qquad
\underbrace{A_i e^{ik_{x,i}x}}_{\text{incident plane wave}}
\xrightarrow{\text{thin-mask product}}
U_{\mathrm{exit}}(x)\\[0.7em]
\xrightarrow{\text{translate by }d}
U_{\mathrm{exit}}(x+d)=e^{ik_{x,i}d}U_{\mathrm{exit}}(x).
\end{gathered}
$$

Identical mask cells need not have identical complex field values. The multiplier is the Bloch translation factor.


(formula-2-3)=
### [2.3 — Allowed orders and propagation angles](2_3_Allowed_Diffraction_Orders.ipynb)

Use the output-medium wavenumber when converting an outgoing component into an angle.

$$
k_{x,m}=k_i\sin\theta_i+mG
\xrightarrow{\text{test }|k_{x,m}|\leq k_{\mathrm{out}}}
\sin\theta_m=\frac{k_{x,m}}{k_{\mathrm{out}}}
=\frac{n_i\sin\theta_i+m\lambda_0/d}{n_{\mathrm{out}}}.
$$

The real angle describes a propagating or grazing order. Beyond that range, retain the evanescent complex longitudinal component.


(formula-2-4)=
### [2.4 — Order coordinates and geometrical markers](2_4_Order_Coordinates_and_Spacing.ipynb)

The same propagation direction intersects different reference surfaces at different coordinates.

$$
\frac{k_{x,m}}{k}=\sin\theta_m
\xrightarrow{\text{choose reference surface}}
\begin{cases}
x_{\mathrm{cap}}=R_{\mathrm{cap}}\sin\theta_m,&\text{sphere},\\
x_{\mathrm{plane}}=L\tan\theta_m,&\text{flat plane}.
\end{cases}
$$

These are geometrical intersections. A physical lens pupil requires its own optical mapping.

## 3 — Phase matching


(formula-3-0)=
### [3.0 — Why orders appear](3_0_Why_Diffraction_Orders_Appear.ipynb)

Matching the incident translation factor across identical cells gives the order rule.

$$
e^{ik_{x,m}d}=e^{ik_{x,i}d}
\overset{\text{phase modulo }2\pi}{\underset{\text{exponentiate}}{\hspace{2.5em}\rightleftharpoons\hspace{2.5em}}}
(k_{x,m}-k_{x,i})d=2\pi m
\xrightarrow{\text{rearrange}}k_{x,m}=k_{x,i}+mG.
$$

The full field can be Bloch periodic while its carrier-removed factor is ordinarily periodic.


(formula-3-1)=
### [3.1 — Phase change and the order rule](3_1_Phase_Change_and_the_Order_Rule.ipynb)

Subtract the incident phase advance from a candidate outgoing phase advance.

$$
\begin{gathered}
\Delta\phi_{\mathrm{relative}}=(k_x-k_{x,i})d
\xrightarrow{\text{reinforcement}}
\Delta\phi_{\mathrm{relative}}=2\pi m\\[0.7em]
\xrightarrow{\text{same medium}}
d(\sin\theta_m-\sin\theta_i)=m\lambda.
\end{gathered}
$$

Reinforcement selects directions. A coefficient can still vanish because of the unit-cell transmission.


(formula-3-2)=
### [3.2 — Selected orders and phase continuity](3_2_Selected_Orders_and_Phase_Continuity.ipynb)

Express a trial direction as a relative number of phase cycles per mask period.

$$
\nu=\frac{k_x-k_{x,i}}{G}
\xrightarrow{\text{one-cell relative phase}}
e^{i2\pi\nu}
\xrightarrow{\text{compatible translation}}
e^{i2\pi\nu}=1\ \Longleftrightarrow\ \nu\in\mathbb Z.
$$

Ordinary repetition of the carrier-removed field supplies the integer condition.


(formula-3-3)=
### [3.3 — Phase cycles and path differences](3_3_Phase_Cycles_and_Path_Differences.ipynb)

In the same surrounding medium, the geometrical path difference gives the same order condition.

$$
\Delta\ell=d(\sin\theta_m-\sin\theta_i)
\xrightarrow{\times\,2\pi/\lambda}\Delta\phi
\xrightarrow{\text{complete cycles}}\Delta\phi=2\pi m
\ \Longleftrightarrow\ \Delta\ell=m\lambda.
$$

For unequal media use optical path, or the wavevector relation, rather than this single-medium path formula.

## 4 — Evanescent fields


(formula-4-0)=
### [4.0 — Propagating, grazing and evanescent orders](4_0_Evanescent_Diffraction_Orders.ipynb)

After finding the tangential component, inspect the longitudinal radicand.

$$
D_m=k^2-|\mathbf k_{\perp,m}|^2
\xrightarrow{\text{outgoing }-z}
k_{z,m}=
\begin{cases}
-\sqrt{D_m},&D_m>0\quad\text{propagating},\\
0,&D_m=0\quad\text{grazing},\\
-i\sqrt{-D_m},&D_m<0\quad\text{evanescent}.
\end{cases}
$$

An evanescent order has a valid transverse spatial frequency; it has no real propagation angle in that medium.


(formula-4-1)=
### [4.1 — Construct an evanescent field](4_1_Constructing_an_Evanescent_Field.ipynb)

Let $s=-z\geq0$ measure distance from the reference plane toward the transmitted half-space.

$$
|k_x|>k
\xrightarrow{\kappa=\sqrt{k_x^2-k^2}}
k_z=-i\kappa
\xrightarrow{e^{ik_zz},\ z=-s}
U(x,s)=Ae^{ik_xx}e^{-\kappa s}.
$$

The transverse factor oscillates; the normal factor decays.


(formula-4-2)=
### [4.2 — Decay lengths and normal power](4_2_Decay_Length_Grazing_and_Energy_Flow.ipynb)

Amplitude and squared amplitude have different decay lengths.

$$
|U(s)|=|U(0)|e^{-\kappa s}
\xrightarrow{\text{square}}
I_{\mathrm{rel}}(s)=I_{\mathrm{rel}}(0)e^{-2\kappa s},
\qquad
\ell_A=\frac1\kappa,\quad \ell_I=\frac1{2\kappa}.
$$

For a single outgoing evanescent component in a lossless exterior, $\operatorname{Re}\beta=0$ gives zero time-averaged normal flux. This spatial decay is not material absorption.

## 5 — Three-dimensional geometry and pupils


(formula-5-0)=
### [5.0 — Grating diffraction in three dimensions](5_0_Grating_Diffraction_in_Three_Dimensions.ipynb)

First count independent periodic directions, then complete each candidate using the dispersion relation.

$$
\text{surface periodicity}
\xrightarrow{\text{reciprocal shifts}}\mathbf k_{\perp,\mathbf m}
\xrightarrow{\text{outgoing dispersion}}
k_{z,\mathbf m}=-\sqrt{k_{\mathrm{out}}^2-|\mathbf k_{\perp,\mathbf m}|^2}.
$$

One periodic direction gives one integer order label, even when propagation is fully three-dimensional.


(formula-5-1)=
### [5.1 — One-dimensional grating, conical diffraction](5_1_Three_Dimensional_Orders_and_Conical_Diffraction.ipynb)

The grating repeats along x and is invariant along y.

$$
\begin{gathered}
\mathbf k_{\perp,i}=(k_{x,i},k_{y,i})
\xrightarrow{\text{grating shift}}
\mathbf k_{\perp,m}=(k_{x,i}+mG,k_{y,i})\\[0.7em]
\xrightarrow{\text{dispersion}}k_{z,m}
=-\sqrt{k_{\mathrm{out}}^2-k_{x,m}^2-k_{y,i}^2}.
\end{gathered}
$$

Nonzero conserved y momentum produces conical geometry; it does not add a second grating order index.


(formula-5-2)=
### [5.2 — Wavevector, frequency and direction coordinates](5_2_Direction_Cosines_and_Spatial_Frequency_Coordinates.ipynb)

These are coordinate conversions for the same component in a stated homogeneous medium.

$$
(k_x,k_y)
\overset{\div\,2\pi}{\underset{\times\,2\pi}{\hspace{1em}\rightleftharpoons\hspace{1em}}}
(f_x,f_y)
\overset{\times\,\lambda}{\underset{\div\,\lambda}{\hspace{1em}\rightleftharpoons\hspace{1em}}}
(\alpha_x,\alpha_y)=\frac{(k_x,k_y)}{k}.
$$

$f_x,f_y$ count cycles per length; $\alpha_x,\alpha_y$ are direction cosines for propagating components. No optical lens assumption is needed.


(formula-5-3)=
### [5.3 — Two-dimensional order lattices](5_3_Two_Dimensional_Gratings_and_Diffraction_Order_Lattices.ipynb)

Two independent surface periods supply two independent integer labels.

$$
\begin{aligned}
A=[\mathbf a_1,\mathbf a_2]
&\xrightarrow{A^{\mathsf T}B=2\pi I}B=[\mathbf b_1,\mathbf b_2],\\
(m,p)\in\mathbb Z^2
&\xrightarrow{\text{reciprocal shift}}
\mathbf k_{\perp,mp}=\mathbf k_{\perp,i}+m\mathbf b_1+p\mathbf b_2
\xrightarrow{\text{dispersion}}k_{z,mp}.
\end{aligned}
$$

For rectangular periods, $\mathbf b_1=(2\pi/d_x,0)$ and $\mathbf b_2=(0,2\pi/d_y)$. The lattice determines candidate locations, not their amplitudes.


(formula-5-4)=
### [5.4 — Lithography sigma coordinates and pupils](5_4_From_Diffraction_Coordinates_to_Pupils.ipynb)

Normalize angular coordinates first. Add a specified sine mapping only when interpreting them as pupil positions.

$$
\begin{aligned}
(k_x,k_y)
&\xrightarrow{\div\,k_0\mathrm{NA}}
(\sigma_x,\sigma_y)_{\mathrm{angular}}
\xrightarrow[\ r_p=C\sin\theta\ ]{\text{assumed sine mapping}}
(x_p/R,y_p/R),\\
\boldsymbol\sigma_s
&\xrightarrow{\text{rectangular grating}}
\boldsymbol\sigma_{s,mp}
=\boldsymbol\sigma_s+
\left(\frac{m\lambda_0}{d_x\mathrm{NA}},
\frac{p\lambda_0}{d_y\mathrm{NA}}\right).
\end{aligned}
$$

Here $\mathrm{NA}=n\sin\theta_{\max}$ and $R=C\sin\theta_{\max}$, all on the same optical side. A real outgoing order is collected by the ideal centered circular pupil when $|\boldsymbol\sigma_{s,mp}|\leq1$. Scalar source extent $\sigma=\mathrm{NA}_{\mathrm{illum}}/\mathrm{NA}_{\mathrm{obj}}$ is distinct from a source point $(\sigma_x,\sigma_y)$. Full aplanatism also requires on-axis correction.

## 6 — Amplitude and intensity


(formula-6-0)=
### [6.0 — From a scalar amplitude to relative intensity](6_0_Amplitude_and_Intensity_of_a_Scalar_Wave.ipynb)

Recover the real optical oscillation, then average its square.

$$
\begin{gathered}
U=Ae^{i\Phi}
\xrightarrow{\operatorname{Re}\{Ue^{-i\omega t}\}}
u=A\cos(\Phi-\omega t)\\[0.7em]
\xrightarrow{\text{cycle average}}\langle u^2\rangle=\frac{A^2}{2}
\xrightarrow{\text{reference normalization}}
I_{\mathrm{rel}}=\frac{|U|^2}{A_{\mathrm{ref}}^2}.
\end{gathered}
$$

The chosen normalization absorbs the common factor. Absolute irradiance requires the relevant electromagnetic and medium factors.


(formula-6-1)=
### [6.1 — Time averaging](6_1_Time_Averaging_and_Relative_Intensity.ipynb)

A detector averages over the fast optical oscillation.

$$
\cos^2(\Phi-\omega t)=\frac12[1+\cos(2\Phi-2\omega t)]
\xrightarrow{\text{average full cycles}}\frac12.
$$

For an envelope or phase varying during exposure, average the resulting squared field over that exposure as well.


(formula-6-2)=
### [6.2 — One order across a detector](6_2_One_Order_Across_a_Detector.ipynb)

At one fixed plane, an ideal propagating plane wave has a varying phase but uniform magnitude.

$$
U_m(x,y)=C_m e^{i(k_{x,m}x+k_{y,m}y)}
\xrightarrow{\text{square magnitude}}
|U_m(x,y)|^2=|C_m|^2.
$$

The stripes in a signed instantaneous field plot are not intensity fringes from one plane wave.


(chapter-7)=
## 7 — Interference and coefficients


(formula-7-0)=
### [7.0 — Overlap and interference](7_0_Introduction_to_Interference_and_Beam_Overlap.ipynb)

For coherent components, add complex fields before forming intensity.

$$
\{U_1,U_2\}
\xrightarrow{\text{field superposition}}U=U_1+U_2
\xrightarrow{|\cdot|^2}
I=|U_1|^2+|U_2|^2+2\operatorname{Re}(U_1^*U_2).
$$

Spatial overlap makes the cross term locally possible. Averaged field correlation determines whether it survives observation.


(formula-7-1)=
### [7.1 — Mutual coherence and visibility](7_1_Interference_and_Mutual_Coherence_of_Two_Plane_Waves.ipynb)

Use the first-field-conjugated convention of the notebook.

$$
\begin{aligned}
\{U_1,U_2\}
&\xrightarrow{\text{average products}}
\Gamma_{12}=\langle U_1^*U_2\rangle
\xrightarrow{\text{normalize}}
\gamma_{12}=\frac{\Gamma_{12}}{\sqrt{\bar I_1\bar I_2}},\\
\bar I&=\bar I_1+\bar I_2+2\operatorname{Re}\Gamma_{12},
\qquad
V=\frac{2\sqrt{\bar I_1\bar I_2}}{\bar I_1+\bar I_2}|\gamma_{12}|.
\end{aligned}
$$

$\bar I_j=\langle|U_j|^2\rangle$. The visibility formula assumes a phase scan at fixed local intensities and coherence magnitude. A fixed phase shift rotates $\gamma_{12}$ without changing its magnitude.


(formula-7-2)=
### [7.2 — Interference between grating orders](7_2_Interference_of_Diffraction_Orders_from_a_Grating.ipynb)

For two distinct, coherent, unredirected orders on a plane parallel to the grating:

$$
\begin{gathered}
\{m,p\}
\xrightarrow{\text{subtract transverse wavevectors}}
\Delta k_x=(p-m)G\\[0.7em]
\xrightarrow{\text{one relative-phase cycle}}
\Lambda_x=\frac{2\pi}{|\Delta k_x|}=\frac{d}{|p-m|}.
\end{gathered}
$$

Complex coefficient phases and propagation change fringe position. They do not change this plane-wave spacing.


(formula-7-3)=
### [7.3 — Finite order beams and recombination](7_3_Finite_Diffraction_Beams_and_Experimental_Recombination.ipynb)

The lesson prescribes finite envelopes around central order directions.

$$
\begin{gathered}
U_j(\mathbf r)\approx A_j(\mathbf r)e^{i\Phi_j(\mathbf r)}\\[0.7em]
\xrightarrow{\text{overlap and coherent addition}}
I=|U_m|^2+|U_p|^2+
2A_mA_p\cos(\Phi_p-\Phi_m).
\end{gathered}
$$

For freely travelling beam centers, $x_j(s)=x_j(0)+s\tan\theta_j$. Envelopes, spreading and curvature affect the observed pattern; these prescribed beams are not a grating scattering solution.


(formula-7-4)=
### [7.4 — Kinematics and complex coefficients](7_4_Diffraction_Kinematics_and_Complex_Order_Coefficients.ipynb)

Two inputs meet at reconstruction: geometry supplies wavevectors, and the field model supplies coefficients.

$$
\begin{aligned}
\{d,\mathbf k_i,k_{\mathrm{out}}\}&\xrightarrow{\text{kinematics}}\{\mathbf k_m\},\\
\{\text{mask or material model},U_i\}&\xrightarrow{\text{prescribe or solve}}\{C_m(z_e)\},\\
\{C_m(z_e),\mathbf k_m\}
&\xrightarrow{\text{synthesis}}
U(x,z)=\sum_m C_m(z_e)e^{i[k_{x,m}x+k_{z,m}(z-z_e)]}.
\end{aligned}
$$

$C_m=|C_m|e^{i\arg C_m}$ contains amplitude and phase. Normal-power efficiencies also require the flux factor given in [0.6](#formula-0-6).

## 8 — Propagation


(formula-8-0)=
### [8.0 — The propagation question](8_0_Introduction_to_Field_Propagation.ipynb)

Hold the starting complex field fixed when comparing propagation methods.

$$
U_{\mathrm{ref}}
\xrightarrow{\mathcal P_s}
U_s
\xrightarrow{\text{observation}}
I_{\mathrm{rel},s}.
$$

Equal starting intensities do not guarantee equal propagated fields: starting phase is also required.


(formula-8-1)=
### [8.1 — Green response and a boundary representation](8_1_Physical_Pictures_Behind_Scalar_Field_Propagation.ipynb)

Use the notebook's normal directed into the propagation region. The boundary data must belong to one consistent outgoing solution.

$$
\begin{aligned}
R=|P-Q|&\xrightarrow{\text{free-space response}}
G(Q,P)=\frac{e^{ikR}}{4\pi R},\\
\{U(Q),h(Q)=\partial_{n_Q}U(Q)\}
&\xrightarrow{\text{Green boundary identity}}
U(P)=\int_S[U\partial_{n_Q}G-Gh]\,dA_Q.
\end{aligned}
$$

The identity is exact for compatible data. Kirchhoff's prescribed aperture/screen values for both field and slope are an additional approximation.


(formula-8-2)=
### [8.2 — Rayleigh–Sommerfeld I](8_2_Rayleigh_Sommerfeld_Propagation.ipynb)

For an infinite reference plane and an outgoing homogeneous half-space, choose a Green function that vanishes on the plane.

$$
\begin{aligned}
G_{\mathrm I}=g_P-g_{P'},\quad G_{\mathrm I}|_S=0
&\xrightarrow{\text{remove the unknown slope term}}
U_s=K_{\mathrm{RS1}}*U_{\mathrm{ref}},\\
K_{\mathrm{RS1}}(\boldsymbol\rho;s)
&=\frac{s\,e^{ikR}(1-ikR)}{2\pi R^3},
\qquad R=\sqrt{|\boldsymbol\rho|^2+s^2}.
\end{aligned}
$$

The convolution is two-dimensional. Equivalently, $\mathcal F_\perp\{K_{\mathrm{RS1}}\}=e^{is\sqrt{k^2-|\mathbf k_\perp|^2}}$: RS I and full ASM represent the same scalar half-space propagation problem.


(formula-8-3)=
### [8.3 — ASM, Fresnel and Fraunhofer](8_3_ASM_Fresnel_and_Fraunhofer_Propagation.ipynb)

Distinguish an exact representation change from a paraxial or far-field approximation.

$$
\begin{aligned}
U_{\mathrm{ref}}
&\xrightarrow{\mathcal F_\perp}\widetilde U_{\mathrm{ref}}
\xrightarrow{\times\,H}\widetilde U_s
\xrightarrow{\mathcal F_\perp^{-1}}U_s,\\
H_{\mathrm{ASM}}=e^{is\sqrt{k^2-|\mathbf k_\perp|^2}}
&\xrightarrow{\text{paraxial approximation}}
H_{\mathrm F}=e^{iks}e^{-is|\mathbf k_\perp|^2/(2k)},\\
U_{\mathrm F}(x,y;s)
&\xrightarrow{\text{far-field limit}}
U_{\mathrm{FF}}(x,y;s)\simeq
\frac{e^{iks}}{i\lambda s}e^{ik(x^2+y^2)/(2s)}
\widetilde U_{\mathrm{ref}}\!\left(\frac{kx}{s},\frac{ky}{s}\right).
\end{aligned}
$$

Fresnel requires paraxial content and small accumulated approximation error. Fraunhofer additionally requires $k(\xi^2+\eta^2)/(2s)\ll1$ over the contributing input. The last line uses a two-dimensional FT; one-dimensional prefactors differ.

(formula-9-1)=
### [9.1 — Define the complex exit field](9_1_Defining_the_Complex_Exit_Field.ipynb)

General thin-mask form first, followed by the special case of a periodic mask under one plane wave.

$$
\begin{aligned}
[a_t(x)e^{i\phi_t(x)}][A_i(x)e^{i\phi_i(x)}]
&\xrightarrow{\text{thin-mask product}}
U_{\mathrm{exit}}(x)=a_t(x)A_i(x)e^{i[\phi_t(x)+\phi_i(x)]},\\
t(x)=\sum_m t_me^{imGx},\quad U_i=A_ie^{ik_{x,i}x}
&\xrightarrow{\text{multiply}}
U_{\mathrm{exit}}(x)=\sum_m
\underbrace{A_it_m}_{C_m}e^{i\underbrace{(k_{x,i}+mG)}_{k_{x,m}}x}.
\end{aligned}
$$

Only the second row assumes periodic transmission and a single plane wave. The incident amplitude/phase reference is at the same plane as $t$. In a material solve, $C_m$ is determined by scattering and need not equal $A_it_m$.


(formula-9-2)=
### [9.2 — Sample fields and recover coefficients](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb)

For one periodic factor sampled over exactly one period, omit the duplicated endpoint.

$$
\begin{gathered}
v(x)=U_{\mathrm{exit}}(x)e^{-ik_{x,i}x},\\[0.5em]
x_j=x_a+j\Delta x,\qquad d=N\Delta x,\\[0.5em]
v(x_j)\xrightarrow{\text{FFT}}D_m
\xrightarrow{\text{scale and correct origin}}
\widehat C_m=\frac{D_m}{N}e^{-imGx_a},
\qquad f_m=\frac{m}{d}.
\end{gathered}
$$

Use signed FFT-bin labels. Sampling gives spacing $\Delta f=1/L$ and a Nyquist interval of width $1/\Delta x$, where $L=N\Delta x$. An FFT round trip recovers stored samples even when they alias an unresolved field. [Continuous-FT scaling](#formula-appendix-4) is different.

(formula-appendix-4)=
### [Appendix 4 — FS, FT, DTFT, DFT and FFT](Appendix_4_How_to_Use_FFTs_in_Python.ipynb)

These names refer to different mathematical inputs or, for FFT, an algorithm.

$$
\begin{aligned}
\text{periodic continuous }u(x)
&\overset{\text{FS analysis}}{\underset{\text{FS synthesis}}{\hspace{1.8em}\rightleftharpoons\hspace{1.8em}}}\{c_m\},\\
\text{continuous }u(x)
&\overset{\text{FT}}{\underset{\text{inverse FT}}{\hspace{1.8em}\rightleftharpoons\hspace{1.8em}}}\widetilde U_f(f_x),\\
\text{sequence }u[j]
&\xrightarrow{\text{DTFT}}\sum_{j\in\mathbb Z}u[j]e^{-i\theta j},\\
\{u[0],\ldots,u[N-1]\}
&\xrightarrow{\text{DFT computed by FFT}}
D_\ell=\sum_{j=0}^{N-1}u[j]e^{-i2\pi j\ell/N}.
\end{aligned}
$$

For $x_j=x_a+j\Delta x$, signed bins $f_\ell=\ell/(N\Delta x)$ give
$$
\widetilde U_f(f_\ell)\approx
\underbrace{\Delta x\,e^{-i2\pi f_\ell x_a}}_{\text{continuous-FT scale and origin}}D_\ell.
$$
This approximates the transform of the sampled window; omitted tails and unresolved structure still matter. Over one period, FS coefficients instead use $D_m/N$ with the same origin correction. With $x$ in nm, $f_x$ is in cycles/nm and $k_x=2\pi f_x$ is in rad/nm. A Gaussian example is
$e^{-(x/w)^2}\xrightarrow{\mathrm{FT}_f}w\sqrt{\pi}e^{-(\pi wf_x)^2}$, with $\operatorname{sinc}(u)=\sin(\pi u)/(\pi u)$ for rectangular-pulse examples.

(formula-companion-binary-masks)=
### [Binary amplitude and phase prescriptions](0_7_Ideal_Binary_Masks_vs_Physical_Gratings.ipynb)

A centered region occupies fraction $D_r$ of each period; the two regions have transmissions $t_r,t_g$.

$$
\begin{gathered}
t(x)=t_g+(t_r-t_g)R(x)
\xrightarrow{\text{FS projection}}
t_m=t_g\delta_{m0}+(t_r-t_g)D_r\operatorname{sinc}(mD_r)\\[0.7em]
\xrightarrow{\text{unit normal incidence}}C_m=t_m.
\end{gathered}
$$

Amplitude masks choose different magnitudes; ideal phase masks may use $t_{r,g}=e^{ik_0n_{r,g}h}$. That phase-only prescription does not include interface reflection or a full dielectric scattering solve.

For the phase prescription, $t$ includes the phase accumulated from $z=0$ to $z=-h$. With $U_i(0)=1$, the thin-field exit coefficient is $C_m^{\mathrm{thin}}(-h)=t_m$. The physical model instead has $C_m^{\mathrm{HBVP}}(z_{\mathrm{ref}})=F_m(z_{\mathrm{ref}})$; $A_m=|C_m|$ is a nonnegative magnitude.


(formula-companion-unit-cell)=
### [Two-dimensional cell coefficients](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb#cell-coefficients-9-2)

For a centered rectangular opening under unit-amplitude normal incidence:

\begin{aligned}
t(x,y)&\xrightarrow{\text{cell projection}}
t_{mp}=D_xD_y\operatorname{sinc}(mD_x)\operatorname{sinc}(pD_y)
=C_{mp},\\
t(x-x_c,y-y_c)&\xrightarrow{\text{translation}}
t'_{mp}=t_{mp}e^{-i(mG_xx_c+pG_yy_c)}.
\end{aligned}
$$

Nonseparable cells use the same double integral or sampled FFT2 projection; their coefficients need not factor into x and y products. The source explanation and this sheet use $(m,p)$. Backend identifiers may retain `n` for the second order; that identifier corresponds to $p$, not refractive index.

(formula-companion-ewald)=
### [Outgoing vector and momentum transfer](5_1_Three_Dimensional_Orders_and_Conical_Diffraction.ipynb)

For the same surrounding medium on each side of this geometrical construction:

$$
\begin{gathered}
\mathbf k_m
\overset{-\mathbf k_i}{\underset{+\mathbf k_i}{\hspace{1em}\rightleftharpoons\hspace{1em}}}
\Delta\mathbf k_m=\mathbf k_m-\mathbf k_i,
\qquad
\\[0.5em]

|\Delta\mathbf k_m+\mathbf k_i|^2=k^2.

(formula-companion-exit-spectra)=
### [Two routes to the same exit spectrum](9_1_Defining_the_Complex_Exit_Field.ipynb#exit-field-fourier-representations)

For an infinite periodic thin mask illuminated by one plane wave, the Fourier shift theorem gives identical endpoints.

$$
\begin{aligned}
t(x)
&\xrightarrow{\times\,A_ie^{ik_{x,i}x}}U_{\mathrm{exit}}(x)
\xrightarrow{\mathcal F_x}\widetilde U_{\mathrm{exit}}(k_x),\\
t(x)
&\xrightarrow{\mathcal F_x}\widetilde t(k_x)
\xrightarrow{\text{shift and scale}}
A_i\widetilde t(k_x-k_{x,i})
=2\pi\sum_m C_m\delta(k_x-k_{x,i}-mG).
\end{aligned}
$$

For a one-cell calculation,
$$
\underbrace{U_{\mathrm{exit}}(x)}_{\text{Bloch periodic}}
\xrightarrow{\times\,e^{-ik_{x,i}x}}
\underbrace{v(x)=A_it(x)}_{v(x+d)=v(x)}
\xrightarrow{\text{cell Fourier analysis}}C_m.
$$
The shift changes component locations from $mG$ to $k_{x,i}+mG$; it does not add a scalar phase to each spectral coefficient.
Ordinary one-cell PBC for the full field agrees only when $e^{ik_{x,i}d}=1$ (apart from a zero field).
At the 9.1 defaults, $d=2\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, $\theta_i=30^\circ$ give $k_{x,i}d=\pi$: adjacent cells differ by a sign.
The displayed $10\,\mu\mathrm m$ window does not itself truncate the illumination. Physically truncating it broadens the spectrum.
Wrapped and unwrapped phase are two displays of the same complex field, not different boundary conditions.

(formula-companion-arrival-geometry)=
### [Finite sums and the order condition](2_3_Allowed_Diffraction_Orders.ipynb)

For N equally spaced contributions with equal weights, use the relative phase advance between neighboring contributions.

$$
\Delta=(k_{x,i}-k_x)d
\xrightarrow{\text{sum }N\text{ contributions}}
S_N=\sum_{\ell=0}^{N-1}e^{i\ell\Delta}
=e^{i(N-1)\Delta/2}\frac{\sin(N\Delta/2)}{\sin(\Delta/2)}.
$$

At $\Delta=2\pi m$, take the limit $S_N=N$. Finite arrays retain finite-width peaks; the infinite periodic limit gives discrete order support.

For mutually incoherent source points, propagate each coherent field separately before adding weighted intensities.

s&\xrightarrow{\text{grating}} \{C_{s,mp},\mathbf k_{\perp,s,mp}\}

(formula-companion-scalar-propagation)=
### [Longitudinal dispersion and approximation error](8_3_ASM_Fresnel_and_Fraunhofer_Propagation.ipynb)

For positive downstream distance, Fourier transformation turns homogeneous Helmholtz propagation into independent ordinary differential equations.

$$
\begin{aligned}
(\nabla^2+k^2)U=0
&\xrightarrow{\mathcal F_\perp}
\partial_s^2\widetilde U+
(k^2-|\mathbf k_\perp|^2)\widetilde U=0
\xrightarrow{\text{outgoing condition}}
\widetilde U_s=e^{i\beta s}\widetilde U_0,\\
\Delta\phi&=
s\left[k-\frac{|\mathbf k_\perp|^2}{2k}
-\sqrt{k^2-|\mathbf k_\perp|^2}\right]
\simeq \frac{s|\mathbf k_\perp|^4}{8k^3}.
\end{aligned}
$$

The last line is the Fresnel-minus-ASM phase error for propagating paraxial content. Whether it matters depends on distance and on the input's spectral support.

\{x_j,U_i(x_j)\}
\xrightarrow{\text{illustrative spreading response}}
