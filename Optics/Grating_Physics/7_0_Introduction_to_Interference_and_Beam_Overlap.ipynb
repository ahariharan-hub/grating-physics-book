## One experiment, three questions

Split a laser beam into two paths and direct both toward a detector. Ask:

Do the fields reach the same detector locations? Does their relative phase remain correlated during the measurement? How does relative phase vary across the detector?

These questions describe different aspects of interference:

- **Spatial overlap:** both fields must be appreciable in the same region for a substantial interference signal there.
- **Mutual coherence:** the averaged correlation of the two fields determines the surviving interference term.
- **Fringe geometry:** spatial variation of relative phase sets the orientation and spacing of the pattern.

A fixed phase shift can move fringes while leaving mutual-coherence magnitude unchanged. Unequal beam strengths, limited overlap or detector averaging can reduce measured contrast, so contrast alone is not a direct measurement of intrinsic coherence.

## The common physical idea

We add fields before calculating intensity:

$$U=U_1+U_2.$$

For stable monochromatic scalar fields with a common polarization, the relative intensity is

$$I=|U_1+U_2|^2=I_1+I_2+2\operatorname{Re}(U_1^*U_2).$$

The last term can be positive or negative. It is the source of interference. When the fields fluctuate during the exposure, the measured cross term involves their averaged correlation instead. Notebook 7.1 develops that connection after you observe the experiment.
