# 3.3 Phase cycles and path differences

One extra wavelength of travel gives one extra phase cycle. We will use this connection to express the grating phase rule as a path difference. Tilting the incoming wave changes that difference, so the signs must follow the directions in the drawing.

:::{note} Surrounding medium

 The incident and output media have the same refractive index in these examples. Refraction between unequal exterior media is outside this model.
:::

The discussion builds on [3.2: phase continuity](3_2_Selected_Orders_and_Phase_Continuity.ipynb). Recall $(k_{x,m}-k_{x,i})d=2\pi m$. The two openings A and B are separated along $x$ by one period $d$, with the grating at $z=0$ and transmission into $z<0$.

## 1. Connect phase cycles to path length

One wavelength of extra travel corresponds to one full phase cycle:

$$\text{extra cycles}=\frac{\Delta L}{\lambda},\qquad
\text{extra phase}=k\Delta L,\qquad k=\frac{2\pi}{\lambda}.$$

**Start at normal incidence:** set $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, and click **Normal incidence** in the view below. The controls belong to this lesson’s explorer; earlier lessons have independent controls. Turn on **Compare non-integer values** and set the comparison value to $0.5$.

- **Left:** the same continuous outgoing plane wave, with $\mathbf{k}_{\mathrm{out}}$ and its projection $k_{\mathrm{out},x}$. Here A and B are at the grating surface, one period apart. Their phase difference is still $k_{\mathrm{out},x}d$.
- **Right:** paths through those same two openings. The green incoming paths begin on a common incident wavefront. The blue outgoing paths end on a common plane perpendicular to their direction. The grey dashed lines join those starting and ending points. This is the parallel-ray geometry of a far-field direction. The path-projection construction follows the grating-equation explanation in [\[1, §2.1, Fig. 2-2\]](#ref-3-3-palmer), [\[2, §2.1, Fig. 2-2\]](#ref-3-3-newport-equation).

A and B sample neighbouring cells of the ideal repeating grating; they are not a model containing only two openings. The route lines show phase/path projections, not the widths of light beams. Likewise, the common wavefront bars are reference planes, not stops that truncate the field. [\[1, §2.1\]](#ref-3-3-palmer), [\[2, §2.1\]](#ref-3-3-newport-equation)


At normal incidence the incoming lengths are equal. The orange segment is the extra outgoing distance for one route. The orange dashed line and right-angle mark show how that distance is obtained by projecting $d$ along the outgoing direction.

We define the **signed** path difference as

$$\Delta L=(\text{total route through A})-(\text{total route through B}).$$

For normal incidence, $\Delta L=d\sin\theta_{\mathrm{out}}$. Since $k_{\mathrm{out},x}=k\sin\theta_{\mathrm{out}}$,

$$\boxed{k\Delta L=k\,d\sin\theta_{\mathrm{out}}=k_{\mathrm{out},x}d.}$$

The left picture compares the phase of one plane wave at A and B. The right compares the travel phases of two contributions on the outgoing reference plane. With the stated sign convention, they give the same cycle count at normal incidence.

**Change just the comparison value:**

| Value | Extra path $\Delta L$ | Phase $k\Delta L$ | Cycles $\Delta L/\lambda$ |
| --- | --- | --- | --- |
| $0.5$ | $1\,\mu\mathrm m=\lambda/2$ | $\pi$ | $0.5$: mismatch |
| $1$ | $2\,\mu\mathrm m=\lambda$ | $2\pi$ | $1$: agreement |
| $-1$ | $-2\,\mu\mathrm m=-\lambda$ | $-2\pi$ | $-1$: agreement |

For the negative direction, the route through B is longer. The highlighted segment has a positive geometric length; the readout carries the sign of A minus B. **An integer number of extra wavelengths is the same condition as an integer number of extra phase cycles.**

### Then tilt the illumination

Keep $d=4\,\mu\mathrm m$ and $\lambda=2\,\mu\mathrm m$. Set $\theta_i=30^\circ$, choose $m=0$ in **Show order**, and click **Use selected order**.

The incoming lengths now differ too. The diagram marks the extra incoming segment in green and the extra outgoing segment in orange. Define their signed projections as

$$\ell_i=d\sin\theta_i,\qquad \ell_{\mathrm{out}}=d\sin\theta_{\mathrm{out}}.$$

For **total route A minus total route B**, the incoming contribution is $-\ell_i$ and the outgoing contribution is $+\ell_{\mathrm{out}}$. Thus

$$\Delta L=\ell_{\mathrm{out}}-\ell_i
=d(\sin\theta_{\mathrm{out}}-\sin\theta_i).$$

Multiplying by $k$ recovers our existing phase condition:

$$\boxed{k\Delta L=(k_{\mathrm{out},x}-k_{x,i})d=\Delta\phi_{\mathrm{out}}-\Delta\phi_i.}$$

In this zero-order example, $\ell_i=\ell_{\mathrm{out}}=2\,\mu\mathrm m$: the two contributions cancel in the total path difference, so $\Delta L=0$. Next select $m=-1$: the outgoing direction is normal, $\ell_{\mathrm{out}}=0$, and $\Delta L=-2\,\mu\mathrm m=-\lambda$. The paths still agree in phase, now with one negative full cycle of difference.

We can therefore express the same order condition in three ways:

$$\boxed{\Delta L=m\lambda
\quad\Longleftrightarrow\quad
\Delta\phi_{\mathrm{out}}-\Delta\phi_i=2\pi m
\quad\Longleftrightarrow\quad
k_{x,m}=k_{x,i}+m\frac{2\pi}{d}.}$$

This gives the conventional grating equation in our signed-angle convention:

$$\boxed{nd(\sin\theta_m-\sin\theta_i)=m\lambda_0}
\quad\Longleftrightarrow\quad d(\sin\theta_m-\sin\theta_i)=m\lambda.$$

Here $n$ is the common exterior index and $\lambda=\lambda_0/n$. For different exterior indices, replace the left-hand side by $d(n_t\sin\theta_m-n_i\sin\theta_i)$.

The Newport path-length explanation [\[1, §2.1\]](#ref-3-3-palmer), [\[2, §2.1\]](#ref-3-3-newport-equation) uses the same integer-wavelength principle, with its own angle convention. Harvey and Pfisterer also explain how angle and order-sign conventions affect the written grating equation [\[3, §2, Eqs. (1)–(3)\]](#ref-3-3-harvey-pfisterer). Here the wavelength is in the common medium on both sides.
- These diagrams explain phase agreement and order directions; they do not calculate order strengths.

### Small calculation 1: convert extra path into phase

Start with $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, normal incidence, and an outgoing angle of $30^\circ$. These are the path view's $m=+1$ settings. Compute both signed projections, subtract them in the **route A minus route B** convention, and compare $k\Delta L$ with $(k_{\mathrm{out},x}-k_{x,i})d$.

This cell begins with angles rather than an order label, so the path lengths come directly from the geometry. **Predict, then try:** keep the outgoing angle at $30^\circ$ and change only `theta_i_deg` to `30.0`. Then keep that incidence and set `theta_out_deg = 0.0`.
- Compare with the two tilted-incidence examples above. 

```python
wavelength, d = 2.0, 4.0  # micrometres
theta_i_deg, theta_out_deg = 0.0, 30.0
k = 2*np.pi / wavelength
sin_i, sin_out = np.sin(np.deg2rad([theta_i_deg, theta_out_deg]))
ell_i = d*sin_i
ell_out = d*sin_out
delta_L = ell_out - ell_i  # total route A minus total route B
phase_from_path = k*delta_L
k_x_i, k_out_x = k*sin_i, k*sin_out
phase_from_components = (k_out_x-k_x_i)*d
print('Incoming projection, outgoing projection, total difference (um):',
      np.round([ell_i, ell_out, delta_L], 3))
print('Phase/pi from path and from k_out_x - k_x_i:',
      np.round(np.array([phase_from_path, phase_from_components])/np.pi, 3))
print(f'Extra wavelengths = extra phase cycles: {delta_L/wavelength:+.3f}')
```

:::{dropdown} Answer for the stated starting values and changes

With normal incidence and $\theta_{\mathrm{out}}=30^\circ$, the projections are $\ell_i=0$ and $\ell_{\mathrm{out}}=2\,\mu\mathrm m$. Thus $\Delta L=\lambda$ and both phase calculations give $2\pi$: one cycle.

With both angles at $30^\circ$, the two projections are equal and $\Delta L=0$, corresponding to $m=0$. Keeping incidence at $30^\circ$ and setting the outgoing angle to zero gives $\Delta L=-2\,\mu\mathrm m=-\lambda$ and a phase difference of $-2\pi$, corresponding to $m=-1$.

:::

## Final checkpoint

1. **Can the outgoing phase change be half a cycle and still belong to an order?**

   Yes—if its difference from the incident phase change is an integer number of cycles.

2. **Does adding more openings move the order directions?**

   No—with wavelength, spacing and incidence fixed, the order directions stay fixed; the finite-array peaks narrow. [\[3, §1, Figs. 1–2\]](#ref-3-3-harvey-pfisterer)

3. **Does a white point in the signed-field snapshot necessarily mean zero intensity?**

   No—it may simply be an instantaneous zero. The distinction between the instantaneous field and time-averaged intensity is discussed in [\[4, §3.4\]](#ref-3-3-delft-wavefield).
