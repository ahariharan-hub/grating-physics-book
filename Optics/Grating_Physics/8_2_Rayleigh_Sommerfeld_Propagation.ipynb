# 8.2 Rayleigh-Sommerfeld propagation and the Weyl bridge

**If we know the field on a plane, must we also guess its normal slope?**

In [8.1](8_1_Physical_Pictures_Behind_Scalar_Field_Propagation.ipynb), Kirchhoff's aperture prescription supplied both. **Rayleigh-Sommerfeld (RS) changes the Green function so that only one kind of boundary data is needed.** We will follow RS I from that choice to a numerical field and intensity pattern, then reconstruct a spherical wave from its plane-wave components.

Keep $e^{-i\omega t}$, $k=2\pi/\lambda$, and propagation coordinates $(x,y,s)$ with $s=-z>0$. The reference plane $S$ is at $s=0$; its normal points into positive $s$. Distances in the explorers are in wavelengths. The medium is homogeneous, isotropic and lossless.

## 1. The RS idea: remove the unknown term

Recall the boundary representation, with $h=\partial_nU$:

$$U(P)=\int_S\left[U(Q)\partial_{n_Q}G(Q,P)-G(Q,P)h(Q)\right]dA_Q.$$

Suppose $U$ is prescribed across the plane but $h$ is unknown. **Choose a Green function that is zero on that plane.** Then the term containing $h$ disappears. We have changed the weights, while keeping the Helmholtz equation and outgoing condition. [1, section 18.3](#ref-1)

### RS I: prescribe the field

Place the mathematical probe at $P=(x,y,s)$ and its mirror at $P'=(x,y,-s)$. Their distances to every $Q$ on $S$ are equal, so subtract their responses:

$$G_{\rm I}(Q,P)=g_P(Q)-g_{P'}(Q),\qquad G_{\rm I}|_S=0,
\qquad g_P(Q)=\frac{e^{ik|Q-P|}}{4\pi|Q-P|}.$$

The responses cancel on $S$; their normal derivatives add. The remaining integral needs only the prescribed field $U_{\mathrm{ref}}$:

$$\boxed{U(P)=\int_S U_{\mathrm{ref}}(Q)\partial_{n_Q}G_{\rm I}(Q,P)\,dA_Q.}$$

These are auxiliary probes, not physical emitters or reflections. Choose **RS I** below: the blue $|G|$ curve vanishes across the entire plane. Switch briefly to **Kirchhoff** to see both weights remain. The dashed orange curve is $|\partial_nG|/k$, scaled for comparable units; these are weight magnitudes, not the propagated field.

### RS II: the short counterpart

If the normal slope is prescribed instead, **add** the mirror responses: $G_{\rm II}=g_P+g_{P'}$. Now $\partial_nG_{\rm II}=0$ on $S$, leaving $U(P)=-\int_S G_{\rm II}h\,dA$. Choose **RS II** to see the orange weight disappear.

| RS choice | Independently prescribe on the whole plane | Determined by the outgoing solution |
| --- | --- | --- |
| **RS I** | Field $U$ | Normal slope $h$ |
| RS II | Normal slope $h$ | Field $U$ |

For an aperture model, RS I may prescribe $U_i$ in opening $A$ and zero in opaque region $B$; it does not additionally set $h=0$ there. Consistent propagation of those data still does not establish that they accurately model a real screen. [13, pp. 8-11](#ref-13)

Here $U_{\mathrm{ref}}$ names prescribed data on a general reference plane. If that plane is the grating exit, $U_{\mathrm{ref}}=U_{\mathrm{exit}}$; if it is downstream, propagate the exit field to that plane first.

**Notation:** use $U_{\mathrm{exit}}$ at the exit plane, $U_{\mathrm{ref}}$ at a general reference plane, and $C_m$ (or $C_{mn}$) for field coefficients. [Shared convention](0_0_Foundations_of_Scalar_Optics.ipynb#exit-field-notation).

## 2. RS I becomes a propagation kernel

For our inward normal on the plane, the surviving derivative is

$$K_{\rm RS1}(\boldsymbol\rho-\boldsymbol\rho';s)
=\partial_{n_Q}G_{\rm I}=-2\partial_sg
=\frac{s\,e^{ikR}(1-ikR)}{2\pi R^3},\qquad
R=\sqrt{|\boldsymbol\rho-\boldsymbol\rho'|^2+s^2}.$$

Here $\boldsymbol\rho=(x,y)$ and $\boldsymbol\rho'=(\xi,\eta)$ locate the observation point and boundary patch transversely. The $s$ derivative holds their transverse separation fixed. **This is the full RS I kernel**; no large-$kR$ or paraxial reduction has been made. [1, sections 18.3-18.5](#ref-1), [11, Eq. (8)](#ref-11)

For a small patch,

$$\Delta U_j(P)=U_{\mathrm{ref}}(Q_j)K_{\rm RS1}(P,Q_j)\Delta A_j,\qquad
U(P)\approx\sum_j\Delta U_j(P),\qquad I(P)\propto|U(P)|^2.$$

The starting field sets an arrow; the kernel changes its length and phase; patch area sets how much surface it represents. Add the complex arrows first, then square the final length. That last step keeps the interference.

### Predict, then move a control

1. **Symmetry:** set **Phase slope = 0** and compare opposite positions on the screen.
2. **Distance:** move the screen. Relative path phases change, so the pattern evolves rather than simply dimming uniformly.
3. **Phase:** change the slope while keeping input amplitude fixed. Equal input intensities can produce different interference patterns. A rigid shift is a paraxial interpretation, not a general property of this full RS calculation.

A detector records intensity, not the individual arrows. Predicting an actual experiment also requires a suitable screen-produced boundary field and detector model.


## 4. Weyl: the bridge from RS patches to plane waves

**RS adds contributions from boundary patches. Can we organize the same calculation by wave direction instead?** Direct patch-to-point sums can become cumbersome on a large observation grid. Weyl supplies the bridge.

Start with the spherical Green response already used by RS. For $Q=(\xi,\eta,0)$ and $P=(x,y,s)$, with $s>0$, take the **two-dimensional Fourier transform over transverse separation** $(x-\xi,y-\eta)$. Distance $s$ remains a parameter. Weyl gives the pair

$$\underbrace{g(P,Q)=\frac{e^{ikR}}{4\pi R}}_{\text{real space}}
\quad\xleftrightarrow{\;\mathcal F_\perp\;}\quad
\underbrace{\widetilde g_s(k_x,k_y)=\frac{i}{2\kappa}e^{i\kappa s}}_{\text{Fourier space}},$$

where $R=|P-Q|$ and $\kappa=\sqrt{k^2-k_x^2-k_y^2}$. **The tilde identifies the transformed Green response.** The spherical response and its continuous plane-wave spectrum describe the same field. Choose nonnegative real $\kappa$ for propagating components and positive imaginary $\kappa$ for evanescent ones. [1, section 18.6](#ref-1), [7](#ref-7), [9, Eqs. (35.1.9)-(35.1.11)](#ref-9)

**Notation.** Here **transverse** means perpendicular to the propagation axis $s$: $(k_x,k_y)$ lie in the reference plane. **Longitudinal** means along $s$: $\kappa$ (so the global $k_z=-\kappa$). Symbols such as $k_\perp$ and $k_\parallel$ depend on whether an author takes the axis or the plane as reference.

Now use Section 2's real-space RS operation, $K_{\rm RS1}=-2\partial_sg$. In Fourier space it acts on $\widetilde g_s$. At fixed $k_x,k_y$, differentiation multiplies each component by $i\kappa$:

$$\begin{aligned}
\widetilde K_{\rm RS1}(k_x,k_y;s)&=-2\partial_s\widetilde g_s(k_x,k_y)\\
&=-2\partial_s\left(\frac{i}{2\kappa}e^{i\kappa s}\right)
=e^{i\kappa s}.
\end{aligned}$$

**This is the full angular-spectrum multiplier**, $H_{\rm ASM}(k_x,k_y;s)=e^{i\kappa s}$. The point-source response $g$ and planar propagation kernel $K_{\rm RS1}$ have different weights; Weyl shows how the RS derivative connects them.

Fourier-transform the prescribed field on the plane:

$$\widetilde U_{\mathrm{ref}}(k_x,k_y)=\iint U_{\mathrm{ref}}(\xi,\eta)
 e^{-i(k_x\xi+k_y\eta)}\,d\xi\,d\eta.$$

Multiply each component by $H_{\rm ASM}=e^{i\kappa s}$, then add them at $P$:

$$U(P)=\frac{1}{(2\pi)^2}\iint\widetilde U_{\mathrm{ref}}(k_x,k_y)
 e^{i(k_xx+k_yy+\kappa s)}\,dk_x\,dk_y.$$

Here $k_x,k_y$ are transverse wavevector components in radians per unit length; the tilde means Fourier transform. The inverse factor $1/(2\pi)^2$ matches that convention. Recall [Goodman's notation in 5.2](5_2_Direction_Cosines_and_Spatial_Frequency_Coordinates.ipynb): $k_x=2\pi f_x$, $k_y=2\pi f_y$.

Full RS I and full angular spectrum give the same outgoing field for the same plane data, including evanescent components. No Fresnel approximation was needed. FFTs can accelerate either formulation; sampling still matters. [11](#ref-11)

The accounting changes from patches to components. The light does not need to stop and change clothes. Section 5 tests this identity by reconstructing a known spherical field.


## 5. Build a spherical wave from plane-wave components

**Can flat waves add up to a curved wave?** Keep the source fixed and build the sum on the left. The spherical field on the right is our target:

$$U_{\rm sphere}(P)=\frac{e^{ik|P-P_0|}}{4\pi|P-P_0|}.$$

Move **Build the sum** from left to right. Start with one plane wave: its phase appears as straight stripes. With more weighted components, interference changes those stripes. With a dense enough sampling of Weyl's spectrum, they follow the curved stripes of the spherical target.

Matching colours mean matching phase, not matching intensity. Both pictures show the same cross-section of three-dimensional space, with the source at $P_0=(0,0,-1.5)$ in $(x,y,s)$ coordinates. All lengths are in wavelengths.

**What should click?** No individual component bends. Each still has flat phase fronts; the curved fronts belong to their **sum**. The source has not moved and the light has not passed through another optical element. We have changed how finely we represent the same field.

The early sums are deliberately incomplete approximations. Each later step refines the weighted sampling of the continuous Weyl integral; the last includes both propagating and evanescent components. Weyl supplies the amplitudes and phases that make the reconstruction work. An arbitrary collection of plane waves would produce a different field.
