## 1. Oscillation and phase

Along one axis, write the real field as $u(x,t)$:

$x$ specifies position; $t$ specifies time. $u$ is a signed field value, which can be positive, zero or negative. Position and field value are different quantities and can have different units.

At the origin, $\mathbf k\cdot\mathbf r=0$. With our chosen $\phi_0=0$, the plane-wave formula becomes a single oscillation. Stay at $x=0$ and let time run:

$$u(0,t)=A\cos\phi(t),\qquad \phi(t)=-\frac{2\pi t}{T}.$$

- **Amplitude $A>0$** is the largest deviation from zero; the reading ranges from $-A$ to $+A$.
- **Period $T$** is the time for one complete repeat of the oscillation.
- **Phase $\phi$** is the angle inside the cosine. It labels where we are in the repeating cycle. It is not a position or a field value.

A complete turn is $360^\circ=2\pi$ **radians**; half a turn is $180^\circ=\pi$ radians. Cosine repeats when its angle changes by $2\pi$. Here the minus sign makes phase decrease with time; we keep this convention throughout. Starting at phase zero means starting at the positive maximum.

| Time $t/T$ | Phase $\phi$ | Reading $u/A$ |
| --- | --- | --- |
| $0$ | $0$ | $+1$ |
| $1/4$ | $-\pi/2$ | $0$ |
| $1/2$ | $-\pi$ | $-1$ |
| $3/4$ | $-3\pi/2$ | $0$ |
| $1$ | $-2\pi$ | $+1$ |

- Phases separated by a whole number of cycles describe the same stage of oscillation: $0$ and $2\pi$ are equivalent in this sense.
- But **equal instantaneous readings do not necessarily mean equal phase**.
- The two zero readings in the table occur at different stages: one heads toward the negative minimum, the other toward the positive maximum.

The plots show the **signed field** $u/A$. Here a field “reading” means a value in the model, not a camera pixel value. Intensity describes energy flow per area and is nonnegative in the usual beam-intensity convention. For the single travelling wave used here in a fixed, simple linear medium, its time-averaged intensity is proportional to $A^2$. A negative field reading does not mean negative intensity.

**Experimental connection.** An ordinary camera averages over the optical oscillations; it does not photograph these moving carrier crests. A single ideal plane wave has uniform intensity even though its signed-field picture has alternating bands. A wavelength measurement (converted to the wavelength in the medium) and a measured propagation direction determine $\mathbf k=(2\pi/\lambda)\hat{\mathbf s}$, where $\hat{\mathbf s}$ is the unit direction vector. The explorer explains the field geometry behind those measurements.

- The finite line segment is a plotting window, not the end of the wave.
- This ideal plane wave also has no finite transverse beam boundary.
- A collimated laser can approximate it locally where its amplitude is nearly uniform and its wavefronts are nearly flat; the laser’s finite beam profile and spreading are not calculated here. [\[1\]](#ref-1-1-gaussian) 

**Predict:** at $t=T/4$ and $t=3T/4$, do the equal zero readings tell you what the field will do next? **Try:** use the table and evaluate the cosine a little later than each time. **Explain:** what information does phase provide that a single reading lacks?

:::{dropdown} Answer and explanation

Just after $T/4$, the reading becomes negative; just after $3T/4$, it becomes positive. Their phases differ by $\pi$, not by a whole $2\pi$ cycle. Phase identifies the stage in the oscillation, including whether the next reading will rise or fall under our time convention.

:::

### Calculate it: two zero readings, different phases

`np.array([...])` stores several numbers in one array. Arithmetic and `np.cos` operate on every entry, so we can calculate all five times together. NumPy's trigonometric functions expect **radians**; `np.pi` supplies $\pi$. Here `phase = -2 * np.pi * tau` is exactly $\phi=-2\pi t/T$.

`np.column_stack` places arrays side by side in a table; `np.round(..., 3)` rounds only the displayed values. The readings should be $1,0,-1,0,1$, with different phases at the two zero crossings.

**Explorer match:** in the 1D explorer introduced next, use wavelength 2, Right, and watch the orange marker at $x=0$ at the listed values of $t/T$.

**Predict and rerun:** replace the times with `np.array([0.26, 0.76])`. Just after each zero crossing, which reading will be negative and which positive?

```python
A = 1.0
tau = np.array([0.0, 0.25, 0.5, 0.75, 1.0])  # normalized time t/T
phase = -2 * np.pi * tau
field = A * np.cos(phase)
print("t/T     phase (rad)     u")
print(np.round(np.column_stack((tau, phase, field)), 3))
```

## 2. Wavelength and wavenumber

Now **freeze time** and walk through a repeating wave. The **wavelength $\lambda$** is the distance from one crest to the next. It is a distance in the medium carrying the wave, not a time interval. A smooth sinusoidal wave accumulates phase uniformly as we walk along its propagation direction.

Walking one wavelength advances the spatial phase by one cycle, $2\pi$. Therefore its phase advance per unit distance is

$$k\lambda=2\pi\qquad\Longrightarrow\qquad
\boxed{k=\frac{2\pi}{\lambda}}.$$

This positive number $k$ is the **angular wavenumber**. Its units are **radians per length unit**: for example, rad/m if $\lambda$ is in metres. Because radians are dimensionless, you will also see the units written as $\mathrm{m}^{-1}$. In contrast, $1/\lambda$ counts *cycles* per metre.

**Worked example.** Let $\lambda=4$ length units. Then $k=2\pi/4=\pi/2\approx1.571$ rad per length unit. Starting at a crest, walking 1 unit along the propagation direction advances phase by $\pi/2$ and reaches a zero crossing; 2 units advances it by $\pi$ and reaches a trough; 4 units completes $2\pi$ and reaches the next crest.
- These are phase changes at one frozen time.

### Let time run: which way does the wave travel?

The positive number $k$ specifies spacing. To describe travel along either direction of the $x$ axis, use a **signed component** $k_x$: choose $k_x=+k$ for rightward travel and $k_x=-k$ for leftward travel with the convention below. Choose phase zero at $x=0,t=0$:

$$u(x,t)=A\cos(k_x x-\omega t),\qquad
\omega=\frac{2\pi}{T}>0.$$

The symbol $\omega$ (omega) is the **angular frequency**, in radians per time unit. The term $k_x x$ changes phase with position, while $-\omega t$ changes it with time. At $x=0$, this is exactly the oscillation in Section 1.

To follow one crest, keep its phase unchanged. Write $\Delta x$ for its change in position during a time interval $\Delta t$:

$$k_x\Delta x-\omega\Delta t=0
\quad\Longrightarrow\quad
\frac{\Delta x}{\Delta t}=\frac{\omega}{k_x}.$$

Since $\omega>0$, positive $k_x$ moves the crest toward $+x$ and negative $k_x$ toward $-x$. The **phase speed**, the positive speed of a crest, is $c=\omega/k=\lambda/T$. A snapshot alone need not reveal the direction: at $t=0$, $\cos(kx)=\cos(-kx)$, but the subsequent motion differs. For more worked 1D wave examples, see [\[2\]](#ref-1-1-openstax-waves).

### Reading the controls

All explorers use **$A=1$ and phase speed $c=1$ length unit per time unit**. The explorer calls the model phase speed $c$; it is the quantity denoted $v$ in [Chapter 0](0_0_Foundations_of_Scalar_Optics.ipynb), not the vacuum speed of light used there in $v=c/n$. Setting it to 1 is a choice of units. Wavelength is measured **in the medium**.
- Consequently $T=\lambda/c$ and $\omega=ck$.

Playback advances in **normalized time** $\tau=t/T$, with $t=\tau T$ and $\omega t=2\pi\tau$. Within each explorer, a cycle has the same nominal on-screen duration for every wavelength, although the physical period $T$ changes. Changing wavelength holds $t/T$ fixed and updates physical time $t$; the readouts show both. For example, at $t/T=1/4$, changing $\lambda$ from 4 to 2 changes $T$ from 4 to 2 and $t$ from 1 to 0.5 time units.

In the spatial plot below, the horizontal axis is **position $x$**, and the vertical axis is **field value $u/A$**. The orange marker samples the fixed location $x=0$; it is not a particle riding a crest.

Halving wavelength halves crest spacing and doubles $k$ from $\pi/2$ to $\pi$ rad per length unit. It does not change the peak field value. With $c=1$, the physical period also halves, even though playback uses the same on-screen cycle duration. Reversing direction changes the sign of $k_x$ and reverses crest motion, while $k$ stays positive and unchanged. :::

### Calculate it: a snapshot along x

Use the worked example $\lambda=4$. `direction` is $+1$ for Right or $-1$ for Left; it sets signed $k_x$, while $k$ and $\omega=ck$ stay positive. The array `x` lists the positions sampled at one time. Each row of the output gives a position and its field value.

**Explorer match:** wavelength 4, Right, and $t/T=0$ in the 1D explorer. Expect $k=k_x=\pi/2$, $T=4$, and readings $1,0,-1,1$ at $x=0,1,2,4$. The calculation uses physical time `t = tau * T`, just as the explorer does.

**Predict and rerun:** change `tau` to `0.25`. Which sampled position becomes a crest after a quarter-period of rightward travel? Edit the parameter and rerun; the explorer itself will not move.

```python
A = 1.0
wavelength = 4.0
c = 1.0
direction = +1  # use -1 for Left
tau = 0.0
x = np.array([0.0, 1.0, 2.0, 4.0])
k = 2 * np.pi / wavelength
kx = direction * k
T = wavelength / c
omega = c * k
t = tau * T
field = A * np.cos(kx * x - omega * t)
print("k, kx (rad/length); T, t (time):", np.round([k, kx, T, t], 3))
print("x      u")
print(np.round(np.column_stack((x, field)), 3))
```
