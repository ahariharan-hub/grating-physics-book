**We know the field leaving a grating. What field reaches the next plane, a collecting lens, or a detector?**

The grating has done its work; the camera is still waiting downstream. Chapters 5-7 established outgoing wavevectors, coordinate representations, interference and complex order coefficients. Chapter 8 follows that known field through space.

## Motivation and prerequisites

Keep two questions distinct: **what field does the grating produce, and how does that field propagate?** A prescribed transmission model or a scattering calculation supplies the starting field. Propagation determines how it continues through the surrounding medium. Predicting an image then requires the collecting optics as well.

Review [5.2: coordinate representations](5_2_Direction_Cosines_and_Spatial_Frequency_Coordinates.ipynb), [7.2: order interference](7_2_Interference_of_Diffraction_Orders_from_a_Grating.ipynb) and [7.4: complex coefficients](7_4_Diffraction_Kinematics_and_Complex_Order_Coefficients.ipynb). Bring complex amplitude, phase and wavevector; we will introduce the propagation mathematics as it becomes useful.

We use scalar fields in a homogeneous, isotropic, lossless medium, with time dependence $e^{-i\omega t}$. Transmission is toward $-z$, with positive propagation distance $s=-z$. The in-medium wavelength is $\lambda=\lambda_0/n$, where $\lambda_0$ is vacuum wavelength and $n$ is the surrounding refractive index.

## A brief historical thread

Huygens (1690) pictured advancing wavelets; Young (1802) and Fresnel (1818) made interference central, while Fraunhofer (1822) measured angular diffraction patterns. Green, Helmholtz and Kirchhoff supplied boundary mathematics; Rayleigh and Sommerfeld developed alternative formulations. Weyl (1919) connected spherical responses to plane-wave spectra. For imaging, Abbe (1873) connected diffraction components admitted by a pupil to image formation. [Historical context and original sources](Appendix_3_Historical_Context_and_Original_Sources.ipynb).

## The central physical idea

A propagating plane-wave component keeps its amplitude while displacement changes its phase:

$$U_m(\mathbf r+\Delta\mathbf r)=U_m(\mathbf r)e^{i\mathbf k_m\cdot\Delta\mathbf r}.$$

Different components generally accumulate different phases. Their changing interference can therefore reshape the total intensity, even in a uniform medium. Outgoing evanescent components decay with propagation distance.

**The starting intensity alone is insufficient: phase helps determine what happens next.**
