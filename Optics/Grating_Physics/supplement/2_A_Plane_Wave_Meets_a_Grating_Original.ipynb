## 1. Start with one isolated slit

**One isolated slit has a continuous range of candidate outgoing directions.** Here we ask only which plane-wave directions are possible in the surrounding medium. We do not yet calculate their amplitudes.

Use an $x$–$z$ cross-section of an opaque screen at $z=0$. The slit runs along $y$ and the incident wave, with wavevector $\mathbf k_i$, approaches from above, toward $-z$. The slit is uniform along $y$ and the incident wave has $k_{iy}=0$, so the outgoing components also lie in the $x$–$z$ plane.

Measure the outgoing angle $\theta$ **from $-z$ toward $+x$**: zero points straight down. This angle reference differs from the $+z$ reference in notebook one; the Cartesian axes and the meaning of $\mathbf k$ stay the same. We write $\mathbf k(\theta)=(k_x,0,k_z)$ for the chosen outgoing candidate. **Bold $\mathbf k$ denotes a vector; plain $k$ denotes its positive magnitude.** At a fixed wavelength in the surrounding medium, each propagating candidate has the same magnitude $|\mathbf k(\theta)|=k$, with

$$k=\frac{2\pi}{\lambda},\qquad k_x=k\sin\theta,\qquad k_z=-k\cos\theta.$$

The outgoing angle $\theta$ can vary continuously between $-90^\circ$ and $+90^\circ$. The endpoints are grazing directions along the screen. In wavevector space, the tips fill a **continuous semicircle**, since $k_x^2+k_z^2=k^2$. This is the plane-wave direction constraint used in the angular-spectrum description. [8, §§3.5–3.6](#ref-8)

**There is no grating period $d$ for one isolated slit.** Moving sideways does not bring us to an identical neighbouring opening. We therefore have no period-to-period phase agreement condition and no integer grating-order number $m$ to assign. We call these a continuum of candidate directions, rather than infinitely many grating orders.

**A possible direction does not guarantee nonzero intensity.** The propagating part of the slit field combines plane-wave components from this continuous set. Their strengths depend on the field across the opening. Contributions from different parts of the slit can reinforce or cancel, so the far-field intensity generally varies continuously with direction and can have dark zeros. [9, Figs. 4.4–4.5](#ref-9)

The semicircle answers **which wavevectors are geometrically possible**. It does not answer **how strongly the slit excites each one**. Its arrows do not represent equal intensity, and choosing a direction does not establish that its amplitude is nonzero.

<details><summary>Example: a possible direction with zero intensity</summary>

For this separate example, take a uniformly illuminated slit of width $a$ at **normal incidence**, using the standard scalar far-field model. Its dark directions satisfy

$$a\sin\theta=p\lambda,\qquad p=\pm1,\pm2,\ldots,$$

when those angles exist. Here $p$ labels **intensity minima**, not grating orders; $a$ is the width of this one opening, not a grating period. [9, Eq. (4.1)](#ref-9)

If $a=2\lambda$, the first zeros occur at $\theta=\pm30^\circ$. Both directions have real wavevectors on our semicircle, yet their far-field intensity is zero in this model because contributions across the slit cancel.

The width assumption belongs to this example. The direction explorer below does not calculate a slit field or its intensity pattern. We will develop the strengths later.

</details>

**Try the direction control below.** The orange arrow selects one candidate, while the green projection shows its $k_x$. The grey arrows are just a few samples of the continuum. The screen drawing shows normal incidence; the orange arrow is a candidate component, not the entire diffracted field.

### Small calculation 1: choose any outgoing direction

The interactive uses the same wavevector components as notebook one, with the angle measured from $-z$. Here, `np.sin` and `np.cos` turn a chosen angle into $k_x$ and $k_z$; `np.hypot` checks that the wavevector length stays $k$.

**Predict, then try:** replace $30^\circ$ with $14.5^\circ$ or $37.2^\circ$. Does the length change? Then keep the angle fixed and double the wavelength. There is no period or integer-order test in this calculation.

```python
import numpy as np

wavelength = 2.0  # micrometres; independent of the sliders
candidate_angle_deg = 30.0
if wavelength <= 0 or not -90 <= candidate_angle_deg <= 90:
    raise ValueError('Use a positive wavelength and an angle between -90 and 90 degrees.')
k = 2*np.pi / wavelength
angle = np.deg2rad(candidate_angle_deg)
kx = k*np.sin(angle)
kz = -k*np.cos(angle)
print(f'Candidate angle: {candidate_angle_deg:+.1f} degrees')
print(f'kx = {kx:+.3f}; kz = {kz:+.3f} rad/um')
print(f'Wavevector length = {np.hypot(kx, kz):.3f}; 2*pi/lambda = {k:.3f} rad/um')
```

## 2. Repeat the opening to make a grating

Now repeat the single opening: imagine an opaque sheet with many straight, equally spaced openings. A wave can pass through the openings; the strips between them block it in our simplest model. More generally, a grating can contain repeated regions with different transmission, or repeated reflecting structures. [1](#ref-1)

Two lengths describe our opening pattern:

- **Period $d$:** the distance between corresponding points in neighbouring openings. You can measure centre to centre or left edge to left edge, as in the diagram. Moving by $d$ reproduces the structure.
- **Opening width $a$:** the width of one transparent strip. It need not equal the period. The incident-wave explorer uses $a=d/2$, so half of each period is open. The small calculations also let you edit $a$.

The long parallel structures in ruled or etched gratings are called **grooves**. Our model uses long slits rather than a detailed groove shape. We still use "along the grooves" to identify their long direction. **One-dimensional grating** means that the pattern repeats in only one direction; the wave and physical structure still occupy three-dimensional space.

Here, all distances, including $a$, $d$, and the **wavelength in the medium** $\lambda$, are in micrometres ($\mu\mathrm m$). We use the same homogeneous, isotropic medium on both sides.

## 3. Name the directions

Walk across the openings along **$x$**: you cross the repeating pattern. Walk along an opening in **$y$**: nothing changes in this ideal model. The **$z$ axis** is perpendicular to the sheet, which lies at $z=0$.

The incoming wave approaches the grating from $z>0$ with a **negative $z$ component** of its propagation direction. At normal incidence it points exactly along $-z$; a tilted wave also has an $x$ component. We keep $+z$ pointing upward in the side view, so the transmitted side is below the sheet, at $z<0$. The diagram shows a few periods of a pattern we idealize as indefinitely repeating. A dot in a circle means an axis points out of the page; a cross means it points into the page.

All examples in this notebook keep $k_y=0$: the incident and outgoing directions lie in the $x$-$z$ plane. The ideal pattern is uniform along $y$.

### Small calculation 2: describe one opening

Use a dimensionless **field-amplitude transmission** $M$: $M=1$ in an opening and $M=0$ in a blocked strip. This specifies our ideal on/off profile, not an intensity measurement. Centre an opening at $x=0$ and inspect one period, from $-d/2$ to $d/2$. [2, §2](#ref-2)

`np.arange(8)` makes the array $[0,1,\ldots,7]$. We use it to place eight sample points inside the period. `np.abs(x) < a/2` asks whether each point is inside the opening; `.astype(float)` turns the resulting true/false answers into 1/0.

**This calculation starts with** $d=2\,\mu\mathrm m$ and $a=1\,\mu\mathrm m$, matching the opening profile of the incident-wave explorer introduced in Section 4. Wavelength and angle do not change the opening profile. **Predict, then try:** set `a = 0.5`. How many of the eight sample positions will have $M=1$? This width change is available in the code; the explorer keeps $a=d/2$.

```python
d, a = 2.0, 1.0  # period and opening width, in micrometres
x = -d/2 + (np.arange(8) + 0.5) * d/8
M = (np.abs(x) < a/2).astype(float)
print(' x (um)   M')
for position, transmission in zip(x, M):
    print(f'{position:7.3f}   {transmission:.0f}')
```

</details>

## 4. Illuminate the surface: straight on, then tilted

At **normal incidence**, the wave travels along $-z$. Its wavefronts are parallel to the grating: every point on the surface is at the same phase at a given time. This does not mean the field is constant in time; all surface points oscillate together.

Now tilt the wave in the $x$-$z$ plane. At a fixed time, different positions on the grating generally have different phases. A wavefront itself is a surface of constant phase; the grating is no longer parallel to those wavefronts. For this grating notebook, define the signed incidence angle $\theta_i$ **from $-z$ toward $+x$**; positive angles lean toward $+x$. Thus $\theta_i=0^\circ$ means downward normal incidence. This reference direction differs from the $+z$ reference used in notebook one; the Cartesian axes and the meaning of $\mathbf k$ stay the same. The incident wavevector is $\mathbf k_i=(k_{ix},k_{iy},k_{iz})$. The subscript $i$ identifies the incident wave at any incidence angle; plain $k=|\mathbf k_i|$ is its magnitude. We use $-90^\circ<\theta_i<90^\circ$ so that the incident wave approaches the surface from above. The components are

$$k=\frac{2\pi}{\lambda},\qquad k_{ix}=k\sin\theta_i,\qquad k_{iy}=0,\qquad k_{iz}=-k\cos\theta_i.$$

Use the same complex scalar plane wave as notebook one, now with the incident components above:

$$U_i(x,z)=A e^{i(k_{ix}x+k_{iz}z)}.$$

Restoring the time factor gives the signed field shown in the plots:

$$u_i(x,z,t)=\operatorname{Re}\{U_i(x,z)e^{-i\omega t}\}
=A\cos(k_{ix}x+k_{iz}z-\omega t).$$

Here $A$ is the scalar amplitude and $\omega$ is the angular frequency. We take the phase to be zero at the origin when $t=0$.

**For a vector electric field, put the direction into the amplitude.** For a linearly polarized plane wave, replace the scalar amplitude $A$ with the vector amplitude
$$\mathbf A_i=A\,\hat{\mathbf e}_i.$$
The electric field is then
$$\boxed{\mathbf E_i(x,z,t)=\mathbf A_i\cos(k_{ix}x+k_{iz}z-\omega t).}$$
Here $A$ gives the amplitude's magnitude, and the fixed unit vector $\hat{\mathbf e}_i$ gives the electric-field oscillation direction. The cosine remains a **scalar**: it determines the field's signed value at each position and time. In our homogeneous, isotropic medium, the amplitude must be transverse: $\mathbf k_i\cdot\mathbf A_i=0$. See [8, §3.3, Eqs. (8)–(11)](#ref-8) for the vector plane wave and scalar-component description.

For example, $\mathbf A_i=(0,A,0)$ gives $\mathbf E_i=(0,u_i,0)$, a **$y$-polarized** wave. The notebook plots $u_i$ without choosing a polarization direction. The condition $k_{iy}=0$ specifies no phase variation along $y$, independently of that choice.

Circular and elliptical polarization use a complex vector amplitude to encode relative phases between electric-field components; those cases can be treated later.

At normal incidence the scalar expression becomes $u_i=A\cos(-kz-\omega t)$. Following a fixed phase as time increases gives $z(t)=z(0)-(\omega/k)t$: the wave moves toward smaller $z$. The time factor remains $e^{-i\omega t}$; the negative $k_{iz}$ sets the downward direction.

At the surface, **set $z=0$**. Its phase is $\phi_i(x,0,t)=k_{ix}x-\omega t$. Thus moving along the grating by one period changes the incident phase by

$$\Delta\phi_i=k_{ix}d=\frac{2\pi d}{\lambda}\sin\theta_i.$$

For a fixed $\mathbf k_i$, this incident phase step is the same for either transverse polarization. It follows from the plane-wave phase, without specifying the grating's polarization-dependent response.

The incident amplitude $A$ is the same at neighbouring openings, but their phases are **generally different** when the wave is tilted. A repeating structure does not imply an identically repeating instantaneous field. A phase offset that is a whole multiple of $2\pi$ is a special case: it returns to the same point in the cycle. [3, §2.3](#ref-3)

### Small calculation 3: compare neighbouring openings

Opening centres are at $x=jd$, where the integer $j$ labels an opening, **not a diffraction order**. We evaluate the time-independent complex incident field $U_i(x,0)=A e^{ik_{ix}x}$. Its magnitude is the wave amplitude; its real part is the signed field at $t=0$.

`np.deg2rad` converts degrees to radians for `np.sin`. As in notebook one, `1j` is the imaginary unit, `np.exp(1j*phase)` gives a complex phase factor of magnitude one, `abs(...)` gives its magnitude, and `.real` gives its real part. `zip` lets us print corresponding array entries together.

**This calculation starts with these incident-wave settings:** $d=2$, $\lambda=2$ (both in $\mu\mathrm m$), $\theta_i=30^\circ$, $t=0$. **Predict, then try:** change `theta_deg` to `0`. Which output columns become constant?

```python
wavelength, d, A = 2.0, 2.0, 1.0  # lengths in micrometres
theta_deg = 30.0
k = 2*np.pi / wavelength
k_ix = k * np.sin(np.deg2rad(theta_deg))
x = np.arange(-2, 3) * d
phase = k_ix * x
incident = A * np.exp(1j * phase)
print(' x (um)   phase/pi   amplitude   u at t=0')
for position, phi, U_value in zip(x, phase, incident):
    print(f'{position:7.1f}   {phi/np.pi:8.2f}   {abs(U_value):9.2f}   {U_value.real:8.2f}')
```

<details><summary>Answer for the stated starting values and changes</summary>

At $30^\circ$, the printed phases advance by $\pi$ between openings. The amplitude column stays at 1 while the real field alternates in sign. At $0^\circ$, all phases become zero and all real fields become 1 at this snapshot; the amplitude column was already constant. At another time they would still oscillate together.

</details>

## 5. What does the grating do right at the surface?

For a first model, imagine the sheet as an **ideal thin mask**: at each position, an opening passes the incident field unchanged and a blocked strip sets it to zero. We prescribe this local action and neglect thickness, reflections, and interactions within the material. It is a scalar approximation, not a complete electromagnetic solution; it does not calculate polarization changes or polarization-dependent order strengths. Small openings or substantial material structure can require a more complete model.

Only with that assumption do we multiply by our transmission profile:

$$U_{\mathrm{after}}(x,0^-)\approx M(x)\,U_i(x,0),\qquad M(x+d)=M(x).$$

Here $0^-$ means **just on the transmission side**. Uppercase $U$ denotes the time-independent complex scalar field; recover the real field with $u_{\mathrm{after}}=\operatorname{Re}\{U_{\mathrm{after}}e^{-i\omega t}\}$. An opening preserves the incoming phase in this mask model. In a blocked region, $U_{\mathrm{after}}=0$ and its phase is undefined. The binary transmission model is also the starting example in [2, §2](#ref-2).

This product specifies a field **at the surface**. It does not calculate the field farther away: propagating that surface field is another step. In particular, it does not say that the space beyond every blocked strip remains dark.

### Small calculation 4: apply the mask

To reuse the opening at $x=0$ across the whole grating, `(x + d/2) % d - d/2` finds each point's position within one centred period. `%` is the remainder operation. We then multiply the complex incident field by the real 0/1 transmission, point by point. The output shows the real and imaginary parts of the transmitted complex field; at $t=0$ its real part is the signed field.

**This calculation starts with these incident-wave settings:** $d=\lambda=2\,\mu\mathrm m$, $a=d/2$, $\theta_i=30^\circ$, $t=0$. The explorer displays the incident wave; this code adds the thin-mask operation. **Predict, then try:** change `theta_deg` to `0`. Will blocked points start transmitting, or will only the values inside openings change?

```python
wavelength, d, a, A = 2.0, 2.0, 1.0, 1.0  # lengths in micrometres
theta_deg = 30.0
x = np.array([0.0, 0.25, 0.75, 1.0, 2.0, 2.25])
k_ix = (2*np.pi / wavelength) * np.sin(np.deg2rad(theta_deg))
local_x = (x + d/2) % d - d/2
M = (np.abs(local_x) < a/2).astype(float)
incident = A * np.exp(1j * k_ix * x)
after = M * incident
print(' x (um)   M    Re(incident)    Re(after)    Im(after)')
for position, mask, incoming, outgoing in zip(x, M, incident, after):
    print(f'{position:7.2f}   {mask:.0f}    {incoming.real:12.3f}    {outgoing.real:9.3f}    {outgoing.imag:9.3f}')
```

<details><summary>Answer for the stated starting values and changes</summary>

With the code's starting mask and $A=1$, the blocked samples ($x=0.75$ and $1.00\,\mu\mathrm m$) remain zero when you change the incidence angle. At normal incidence every opening receives $U_i=1$, so $U_{\mathrm{after}}$ is 1 inside an opening and 0 outside. At $30^\circ$, the opening centres $x=0$ and $2\,\mu\mathrm m$ receive opposite real values at $t=0$. Multiplication by the repeating mask preserves that phase offset; it does not make the tilted field identical from period to period.

</details>

## 6. Meet the diffraction orders

At the start of this notebook, one isolated slit had a continuous range of candidate outgoing directions. Repeating the slit indefinitely adds a constraint: only discrete directions fit the periodic structure and its illumination.

Beyond the grating, the transmitted field can be described using outgoing plane-wave components called **diffraction orders**. For an indefinitely repeating grating illuminated by one plane wave, periodicity selects discrete possible values of the wavevector component across the grooves. We will determine which of those values give real outgoing directions. [1, §2.2.1](#ref-1) In a far-field pattern from a finite illuminated grating, orders that are present have finite angular width around the ideal directions. [7, §1, Figs. 1–4](#ref-7)

The integer $m$ labels an order: $m=0$ is the zero order, $m=\pm1$ the first orders, and $m=\pm2$ the second orders. It does not count openings. With the same medium on both sides, the transmitted zero-order direction is the incident direction. [7, §3, Fig. 8](#ref-7)

### Describe an order with its wavevector

Notebook one used a wavevector to describe a plane wave's direction and spatial phase. Here the subscripts identify which wave we mean:

| Symbol | Meaning |
| --- | --- |
| $\mathbf k_i=(k_{ix},k_{iy},k_{iz})$ | Incident wavevector, at normal or tilted incidence |
| $\mathbf k_m=(k_{mx},k_{my},k_{mz})$ | Outgoing wavevector of diffraction order $m$ |
| $k=2\pi/\lambda$ | Common scalar magnitude for these real wavevectors in the same medium |

Thus $|\mathbf k_i|=|\mathbf k_m|=k$: their directions and components can differ while their lengths agree. For the transmitted zero order in this geometry, $\mathbf k_0=\mathbf k_i$. In $k_{mx}$, the $m$ identifies the order and the $x$ identifies its component across the grooves. Each component is a signed scalar. The single-slit notation $\mathbf k(\theta)$ earlier labels a continuous candidate direction; an integer order label becomes appropriate only after introducing the periodic grating.

The grating equation can be written directly in these components:

$$\boxed{k_{mx}=k_{ix}+mG,\qquad k_{my}=k_{iy}},\qquad G=\frac{2\pi}{d}.$$

Across the grooves, the candidate values of $k_x$ are separated by $G$. Like $k$, $G$ has units of radians per unit length: a step of $G$ adds $Gd=2\pi$ to the phase advance over one period. Along the grooves, the structure is uniform, so $k_y$ is preserved. In our current $x$-$z$ examples, $k_{iy}=k_{my}=0$. We will explain why the steps must be integer multiples next time. [3, §2.3, Eq. (2.50), and §2.4](#ref-3)

The same rule has a compact **vector form at the surface**:

$$\boxed{\mathbf k_{m,\parallel}=\mathbf k_{i,\parallel}+m\mathbf G},\qquad \mathbf G=\frac{2\pi}{d}\,\hat{\mathbf x}.$$

Here $\parallel$ means **parallel to the grating surface**: it collects the $x$ and $y$ components, excluding $z$. The symbol $\hat{\mathbf x}$ is a unit vector along $+x$, so $\mathbf G$ points across the grooves.

### Complete the outgoing wavevector

The surface rule supplies $k_{mx}$ and $k_{my}$. In the same surrounding medium, an outgoing propagating plane wave still has magnitude $k=2\pi/\lambda$, so

$$k_{mx}^2+k_{my}^2+k_{mz}^2=k^2,\qquad
\boxed{k_{mz}=-\sqrt{k^2-k_{mx}^2-k_{my}^2}}.$$

For a positive quantity under the square root, the **negative** root gives $k_{mz}<0$ and selects transmission into $z<0$. Zero gives $k_{mz}=0$: a **grazing** direction parallel to the grating. A negative value gives no real outgoing direction; the corresponding evanescent field will be discussed later. The normal component is determined by this magnitude condition and the outgoing side, separately from the vector rule at the surface. [3, Eqs. (2.54)–(2.58) and §2.4](#ref-3)

In the plots below, **allowed directions** means real directions, including grazing ones. The order explorer reports both the total and the split between orders travelling into $z<0$ and grazing orders. This is a count of possible directions, not a count of bright or nonzero-amplitude beams.

### Recover the angle equation

For our $x$-$z$ examples, $k_{ix}=k\sin\theta_i$ and $k_{mx}=k\sin\theta_m$. Substituting these into the $x$-component rule and dividing by $k$ gives

$$k\sin\theta_m=k\sin\theta_i+m\frac{2\pi}{d}
\quad\Longrightarrow\quad
\boxed{\sin\theta_m=\sin\theta_i+m\frac{\lambda}{d}}.$$

Both angles are measured from $-z$ toward $+x$, with $-90^\circ\leq\theta_m\leq90^\circ$ for transmitted or grazing directions. Positive $m$ increases $k_{mx}$ and negative $m$ decreases it; at tilted incidence the sign of $m$ need not be the sign of $\theta_m$. Requiring the right-hand side to lie between $-1$ and $+1$ is the same check for a real direction as the square-root condition above. [3, §2.4](#ref-3)

The **fixed preview** below uses **$\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, and $\theta_i=0^\circ$**. For these values, the real orders are $m=-2,-1,0,+1,+2$, all with $k_{mz}<0$. This preview does not change when you use the later interactive controls. Arrows share an origin only to compare directions; each represents the direction of an extended plane-wave component.

**Direction is not strength.** The relation does not give the amplitude of an order, and an allowed direction can have zero amplitude. ### Small calculation 5: build one outgoing wavevector

Use the preview's settings: $\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, and normal incidence. Construct order $m=+1$ by adding $mG$ to the incident $x$ component, retaining $k_y=0$, and choosing the negative $k_z$ branch. In the code, `k_i` and `k_m` are three-component vectors, while `k` is their common scalar magnitude. All printed components are in radians per micrometre.

`np.sqrt` takes a square root and `np.linalg.norm` gives a vector's length. The `if` statement checks for a real direction before taking the square root. A small tolerance treats numerical rounding at grazing as zero. The final line distinguishes transmission into $z<0$ from grazing. This calculation uses the values written in its code cell, independently of the interactive controls.

**Predict, then try:** keep wavelength, period, and incidence angle fixed, and change `m` to `-1`, then `0`, then `3`. Which components change, and which choice has no propagating direction?

```python
wavelength, d, theta_deg = 1.0, 2.5, 0.0  # lengths in micrometres
m = 1  # integer diffraction-order label
k, G = 2*np.pi / wavelength, 2*np.pi / d
theta = np.deg2rad(theta_deg)
k_i = k * np.array([np.sin(theta), 0.0, -np.cos(theta)])
kmx, kmy = k_i[0] + m*G, k_i[1]
remaining = k**2 - kmx**2 - kmy**2
tolerance = 1e-12 * k**2  # rounding tolerance at grazing
if remaining < -tolerance:
    print(f'm = {m:+d}: no real propagating direction')
else:
    kmz = 0.0 if abs(remaining) <= tolerance else -np.sqrt(remaining)
    k_m = np.array([kmx, kmy, kmz])
    print('Incident k_i:', np.round(k_i, 3))
    print(f'Order {m:+d} k_m:', np.round(k_m, 3))
    print('|k_i|, |k_m| (rad/um):', np.round([np.linalg.norm(k_i), np.linalg.norm(k_m)], 3))
    print('Direction:', 'grazing along z=0' if kmz == 0 else 'into z<0')
```

Use the same signed angle convention as before: $\theta_m$ is measured from $-z$, positive toward $+x$ and negative toward $-x$. The **middle panel** shows the order wavevector $\mathbf k_m$ and its signed components:

$$k_{mx}=k\sin\theta_m,\qquad k_{mz}=-k\cos\theta_m.$$

The diagonal arrow is labelled $\mathbf k_m$; its length is the scalar $|\mathbf k_m|=k$. The subscript identifies the selected order and does not imply a different wavelength or magnitude.

The horizontal component is positive when the order points right and negative when it points left. The vertical arrow's **length** is $|k_{mz}|$; the signed component is $k_{mz}<0$ for transmission toward $-z$. At grazing, $k_{mz}=0$ and the triangle becomes flat. $$x_{m,\mathrm{sphere}}=R\sin\theta_m,\qquad
z_{m,\mathrm{sphere}}=-R\cos\theta_m.$$

The **right panel** expresses phase variation in **cycles per unit length**. One full phase cycle is $2\pi$ radians, so the spatial-frequency components are

$$\boxed{f_{x,m}=\frac{k_{mx}}{2\pi}=\frac{\sin\theta_m}{\lambda},\qquad
f_{z,m}=\frac{k_{mz}}{2\pi}=-\frac{\cos\theta_m}{\lambda}.}$$

The symbol $f$ here describes variation with position, not oscillations per second. With $k_y=0$, these components satisfy

$$f_{x,m}^2+f_{z,m}^2=\frac{1}{\lambda^2}.$$

The third semicircle therefore has radius $1/\lambda$ in **cycles per micrometre**. It is a coordinate-space drawing of the same order. All three triangles have the same angle and proportions; their diagonal lengths are $R$, $k$, and $1/\lambda$, in their respective units. ### Give the normalized horizontal direction a name

Divide the **signed horizontal component** by the positive diagonal length in each triangle:

$$\boxed{\frac{x_{m,\mathrm{sphere}}}{R}
=\frac{k_{mx}}{k}
=\frac{f_{x,m}}{1/\lambda}
=\sin\theta_m.}$$

All these expressions give the same **dimensionless** number. We call it the horizontal direction cosine and write

$$\boxed{\alpha_m=\frac{k_{mx}}{k}.}$$

Why a *cosine* when the formula contains a sine? A direction cosine is the cosine of the angle with a coordinate axis. Here that axis is $+x$, whereas our $\theta_m$ is measured from $-z$. For this geometry, the cosine of the angle with $+x$ equals $\sin\theta_m$. Projecting sphere points onto transverse coordinates and normalizing by the radius is also the construction in [7, §3, Fig. 6](#ref-7).

For the worked example restored by **Reset example**, with $m=+1$, $\theta_m\approx23.58^\circ$ and $\alpha_m=0.4$. On the sphere, $x_{m,\mathrm{sphere}}/R=0.800/2=0.4$; from the wavevector, $k_{mx}/k\approx2.513/6.283=0.4$; and from the spatial-frequency panel, $f_{x,m}/(1/\lambda)=0.400/1=0.4$.

The quantity $k_{mx}$ describes the phase change per unit distance along $x$. Dividing it by $k$ isolates the horizontal component of the **unit direction**. This is not an extra diffraction rule; it is a normalized description of the same wavevector.

### From one order to neighbouring orders

Keep wavelength, period, and incidence angle fixed while comparing the order labels $m$ and $m+1$. All real orders in this medium have the same wavevector magnitude $k$. Their individual horizontal components vary. What stays constant between neighbouring orders is the **step**:

$$k_{m+1,x}-k_{mx}=G=\frac{2\pi}{d}.$$

Dividing instead by $2\pi$ expresses each order in spatial-frequency coordinates:

$$f_{x,m}=\frac{\sin\theta_i}{\lambda}+\frac{m}{d},\qquad
\boxed{f_{x,m+1}-f_{x,m}=\frac{1}{d}.}$$

The illumination sets the common offset; the period sets the spatial-frequency step.

Divide the wavevector step by $k=2\pi/\lambda$ to express it as a direction cosine:

$$\alpha_{m+1}-\alpha_m
=\sin\theta_{m+1}-\sin\theta_m
=\frac{G}{k}
=\frac{\lambda}{d}.$$

When both neighbouring orders have real directions, multiplying by $R$ gives their step on the sphere's horizontal ruler:

$$\boxed{x_{m+1,\mathrm{sphere}}-x_{m,\mathrm{sphere}}
=R\frac{\lambda}{d}.}$$

| Description | Coordinate | Step between neighbouring orders |
| --- | --- | --- |
| Transverse wavevector | $k_{mx}$ | $2\pi/d$ |
| Transverse spatial frequency | $f_{x,m}=k_{mx}/(2\pi)$ | $1/d$ |
| Horizontal direction cosine | $\alpha_m=\sin\theta_m$ | $\lambda/d$ |
| Sphere's horizontal position | $x_{m,\mathrm{sphere}}=R\alpha_m$ | $R\lambda/d$ |

These are the same equal-spacing property expressed on different scales. The angles $\theta_m=\arcsin\alpha_m$, the distances along the circle, and the straight-line distances between neighbouring blue points are generally unequal. The direction-cosine and angular spacings are compared in [7, §3, Figs. 9–10](#ref-7).

For the specific ruler exercise with $\lambda=1\,\mu\mathrm m$, $d=4\,\mu\mathrm m$, $R=4\,\mu\mathrm m$, and normal incidence, $\alpha_m=m/4$ and $x_{m,\mathrm{sphere}}=m\,\mu\mathrm m$. Every horizontal gap is $1\,\mu\mathrm m$, including the gap from $m=+3$ to the grazing order $m=+4$.

**What changes the spacing?** Changing $R$ only scales the sphere positions. Changing wavelength at fixed $d$ changes $\lambda/d$ and $k$, while the $k_x$ step remains $2\pi/d$ and the $f_x$ step remains $1/d$. Changing $d$ changes both of those steps. Tilting incidence changes the common offset $k_{ix}$; it leaves the step between neighbouring orders unchanged, although the set of allowed orders can change.

<details><summary>Reading the 2019 paper: its Δβ and our Δα</summary>

In Figs. 9–10 of Harvey and Pfisterer [7](#ref-7), **$\Delta\beta$ means the direction-cosine spacing**. It corresponds to our $\Delta\alpha=\Delta(\sin\theta)=\lambda/d$ as a spacing magnitude. It is dimensionless.

The paper's earlier Eq. (4) and Figs. 7–8 use $\alpha$ for the coordinate along which orders are spaced. Thus the spacing label changes within the paper. This notebook consistently uses $\alpha_m=k_{mx}/k$.

Their **$\Delta x$ in Fig. 9 is spacing on the flat screen**, corresponding to our $\Delta x_{\mathrm{plane}}$. Our sphere's horizontal spacing is instead $\Delta x_{\mathrm{sphere}}=R\Delta\alpha$. Both are lengths, but they are measured on different reference surfaces. The next section derives the flat-screen coordinate.

</details>


### Compare with the flat observation plane

The representative ray meets the sphere at

$$\mathbf r_{m,\mathrm{sphere}}=R\frac{\mathbf k_m}{k}.$$

For an order with $k_{mz}<0$, follow the ray until its $z$ coordinate reaches the plane at $-L$:

$$x_{m,\mathrm{plane}}=-L\frac{k_{mx}}{k_{mz}},\qquad
y_{m,\mathrm{plane}}=-L\frac{k_{my}}{k_{mz}},\qquad
z_{m,\mathrm{plane}}=-L.$$

For our $x$-$z$ geometry, substituting the components from the triangle gives

$$x_{m,\mathrm{plane}}
=-L\frac{k\sin\theta_m}{-k\cos\theta_m}
=L\tan\theta_m.$$

Thus the dimensionless plane coordinate $X_m=x_{m,\mathrm{plane}}/L$ follows a **tangent mapping**:

$$\boxed{X_m=\tan\theta_m},\qquad
\boxed{\alpha_m=\frac{x_{m,\mathrm{sphere}}}{R}=\sin\theta_m}.$$

The plane positions are generally unequally spaced because the extra division by $\cos\theta_m$ varies from order to order. The same direction gives different coordinates on the two surfaces; both follow from ordinary straight-ray geometry. At grazing, $k_{mz}=0$: the ray runs parallel to the plane and has no plane intersection. The same flat-screen relation $x_m=L\tan\theta_m$ appears in [7, §3, Eq. (9) and Fig. 9](#ref-7).

**Experimental connection: infer a direction from a position.** In a finite
illuminated grating experiment with separated far-field order peaks, measure
a peak centre $x_m$ relative to the grating normal on a screen a known distance
$L$ away. In this geometry,

$$\theta_m=\arctan(x_m/L),\qquad
k_{mx}=\frac{2\pi}{\lambda}\sin\theta_m.$$

The measured position and known wavelength give a wavevector component.
The ideal plane-wave markers above specify directions; they do not predict
the finite experiment's spot widths or brightness. A missing peak alone does
not establish that an order is forbidden: its excitation and the detector's
collection range also matter.

### Small calculation 6: where does one order meet each surface?

This code starts with $m=+1$, $\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, normal incidence, $R=2\,\mu\mathrm m$, and $L=3\,\mu\mathrm m$. It rebuilds the wavevector, then prints both intersections, the spatial-frequency components, and $\alpha_m$. Its wavelength, period, incidence angle, radius, and plane distance can be edited to match the controls; `m` selects the order. Because this calculation needs **both** intersections, it rejects orders with no real direction and those at grazing within the numerical tolerance. The interactive view can still show a grazing order on the sphere alone.

`direction` is a unit vector; multiplying it by a distance moves along that ray. **Predict, then try:** first change $L$ from $3$ to $6\,\mu\mathrm m$. Then restore $L=3\,\mu\mathrm m$ and change $m$ from $+1$ to $-1$, keeping the other code parameters fixed. This code does not change either explorer.

```python
wavelength, d, m = 1.0, 2.5, 1  # lengths in micrometres
theta_deg = 0.0  # signed incidence angle from -z toward +x
R, L = 2.0, 3.0
k = 2*np.pi / wavelength
kmx = k*np.sin(np.deg2rad(theta_deg)) + m*2*np.pi/d
remaining = 1 - (kmx/k)**2
if remaining <= 1e-12:  # allow for rounding near grazing
    raise ValueError('Choose an order that travels toward -z, away from grazing.')
k_m = np.array([kmx, 0.0, -k*np.sqrt(remaining)])
direction = k_m / np.linalg.norm(k_m)
spatial_frequency = k_m / (2*np.pi)  # cycles per micrometre
on_sphere = R * direction
on_plane = (-L / direction[2]) * direction
print('Sphere point (um):', np.round(on_sphere, 3))
print('Plane point  (um):', np.round(on_plane, 3))
print('Spatial frequency (cycles/um):', np.round(spatial_frequency, 3))
print('Horizontal direction cosine alpha:', round(direction[0], 3))
```

The spatial-frequency vector is approximately $(0.400,\,0,\,-0.917)$ cycles per micrometre, with length $1/\lambda=1$ cycle per micrometre. Changing $L$ or $R$ leaves it unchanged.

The order angle is $\theta_{+1}\approx23.58^\circ$. Its normalized plane coordinate is $X_{+1}=\tan\theta_{+1}\approx0.436$, whereas its normalized sphere $x$ coordinate is $\sin\theta_{+1}=0.400$. Doubling $L$ leaves $X_{+1}$ fixed even though the physical plane coordinate doubles.

Keeping normal incidence and the other parameters fixed, changing from $m=+1$ to $m=-1$ reverses the $x$ coordinates and leaves the $z$ coordinates unchanged. This symmetry about the $z$ axis generally does not hold at tilted incidence.

For either selected order, its sphere point and plane point lie on the same ray. The two points need not be the same distance from the origin.

<details><summary>Looking ahead: from directions to an optical pupil (optional)</summary>

The grating determines the allowed directions. Later, a collecting optical system can give the collected directions positions in a specified Fourier/pupil plane. For this preview, align the optical axis with $-z$ and choose the pupil's horizontal axis to match our $+x$ direction. In the **ideal sine-mapped pupil model**,

$$x_{p,m}=C\sin\theta_m
=\frac{C}{k}k_{mx}
=\frac{C}{R}x_{m,\mathrm{sphere}}.$$

Here $x_{p,m}$ is a physical pupil coordinate and $C$ is a length set by the optical system and the chosen pupil plane. At fixed wavelength, this mapping makes the pupil coordinate proportional to $k_x$. Neighbouring order labels therefore have equally spaced mapped coordinates; a finite pupil may admit only some of those orders. It is an optical-model assumption: our reference sphere already has a sine dependence through geometry, but does not itself supply a lens or a physical pupil. [5, Appendix A, Eqs. 41–43](#ref-5)

This sine mapping is used in an ideal **aplanatic** model and is associated with the Abbe sine condition. Calling the complete system aplanatic also concerns correction of spherical aberration and coma; mapping directions to a plane alone is not its definition. [6](#ref-6)

The pupil plane and its mapping must be specified; this is not a universal rule for every entrance pupil. We will develop the optical system, its collection aperture, and its imaging conditions later. The present notebook establishes the grating directions and their geometrical representations.
