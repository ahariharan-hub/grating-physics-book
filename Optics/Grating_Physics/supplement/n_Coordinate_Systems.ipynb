# Coordinate systems for the wavevector

**One plane wave, several useful descriptions of its direction.** This notebook follows [Plane waves and wavevectors](../1_0_Plane_Waves_and_Wavevectors.ipynb). We start with components, then use angles and normalized coordinates to make optical geometry easier to read.

**Medium and wavelength convention.** The surrounding medium is homogeneous, isotropic and lossless.

- $\lambda$ is wavelength **in the medium**: $\lambda=\lambda_0/n$ and $k=nk_0=2\pi/\lambda$.
- The controls vary medium wavelength; there is no separate index control.

See the [course medium convention](../0_0_Foundations_of_Scalar_Optics.ipynb#medium-and-wavelength).

We consider real, propagating wavevectors in a homogeneous, isotropic, lossless medium with positive refractive index $n$. Let $\lambda_0$ be the **vacuum wavelength**, while $\lambda=\lambda_0/n$ is the wavelength in the medium:

$$k_0=\frac{2\pi}{\lambda_0},\qquad k=|\mathbf k|=n k_0=\frac{2\pi}{\lambda}.$$

The optical axis is $z$. All diagrams below are in **wavevector or direction space**; they do not show the wave's position in real space. Choose **Run All** to activate the explorer. It creates no image files.

## 1. Cartesian coordinates: phase change along each axis

$$\mathbf k=(k_x,k_y,k_z),\qquad
\mathbf k\cdot\mathbf r=k_xx+k_yy+k_zz.$$

Each component measures phase change per distance along its axis. Cartesian components are convenient for plane-wave calculations and for separating directions parallel and normal to an optical surface [\[1\]](#ref-1).

For a surface in the $xy$-plane, $(k_x,k_y)$ are tangential components and $k_z$ is normal. A tilted surface needs a **local Cartesian frame**:

$$k_{t_1}=\mathbf k\cdot\hat{\mathbf t}_1,\qquad
k_{t_2}=\mathbf k\cdot\hat{\mathbf t}_2,\qquad
k_n=\mathbf k\cdot\hat{\mathbf n}.$$

The three unit vectors form an orthonormal frame attached to the surface. Rotating the frame changes the components, while preserving the physical wavevector and its magnitude.

## 2. Spherical coordinates: magnitude and direction

Use $(k,\theta,\varphi)$, where $\theta$ is measured **from $+z$** and $\varphi$ is measured **from $+x$ toward $+y$**:

$$\boxed{
k_x=k\sin\theta\cos\varphi,\qquad
k_y=k\sin\theta\sin\varphi,\qquad
k_z=k\cos\theta.}$$

For fixed wavelength and medium, changing the angles moves the tip of $\mathbf k$ over a sphere of radius $k$.

$\theta=0$: along $+z$. $\theta=90^\circ$: in the transverse plane. $\theta>90^\circ$: toward the $-z$ hemisphere.

At either pole, azimuth is undefined: changing $\varphi$ leaves the direction unchanged. Specifying a plane wave with spherical coordinates does **not** make it a spherical wave.

## 3. Cylindrical coordinates: transverse and longitudinal parts

Use $(k_\perp,\varphi,k_z)$, with

$$k_\perp=\sqrt{k_x^2+k_y^2}=k\sin\theta,\qquad
k_x=k_\perp\cos\varphi,\quad k_y=k_\perp\sin\varphi.$$

This is useful when the optical axis matters: $k_\perp$ gives the transverse magnitude, while $k_z$ retains the forward/backward sign. The allowed components satisfy

$$k_\perp^2+k_z^2=k^2.$$

Here $\perp$ means **perpendicular to the optical axis**. Relative to an $xy$ interface, those same transverse components are often called $\mathbf k_\parallel$, because they are parallel to the surface.

## 4. Direction cosines: remove the wavelength scale

Normalize the vector by its magnitude:

$$\boxed{(\alpha,\beta,\gamma)=\frac{\mathbf k}{k}
=(\sin\theta\cos\varphi,\sin\theta\sin\varphi,\cos\theta).}$$

These are the cosines of the angles with the three Cartesian axes. In particular, $\theta$ is the angle with $z$, so $\gamma=\cos\theta$.

$$\alpha^2+\beta^2+\gamma^2=1,\qquad
\gamma=\pm\sqrt{1-\alpha^2-\beta^2}.$$

The unit sphere describes all directions. Projecting onto $(\alpha,\beta)$ produces a unit disk, but **loses the sign of $\gamma$**. For example, $\theta=40^\circ$ and $140^\circ$ at the same azimuth produce the same disk point. Choose a hemisphere to recover a unique direction.

This is the direction-space representation used in Harvey's treatment of grating geometry [\[2\]](#ref-2). We will add the grating constraints in a later notebook.

## 5. Spatial-frequency coordinates: count cycles instead of radians

Define $f_x=k_x/(2\pi)$, $f_y=k_y/(2\pi)$ and $f_z=k_z/(2\pi)$. The plane-wave phasor becomes

$$\widetilde U(\mathbf r)=A\exp\!\left[i2\pi(f_xx+f_yy+f_zz)\right].$$

Thus $\mathbf k$ has units of **radians per length**, while spatial frequency has units of **cycles per length**. The transverse pair $(f_x,f_y)$ labels plane-wave components in Fourier optics [\[1\]](#ref-1):

$$f_x=\frac{n\alpha}{\lambda_0},\qquad
f_y=\frac{n\beta}{\lambda_0}.$$

Spatial frequency describes variation across space; it is different from temporal frequency in hertz.

## 6. Normalized pupil coordinates: direction relative to an aperture

The coordinates so far describe the wavevector directly. Introducing a physical pupil adds a question: **how does a position in the pupil correspond to a propagation direction?** We start with a centered circular pupil and use an ideal rotationally symmetric imaging model.

### 6.1 A pupil point labels a direction at a field point

> A point in the pupil corresponds to a propagation direction at a given point in the field plane.

The field coordinate tells us **where** we are looking; the pupil coordinate tells us **which direction** the light travels there. In the ideal Fourier-optics picture, a pupil point labels a plane-wave component extending across the field. It does not map to a single spatial point in that field.

Let $(x_p,y_p)$ be physical pupil coordinates and $R_p$ the pupil's outer radius. Normalize them as

$$\sigma_x=\frac{x_p}{R_p},\qquad \sigma_y=\frac{y_p}{R_p}.$$

Their polar representation is

$$\boxed{\rho=\sqrt{\sigma_x^2+\sigma_y^2},\qquad
\phi=\operatorname{atan2}(\sigma_y,\sigma_x),}$$

$$\sigma_x=\rho\cos\phi,\qquad \sigma_y=\rho\sin\phi.$$

Here $\rho$ is dimensionless: zero at the center and one at the edge. The two-argument $\operatorname{atan2}$ preserves the quadrant; azimuth is undefined at the center. The symbol $\phi$ denotes the same azimuth called $\varphi$ earlier.

### 6.2 The key assumption: how $\rho$ maps to $\theta$

**A pupil disk alone does not specify how its radius encodes propagation angle.** Turning a physical pupil coordinate into a direction requires a mapping $\rho\mapsto\theta$.

For our ideal aplanatic model, we adopt the **sine mapping**:

$$\boxed{\rho=\frac{\sin\theta}{\sin\theta_{\max}},\qquad
\sin\theta=\rho\sin\theta_{\max}.}$$

For the forward collection cone, $0\leq\theta\leq\theta_{\max}\leq\pi/2$, the inverse is

$$\theta=\arcsin\!\left(\rho\sin\theta_{\max}\right).$$

This is a **choice of optical model**: a real lens design determines the physical mapping that the model must represent. If $\rho$ is instead defined from the start as a normalized angular coordinate, the sine relationship is its definition. The optical assumption enters when we identify that angular coordinate with normalized physical pupil radius.

To see the normalization explicitly, write the dimensional sine mapping as

$$\rho_p=C\sin\theta,\qquad R_p=C\sin\theta_{\max}.$$

Then

$$\boxed{\rho=\frac{\rho_p}{R_p}
=\frac{C\sin\theta}{C\sin\theta_{\max}}
=\frac{\sin\theta}{\sin\theta_{\max}}.}$$

**$\rho_p$ and $R_p$ are lengths; $\rho$ is dimensionless.** The unknown scale $C$ cancels. This sine mapping is associated with the Abbe sine condition; telecentricity alone does not establish it [\[3\]](#ref-3).

### 6.3 Connect the pupil to direction cosines and wavevectors

Use $\mathrm{NA}=n\sin\theta_{\max}$. In the centered sine-mapped model, pupil and propagation azimuths agree, giving

$$\boxed{
\sigma_x=\frac{n\sin\theta\cos\phi}{\mathrm{NA}}
=\frac{n\alpha}{\mathrm{NA}}=\frac{k_x}{k_0\mathrm{NA}}
=\frac{\lambda_0 f_x}{\mathrm{NA}},
}$$
$$\boxed{
\sigma_y=\frac{n\sin\theta\sin\phi}{\mathrm{NA}}
=\frac{n\beta}{\mathrm{NA}}=\frac{k_y}{k_0\mathrm{NA}}
=\frac{\lambda_0 f_y}{\mathrm{NA}}.
}$$

Conversely, a pupil point supplies the forward direction:

$$\alpha=\frac{\mathrm{NA}}{n}\sigma_x,\qquad
\beta=\frac{\mathrm{NA}}{n}\sigma_y,\qquad
\gamma=+\sqrt{1-\left(\frac{\mathrm{NA}}{n}\right)^2\rho^2}.$$

Multiplying $(\alpha,\beta,\gamma)$ by $k=nk_0$ gives the full wavevector. Use the refractive index, NA, and wavevector components from the **same side** of the optical system.

The direction-cosine disk ends at the **propagating limit**. The normalized pupil ends at the **collection limit**, $\sigma_x^2+\sigma_y^2=1$. A point outside the pupil can still represent a propagating wave. The transverse coordinates alone still require a hemisphere choice; a backward wave is not collected by the $+z$ cone.

### 6.4 What does aplanatic mapping imply?

At fixed wavelength and refractive index, the sine mapping gives

$$\boxed{x_p=\frac{C}{k}k_x,\qquad y_p=\frac{C}{k}k_y.}$$

**The physical pupil is a scaled map of transverse wavevector space.** Several consequences follow:

Equal changes in transverse wavevector produce equal changes in pupil position. Thus equally spaced grating orders in $k_x$ appear equally spaced across this ideal pupil, even though their angular spacing is generally unequal. An aperture selects a range of spatial frequencies through $\sigma_x=\lambda_0 f_x/\mathrm{NA}$ and $\sigma_y=\lambda_0 f_y/\mathrm{NA}$. This is why placing diffraction orders in the pupil helps determine which ones contribute to an image. Retaining $\sin\theta$ supports large propagation angles. Approximating it by $\theta$ is an additional paraxial approximation.

There is also an image-quality implication. With spherical aberration corrected, the Abbe sine condition makes the magnification consistent across pupil zones and removes coma to first order in field displacement around the chosen imaging condition [\[3\]](#ref-3). Aplanatism does not imply correction of every other aberration.

For lithography, the pupil geometry identifies collected orders. Their amplitudes and relative phases are then needed to calculate their interference in the image.

### 6.5 What is the scale factor $C$?

From the pupil edge,

$$\boxed{C=\frac{R_p}{\sin\theta_{\max}}
=\frac{nR_p}{\mathrm{NA}}.}$$

$C$ has units of length. Its relation to a quoted focal length depends on which pupil plane and focal-length convention are used.

For a simple ideal Fourier lens in air, $\rho_p=f\sin\theta$, so **$C=f$**. In a common microscope-objective convention, the pupil radius is written $R_p=f\,\mathrm{NA}$ using the objective's nominal focal length. Combining that convention with the edge formula gives **$C=nf$** and $\rho_p=nf\sin\theta$ [\[4\]](#ref-4).

**For lithography, specify the mask or wafer side.** Under the ideal sine mapping, using a chosen finite pupil plane of radius $R_p$,

$$C_{\mathrm{wafer}}=\frac{n_{\mathrm{wafer}}R_p}{\mathrm{NA}_{\mathrm{wafer}}},\qquad
C_{\mathrm{mask}}=\frac{n_{\mathrm{mask}}R_p}{\mathrm{NA}_{\mathrm{mask}}}.$$

The same pupil position can correspond to different angles on the two sides. NA alone does not determine $C$; the physical pupil scale is also required.

For an illustrative **ideal aplanatic 4f relay in air**, let $f_1$ belong to the mask-side group and $f_2$ to the wafer-side group. At their common Fourier pupil, using positive radial angles,

$$\rho_p=f_1\sin\theta_{\mathrm{mask}}
=f_2\sin\theta_{\mathrm{wafer}},\qquad
\boxed{C_{\mathrm{mask}}=f_1,\quad C_{\mathrm{wafer}}=f_2.}$$

This is an idealized relay model for understanding projection imaging, not a prescription for an entire scanner. Lithographic imaging is often introduced using a double-telecentric projection model [\[5\]](#ref-5). The finite pupil here is the internal Fourier pupil or a specified image of it; a telecentric system's formal entrance or exit pupil can lie at infinity.

With normalized coordinates, $C$ disappears. For wafer-side angles,

$$\rho=\frac{n_{\mathrm{wafer}}\sin\theta_{\mathrm{wafer}}}
{\mathrm{NA}_{\mathrm{wafer}}}.$$

We can therefore study order directions and collection without knowing the physical pupil size.

### 6.6 Ideal mapping, real lithography systems, and Hopkins

**We use an ideal aplanatic reference system to introduce the geometry of lithographic imaging.** Real projection systems are highly corrected but have finite residual aberrations over their working field and aperture. These also change during operation: ASML describes measurements across the projection slit and lens manipulators that correct aberrations, including changes associated with heating [\[6\]](#ref-6).

Two aspects of the model can be treated separately. We can retain the ideal sine mapping while adding residual phase errors to the pupil function. When needed, a more detailed model can also include deviations in the physical pupil mapping.

The definitions $\sigma_x=k_x/(k_0\mathrm{NA})$ and $\sigma_y=k_y/(k_0\mathrm{NA})$ remain useful as **angular coordinates**. Equating them to $x_p/R_p$ and $y_p/R_p$ is the step that assumes the ideal physical mapping.

**Hopkins' canonical pupil coordinates fit here:** they organize pupil directions for diffraction-based image calculations. His general construction also handles local reference directions and generalized numerical apertures. The centered circular-pupil model above is the starting point we need; the general theory accommodates more complex systems [\[7\]](#ref-7).
