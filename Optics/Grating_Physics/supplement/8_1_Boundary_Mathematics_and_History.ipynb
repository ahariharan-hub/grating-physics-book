## 1. From a closed boundary to the aperture

Move **Transformation** and watch the same closed surface become $S=A\cup B\cup C$: opening $A$, the boundary immediately behind the opaque region $B$, and distant closure $C$. The intermediate shapes are geometric schematics; the integral plotted at the endpoint uses a true hemisphere.

Then increase **R_C**. The closure recedes while $P$ stays at the same distance from the plane. This changes a mathematical boundary, not the detector's far-field distance. Under the outgoing radiation condition,

$$\int_C(U\partial_nG-G\partial_nU)\,dA\longrightarrow0.$$

It is the **combined contribution** on $C$ that vanishes. The representation over $A+B$ still uses actual boundary data. Its plotted reference check has no physical screen and does not claim to solve screen scattering. [1, sections 18.2-18.3](#ref-1)

On the destination plane, $Q=(\xi,\eta,0)$ and $P=(x,y,s)$, with $s=-z>0$; $\xi,\eta$ sweep the reference plane. The normal points toward positive $s$.

## 2. Shared-face cancellation: why only the boundary remains

Remove a small sphere $S'$ around $P$ and call the remaining volume $V$. Neither field has a source in this punctured region. Multiplying their equations crosswise and subtracting cancels the matching $k^2UG$ terms:

$$U\nabla^2G-G\nabla^2U=0\quad\text{in }V.$$

Green's identity, an integration-by-parts statement, gives our inward-normal form:

$$\int_V(U\nabla^2G-G\nabla^2U)\,dV
=-\int_{S\cup S'}(U\partial_nG-G\partial_nU)\,dA=0.$$

Each internal face is counted twice with opposite normals. Those contributions cancel, leaving the exterior surface **and the hole**. This is mathematical bookkeeping, not destructive interference of light. The cells and enlarged hole are schematic. We have not discarded the singularity; we have given it its own boundary. [1, sections 18.2-18.3](#ref-1), [13, pp. 3-6](#ref-13)

## 3. The small sphere: how the value at P appears

On $S'$, the normal into $V$ points **away from $P$**. As its radius $\epsilon$ shrinks,

$$\operatorname{area}(S')=4\pi\epsilon^2,\qquad
\partial_nG\sim-\frac{1}{4\pi\epsilon^2},$$

$$\int_{S'}U\partial_nG\,dA\longrightarrow-U(P),\qquad
-\int_{S'}G\partial_nU\,dA\longrightarrow0.$$

The sphere is shrinking, but its contribution is not disappearing. The derivative is doing some compensating. The plot divides each complex term by the known $U(P)$: solid curves show real parts, dotted curves show imaginary parts. **Their complete sum already equals $-U(P)$ at every admissible finite radius.** Shrinking explains the separate limits; it does not repair an initially incorrect total.

Thus the outer boundary reconstructs the hidden field:

$$\boxed{U(P)=\int_S\left[U(Q)\partial_{n_Q}G(Q,P)
-G(Q,P)\partial_{n_Q}U(Q)\right]dA_Q.}$$

This is Kirchhoff's integral theorem for the free-space Green function. The two data are the complex field value and its spatial normal slope, $h=\partial_nU$. A slope may describe phase change even when field magnitude is constant. They must belong to one compatible solution.

**Try it:** change **epsilon**. The actual exclusion sphere and its normal move with the selected radius. Read its area, derivative weight, and both complex contributions at that radius; the total remains $-U(P)$.

### Why the distant closure disappears

The outgoing condition has the form $R_C(\partial_{R_C}U-ikU)\to0$, with the same condition for $G$ and both fields spreading as $1/R_C$. Their matching leading derivative terms cancel in $U\partial_nG-G\partial_nU$. The remaining combined surface integral tends to zero as the closure expands. This is the limiting argument used in the main geometry explorer; it introduces no paraxial or far-field approximation at the fixed observation point.

## Boundary compatibility and historical notation

The main text needs only the distinction between a valid representation and independently guessed data. The following gives the detailed compatibility illustration and the source-specific notation.

**See why the second zero is an extra assumption.** The left panel has zero field at the plane but a nonzero normal slope. It obeys Helmholtz's equation. The right panel imposes both zeros: only the everywhere-zero solution survives. Trying to make the field become nonzero while retaining both zeros, as the orange cross-marked trial does, violates that equation.

![Two Helmholtz boundary examples: zero field with nonzero slope is possible; zero value and zero slope force the zero solution, while a nonzero trial curve fails the equation.](figures/8_1_boundary_consistency.svg)

**Read the curves:** here $A,B$ are coefficients, not names of screen regions. $U=A\cos(ks)+B\sin(ks)$ has $U(0)=A$ and $h(0)=kB$. One zero fixes one coefficient; both zeros remove both coefficients. This is a one-dimensional illustration, not the outgoing aperture solution. The same compatibility issue extends to data over a finite surface region of the opaque screen $B$, away from its edge, in the source-free aperture problem.

**The trial curve is not the Kirchhoff reconstruction.** The field calculated by Kirchhoff's integral still satisfies Helmholtz's equation away from the boundary. Its problem is that it cannot reproduce **both assumed boundary data** exactly. The orange trial only illustrates why arbitrary boundary assignments need not belong to a valid solution.

**Where is the contradiction?** Setting both $U=0$ and $h=\partial_nU=0$ on a finite surface region of the opaque screen $B$, away from its edge, forces the source-free Helmholtz solution to vanish throughout the connected propagation region. This contradicts the nonzero field prescribed in the aperture opening $A$. The Kirchhoff reconstruction therefore cannot reproduce both assumed boundary values exactly. The theorem remains valid; the approximation to its boundary data causes the conflict. [1, §18.2](#ref-1), [13, p. 7](#ref-13)

**Reading the historical paper [5].** Our sine-curve example is a modern compatibility illustration, not Poincare's original proof. Section 5 reconstructs his argument using Green integrals for illuminated and shadowed surfaces; section 8 discusses the separate test of whether the calculated field recovers its assumed boundary values. In its Figs. 6-8, $B$ is the illuminated screen face and $C$ the shadowed face. Our simpler planar notation uses $B$ for the opaque screen region, viewed from the propagation side; its role corresponds to that shadowed face $C$. The opening is $A$ in both descriptions. [5, sections 5, 8](#ref-5)
