## Geometry for the arrival experiment

A one-dimensional grating has identical openings spaced by a period $d$ along $x$, with grooves along $y$. The transmitted normal is $+s$, corresponding to $-z$ in notebook two. Both exterior media are the same homogeneous, isotropic, lossless medium. The drawings put $+s$ upward and $+x$ to the right.

$\lambda$ is the wavelength **in that medium**; all lengths in an explorer use micrometres. $\theta$ is the signed **observation angle**, measured from $+s$ toward $+x$. $\theta_i$ is the incident propagation angle with the same convention. Begin with $\theta_i=0$. All waves and observation directions lie in the $x$–$s$ plane.

For each chosen direction, compare contributions at the **same far-field point $P$**. Changing $\theta$ selects a different observation point. Distance to $P$ is much greater than the illuminated array size, so neighboring path differences can be calculated using the far-field approximation.
- The interactive path drawing compresses that distance to keep both the openings and $P$ visible; its drawn lengths are schematic.

## 1. Geometric path difference: begin with two openings

Take neighboring openings $A$ at $x=0$ and $B$ at $x=d$. Illuminate them coherently with a plane wave. Compare their contributions at the **same distant observation point**, in a direction $\theta$ from the transmitted normal. In the far field, the outgoing paths are approximately parallel for calculating their difference.

### Normal incidence

The incident phase is the same at both openings. Measured as the path through $A$ minus the path through $B$, the outgoing path difference is

$$\Delta L=d\sin\theta.$$

A path difference corresponds to an arrival-phase difference. With $k=2\pi/\lambda$,

$$\delta=k\Delta L=\frac{2\pi}{\lambda}d\sin\theta.$$

The two contributions agree in phase when $\delta=2\pi m$, with $m$ any integer. Therefore

$$d\sin\theta_m=m\lambda.$$

**The order number $m$ counts signed complete phase cycles between neighboring contributions.** Zero cycles and one complete cycle both give phase agreement. A half-cycle gives opposite phases and, for equal amplitudes, cancellation. The two-opening interference pattern varies continuously with angle; the equation identifies its constructive-interference directions.



### Put numbers on the paths

Keep $\lambda=1\,\mu\mathrm m$ and $d=2.5\,\mu\mathrm m$ at normal incidence:

| Observation direction | Neighboring path difference | Contributions at $P$ |
| --- | --- | --- |
| $0^\circ$ | $0$ | Same phase |
| $\arcsin(0.2)\approx11.54^\circ$ | $\lambda/2$ | Opposite phases; two equal contributions cancel |
| $\arcsin(0.4)\approx23.58^\circ$ | $\lambda$ | Same phase again; order $m=+1$ |
| $15^\circ$ | About $0.647\lambda$ | Different phases; partial cancellation for two openings |

We are selecting where to observe waves from the openings. Each opening contributes a wave to $P$; their relative arrival phases determine how they add there.

## 2. Add more openings: repeat the arrival-phase step

Return to normal incidence. Add opening C another distance $d$ along the grating, then D, and so on. For any fixed far-field point $P$, each extra step along the grating gives the **same neighboring path difference** $d\sin\theta$.

If A and B arrive in phase, C agrees with B, D agrees with C, and all contributions reinforce at $P$. If neighboring arrivals differ in phase, that offset accumulates across the array.

### Express the arrival phases mathematically

Number the openings by $j=0,1,\ldots,N-1$, with A at $j=0$. At normal incidence,

$$\delta=\frac{2\pi}{\lambda}d\sin\theta.$$

With time convention $e^{-i\omega t}$ and A as the phase reference, the complex contribution from opening $j$ has relative phase factor

$$a_j=e^{-ij\delta}.$$

The minus sign follows our path convention: $\delta$ is the phase through A minus the phase through B. The real curves at $P$, scaled to unit single-opening amplitude, are therefore

$$u_j(P,t)=\operatorname{Re}\{a_je^{-i\omega t}\}=\cos(\omega t+j\delta).$$

If $B(\theta)$ contains the common single-opening field contribution, the sum is

$$U_N(P)=B(\theta)\sum_{j=0}^{N-1}e^{-ij\delta}.$$

For two openings the normalized array intensity is $\cos^2(\delta/2)$. For $N$ openings it is

$$\mathcal I_N(\theta)=\frac{1}{N^2}\left|\sum_{j=0}^{N-1}e^{-ij\delta}\right|^2
=\left[\frac{\sin(N\delta/2)}{N\sin(\delta/2)}\right]^2,$$

with limiting value $1$ at $\delta=2\pi m$. We evaluate the coherent sum directly to avoid dividing by zero.

### Try two, four, eight, and thirty-two openings

Use the **same arrival explorer above**. First follow $m=+1$ and increase $N$ from $2$ to $4$, $8$, and $32$. Every arrival remains in phase. The principal peak stays at the same angle and becomes narrower.

Then set the observation angle to $15^\circ$ and repeat. Additional openings bring additional phase offsets; inspect the time curves and the angular pattern. Intensity at a fixed off-order angle need not decrease monotonically with $N$, because finite arrays have sidelobes.

A **finite** illuminated array has a continuous angular pattern with finite-width peaks. The integer $m$ labels its principal phase-agreement directions. Exactly discrete spatial orders belong to the ideal **infinite** periodic grating.

The plotted normalization compares shape: each array-factor principal peak is scaled to $1$. The single-opening envelope $|B(\theta)|^2$ is omitted. Order strength also depends on the physical opening or groove response, and an allowed order can have zero amplitude.

### Now tilt the illumination: include the incoming path

At tilted incidence, the two openings are already driven with different phases. Compare total paths from a **common incident wavefront** to the observation point. Using the same $A$-minus-$B$ convention, the incident contribution is $-d\sin\theta_i$ and the outgoing contribution is $+d\sin\theta$. Thus

$$\Delta L=d(\sin\theta-\sin\theta_i),\qquad
\delta=kd(\sin\theta-\sin\theta_i).$$

Phase agreement requires

$$\boxed{d(\sin\theta_m-\sin\theta_i)=m\lambda}.$$

The minus sign accounts for the phase offset that the incident wave has already supplied. In particular, $m=0$ has $\theta_0=\theta_i$ in our common-medium transmission geometry.

The incoming phase at opening $j$ supplies $e^{ikjd\sin\theta_i}$. Propagation from that opening to $P$ supplies $e^{-ikjd\sin\theta}$, after removing a common factor. Together they give $e^{-ij\delta}$ with the new $\delta$ above. The arrival-sum calculation is otherwise unchanged.

## 3. Fourier series: describe the repeated field

We have established the order directions by adding arrivals at P. Now describe the same selection through the translation symmetry of an infinite grating. Shifting the structure by one period reproduces it; the incident plane wave acquires the phase factor $e^{ik_{x,i}d}$. The driven field inherits that same translation rule. Here $k_{x,i}=k\sin\theta_i$, and an outgoing plane-wave component has $k_{x,m}=k\sin\theta_m$. At the surface:

$$U(x+d,0)=e^{ik_{x,i}d}U(x,0).$$

This is Floquet quasi-periodicity. Removing the incident phase gives a periodic function $P(x)=U(x,0)e^{-ik_{x,i}x}$, so

$$P(x)=\sum_{m=-\infty}^{\infty}c_m e^{imGx},\qquad G=\frac{2\pi}{d}.$$

Restore the incident phase:

$$U(x,0)=\sum_{m=-\infty}^{\infty}c_m e^{i(k_{x,i}+mG)x}.$$

Each term has tangential wavevector $k_{x,m}=k_{x,i}+mG$. Using the angle definitions,

$$k\sin\theta_m=k\sin\theta_i+m\frac{2\pi}{d},$$

and therefore

$$\boxed{d(\sin\theta_m-\sin\theta_i)=m\lambda}.$$

The coefficients $c_m$ determine amplitude and phase. Their values are unnecessary for identifying the allowed wavevectors.

### Optional: interpret the EMPossible copied-wavefront illustration

This section connects the Fourier description to the [EMPossible animation by Raymond Rumpf](https://empossible.net/why-diffraction-order-exist-only-at-discrete-angles/). The arrival argument above already explains why the principal order directions occur; this is a second representation of the translation condition.

Take one proposed Fourier component,

$$U(x,s)=e^{i(k_xx+k_ss)},\qquad k_x=k\sin\theta,\quad k_s=k\cos\theta.$$

It is a continuous plane wave for any real $\theta$. To be a component of the infinite illuminated grating's field, it must also acquire the correct factor under a shift by $d$:

$$e^{ik_xd}=e^{ik_{x,i}d}
\quad\Longleftrightarrow\quad(k_x-k_{x,i})d=2\pi m.$$

The optional drawing compares that continuous component (left) with a construction made by copying one cell (right). Write $x=jd+\xi$, with $0\leq\xi<d$. Each copy is assigned the incident phase offset:

$$U_{\rm copy}(jd+\xi,s)=e^{ik_{x,i}jd}\,e^{i(k_x\xi+k_ss)}.$$

At an order angle the copies join and reproduce the continuous component. At other angles they have mismatches at their edges. **Those seams belong to the constructed picture.** They are imaginary cell boundaries in a homogeneous medium, and the mismatched image does not represent a physical scattered field.

The complex mismatch $|e^{i\delta}-1|$ compares the complete phase, even when just the displayed real part happens to agree at a particular instant. Colors show $\operatorname{Re}U$ at $t=0$; arrows show the outgoing component's direction. The upper panels always show five cells, independently of $N$ in the lower finite-array plot.

## 4. Reciprocal-space wave-vector matching

A change in tangential wavevector must accumulate an integer multiple of $2\pi$ across a period:

$$e^{i(k_{x,m}-k_{x,i})d}=1
\quad\Longrightarrow\quad
(k_{x,m}-k_{x,i})d=2\pi m.$$

The reciprocal-grating spacing is $G=2\pi/d$, giving

$$\boxed{k_{x,m}-k_{x,i}=mG}.$$

This is tangential wave-vector conservation modulo a reciprocal-grating vector. Substituting $k_{x,m}=k\sin\theta_m$, $k_{x,i}=k\sin\theta_i$, and $k=2\pi/\lambda$ gives

$$\boxed{d(\sin\theta_m-\sin\theta_i)=m\lambda}.$$

The grating supplies a discrete tangential wavevector change. The normal component is determined separately by the surrounding medium's wave equation.

## One equation, and the step toward the Ewald sphere

The four arguments express the same condition:

$$\underbrace{d(\sin\theta_m-\sin\theta_i)=m\lambda}_{\text{path lengths}}
\quad\Longleftrightarrow\quad
\underbrace{kd(\sin\theta_m-\sin\theta_i)=2\pi m}_{\text{phases}}
\quad\Longleftrightarrow\quad
\underbrace{k_{x,m}-k_{x,i}=mG}_{\text{wavevectors}}.$$

The wave equation adds the in-plane dispersion circle

$$k_{x,m}^{\,2}+k_{s,m}^{\,2}=k^2.$$

Each candidate $k_{x,m}=k_{x,i}+mG$ defines a line parallel to the $k_s$ axis. Its intersection with the circle gives a real outgoing direction; choose $k_{s,m}>0$ for transmission. Tangency is grazing propagation. No real intersection means an evanescent order.

In three dimensions the circle becomes a sphere. Equivalently, define the scattering vector $\Delta\mathbf k=\mathbf k_m-\mathbf k_i$. Then

$$|\Delta\mathbf k+\mathbf k_i|=k.$$

This is an **Ewald sphere centered at $-\mathbf k_i$ in scattering-vector space**. A surface grating supplies reciprocal-space rods $\Delta k_x=mG$, $\Delta k_y=0$, with unrestricted normal component; their intersections with the sphere identify the orders.

## A small numerical example

Change the values below and run the two code cells to evaluate the standard 1D equation.

```python
import numpy as np

# Edit these values, then run all cells. Use the same units for d and wavelength.
wavelength_nm = 193.0
period_nm = 800.0
theta_i_deg = 0.0
orders = np.arange(-5, 6)
```

Calculate each direction using

$$\theta_m=\sin^{-1}\!\left(\sin\theta_i+\frac{m\lambda}{d}\right).$$

A real outgoing angle exists only when the inverse-sine argument lies between $-1$ and $+1$. Equality means grazing propagation; outside that range, the order is evanescent.

```python
if not np.isfinite(wavelength_nm) or wavelength_nm <= 0:
    raise ValueError('Wavelength must be positive and finite.')
if not np.isfinite(period_nm) or period_nm <= 0:
    raise ValueError('Period must be positive and finite.')
if not np.isfinite(theta_i_deg) or abs(theta_i_deg) >= 90:
    raise ValueError('The incident angle must lie between -90 and +90 degrees.')

sin_theta_m = np.sin(np.deg2rad(theta_i_deg)) + orders * wavelength_nm / period_nm
tolerance = 1e-12

print(f'{"Order m":>8}  {"Angle (degrees)":>16}  Status')
for m, sine in zip(orders, sin_theta_m):
    if abs(sine) > 1 + tolerance:
        print(f'{m:8d}  {"--":>16}  Evanescent')
    else:
        angle_deg = np.rad2deg(np.arcsin(np.clip(sine, -1, 1)))
        status = 'Grazing' if abs(abs(sine) - 1) <= tolerance else 'Propagating'
        print(f'{m:8d}  {angle_deg:16.2f}  {status}')
```
