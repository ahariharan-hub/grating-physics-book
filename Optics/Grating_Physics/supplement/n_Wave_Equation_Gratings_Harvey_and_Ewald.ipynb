# From the wave equation to Harvey and Ewald

**Where can a diffracted order travel?** The answer combines two constraints: the wave equation determines which plane waves the surrounding medium supports, and the grating's periodicity determines which wavevector changes are allowed. Harvey's direction-cosine diagram and the Ewald construction express these same constraints in different coordinates.

**Medium and wavelength convention.** The surrounding medium is homogeneous, isotropic and lossless.

- $\lambda$ is wavelength **in the medium**: $\lambda=\lambda_0/n$ and $k=nk_0=2\pi/\lambda$.
- The controls vary medium wavelength; there is no separate index control.

See the [course medium convention](../0_0_Foundations_of_Scalar_Optics.ipynb#medium-and-wavelength).

This is the **kinematics of diffraction**: we identify possible orders before calculating their amplitudes or powers. An allowed order can have zero amplitude. Here, “kinematics” does not assume weak scattering.

## Geometry and notation

An infinite, stationary grating lies in the $xy$-plane, with period $d$ along $x$ and uniform grooves along $y$. Light arrives from $z<0$; transmitted waves travel toward $+z$.

Both exterior regions contain the **same homogeneous, isotropic, lossless, nondispersive medium**, with wave speed $v$.

$\lambda$ is the wavelength in that medium; $k=2\pi/\lambda=\omega/v$. $\mathbf{k}_i$ and $\mathbf{k}_m$ are actual incident and outgoing **propagation vectors**. $\mathbf{G}=(2\pi/d)\hat{\mathbf{x}}$, and our order label satisfies $k_{x,m}-k_{x,i}=mG$. The real scalar field is $u(\mathbf{r},t)=\operatorname{Re}\{U(\mathbf{r})e^{-i\omega t}\}$.

We use $z$ for the surface normal, as in the plane-wave notebook; it plays the role called $s$ in the earlier grating-equation notebook.

## 1. The wave equation supplies the sphere

For a scalar wave in a medium with constant speed $v$,
$$
\nabla^2u-\frac{1}{v^2}\frac{\partial^2u}{\partial t^2}=0.
$$

With time dependence $e^{-i\omega t}$ this becomes the **Helmholtz equation**,
$$
(\nabla^2+k^2)U=0,\qquad k=\frac{\omega}{v}.
$$

Substitute the plane wave $U=Ae^{i\mathbf{k}_m\cdot\mathbf{r}}$. Since each spatial derivative brings down a component of $i\mathbf{k}_m$,
$$
\boxed{k_{x,m}^2+k_{y,m}^2+k_{z,m}^2=k^2.}
$$

For real components, this is a sphere in wavevector space. Its radius is set by wavelength. **The sphere exists even before a grating is introduced:** it is the fixed-frequency dispersion surface of the medium. A stationary grating preserves frequency.

The grating must select outgoing waves from these solutions. [Background: TU Delft, angular-spectrum propagation](https://interactivetextbooks.tudelft.nl/tn2421/content/Chap6_Diffraction/DiffractiveOptics_2022_01Clean.html#angular-spectrum-method).

## 2. Grating periodicity supplies discrete tangential changes

Moving one period along the grating changes the incident phase by $k_{x,i}d$. The total field has the same translation factor:
$$
U(x+d,y,0)=e^{ik_{x,i}d}U(x,y,0).
$$

This is **Floquet quasi-periodicity**. Each plane-wave order must satisfy it, so
$$
e^{i(k_{x,m}-k_{x,i})d}=1
\quad\Longrightarrow\quad
\boxed{k_{x,m}=k_{x,i}+mG,\qquad G=\frac{2\pi}{d}.}
$$

The structure is uniform along $y$, giving $k_{y,m}=k_{y,i}$. Equivalently, its surface field has a Fourier expansion
$$
U(x,y,0)=e^{ik_{y,i}y}\sum_{m=-\infty}^{\infty}
c_m e^{i(k_{x,i}+mG)x}.
$$

The coefficients $c_m$ contain amplitude and phase. The order directions follow without knowing them.

For incidence confined to the $xz$-plane, signed angles measured from $+z$ toward $+x$ give the familiar equation
$$
\boxed{d(\sin\theta_m-\sin\theta_i)=m\lambda.}
$$

## 3. An allowed order satisfies both constraints

For each integer $m$, periodicity fixes $k_{x,m}$ and $k_{y,m}$. The wave equation then gives the transmitted normal component:
$$
\boxed{k_{z,m}=+\sqrt{k^2-(k_{x,i}+mG)^2-k_{y,i}^2}.}
$$

The positive real root selects transmission toward $+z$.

Positive radicand: a propagating transmitted order. Zero radicand: a grazing order, along the grating plane. Negative radicand: choose $k_{z,m}=+i\kappa_m$, $\kappa_m>0$, so $e^{ik_{z,m}z}=e^{-\kappa_m z}$. This order is evanescent and has no real outgoing direction on the sphere.

Geometrically, each order defines a line parallel to $k_z$. Intersections with the upper hemisphere give the propagating transmitted directions.

## 4. Harvey: describe the outgoing directions

Normalize each propagating wavevector:
$$
\hat{\mathbf{k}}_m=\frac{\mathbf{k}_m}{k}
=(\alpha_m,\beta_m,\gamma_m),\qquad
\alpha_m^2+\beta_m^2+\gamma_m^2=1.
$$

These are **direction cosines**: cosines of the angles with the $x$, $y$, and $z$ axes. Projecting the transmitted hemisphere onto the $\alpha\beta$-plane gives the unit disk:
$$
\boxed{
\alpha_m=\alpha_i+m\frac{\lambda}{d},\qquad
\beta_m=\beta_i,\qquad
\gamma_m=+\sqrt{1-\alpha_m^2-\beta_m^2}.
}
$$

Harvey's useful observation is that order positions are equally spaced by $\lambda/d$ along a straight row in this disk. Their angular spacing is generally unequal. The disk boundary marks grazing; points outside it cannot be lifted to real points on the sphere.

On the sphere, $\beta_m=\beta_i$ defines a circle. When $\beta_i\ne0$, it is a latitude circle and the rays lie on a cone. Incidence within the $xz$-plane has $\beta_i=0$ and gives a great-circle section.

An observation hemisphere can be centred at the illuminated point on the grating. Dividing ray-intersection positions by its radius gives this direction-space construction.

See [Harvey and Vernold (1998)](https://doi.org/10.1364/AO.37.008158) and [Harvey and Pfisterer (2019), Section 3](https://doi.org/10.1117/1.OE.58.8.087105).

## 5. Ewald: describe the change in wavevector

Define the **scattering vector**
$$
\Delta\mathbf k_m=\mathbf{k}_m-\mathbf{k}_i.
$$

Writing the same wave equation in these coordinates gives
$$
\boxed{|\Delta\mathbf k_m+\mathbf{k}_i|^2=k^2.}
$$

This is an Ewald sphere of radius $k$, centred at $-\mathbf{k}_i$ in $\Delta\mathbf k$-space. Because $|\mathbf{k}_i|=k$, it passes through $\Delta\mathbf k=0$. That point is the directly transmitted, zero-order beam. [IUCr: Ewald sphere](https://dictionary.iucr.org/Ewald_sphere).

For our surface grating, periodicity requires
$$
\Delta k_{x,m}=mG,\qquad \Delta k_{y,m}=0,
$$
while it does not fix $\Delta k_z$. These constraints define parallel **reciprocal-space rods**. Their intersections with the Ewald sphere identify the orders; transmission selects $\Delta k_z+k_{z,i}>0$, with equality at grazing. [Example of the rod construction](https://journals.iucr.org/m/issues/2017/04/00/hf5342/hf5342fig3.html).

A surface grating therefore constrains the **tangential** wavevector change. Writing the full vector equation $\Delta\mathbf k_m=m\mathbf{G}$ would incorrectly fix its normal change to zero.

Crystallography often uses radius $1/\lambda$ instead of $2\pi/\lambda$. Both conventions work when reciprocal vectors use the same scaling.

## 6. The exact connection

| Representation | Coordinate of an order | Sphere centre | Radius |
| --- | --- | --- | --- |
| Wavevector space | $\mathbf{k}_m$ | $\mathbf{0}$ | $k$ |
| Harvey's direction space | $\hat{\mathbf{k}}_m=\mathbf{k}_m/k$ | $\mathbf{0}$ | $1$ |
| Ewald scattering-vector space | $\Delta\mathbf k_m=\mathbf{k}_m-\mathbf{k}_i$ | $-\mathbf{k}_i$ | $k$ |

The coordinate transformation is
$$
\boxed{\hat{\mathbf{k}}_m=\frac{\Delta\mathbf k_m+\mathbf{k}_i}{k}.}
$$

Translate the Ewald sphere by $\mathbf{k}_i$, then scale by $1/k$: it becomes the direction-cosine sphere. Project its transmitted hemisphere to obtain Harvey's disk. **The allowed waves are the same in all three descriptions.**

**Reading signs in Harvey's figures.** This notebook always uses actual propagation vectors. In a reflection observation hemisphere, the outward radius to the incident entry point points opposite to the incoming propagation vector; incident-point coordinates therefore have reversed signs. Also, Harvey and Pfisterer's 2019 Eq. (3) numbers planar transmission orders oppositely to our equation when the same signed angles are used: $m_{\mathrm{H}}=-m$.
- Keep coordinate directions and order labels explicit when comparing diagrams.

## 7. One oblique-transmission example

Use polar angle $\theta_i$ from $+z$ and azimuth $\varphi_i$ from $+x$ toward $+y$:
$$
\hat{\mathbf{k}}_i=
(\sin\theta_i\cos\varphi_i,\;
 \sin\theta_i\sin\varphi_i,\;
 \cos\theta_i).
$$

The table compares each real direction with its normalized scattering vector $\Delta\mathbf k_m/k$. For an evanescent order, it lists the candidate disk position instead of assigning a real direction.

Select the **Anaconda Python** kernel and choose **Run All**. Wavelength and period must use the same length unit.

```python
# Find the shared wave_optics package in the project root.
from pathlib import Path
import sys
import importlib

_support_dir = next((base
    for base in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (base / 'wave_optics' / '__init__.py').is_file()), None)
if _support_dir is None:
    raise FileNotFoundError('Open this notebook within the Grating_Physics folder.')
if str(_support_dir) not in sys.path:
    sys.path.insert(0, str(_support_dir))
# Relocate a cached package without changing the lesson's module reload order.
importlib.invalidate_caches()
if 'wave_optics' in sys.modules:
    sys.modules['wave_optics'].__path__[:] = [str(_support_dir / 'wave_optics')]

from IPython.display import Markdown, display
from wave_optics.diffraction import diffraction_orders
from wave_optics.plotting import diffraction_order_table, diffraction_residual_report

# Edit these inputs, then choose Run All.
wavelength_nm = 500.0
period_nm = 1000.0
theta_i_deg = 30.0
phi_i_deg = 30.0
orders = range(-3, 4)
```

```python
example = diffraction_orders(
    wavelength_nm, period_nm, theta_i_deg, phi_i_deg, orders=orders)
display(Markdown(diffraction_order_table(example)))
print(diffraction_residual_report(example))
```

The default inputs give four propagating orders: $m=-2,-1,0,+1$. All have the same $\beta_m=0.25$, and their $\alpha_m$ values are separated by $0.5$. The zero order has $\hat{\mathbf{k}}_0=\hat{\mathbf{k}}_i$ and $\Delta\mathbf k_0=\mathbf{0}$.

**Try two changes:**

1. Set $\varphi_i=0^\circ$: the orders move to the $\beta=0$ great-circle section.
2. Set $\theta_i=0^\circ$ and $\lambda=d$: $m=\pm1$ reach grazing, while $|m|>1$ is evanescent.

For different incident and transmission media, use $k_i$ and $k_t$ separately. The transmitted sphere has radius $k_t$, and $\alpha_m=(k_i/k_t)\alpha_i+mG/k_t$. The equal-radius construction above assumes the same medium on both sides.
