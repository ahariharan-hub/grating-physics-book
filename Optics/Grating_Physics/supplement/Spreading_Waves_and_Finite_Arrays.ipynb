## 1. From one opening to many: how orders emerge

**Spreading contributions from individual openings collectively produce the directional diffraction orders.** We can now watch that happen by changing just the number of illuminated openings, $N$.

For this illustration, each opening is approximated as a very narrow source of a cylindrical wave: its wavefronts look circular in our 2D picture. All sources have the same drive amplitude. Their phases come from the incident wave, so neighbouring source phases differ by the familiar $k_{x,i}d$.

This notebook has its own **wavelength, spacing, incidence and number of openings** controls. It starts with one opening and normal incidence. Start with $\lambda=2\,\mu\mathrm m$, $d=4\,\mu\mathrm m$, and click **Normal incidence**, so every opening is driven at the same phase. Then keep those three settings fixed.

- **Left:** a snapshot of the **sum** of the spreading fields from all illuminated openings. Orange dots mark the openings. The view keeps the same physical size as you change $N$.
- **Right:** the resulting **far-field directional pattern**. Height indicates relative strength, obtained by squaring the magnitude of the summed field. Orange dashed lines mark the integer-order directions from the phase condition $(k_x-k_{x,i})d=2\pi m$.

The right curve is the far-field limit; it is not a measurement along the bottom edge of the left picture. Every illuminated opening contributes to this sum.

**Change only N:**

1. **One opening:** follow the curved wavefronts. There is no neighbouring opening to interfere with it. Our ideal narrow source has a broad, uniform far-field response.
2. **Two openings:** the two fields overlap and add. At normal incidence, a path difference of $m\lambda$ gives phase agreement; a half-cycle difference gives cancellation in the far field. Both effects appear in the angular curve.
3. **Five, then fifteen openings:** every new contribution must agree with the others to reinforce strongly. The principal peaks become narrower around the **same order directions**. For these settings they are $0^\circ$, $\pm30^\circ$, and the grazing limits $\pm90^\circ$.

- The plots normalize their values to compare shapes.
- The left uses enhanced colour contrast and omits the first wavelength around each source, where the simple spreading-wave approximation is unsuitable.
- This illustration treats the openings as prescribed narrow emitters; it does not calculate the finite-width slit envelope or actual order efficiencies.

## 2. How the spreading-wave figure is calculated

**Every plotted point is calculated by adding the contributions from the openings.** Consider a point $P=(x,z)$ and an opening at $(x_j,0)$.

**First find the travel distance:**

$$r_j=\sqrt{(x-x_j)^2+z^2}.$$

**Then calculate the contribution from that opening.** The illustration uses an outgoing cylindrical-wave approximation, away from the source:

$$u_j(x,z,t)=\frac{\cos\!\left(kr_j+k_{x,i}x_j-\omega t\right)}{\sqrt{r_j}}.$$

| Term | Meaning in this picture |
| --- | --- |
| $k_{x,i}x_j$ | Phase supplied by the incident wave at opening $j$ |
| $kr_j$ | Phase accumulated while travelling from that opening to $P$ |
| $1/\sqrt{r_j}$ | Decrease in amplitude as this cylindrical wave spreads |
| $-\omega t$ | Common time evolution of every contribution |

The source amplitude has an arbitrary common scale here; we compare normalized patterns. The approximation omits a common cylindrical-wave phase offset, which only changes the time origin of the snapshot, not the relative phases or intensities.

**Finally add all contributions** and take $t=0$:

$$u_{\mathrm{total}}(x,z,0)=\sum_j u_j(x,z,0).$$

The code keeps amplitude and phase together using the phasor notation from notebook one:

$$\widetilde u_j=\frac{e^{i(kr_j+k_{x,i}x_j)}}{\sqrt{r_j}},\qquad
\widetilde u_{\mathrm{total}}=\sum_j\widetilde u_j,\qquad
u_{\mathrm{total}}(x,z,0)=\operatorname{Re}\widetilde u_{\mathrm{total}}.$$

- **How this becomes colour:** repeat the calculation across the $x$–$z$ grid, then divide the real summed field by the largest magnitude $|\widetilde u_{\mathrm{total}}|$ over the displayed, unmasked grid.
- Red is positive, blue negative and white near zero.
- The colour map uses a symmetric logarithmic contrast scale to make weaker ripples visible; colour spacing is therefore not linear in field value.
- **White at one instant does not necessarily mean zero intensity.** Blank regions within $r_j<\lambda$ of any opening are omitted because the spreading approximation is unsuitable close to a source.

For the two-opening example with $d=4\,\mu\mathrm m$, $\lambda=1\,\mu\mathrm m$ and $\theta_i=30^\circ$,

$$k_{x,i}d=\frac{2\pi}{1}\sin30^\circ\times4=4\pi.$$

The drive phases differ by **two full cycles**. Thus those two openings are driven in phase despite the tilted incidence, giving the same summed pattern as normal incidence for these parameters. This is a prescribed scalar-source illustration, not a full solution of fields inside finite-width slits.


### Small calculation 1: add the waves at one point

Reproduce the calculation behind **one point** of the summed-field image above. Use two openings at $x=0$ and $4\,\mu\mathrm m$, $\lambda=1\,\mu\mathrm m$, $\theta_i=30^\circ$, and choose $P=(2,-10)\,\mu\mathrm m$. This point lies on the perpendicular bisector of the two openings, so their travel distances are equal.

`np.hypot` computes $\sqrt{(x-x_j)^2+z^2}$ for every opening. `np.exp(1j*phase)` keeps each contribution's phase, `np.sum` adds the complex contributions, and `.real` gives the signed field at $t=0$. The cell prints the **raw sum**, before the full-image normalization and contrast mapping. The guard keeps the sample outside the excluded source neighbourhoods.

**Predict, then try:** change `theta_i_deg` to `0.0` without changing the other values. Then restore $30^\circ$ and change `wavelength` to `4.0`. At this equidistant point, will the two contributions reinforce or cancel?


```python
wavelength, d, theta_i_deg = 1.0, 4.0, 30.0
positions = np.array([0.0, d])  # two opening centres, in micrometres
x, z = 2.0, -10.0  # chosen point P
k = 2*np.pi / wavelength
k_x_i = k * np.sin(np.deg2rad(theta_i_deg))
distance = np.hypot(x-positions, z)
if np.any(distance < wavelength):
    raise ValueError('Choose a point at least one wavelength from every opening.')
source_phase = k_x_i*positions
travel_phase = k*distance
contributions = np.exp(1j*(source_phase+travel_phase)) / np.sqrt(distance)
total = np.sum(contributions)
print(' x_j (um)   r_j (um)   drive cycles    Re(part)    Im(part)')
for position, r, phi, part in zip(positions, distance, source_phase, contributions):
    print(f'{position:9.2f}   {r:8.3f}   {phi/(2*np.pi):12.3f}   {part.real:9.6f}   {part.imag:9.6f}')
print('Summed phasor:', np.round(total, 6))
print(f'Signed field at t=0: {total.real:+.6f}')
print(f'Magnitude of the summed phasor: {abs(total):.6f}')
```

<details><summary>Answer for the stated starting values and changes</summary>

Both distances are approximately $10.198\,\mu\mathrm m$. The source phases differ by two full cycles, so the contributions are identical: each is approximately $0.100428+0.296601i$. Their sum is $0.200857+0.593202i$, and its real part, approximately **$0.200857$**, is the raw signed field at this point at $t=0$.

Changing the incidence to zero gives the same result: replacing a two-cycle phase offset by zero does not change a phasor. With incidence restored to $30^\circ$ and wavelength changed to $4\,\mu\mathrm m$, the drive difference becomes half a cycle. The equal-distance contributions then cancel, leaving a sum of zero apart from numerical rounding. A single instantaneous real value and the magnitude of the summed phasor are different quantities.

</details>


## 3. Connect the sum to the phase-cycle condition

In the far-field limit, use parallel outgoing paths at angle $\theta$ measured from $-z$ toward $+x$. Let $k=2\pi/\lambda$, $k_x=k\sin\theta$, and $k_{x,i}=k\sin\theta_i$. For neighbours A and B separated by $d$, define $\Delta L$ as total route A minus total route B, including incoming and outgoing segments. Then $\Delta L=d(\sin\theta-\sin\theta_i)$, so

$$\frac{\Delta L}{\lambda}
=\frac{(k_x-k_{x,i})d}{2\pi}.$$

When that value is an integer, each added opening contributes at the same phase in the cycle. Between those directions the contributions have different phases and partially cancel. More openings make the principal peaks narrower; they do not change the integer-order angles when wavelength, spacing and incidence stay fixed. [Multi-slit interference](https://farside.ph.utexas.edu/teaching/315/Waveshtml/node93.html)

**A finite number of openings produces peaks with angular width, including some field between the ideal orders.** The exact discrete plane-wave orders of [notebook three](../3_0_Why_Diffraction_Orders_Appear.ipynb) belong to the ideal infinitely periodic grating. The present picture shows the finite-array approach to that limit; its full field is a sum of spreading contributions, not one plane wave.

After trying normal incidence, set $\theta_i=30^\circ$. The incident phase offsets now shift the reinforced directions according to the same condition $(k_x-k_{x,i})d=2\pi m$.

**Check:** if we keep $\lambda$, $d$ and $\theta_i$ fixed and increase $N$, what changes: the principal peak directions, or their widths?

<details><summary>Check this point</summary>

Their widths decrease. The principal peak directions still satisfy $(k_x-k_{x,i})d=2\pi m$; $N$ does not appear in that condition.

</details>


### Small calculation 2: reproduce the directional reinforcement

Use the interactive view's starting parameters $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$ and normal incidence. Compare the directions $0^\circ$, approximately $14.48^\circ$ and $30^\circ$, corresponding to $m_{\mathrm{test}}=0$, $0.5$ and $1$.

Far from the array, the common propagation phase and common spreading factor can be removed when comparing relative angular strengths. What remains from opening $j$ is

$$e^{i(k_{x,i}-k\sin\theta)x_j}.$$

The sign follows the **B-relative-to-A arrival phase**; it is the opposite of the route-A-minus-route-B convention used for $\Delta L$. Both give the same integer-cycle agreement condition. `np.sum` adds these remaining factors. Dividing by $N$ gives amplitude 1 when all agree; squaring its magnitude gives the normalized curve on the right of the interactive view. This normalization compares shape, not absolute power.

The opening positions use the same rule as the explorer: keep $x=0$ and add neighbours on the grating sites. `//` is integer division. **Predict, then run:** which columns stay at 1 as $N$ increases? Does a half-cycle difference cancel exactly for both even and odd $N$?


```python
wavelength, d, theta_i_deg = 2.0, 4.0, 0.0
k = 2*np.pi / wavelength
k_x_i = k * np.sin(np.deg2rad(theta_i_deg))
half_cycle_angle = np.rad2deg(np.arcsin(wavelength/(2*d)))
angles_deg = np.array([0.0, half_cycle_angle, 30.0])
print('Angles (degrees):', np.round(angles_deg, 3))
print(' N     I at 0 deg     I at half-cycle     I at 30 deg')
for N in (1, 2, 5, 15):
    positions = (np.arange(N)-(N-1)//2)*d
    intensities = []
    for angle_deg in angles_deg:
        kx = k*np.sin(np.deg2rad(angle_deg))
        phase = (k_x_i-kx)*positions
        amplitude = np.sum(np.exp(1j*phase)) / N
        intensities.append(abs(amplitude)**2)
    print(f'{N:2d}   {intensities[0]:12.6f}   {intensities[1]:17.6f}   {intensities[2]:13.6f}')
```

<details><summary>Answer for the stated starting values and changes</summary>

For the written normal-incidence parameters, the $0^\circ$ and $30^\circ$ columns are 1 for every $N$: all contributions agree at these integer-order directions. For $N=1$ the middle column is also 1, because there is only one contribution.

At approximately $14.48^\circ$, successive contributions differ by half a cycle. Two contributions cancel exactly. For odd $N$, opposite pairs cancel but one contribution remains: the normalized intensity is $1/N^2$, giving **$0.04$ for $N=5$** and approximately **$0.004444$ for $N=15$**. This illustrates why a finite array can retain some field between the ideal orders.

</details>
