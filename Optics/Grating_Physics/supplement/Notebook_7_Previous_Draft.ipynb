## 1. Add readings at the same point and time

Here “readings” are instantaneous values of the model fields. An ordinary camera records an averaged intensity-related signal; Section 4 connects that measurement to the fields.

In a linear medium, **superposition** means adding fields at the same position and time:

$$u(t)=u_1(t)+u_2(t).$$

For two unit-amplitude waves in phase, with common period $T$,

$$u_1(t)=\cos(-2\pi t/T),\qquad u_2(t)=\cos(-2\pi t/T).$$

At $t=0$, $u=1+1=2$; at $t=T/2$, $u=-1-1=-2$. The combined amplitude is 2.

## 3. Find the amplitude of the combined wave

At the chosen point, expand the two real waves using
$\cos(\phi-\omega t)=\cos\phi\cos\omega t+\sin\phi\sin\omega t$:

$$\begin{aligned}
u(t)&=A_1\cos(\phi_1-\omega t)+A_2\cos(\phi_2-\omega t)\\
&=\underbrace{(A_1\cos\phi_1+A_2\cos\phi_2)}_{a}\cos\omega t
+\underbrace{(A_1\sin\phi_1+A_2\sin\phi_2)}_{b}\sin\omega t.
\end{aligned}$$

Write the same oscillation as $u(t)=B\cos(\psi-\omega t)$.
Matching coefficients gives $a=B\cos\psi$ and $b=B\sin\psi$, so

$$\begin{aligned}
B^2&=a^2+b^2\\
&=A_1^2(\cos^2\phi_1+\sin^2\phi_1)
+A_2^2(\cos^2\phi_2+\sin^2\phi_2)\\
&\quad+2A_1A_2(\cos\phi_1\cos\phi_2+\sin\phi_1\sin\phi_2)\\
&=A_1^2+A_2^2+2A_1A_2\cos(\phi_2-\phi_1).
\end{aligned}$$

The **resultant amplitude** $B$ depends on the alignment of the cycles.
For equal unit amplitudes and $\phi_1=0$:

| $\phi_2$ | Combined real field | Amplitude $B$ |
| --- | --- | --- |
| $0^\circ$ | $2\cos\omega t$ | $2$ |
| $90^\circ$ | $\cos\omega t+\sin\omega t=\sqrt2\cos(\pi/4-\omega t)$ | $\sqrt2$ |
| $180^\circ$ | $0$ | $0$ |

The phase and addition explorers above show these same cycles. At $90^\circ$,
their peaks occur a quarter cycle apart; they neither reinforce maximally nor
cancel throughout the cycle.

### Connect back to the central quantity $U$

The scalar complex field introduced in Notebook 0 represents the same real wave:

$$U_j=A_je^{i\phi_j},\qquad
u_j(t)=\operatorname{Re}\{U_je^{-i\omega t}\}.$$

Because taking the real part is linear,

$$\begin{aligned}
u_1+u_2
&=\operatorname{Re}\{U_1e^{-i\omega t}+U_2e^{-i\omega t}\}\\
&=\operatorname{Re}\{(U_1+U_2)e^{-i\omega t}\}.
\end{aligned}$$

Thus $U=U_1+U_2=a+ib$ and $|U|=\sqrt{a^2+b^2}=B$.
This is the compact algebraic form of the real-wave addition just derived.

### Small calculation 3: recover the resultant amplitude from the cycle coefficients

**Predict:** calculate the amplitudes for phase differences of $0^\circ$,
$90^\circ$, and $180^\circ$. Then change $A_2$ to 0.5 and repeat.

````python
A1, A2 = 1.0, 1.0
phi1_deg = 0.0
phi2_deg = np.array([0.0, 90.0, 180.0])
phi1, phi2 = np.deg2rad(phi1_deg), np.deg2rad(phi2_deg)
a = A1 * np.cos(phi1) + A2 * np.cos(phi2)
b = A1 * np.sin(phi1) + A2 * np.sin(phi2)
B = np.hypot(a, b)
print('phase difference (degrees)    a      b      amplitude B')
print(np.round(np.column_stack((phi2_deg - phi1_deg, a, b, B)), 3))
````

<details><summary>Answer</summary>

For equal unit amplitudes, $B=2,\sqrt2,0$. With $A_2=0.5$, the amplitudes
are $1.5,\sqrt{1.25},0.5$. Unequal waves leave a residual oscillation even
at opposite phases. A zero resultant has no defined phase.

</details>

## 4. Add fields first, then calculate intensity

**Superposition** adds fields, $U=U_1+U_2$. **Interference** is the resulting phase-dependent enhancement or reduction of intensity.

### 4.1 Why the detector leads us to a squared magnitude

An ordinary optical detector averages over rapid field oscillations. In our scalar model, $I=C\langle u^2\rangle_T$, with fixed conversion factor $C$. For a periodic field,

$$\langle f\rangle_T=\frac{1}{T}\int_0^T f(t)\,dt.$$

For $U=Be^{i\psi}$ and $u(t)=B\cos(\psi-\omega t)$, use $\cos^2 q=(1+\cos 2q)/2$:

$$u^2(t)=\frac{B^2}{2}\left[1+\cos(2\psi-2\omega t)\right].$$

The oscillating term averages to zero over a full period:

$$\langle u^2\rangle_T=\frac{B^2}{2}(1+0)=\frac{|U|^2}{2}.$$

A many-cycle exposure approaches this average for our stable waves.

Normalize by the fixed unit-amplitude reference, $A_{\rm ref}=1$ and $I_{\rm ref}=CA_{\rm ref}^2/2$:

$$\frac{I}{I_{\rm ref}}
=\frac{C|U|^2/2}{CA_{\rm ref}^2/2}
=\frac{|U|^2}{A_{\rm ref}^2}.$$

Now substitute $U=U_1+U_2$:

$$\boxed{\frac{I}{I_{\rm ref}}=\frac{|U_1+U_2|^2}{A_{\rm ref}^2}.}$$

The field can be signed or complex; the measured intensity is nonnegative. We next expand $|U_1+U_2|^2$.

### 4.2 Squared magnitude and conjugation

For $z=a+ib$ and its conjugate $z^*=a-ib$,

$$\begin{aligned}
zz^*&=(a+ib)(a-ib)\\
&=a^2-iab+iab-i^2b^2\\
&=a^2+b^2=|z|^2,
\end{aligned}$$

Thus $|z|^2=zz^*$. For comparison, $(1+i)^2=2i$, while $|1+i|^2=(1+i)(1-i)=2$.

Also, $(U_1+U_2)^*=U_1^*+U_2^*$.

### 4.3 Expand the squared sum

Apply $|z|^2=zz^*$ to the summed field:

$$\begin{aligned}
|U_1+U_2|^2
&=(U_1+U_2)(U_1+U_2)^*\\
&=(U_1+U_2)(U_1^*+U_2^*)\\
&=U_1U_1^*+U_1U_2^*+U_2U_1^*+U_2U_2^*\\
&=|U_1|^2+|U_2|^2+U_1U_2^*+U_2U_1^*.
\end{aligned}$$

The mixed products $U_1U_2^*$ and $U_2U_1^*$ are the **cross terms**.

### 4.4 Combine the cross terms

The cross terms are conjugates: $Q=U_1U_2^*$ and $Q^*=U_2U_1^*$. Writing $Q=a+ib$,

$$Q+Q^*=(a+ib)+(a-ib)=2a=2\operatorname{Re}Q.$$

Therefore

$$\boxed{|U_1+U_2|^2
=|U_1|^2+|U_2|^2+2\operatorname{Re}(U_1U_2^*).}$$

After division by $A_{\rm ref}^2$, the last term is the **interference contribution to relative intensity**.

### 4.5 Where the cosine of phase difference comes from

For $U_j=A_je^{i\phi_j}$ with real nonnegative amplitudes,

$$U_2^*=A_2(\cos\phi_2-i\sin\phi_2)=A_2e^{-i\phi_2}.$$

Hence

$$\begin{aligned}
U_1U_2^*
&=A_1A_2e^{i\phi_1}e^{-i\phi_2}\\
&=A_1A_2e^{i(\phi_1-\phi_2)}\\
&=A_1A_2\left[\cos(\phi_1-\phi_2)+i\sin(\phi_1-\phi_2)\right].
\end{aligned}$$

Since $\phi_1-\phi_2=-\delta$ and $\cos(-\delta)=\cos\delta$,

$$2\operatorname{Re}(U_1U_2^*)=2A_1A_2\cos\delta.$$

Using $U_jU_j^*=A_j^2e^{i\phi_j}e^{-i\phi_j}=A_j^2$,

$$\boxed{\frac{I}{I_{\rm ref}}
=\underbrace{\frac{A_1^2}{A_{\rm ref}^2}}_{\text{wave 1 alone}}
+\underbrace{\frac{A_2^2}{A_{\rm ref}^2}}_{\text{wave 2 alone}}
+\underbrace{\frac{2A_1A_2\cos\delta}{A_{\rm ref}^2}}_{\text{interference term}}.}$$

At fixed amplitudes, relative phase controls the interference term.

**Experimental connection: three intensity measurements.** With stable beams
and the same calibrated detector settings, record $I_1$ with only wave 1,
$I_2$ with only wave 2, and $I_{12}$ with both present, at the same resolved
position. For our coherent scalar pair,

$$I_{12}-I_1-I_2=2\sqrt{I_1I_2}\cos\delta,
\qquad
\cos\delta=\frac{I_{12}-I_1-I_2}{2\sqrt{I_1I_2}}.$$

The subtraction isolates the interference contribution. It gives $\cos\delta$
when both beams are nonzero, not a unique signed phase: for example,
$+90^\circ$ and $-90^\circ$ give the same value. Additional measurements with a
known reference phase change can resolve that ambiguity. This is how a phase
quantity enters an intensity measurement; the detector never needs to follow
each optical cycle. See the
[TU Delft discussion of separate and combined intensities](https://interactivetextbooks.tudelft.nl/tn2421/content/Chap5_Interference/InterferenceCoherence_2022_01Clean.html#interference-of-monochromatic-fields-of-the-same-frequency).


### 4.6 Connect the resultant amplitude to intensity

For $A_1=A_2=A_{\rm ref}=1$ and $\phi_1=0^\circ$, square the resultant amplitudes from Section 3.

Work through the $90^\circ$ case: $U_1=1$, $U_2=i$, and $U=1+i$. Thus

$$|U|^2=(1+i)(1-i)=1-i+i+1=2.$$

The cosine form gives the same result: $1+1+2\cos90^\circ=1+1+0=2$.

| Phase difference | Sum $U$ | Resultant amplitude $\lvert U\rvert$ | Relative intensity |
| --- | --- | --- | --- |
| $0^\circ$ | $2$ | $2$ | $4$ |
| $90^\circ$ | $1+i$ | $\sqrt2$ | $2$ |
| $180^\circ$ | $0$ | $0$ | $0$ |

Each wave alone has relative intensity 1. Their combined intensity depends on how their fields add.

### 4.7 Can the intensity become negative? Can unequal waves cancel?

The minimum occurs at $\delta=180^\circ$:

$$\frac{I_{\min}}{I_{\rm ref}}
=\frac{A_1^2+A_2^2-2A_1A_2}{A_{\rm ref}^2}
=\frac{(A_1-A_2)^2}{A_{\rm ref}^2}\geq0.$$

The maximum occurs at $\delta=0^\circ$:

$$\frac{I_{\max}}{I_{\rm ref}}
=\frac{A_1^2+A_2^2+2A_1A_2}{A_{\rm ref}^2}
=\frac{(A_1+A_2)^2}{A_{\rm ref}^2}.$$

Total intensity remains nonnegative. Complete cancellation requires equal amplitudes and opposite phase. For $A_1=1$, $A_2=0.5$, the extrema are $(1+0.5)^2=2.25$ and $(1-0.5)^2=0.25$.

### 4.8 What changes if we move both phase controls together?

Add the same phase $\alpha$ to both waves. Their difference stays fixed:

$$(\phi_2+\alpha)-(\phi_1+\alpha)=\phi_2-\phi_1=\delta.$$

The combined complex field transforms as

$$U'=e^{i\alpha}(U_1+U_2),\qquad
|U'|^2=|e^{i\alpha}|^2|U_1+U_2|^2=|U_1+U_2|^2.$$

Its phase changes, while its intensity stays fixed.

If either amplitude is zero, the interference term vanishes. If relative phase varies during an exposure, the detector averages that term. For equal durations at four phases,

$$\langle\cos\delta\rangle=\frac{1+0-1+0}{4}=0
\quad\text{for}\quad\delta=0^\circ,90^\circ,180^\circ,270^\circ.$$

Then the recorded intensities add. This leads to coherence, a later topic; our examples use stable phases.

A fixed $90^\circ$ phase difference also gives a zero cross term, but does **not** imply incoherence: changing that stable phase relation changes the intensity.

### Small calculation 4a: inspect all four products

For $U_1=1$ and $U_2=0.5i$, predict all four products and the total relative intensity, then check the code.

````python
A1, A2, A_ref = 1.0, 0.5, 1.0
phi1_deg, phi2_deg = 0.0, 90.0
phi1, phi2 = np.deg2rad([phi1_deg, phi2_deg])
U1 = A1 * np.exp(1j * phi1)
U2 = A2 * np.exp(1j * phi2)
own1 = U1 * np.conj(U1)
cross12 = U1 * np.conj(U2)
cross21 = U2 * np.conj(U1)
own2 = U2 * np.conj(U2)
for label, value in [('U1 U1*', own1), ('U1 U2*', cross12),
                     ('U2 U1*', cross21), ('U2 U2*', own2)]:
    print(f'{label:8s} = {value.real:+.3f} {value.imag:+.3f}j')
print(f'Cross terms together: {np.real(cross12 + cross21):.3f}')
print(f'Four-product intensity: {np.real(own1 + cross12 + cross21 + own2)/A_ref**2:.3f}')
print(f'Squared-magnitude intensity: {abs(U1 + U2)**2/A_ref**2:.3f}')
print(f'Cosine-form intensity: {(A1**2 + A2**2 + 2*A1*A2*np.cos(phi2-phi1))/A_ref**2:.3f}')
````

<details><summary>Answer</summary>

The products are $1,-0.5i,+0.5i,0.25$. The cross terms cancel, leaving relative intensity $1+0.25=1.25$ in all three calculations.

</details>

### Small calculation 4b: recover the interference term using real fields

The real-field calculation provides an independent route:

$$\langle(u_1+u_2)^2\rangle_T
=\langle u_1^2\rangle_T+\langle u_2^2\rangle_T+2\langle u_1u_2\rangle_T.$$

Section 4.1 gives the first two averages as $A_1^2/2$ and $A_2^2/2$. To obtain the cross term, use

$$2\cos a\cos b=\cos(a-b)+\cos(a+b).$$

With $a=\phi_1-\omega t$ and $b=\phi_2-\omega t$, $a-b=\phi_1-\phi_2$ and $a+b=\phi_1+\phi_2-2\omega t$. Therefore

$$2u_1u_2=A_1A_2\left[\cos(\phi_1-\phi_2)+\cos(\phi_1+\phi_2-2\omega t)\right].$$

The second cosine averages to zero, while $\cos(\phi_1-\phi_2)=\cos\delta$ is constant:

$$2\langle u_1u_2\rangle_T=A_1A_2\cos\delta,$$

and therefore

$$\begin{aligned}
\langle(u_1+u_2)^2\rangle_T
&=\frac{A_1^2}{2}+\frac{A_2^2}{2}+A_1A_2\cos\delta\\
&=\frac{A_1^2+A_2^2+2A_1A_2\cos\delta}{2}
=\frac{|U_1+U_2|^2}{2}.
\end{aligned}$$

Dividing by $A_{\rm ref}^2/2$ recovers the relative intensity. The order matters: $\langle u^2\rangle_T$ can be nonzero even though $\langle u\rangle_T^2=0$.

The code compares both routes using 600 equally spaced samples over one period.

**Predict:** for phase differences $0^\circ,90^\circ,180^\circ$, what values should appear in the mean-square and intensity columns? Then run both calculation routes and compare.

````python
A1, A2, A_ref = 1.0, 1.0, 1.0
phi1_deg = 0.0
phi2_values_deg = [0.0, 90.0, 180.0]
tau = np.linspace(0.0, 1.0, 600, endpoint=False)
print('delta (degrees)   mean(u)   mean(u²)   2 mean(u²)/A_ref²   |U|²/A_ref²')
for phi2_deg in phi2_values_deg:
    delta_deg = phi2_deg - phi1_deg
    phi1, phi2 = np.deg2rad([phi1_deg, phi2_deg])
    u1 = A1 * np.cos(phi1 - 2 * np.pi * tau)
    u2 = A2 * np.cos(phi2 - 2 * np.pi * tau)
    u = u1 + u2
    U = A1 * np.exp(1j * phi1) + A2 * np.exp(1j * phi2)
    print(f'{delta_deg:7.0f}          {np.mean(u):+.3f}      {np.mean(u**2):.3f}'
          f'           {2*np.mean(u**2)/A_ref**2:.3f}              {abs(U)**2/A_ref**2:.3f}')
````

<details><summary>Answer</summary>

The mean-square values are $2,1,0$; the relative intensities are $4,2,0$. Both intensity routes agree. The mean signed field is zero in all three cases, so that mean cannot distinguish a bright point from a dark one.

</details>

## 5. Different directions: relative phase changes with position

Two broad, mutually coherent scalar plane waves overlap at a detector in the $xy$ plane, at $z=-1\,\mu\mathrm m$. Both travel toward decreasing $z$, arriving from opposite sides in $x$. The $xz$ side view shows the detector edge-on in purple and one propagation arrow per wave. Their overlap does not represent focusing; the model includes no lens, slit, or grating.

The coloured lines are **positive-crest wavefronts**: parallel planes perpendicular to each wavevector, seen edge-on. Their finite drawn lengths are only a display choice; the calculated plane waves have constant amplitudes across the detector.

### 5.1 Write the phase of each arriving wave

The phases are

$$\Phi_1=\mathbf k_1\cdot\mathbf r-\omega t+\phi_1,
\qquad
\Phi_2=\mathbf k_2\cdot\mathbf r-\omega t+\phi_2.$$

The offsets $\phi_j$ refer to the spatial origin at $t=0$.

A positive crest has $\Phi_j=2\pi n$, so

$$\mathbf k_j\cdot\mathbf r=\omega t-\phi_j+2\pi n.$$

At fixed time this defines a plane perpendicular to $\mathbf k_j$. Adjacent crests differ by $2\pi$, giving their separation along propagation:

$$ks=2\pi\quad\Rightarrow\quad s=\frac{2\pi}{k}=\lambda.$$

Following one crest for a time increment $\Delta t$ gives

$$ks=\omega\Delta t\quad\Rightarrow\quad
s=\frac{\omega}{k}\Delta t=\lambda\frac{\Delta t}{T}.$$

The wavefronts advance along their fixed wavevectors.

At the selected point, the two cycle traces are

$$u_j(\mathbf r_P,t)=A_j\cos\Phi_j(\mathbf r_P,t)
=\operatorname{Re}\{U_j(\mathbf r_P)e^{-i\omega t}\},\qquad
U_j(\mathbf r_P)=A_je^{i(\mathbf k_j\cdot\mathbf r_P+\phi_j)}.$$

Use the same position $\mathbf r_P$ and time $t$ in both phases:

$$\begin{aligned}
\Phi_2(\mathbf r_P,t)-\Phi_1(\mathbf r_P,t)
&=(\mathbf k_2\cdot\mathbf r_P-\omega t+\phi_2)
 -(\mathbf k_1\cdot\mathbf r_P-\omega t+\phi_1)\\
&=\mathbf k_2\cdot\mathbf r_P-\mathbf k_1\cdot\mathbf r_P
 +\underbrace{(-\omega t)-(-\omega t)}_{0}+\phi_2-\phi_1.
\end{aligned}$$

$$\Delta\Phi(\mathbf r_P)
=(\mathbf k_2-\mathbf k_1)\cdot\mathbf r_P+\delta,
\qquad \delta=\phi_2-\phi_1.$$

The equal frequencies make the common time contribution cancel. The local difference $\Delta\Phi(P)$ also contains the propagation phase, whereas $\delta$ compares the phase offsets at the spatial origin.

**We subtract phases to compare alignment; the fields themselves add.** Applying Section 4 to $U_1(\mathbf r_P)+U_2(\mathbf r_P)$ gives

$$\frac{I(\mathbf r_P)}{I_{\rm ref}}
=\frac{A_1^2+A_2^2+2A_1A_2\cos[\Delta\Phi(\mathbf r_P)]}{A_{\rm ref}^2}.$$

Maxima occur at $\Delta\Phi=2\pi n$ and minima at $\Delta\Phi=(2n+1)\pi$. For nonzero components, only equal amplitudes give zero minima. The spatial variation of relative phase creates the **fringe pattern**.

### 5.3 Find the fringe spacing for symmetric arrival directions

For directions symmetric about $-z$ in the $xz$ plane, with $k=2\pi/\lambda$,

$$\begin{aligned}
\mathbf k_1&=(-k\sin\theta,0,-k\cos\theta),\\
\mathbf k_2&=(+k\sin\theta,0,-k\cos\theta),\\
\mathbf k_2-\mathbf k_1&=(2k\sin\theta,0,0).
\end{aligned}$$

Only the transverse phase difference remains:

$$\Delta\Phi(x)=\Delta k_x\,x+\delta,
\qquad \Delta k_x=k_{2x}-k_{1x}=2k\sin\theta.$$

Over one fringe spacing $\Lambda$, the relative phase increases by $2\pi$:

$$\begin{aligned}
\Delta\Phi(x+\Lambda)-\Delta\Phi(x)
&=[\Delta k_x(x+\Lambda)+\delta]-[\Delta k_x x+\delta]\\
&=\Delta k_x\Lambda.
\end{aligned}$$

Therefore

$$\Delta k_x\Lambda=2\pi,
\qquad \Lambda=\frac{2\pi}{\Delta k_x}.$$

Here $\Delta k_x>0$; for either sign, the spacing is $2\pi/|\Delta k_x|$. Substituting the symmetric wavevectors,

$$\Lambda=\frac{2\pi}{\Delta k_x}
=\frac{2\pi}{2k\sin\theta}
=\frac{2\pi}{2(2\pi/\lambda)\sin\theta}
=\frac{\lambda}{2\sin\theta}.$$

$$\boxed{\Lambda=\frac{\lambda}{2\sin\theta}.}$$

Here $\lambda$ is measured along propagation and $\Lambda$ along the detector. Each wave makes angle $\theta$ with $-z$; the angle between them is $2\theta$.

The wavevector difference describes the fringe pattern, not an additional light wave.

At reset, $\lambda=2\,\mu\mathrm m$ and $\theta=30^\circ$. Thus $k=2\pi/2=\pi\,\mathrm{rad}/\mu\mathrm m$ and $\Delta k_x=2k\sin30^\circ=2\pi(1/2)=\pi\,\mathrm{rad}/\mu\mathrm m$, giving $\Lambda=2\pi/\pi=2\,\mu\mathrm m$. With equal unit amplitudes and zero phase offset, $x=0$ is bright ($\Delta\Phi=0$), $x=1\,\mu\mathrm m$ is dark ($\Delta\Phi=\pi$), and $x=2\,\mu\mathrm m$ is bright again ($\Delta\Phi=2\pi$). The interactive plot displays the spacing and updates it when the half-angle changes.

Bright positions satisfy

$$\Delta k_xx_n+\delta=2\pi n
\quad\Rightarrow\quad x_n=\frac{2\pi n-\delta}{\Delta k_x}.$$

Changing phase while fixing the wavevectors gives

$$\begin{aligned}
x_n(\delta+\Delta\delta)-x_n(\delta)
&=\frac{2\pi n-\delta-\Delta\delta}{\Delta k_x}
-\frac{2\pi n-\delta}{\Delta k_x}
=-\frac{\Delta\delta}{\Delta k_x},\\
x_{n+1}-x_n
&=\frac{2\pi(n+1)-\delta-(2\pi n-\delta)}{\Delta k_x}
=\frac{2\pi}{\Delta k_x}.
\end{aligned}$$

Thus relative phase shifts the pattern while preserving its spacing; $\Delta\delta$ is in radians.

**Graphical check:** the purple arrow joins the tips of $\mathbf k_1/k$ and $\mathbf k_2/k$. Change the half-angle from $30^\circ$ to $15^\circ$ and compare the fringe spacing. Then change either phase: the pattern shifts while its spacing stays fixed.

wavevectors. If the sign of $\Delta k_x$ is known from the setup, a tracked
fringe displacement $\Delta x$ gives

$$\Delta\delta=-\Delta k_x\,\Delta x.$$

For the reset geometry, a shift by $+0.5\,\mu\mathrm m$ corresponds to
$\Delta\delta=-\pi/2=-90^\circ$. Without tracking or additional phase
information, identical fringe patterns leave a whole-cycle ambiguity.

### Small calculation 5: walk from a bright point to a dark point

The positions lie on the line $z=-1\,\mu\mathrm m$. Here the equal $z$ components cancel in the phase difference, so only $x$ is needed to compute intensity.

**Predict:** what are the relative intensities at the four listed positions? Keep `phi1_deg = 0`, change `phi2_deg` from 0 to 180, and rerun.

````python
wavelength = 2.0  # micrometres
theta_deg = 30.0
A1, A2, A_ref = 1.0, 1.0, 1.0
phi1_deg, phi2_deg = 0.0, 0.0  # try phi2_deg = 180.0
x = np.array([0.0, 0.5, 1.0, 2.0])  # micrometres
k = 2 * np.pi / wavelength
delta_kx = 2 * k * np.sin(np.deg2rad(theta_deg))
phase_difference = delta_kx * x + np.deg2rad(phi2_deg - phi1_deg)
I_relative = (A1**2 + A2**2 + 2*A1*A2*np.cos(phase_difference)) / A_ref**2
print(f'Fringe spacing: {2*np.pi/delta_kx:.3f} um')
print('x (um)   phase difference (degrees)   I/I_ref')
print(np.round(np.column_stack((x, np.rad2deg(phase_difference), I_relative)), 3))
````

<details><summary>Answer</summary>

At zero offset, the phases are $0^\circ,90^\circ,180^\circ,360^\circ$ and the intensities are $4,2,0,4$. At an offset of $180^\circ$, the intensities become $0,2,4,0$. The spacing remains $2\,\mu\mathrm m$.

</details>

At a bright fringe, the local intensity can exceed $I_1+I_2$. Between fringes it is reduced. For this symmetric plane-wave example, averaging across one complete fringe period makes the cosine term vanish, leaving the spatial average $I_1+I_2$. Local enhancement is accompanied by redistribution across the pattern.

## Recap: from adding fields to spatial fringes

- Fields add: $u=u_1+u_2$, or $U=U_1+U_2$ for the complex fields of equal-frequency waves.
- Relative intensity follows from $|U_1+U_2|^2/A_{\rm ref}^2$, including the cross terms.
- The spatial phase difference is $\Delta\Phi=(\mathbf k_2-\mathbf k_1)\cdot\mathbf r+\delta$.
- For symmetric incidence, $\Lambda=\lambda/(2\sin\theta)$. Relative phase shifts the fringes; a common phase change or advancing time preserves their intensity pattern.
