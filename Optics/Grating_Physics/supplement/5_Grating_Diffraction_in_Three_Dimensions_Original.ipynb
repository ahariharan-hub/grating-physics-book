## 1. Let the incident wave travel along the grooves

Notebook 4 completed each order using the two-dimensional relation
$$k_{mx}^2+k_{mz}^2=k^2.$$
Allowing phase variation along a groove introduces $k_{my}$, so the same wave equation gives
$$k_{mx}^2+k_{my}^2+k_{mz}^2=k^2.$$
We will determine the two surface components first, then use this relation to complete the normal component.

Keep the earlier geometry:

- The grating lies at $z=0$, repeats along $x$ with period $d$, and has uniform grooves along $y$.
- Incidence is from $z>0$; transmission is into $z<0$.
- Both exterior regions have the same homogeneous, isotropic, lossless medium.
- $\lambda$ is the wavelength **in that medium**, $k=2\pi/\lambda$, and lengths are in micrometres.

Use $\mathbf k_i=(k_{ix},k_{iy},k_{iz})$ for the incident wavevector and $\mathbf k_m=(k_{mx},k_{my},k_{mz})$ for diffraction order $m$. Plain $k$ is the positive scalar wavenumber of the medium. For the real incident vector, $|\mathbf k_i|=k$.

The incident complex scalar field and its real, time-dependent field are
$$U_i(x,y,z)=A_i e^{i(k_{ix}x+k_{iy}y+k_{iz}z)},\qquad
u_i=\operatorname{Re}\{U_i e^{-i\omega t}\}.$$

Previously $k_{iy}=0$: the phase did not vary along a groove. With $k_{iy}\ne0$, the surface field is
$$U_i(x,y,0)=A_i e^{i(k_{ix}x+k_{iy}y)}.$$

The controls use the component fractions
$$a_i=\frac{k_{ix}}k,\qquad b_i=\frac{k_{iy}}k,\qquad
\frac{k_{iz}}k=-\sqrt{1-a_i^2-b_i^2}.$$
We require $a_i^2+b_i^2<1$ for incidence onto the surface. These fractions will later become the direction-cosine coordinates $\alpha_i,\beta_i$.

The next view extends the earlier $xz$ wave picture into three dimensions. A wavefront is a plane of constant phase, perpendicular to $\mathbf k_i$. At $t=0$, the orange crest planes satisfy
$$\mathbf k_i\cdot\mathbf r=2\pi n,\qquad n\in\mathbb Z.$$
Here $n$ labels successive **incident wave crests**, not diffraction orders. Adjacent planes are separated by $\lambda$ measured along their normal. Orange lines mark their intersections with $z=0$; the same lines appear in the surface map. At normal incidence a crest coincides with the whole grating plane, so there is no distinct intersection line.

Begin with **Normal incidence**, then **Reset: in-plane**, then **Conical: β = 0.4**. The last two keep $a_i=0.3$ fixed, isolating the new along-groove variation. All three presets use $\lambda=2$ and $d=4$.

### Small calculation 1: complete the incident vector

Construct the incident vector and check its length. Polar angle $\theta_i$ is measured from $-z$; azimuth $\varphi_i$ is measured in the $xy$-plane from $+x$ toward $+y$:
$$\frac{\mathbf k_i}{k}=
(\sin\theta_i\cos\varphi_i,\ \sin\theta_i\sin\varphi_i,\ -\cos\theta_i).$$

Unlike the signed $xz$-plane angle in the preceding notebooks, this **polar angle is nonnegative**; the azimuth identifies the transverse direction. For example, the earlier planar angle $-30^\circ$ is now $\theta_i=30^\circ$, $\varphi_i=180^\circ$. Both descriptions give $\mathbf k_i/k=(-1/2,0,-\sqrt3/2)$. The Cartesian axes and outgoing $-z$ convention are unchanged.

**Predict and rerun:** change only $b_i$ from $0.4$ to $0$. Which Cartesian component remains fixed? Does the polar angle remain fixed?

```python
wavelength = 2.0  # micrometres
a_i, b_i = 0.3, 0.4
k = 2*np.pi/wavelength

if a_i**2+b_i**2 >= 1:
    print("Choose a_i**2 + b_i**2 < 1 for incidence onto the surface.")
else:
    s_i = np.array([a_i, b_i, -np.sqrt(1-a_i**2-b_i**2)])
    theta_i = np.rad2deg(np.arccos(-s_i[2]))
    phi_i = (np.rad2deg(np.arctan2(b_i, a_i))
             if np.hypot(a_i, b_i) > 0 else None)
    print("k_i/k =", np.round(s_i, 6))
    print("k_i (rad/um) =", np.round(k*s_i, 6))
    print("|k_i|/k =", np.linalg.norm(s_i))
    print("theta_i from -z (degrees) =", round(theta_i, 6))
    print("azimuth phi_i (degrees) =", phi_i)
```

## 2. What does the grating change, and what does it preserve?

Across the periodic direction, Notebook 3's argument still applies:
$$e^{i(k_{mx}-k_{ix})d}=1
\quad\Longrightarrow\quad
\boxed{k_{mx}=k_{ix}+mG,\qquad G=\frac{2\pi}{d},\quad m\in\mathbb Z.}$$

Along $y$, the grating is unchanged by **any** displacement $\Delta y$, not just one period. Under plane-wave illumination its field has the incident translation factor $e^{ik_{iy}\Delta y}$. Each outgoing component must satisfy
$$e^{i(k_{my}-k_{iy})\Delta y}=1\quad\text{for every }\Delta y,
\quad\Longrightarrow\quad
\boxed{k_{my}=k_{iy}.}$$

If $k_{my}-k_{iy}$ were nonzero, a sufficiently small displacement would give a nonzero phase change smaller than one full cycle, so the factor could not remain 1 for every displacement. Thus the difference must be zero.

The key distinction is **discrete periodicity across the grooves** and **continuous uniformity along the grooves**. This is still a linear grating with **one order integer**; allowing three-dimensional propagation has not introduced a second grating period.

Maystre's pseudo-periodicity and Rayleigh-expansion derivation provides the foundation for the discrete components [3](https://www.fresnel.fr/files/gratings/Second-Edition/GratingsTheoryandNumericApplicationsSecondEdition.pdf).

### Small calculation 2: check both translation factors

Compare incident and outgoing surface phases across one period along $x$, then across an arbitrary displacement along $y$.

**Predict and rerun:** change the integer $m$ and the along-groove displacement. Should the two complex translation factors still agree?

```python
wavelength, d = 2.0, 4.0
a_i, b_i, m = 0.3, 0.4, 1
delta_y = 0.7  # any displacement in micrometres
k = 2*np.pi/wavelength
kix, kiy = k*a_i, k*b_i
kmx, kmy = kix+m*2*np.pi/d, kiy

incident_x, outgoing_x = np.exp(1j*kix*d), np.exp(1j*kmx*d)
incident_y, outgoing_y = np.exp(1j*kiy*delta_y), np.exp(1j*kmy*delta_y)
print("Across one period: incident =", incident_x, "; outgoing =", outgoing_x)
print("Along a groove:    incident =", incident_y, "; outgoing =", outgoing_y)
print("Across-groove phase difference / (2*pi) =", (kmx-kix)*d/(2*np.pi))
print("Along-groove phase difference =", (kmy-kiy)*delta_y)
```

<details><summary>Answer</summary>

The across-groove phase difference is $2\pi m$; the along-groove difference is zero. Both translation factors match. A trial $k_{my}=k_{iy}+2\pi/L$ would match translations by $L$, but generally fail other displacements. Uniform grooves require the condition for **every** displacement.

</details>

## 3. Complete the third component

The homogeneous medium requires
$$k_{mx}^2+k_{my}^2+k_{mz}^2=k^2.$$

Insert the tangential components:
$$D_m=k^2-(k_{ix}+mG)^2-k_{iy}^2.$$

As in Notebook 4, choose the outgoing or decaying branch for $z<0$:
$$\boxed{k_{mz}=
\begin{cases}
-\sqrt{D_m},&D_m>0\quad\text{(propagating)},\\
0,&D_m=0\quad\text{(grazing)},\\
-i\sqrt{-D_m},&D_m<0\quad\text{(evanescent)}.
\end{cases}}$$

The new term is $k_{iy}^2$. A real outgoing direction requires
$$\left(a_i+m\frac{\lambda}{d}\right)^2+b_i^2\le1.$$
Equality is grazing. The condition $|k_{mx}|\le k$ alone is no longer sufficient.

For an evanescent component, write $\kappa_m=\sqrt{-D_m}$ and $s=-z\ge0$:
$$U_m(x,y,-s)=A_m e^{i(k_{mx}x+k_{iy}y)}e^{-\kappa_m s}.$$
The phase varies along the surface; the amplitude decays normally with length $\ell_m=1/\kappa_m$. Use algebraic squares in the dispersion relation when $k_{mz}$ is complex: $\mathbf k_m\cdot\mathbf k_m=k^2$, without complex conjugation. The ordinary length statement $|\mathbf k_m|=k$ applies only to real propagating or grazing vectors; $k$ remains the medium wavenumber for evanescent orders too.

TU Delft derives the 3D square-root relation [5, Eq. (375)](https://interactivetextbooks.tudelft.nl/tn2421/content/Chap6_Diffraction/DiffractiveOptics_2022_01Clean.html#angular-spectrum-method). Its outgoing side is $+z$; our branch signs are reversed.

### Small calculation 3: move an order through the 3D boundary

Keep $a_i=0.3$ and $\lambda/d=0.5$. For $m=+1$, $k_{mx}/k=0.8$, so grazing occurs at
$$b_i^2=1-0.8^2=0.36.$$

**Predict:** what happens to $m=+1$ at $b_i=0.4,0.6,0.8$? At the last value, $k_{mx}$ is still smaller than $k$. Is that enough for propagation?

```python
wavelength, d = 2.0, 4.0
a_i = 0.3
orders = np.arange(-3, 4)

for b_i in [0.4, 0.6, 0.8]:
    a_m = a_i+orders*wavelength/d
    r = 1-a_m**2-b_i**2  # D_m / k**2
    r[np.abs(r) <= 1e-12] = 0.0
    gamma = -np.sqrt(r+0j)
    states = np.where(r == 0, "grazing", np.where(r > 0, "propagating", "evanescent"))
    print(f"\nb_i = {b_i:.1f}")
    print(" m   k_mx/k  k_my/k       k_mz/k            state")
    for m, a, g, state in zip(orders, a_m, gamma, states):
        print(f"{m:+2d}  {a:+6.2f}  {b_i:+6.2f}  {g.real:+.4f}{g.imag:+.4f}j   {state}")
```

<details><summary>Answer</summary>

- At $b_i=0.4$, $m=-2,-1,0,+1$ propagate.
- At $b_i=0.6$, $m=+1$ is grazing; $m=-2,-1,0$ propagate.
- At $b_i=0.8$, only $m=-1,0$ propagate.

For $m=+1$ at $b_i=0.8$, $k_{mx}^2+k_{my}^2=1.28k^2$, although each tangential component is smaller than $k$. Here $k_{mz}/k=-i\sqrt{0.28}$ and, for $\lambda=2\,\mu\mathrm m$, $\ell\approx0.602\,\mu\mathrm m$.

</details>

## 5. Why is the diffraction called conical?

For every real outgoing order,
$$\frac{\mathbf k_m}{k}\cdot\hat{\mathbf y}
=\frac{k_{my}}k=b_i.$$
Its angle $\chi_m$ to the positive groove axis satisfies $\cos\chi_m=b_i$. All orders make the **same angle to the groove axis**; their rays lie on a cone with its vertex at the grating origin.

Press **Conical: β = 0.4**, then enable **Show cone / planar limit** in the spatial view. The translucent surface connects candidate directions; the discrete marked rays are the propagating orders. Only the $z\le0$ portion is drawn.

Distinguish two pictures:

- **Rays from a common origin** lie on the cone.
- **Their endpoints at a common radius** lie on a circle in a plane of constant $y$. In wavevector space, each real order obeys
  $$k_{my}=k_{iy},\qquad k_{mx}^2+k_{mz}^2=k^2-k_{iy}^2.$$

The cone axis is along the grooves, not the surface normal. At $b_i=0$, the ray locus becomes the $xz$-plane. For negative $b_i$, it opens toward $-y$; the angle to $+y$ is obtuse, while its semi-angle about $-y$ is $\arccos|b_i|$.

Any nonzero along-groove component gives conical geometry under our assumptions; large angles merely make it easier to see. See [1](https://doi.org/10.1364/AO.37.008158), [2, Fig. 6](https://doi.org/10.1117/1.OE.58.8.087105), and the handbook's conical equation [4](https://www.newport.com/mam/celum/celum_assets/np/resources/MKS_Diffraction_Grating_Handbook.pdf).

### Small calculation 4: verify the common cone angle

Compute the real outgoing unit vectors and measure their angles to $+y$. Their endpoint circle on a unit sphere has radius
$$\rho=\sqrt{1-b_i^2}.$$

**Predict and rerun:** replace $b_i=0.4$ with $-0.4$, then $0$. Which results change?

```python
wavelength, d = 2.0, 4.0
a_i, b_i = 0.3, 0.4
orders = np.arange(-3, 4)
a_m = a_i+orders*wavelength/d
r = 1-a_m**2-b_i**2
real = r >= -1e-12
directions = np.column_stack([a_m[real], np.full(np.count_nonzero(real), b_i),
                              -np.sqrt(np.maximum(r[real], 0))])
chi = np.rad2deg(np.arccos(np.clip(directions[:, 1], -1, 1)))
circle_radii = np.hypot(directions[:, 0], directions[:, 2])
print("  m    angle to +y (degrees)   circle radius")
print(np.column_stack([orders[real], chi, circle_radii]))
print("Expected radius =", np.sqrt(1-b_i**2))
```

<details><summary>Answer</summary>

At $b_i=0.4$, all four real orders make an angle $\chi\approx66.422^\circ$ to $+y$, with circle radius $\rho\approx0.916515$. Their individual directions differ despite this common angle.

At $b_i=-0.4$, the real order labels and circle radius are unchanged; the angle to $+y$ is $113.578^\circ$. At $b_i=0$, $\chi=90^\circ$ and the circle is a great-circle section: the rays lie in the $xz$-plane.

</details>

## 6. Harvey: make the directions easy to read

We already know the waves. Give each real wavevector normalized coordinates:
$$\hat{\mathbf k}_m=\frac{\mathbf k_m}{k}
=(\alpha_m,\beta_m,\gamma_m),\qquad
\alpha_m^2+\beta_m^2+\gamma_m^2=1.$$

These are **direction cosines**: components of a unit vector, or cosines of its angles with the coordinate axes. Here $\gamma_m<0$ for transmitted propagating waves. The incident fractions $a_i,b_i$ are now written $\alpha_i,\beta_i$.

With **Coordinates → Direction cosines**, the plot axes use unsubscripted $\alpha=k_x/k$, $\beta=k_y/k$ and $\gamma=k_z/k$ to label the coordinate system. An order point has coordinates $(\alpha_m,\beta_m,\gamma_m)$; the incident point uses subscript $i$.

Project the outgoing hemisphere onto the $\alpha\beta$-plane:
$$\boxed{\alpha_m=\alpha_i+m\frac{\lambda}{d},\qquad
\beta_m=\beta_i,\qquad
\gamma_m=-\sqrt{1-\alpha_m^2-\beta_i^2}.}$$

The **Harvey direction-cosine diagram** has a simple structure:

- Candidates form a straight row at $\beta=\beta_i$.
- Neighbouring orders have exactly equal horizontal spacing $\lambda/d$.
- The interior of the unit disk permits transmission; the boundary is grazing.
- Outside the disk, tangential coordinates remain meaningful, but there is no real point on the hemisphere.

Harvey's papers [1](https://doi.org/10.1364/AO.37.008158), [2](https://doi.org/10.1117/1.OE.58.8.087105) develop this representation. Our diagrams always use actual propagation vectors; the references explain the different signs in reflection observation diagrams.

### 6.1. Spatial-frequency coordinates: Goodman's convention

[Notebook 2](../2_4_Order_Coordinates_and_Spacing.ipynb#one-order-two-triangles) expressed one wavevector in **cycles per unit length**. Extend that definition to all three components:
$$\boxed{\mathbf f_i=\frac{\mathbf k_i}{2\pi},\qquad
\mathbf f_m=\frac{\mathbf k_m}{2\pi}=(f_{x,m},f_{y,m},f_{z,m}).}$$
Here $f$ means spatial frequency, not oscillations per second. With lengths in micrometres, its units are cycles per micrometre.

This follows **Goodman's spatial-frequency convention** in *Introduction to Fourier Optics*: $f_x=\alpha/\lambda$ and $f_y=\beta/\lambda$ [13, §3.10.1, Eq. (3-62)](#ref-13). We add the order subscript $m$ and extend the notation to the normal component $f_{z,m}$. Goodman develops propagation toward $+z$; our outgoing side remains $-z$, so real outgoing orders have $f_{z,m}\le0$.

At the grating surface, one order has the complex scalar field
$$U_m(x,y,0)=A_m\exp\!\left[i2\pi(f_{x,m}x+f_{y,m}y)\right].$$
Moving along $x$ samples phase cycles **across** the grooves; moving along $y$ samples phase cycles **along** them. For example, positive $f_{y,m}=0.2$ cycles/µm means a $1\,\mu\mathrm m$ step along $+y$ advances the phase by $0.2$ cycles. TU Delft uses the symbols $\xi,\eta$ for these transverse frequencies [5, Eqs. (371)–(376)](https://interactivetextbooks.tudelft.nl/tn2421/content/Chap6_Diffraction/DiffractiveOptics_2022_01Clean.html#angular-spectrum-method).

**What does the grating do to those frequencies?** Divide section 2's component equations by $2\pi$:
$$\boxed{f_{x,m}=f_{x,i}+\frac{m}{d},\qquad f_{y,m}=f_{y,i}.}$$
The orders form a row with horizontal step $1/d$ and a common along-groove frequency. Conical incidence shifts the row away from $f_y=0$; there is still only one order integer.

**Where can the real directions lie?** For propagating or grazing orders,
$$f_{x,m}^2+f_{y,m}^2+f_{z,m}^2=\frac{1}{\lambda^2},\qquad f_{z,m}\le0.$$
The outgoing hemisphere and its transverse disk have radius $1/\lambda$. Inside the disk an order propagates; on the boundary it grazes the surface. Outside, $f_{z,m}=k_{mz}/(2\pi)$ is imaginary and describes normal decay. Such a candidate has a real transverse point but no real point on the hemisphere. The complex vector still obeys the algebraic sum of squares; its ordinary length is not $1/\lambda$.

The link to Harvey coordinates is
$$\boxed{(\alpha_m,\beta_m,\gamma_m)=\lambda(f_{x,m},f_{y,m},f_{z,m}).}$$
For real orders these are direction cosines; for evanescent candidates only the transverse coordinates remain real. The three scales describe the same order candidates:

| Coordinates | Units | Hemisphere/disk radius | Step between neighbouring orders along $x$ |
| --- | --- | --- | --- |
| Wavevector $\mathbf k_m$ | rad/µm | $2\pi/\lambda$ | $2\pi/d$ |
| Spatial frequency $\mathbf f_m$ (Goodman convention) | cycles/µm | $1/\lambda$ | $1/d$ |
| Normalized components $\mathbf k_m/k$ | dimensionless | $1$ | $\lambda/d$ |

Use **Coordinates** below to switch scales. The same controls, order labels, colours and selection apply to both panels. The axes rescale to fit; read their numbers and the radius/spacing readout when comparing units. These are coordinate representations, with no optical mapping or order strengths assigned.

### Small calculation 5: one order on three coordinate scales

Use the conical preset: $\lambda=2\,\mu\mathrm m$, $d=4\,\mu\mathrm m$, $(a_i,b_i)=(0.3,0.4)$, and $m=+1$. Build $\mathbf k_m$, divide by $2\pi$ to obtain $\mathbf f_m$, and divide by $k$ to obtain the normalized components.

```python
wavelength, d = 2.0, 4.0  # micrometres; wavelength in the medium
a_i, b_i, m = 0.3, 0.4, 1
if wavelength <= 0 or d <= 0 or a_i**2+b_i**2 >= 1 or m != int(m):
    raise ValueError("Use positive lengths, physical incidence and an integer m.")
m = int(m)
k = 2*np.pi/wavelength
k_i = k*np.array([a_i, b_i, -np.sqrt(1-a_i**2-b_i**2)])
k_m = k_i.astype(complex)
k_m[0] += m*2*np.pi/d
r = 1-(k_m[0].real/k)**2-(k_m[1].real/k)**2
if abs(r) <= 1e-12:
    r = 0.0
k_m[2] = -k*np.sqrt(r+0j)
f_m = k_m/(2*np.pi)
normalized = k_m/k
state = "grazing" if r == 0 else ("propagating" if r > 0 else "evanescent")
print(f"m = {m:+d}: {state}")
print("k_m (rad/um):", np.round(np.real_if_close(k_m), 6))
print("f_m (cycles/um):", np.round(np.real_if_close(f_m), 6))
print("k_m/k = wavelength*f_m:", np.round(np.real_if_close(normalized), 6))
print("Order steps: 2*pi/d, 1/d, wavelength/d =",
      np.round([2*np.pi/d, 1/d, wavelength/d], 6))
if r < 0:
    print("The imaginary normal component describes decay; no real hemisphere point.")
```

## 7. Optional application: collection, pupil mapping, and lithography

The order geometry in sections 1-6 is the prerequisite for the next notebook. This section adds an optical system that collects some of those orders. It introduces NA and sigma coordinates, then connects directions to pupil positions and a lithography example. For a first pass, you may go directly to the [Final checkpoint](#final-checkpoint) and answer questions 1-5.

Harvey's disk already gives us two coordinates for each real outgoing order $m$:
$$\alpha_m=\frac{k_{mx}}{k},\qquad \beta_m=\frac{k_{my}}{k}.$$
The mapping below applies to any real outgoing direction. We write $(\alpha,\beta)$ without an order subscript when discussing the coordinate system; for a particular order, use $(\alpha_m,\beta_m)$.
Let $n$ be the refractive index of the medium containing the grating. The wavelength $\lambda$ used earlier remains the wavelength **in that medium**. With the collecting objective's optical axis along $-z$, the ray's angle $\theta$ to that axis satisfies
$$\sqrt{\alpha^2+\beta^2}=\sin\theta.$$

### 7.1. The NA diagram is a change of scale

Multiply both axes by $n$. The coordinates become $(n\alpha,n\beta)$, and their radial distance is $n\sin\theta$. These are signed, dimensionless coordinates. Labels such as $\mathrm{NA}_x$ and $\mathrm{NA}_y$ can be used as shorthand for them; **the objective's numerical aperture is one scalar**, $\mathrm{NA}_{\rm obj}=n\sin\theta_{\max}$.

For an ideal centered circular collection aperture, a propagating transmitted order is accepted when
$$\boxed{(n\alpha_m)^2+(n\beta_m)^2\le\mathrm{NA}_{\rm obj}^{\,2}.}$$
Harvey's full propagation disk has radius $1$; the same disk in NA coordinates has radius $n$. The objective collects the circle of radius $\mathrm{NA}_{\rm obj}\le n$ within it. In air, $n\approx1$, so the two coordinate scales are essentially identical.

**This rescaling requires no optical system.** A real order can lie inside the propagation disk yet outside the objective's collection aperture.

### 7.2. From an order direction to a pupil position

Section 7.1 tells us which directions an objective accepts. To locate those orders in a **physical pupil plane**, we also need the optical mapping. For the ideal sine mapping used here,
$$r=C\sin\theta,\qquad R=C\sin\theta_{\max},$$
where $r$ is the distance from the pupil center, $R$ its outer radius, and $C$ a length scale set by the chosen optics. Dividing cancels that scale. With pupil axes aligned to the transverse direction coordinates,
$$\boxed{\left(\frac{x_p}{R},\frac{y_p}{R}\right)
=\frac{n}{\mathrm{NA}_{\rm obj}}(\alpha,\beta).}$$
Tran and Oldenbourg use this relation to calibrate pupil positions with known grating directions [7, Eqs. (1)–(2), Fig. 2](https://pmc.ncbi.nlm.nih.gov/articles/PMC8005462/).

This pupil follows a **sine law**, whereas the bare observation plane at $z=-L$ follows $r_D=L\tan\theta$. The optical system supplies the different mapping. Aplanatic imaging combines the sine condition with stigmatic on-axis imaging, correcting spherical aberration and coma in the ideal design [10](https://doi.org/10.1364/OE.25.010710). Here we use its geometrical mapping part at a specified pupil or relay; the object-to-image sine condition alone does not establish the mapping at every internal aperture [12](https://patents.google.com/patent/WO2005015316A2/en).

### 7.3. Sigma coordinates: put the collection edge at one

The normalized pair above is the **sigma-space coordinate** used in lithography. Define it for each order by
$$\boxed{(\sigma_{x,m},\sigma_{y,m})
=\frac{n}{\mathrm{NA}_{\rm obj}}(\alpha_m,\beta_m).}$$
Using $\lambda_0=n\lambda$ for vacuum wavelength and $k_0=2\pi/\lambda_0$, the same pair is
$$(\sigma_{x,m},\sigma_{y,m})
=\frac{(k_{mx},k_{my})}{k_0\mathrm{NA}_{\rm obj}}
=\frac{\lambda_0}{\mathrm{NA}_{\rm obj}}(f_{x,m},f_{y,m}).$$
All quantities must refer to the **same side** of the optical system. Under section 7.2's sine mapping, these angular coordinates equal $(x_{p,m}/R,y_{p,m}/R)$; their angular definition remains useful even when a physical pupil has a different mapping.

The dimensionless radius is
$$\boxed{\rho_m=\sqrt{\sigma_{x,m}^2+\sigma_{y,m}^2}
=\frac{\sin\theta_m}{\sin\theta_{\max}}.}$$
For a real order travelling toward the objective, $\rho_m\le1$ means it is collected. **Radius one is the collection edge**, not generally the grazing boundary. Grazing occurs at $\alpha_m^2+\beta_m^2=1$, or $\rho_m=n/\mathrm{NA}_{\rm obj}$. An order outside the unit collection disk can therefore still propagate.

The grating equation becomes
$$\boxed{\sigma_{x,m}=\sigma_{x,i}+m\frac{\lambda_0}{d\,\mathrm{NA}_{\rm obj}},\qquad
\sigma_{y,m}=\sigma_{y,i}.}$$
Orders remain equally spaced; changing the incident direction shifts their row relative to the unit collection disk [14, problem 3](#ref-14).

The scalar illumination setting **$\sigma$ describes the source extent**. For centered circular illumination,
$$\sigma=\frac{\mathrm{NA}_{\rm illum}}{\mathrm{NA}_{\rm obj}},$$
with both NAs referred to the mask side. Thus $\sigma=0.6$ describes a source disk of radius $0.6$, while $(\sigma_{x,m},\sigma_{y,m})$ locates one outgoing order. Each source point represents an incident plane wave with its own shifted order row. This source-size ratio is conventionally called the *partial-coherence factor* [15, Eq. (2.4)](#ref-15); annular illumination instead specifies inner and outer source radii.

### 7.4. One lithography example: mask orders, pupil, wafer

Consider an ideal projection system with vacuum wavelength $\lambda_0=193\,\mathrm{nm}$, wafer-side NA $1.35$, and an inverted image with $4\times$ reduction. Its signed transverse magnification is $M=-1/4$, meaning $x_{\rm wafer}=Mx_{\rm mask}$ and likewise for $y$. The positive reduction ratio is $D=1/|M|=4$.

The ideal sine condition relates corresponding order directions and collection apertures:
$$n_{\rm wafer}(\alpha_{\rm wafer},\beta_{\rm wafer})
=\frac1M\,n_{\rm mask}(\alpha_{\rm mask},\beta_{\rm mask}),\qquad
\mathrm{NA}_{\rm wafer}=\frac{\mathrm{NA}_{\rm mask}}{|M|}.$$
A smaller spatial image therefore has a larger spread in NA coordinates [11, Eqs. (27)–(30)](https://doi.org/10.1117/12.848252). Here the signed coordinate multiplier is $-4$, and the mask-side collection NA is $1.35/4=0.3375$.

Illuminate a mask grating of pitch $d_{\rm mask}=800\,\mathrm{nm}$ in air at normal incidence. Its order coordinates are
$$n_{\rm mask}(\alpha_m,\beta_m)
=\left(m\frac{\lambda_0}{d_{\rm mask}},0\right)=(0.24125m,0),$$
$$(\sigma_{x,m},\sigma_{y,m})_{\rm mask}
=\left(m\frac{\lambda_0}{d_{\rm mask}\,\mathrm{NA}_{\rm mask}},0\right)
\approx(0.7148m,0).$$

| Mask order $m$ | Mask-side NA coordinates | Sigma-space radius $\rho_m$ | Outcome |
| --- | --- | --- | --- |
| $0$ | $(0,0)$ | $0$ | Collected; wafer-side NA coordinates $(0,0)$. |
| $+1$ | $(+0.24125,0)$ | $0.715$ | Collected; wafer-side NA coordinates $(-0.965,0)$. |
| $-1$ | $(-0.24125,0)$ | $0.715$ | Collected; wafer-side NA coordinates $(+0.965,0)$. |
| $\pm2$ | $(\pm0.48250,0)$ | $1.430$ | Propagating on the mask side, but excluded by the objective. |

Because each side is normalized by its own NA, corresponding sigma coordinates differ only by the image inversion. For a chosen sine-mapped pupil with axes aligned to the mask directions, the same collected order has
$$\boxed{\frac{(x_{p,m},y_{p,m})}{R}
=(\sigma_{x,m},\sigma_{y,m})_{\rm mask}
=-(\sigma_{x,m},\sigma_{y,m})_{\rm wafer}.}$$
For $m=+1$, the mask-side pair is approximately $(0.7148,0)$ and the wafer-side pair is $(-0.7148,0)$. An illustrative pupil radius $R=3\,\mathrm{mm}$ puts that order at $(x_{p,+1},y_{p,+1})\approx(2.144,0)\,\mathrm{mm}$. The physical radius $R$ comes from the optical design or relay; NA and image magnification alone do not determine it.

The geometrical image pitch is $|M|d_{\rm mask}=200\,\mathrm{nm}$. The pupil separates the accepted directions, and the image-forming optics recombine their fields at the wafer. Their amplitudes, relative phases and polarization are needed to calculate image intensity; the geometry above establishes only direction, collection and position.

### 7.5. Other optical systems designed toward the sine condition

| Optical system | Concrete example and role of the condition |
| --- | --- |
| **Corrected immersion microscope objectives** | Tran and Oldenbourg use 60×/1.2 NA water-immersion and 60×/1.4 NA oil-immersion Plan Apo objectives, calibrating angle-to-aperture position using known grating orders [7, §2.1 and Table 2](https://pmc.ncbi.nlm.nih.gov/articles/PMC8005462/). |
| **Aplanatic high-NA condensers** | Farré and colleagues analyze a Nikon T-CHA oil-immersion condenser, NA 1.4, for optical-trap detection. Their ray tracing and measurements check the sine mapping needed to relate detector position to transverse optical momentum [8, Fig. 11](https://doi.org/10.1364/OE.20.012270). |
| **Schwarzschild-based aplanatic mirror systems** | Wolter–Schwarzschild telescope designs use mirror shapes constrained by the sine condition together with on-axis focusing. They illustrate deliberate aplanatic design using reflecting surfaces [9](https://openaccess.inaf.it/entities/publication/0c76045c-44a4-4277-8fc0-85e1c57d0f97). |

These are design conditions at specified wavelengths, conjugates and surrounding media. A manufactured system's mapping accuracy is something to verify under its actual operating conditions. The circular pupil assumed above is the simple introductory case; mirror systems can also have a central obscuration.

For now, retain the sequence **Harvey coordinates → NA coordinates → sigma coordinates → physical pupil positions when the chosen pupil has the sine mapping**. Deriving the lens-design condition and the field-amplitude transformation belongs to a later treatment.


### Optional small calculation 7: locate an order in sigma space and check collection

Use the earlier real order with $(\alpha,\beta)=(0.8,0.4)$. Put the grating in a medium with $n=1.33$ and collect with an ideal objective of $\mathrm{NA}_{\rm obj}=1.2$. Take an illustrative pupil radius $R=3\,\mathrm{mm}$.

This calculation holds the outgoing direction fixed and applies the collection and mapping rules; it does not recompute diffraction for a changed medium at fixed vacuum wavelength. Physical pupil positions below are in **millimetres**.

**Predict and rerun:** reduce only the objective NA to $1.0$. Does the sigma-space radius become larger or smaller? Does the order become evanescent, or does a real order simply fall outside the collection aperture?


```python
alpha, beta = 0.8, 0.4  # one outgoing order's direction cosines
n, NA_obj = 1.33, 1.2
R_mm = 3.0  # illustrative physical pupil radius

if (not np.all(np.isfinite([alpha, beta, n, NA_obj, R_mm]))
        or not 0 < NA_obj <= n or R_mm <= 0):
    raise ValueError("Use finite parameters, 0 < NA_obj <= n, and R_mm > 0.")

sin_theta_squared = alpha**2+beta**2
if sin_theta_squared > 1+1e-12:
    print("Evanescent candidate: there is no real outgoing direction to collect.")
elif abs(sin_theta_squared-1) <= 1e-12:
    print("Grazing candidate: no propagation toward an objective below the grating.")
else:
    na_coordinates = n*np.array([alpha, beta])
    radial_na = np.linalg.norm(na_coordinates)
    print("NA coordinates (n*alpha, n*beta) =", np.round(na_coordinates, 6))
    print("Radial NA coordinate =", round(radial_na, 6))
    sigma_coordinates = na_coordinates/NA_obj
    rho = np.linalg.norm(sigma_coordinates)
    print("Sigma coordinates (sigma_x, sigma_y) =", np.round(sigma_coordinates, 6))
    print("Sigma-space radius rho =", round(rho, 6), "(collection edge: 1)")
    collected = rho <= 1+1e-12
    print("Real outgoing order collected:", bool(collected))
    if collected:
        print("Sine-mapped pupil (xp/R, yp/R) =", np.round(sigma_coordinates, 6))
        print("Physical pupil (xp, yp), mm =", np.round(R_mm*sigma_coordinates, 6))
    else:
        print("The order still propagates, but it has no collected point inside this pupil.")
```

$$\left(\frac{x_p}{R},\frac{y_p}{R}\right)\approx(0.8867,0.4433),\qquad

Reducing the objective NA to $1.0$ gives sigma coordinates $(1.064,0.532)$ and $\rho_m\approx1.1896>1$, so the order is excluded.
