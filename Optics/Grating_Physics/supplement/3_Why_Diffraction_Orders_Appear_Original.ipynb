## 1. Recall the phase at the surface

The grating lies at $z=0$, and its pattern repeats along $x$ with period $d$.

From notebook two, the incident phase at the surface is

$$\phi_i(x,0,t)=k_{ix}x-\omega t.$$

Keep $\mathbf k_i=(k_{ix},0,k_{iz})$ and $k=2\pi/\lambda$. The signed component $k_{ix}$ measures phase change along $x$, in radians per micrometre when $x$ is in micrometres.

The plane-wave phase and the choice of time convention are reviewed in [1, §3.3, Eq. (8)](#ref-1).


## 2. Compare two positions at the same time

Choose A at position $x$ and B at $x+d$. Both are on the grating surface. Compare their phases at the **same instant**:

$$\phi_{i,A}=k_{ix}x-\omega t,$$
$$\phi_{i,B}=k_{ix}(x+d)-\omega t.$$

Define the phase difference as the phase at B minus the phase at A. The common time term cancels, giving

$$\boxed{\Delta\phi_i=\phi_{i,B}-\phi_{i,A}=k_{ix}d.}$$

The meaning is direct: **phase change per unit distance × one period = phase change across that period**.

### Small calculation 1: phase across one period

Use section 3's starting settings: $\lambda=d=2\,\mu\mathrm m$ and $\theta_i=30^\circ$. Sample A at $x=0$ and B at $x=d$, at the same instant and at $z=0$.

`np.deg2rad` converts degrees to radians before `np.sin` is applied. `np.diff` subtracts neighbouring array entries, giving the B-minus-A phase change. Dividing by $2\pi$ expresses that change in full cycles.

**Predict, then try:** double `d` to `4.0`. Restore `d = 2.0`, then set `theta_i_deg = 0.0`. Compare the printed changes with the incident-phase explorer. This cell has its own editable settings; it does not move the sliders.


```python
wavelength, d = 2.0, 2.0  # micrometres
theta_i_deg = 30.0
k = 2*np.pi / wavelength
k_ix = k * np.sin(np.deg2rad(theta_i_deg))
x = np.array([0.0, d])  # A, B
phase = k_ix * x
delta_phi_i = np.diff(phase)[0]
print(' x (um)   incident phase/pi')
for position, phi in zip(x, phase):
    print(f'{position:7.2f}   {phi/np.pi:17.3f}')
print(f'B minus A: {delta_phi_i/np.pi:+.3f} pi rad')
print(f'Cycles across d: {delta_phi_i/(2*np.pi):+.3f}')
```

Doubling $d$ gives $\Delta\phi_i=2\pi$, one full cycle. Restoring $d=2\,\mu\mathrm m$ and choosing normal incidence gives $k_{ix}=0$ and no spatial phase change across the period. These are phase changes of the incident wave; they are not yet order labels.

## 4. Connect the phase change to diffraction orders

Recall the [single-slit view at the start of notebook two](../2_1_From_a_Slit_to_a_Grating.ipynb#single-slit-kinematics): one isolated opening has a continuous range of candidate outgoing directions.

**The relation $\Delta\phi_i=k_{ix}d$ alone does not select discrete orders.** Repeating the opening introduces an additional condition: the outgoing field must fit the repeating grating and its illumination.

Assume an ideal linear grating that repeats indefinitely with period $d$.

**First, keep track of the incident phase change.** Keep the incident subscript $i$:

$$\Delta\phi_i=k_{ix}d.$$

Move the whole description one period along $x$. The grating geometry looks identical, while the incident wave acquires this phase shift. Since the grating is unchanged and its response is linear, the outgoing field must acquire the same phase shift, up to complete cycles. This is the phase-repeat property called *pseudo-periodicity* in [2, §2.3, Eq. (2.46)](#ref-2).

**Now describe a candidate outgoing plane wave** with vector $\mathbf k_{\mathrm{out}}=(k_{\mathrm{out},x},0,k_{\mathrm{out},z})$. Its phase change obeys the same relation:

$$\Delta\phi_{\mathrm{out}}=k_{\mathrm{out},x}d.$$

To fit the repeating response, its phase change must agree with the incident phase change up to an integer number of full cycles. One full cycle is $2\pi$ radians, so

$$\boxed{\Delta\phi_{\mathrm{out}}-\Delta\phi_i=2\pi m},
\qquad m=0,\pm1,\pm2,\ldots$$

The repeat condition is the new physical constraint. Substituting the two expressions above gives

$$\left(k_{\mathrm{out},x}-k_{ix}\right)d=2\pi m.$$

When the candidate satisfies this condition for an integer $m$, it is diffraction order $m$: write $\mathbf k_{\mathrm{out}}=\mathbf k_m$ and $k_{\mathrm{out},x}=k_{mx}$. Then

$$\boxed{k_{mx}=k_{ix}+m\frac{2\pi}{d}.}$$

**The integer $m$ is the diffraction-order number.** It counts the extra complete spatial phase cycles per period relative to the incident wave. Once the illumination and period are fixed, the candidate horizontal wavevector components come in steps of $2\pi/d$. [2, §2.3, Eqs. (2.48)–(2.50)](#ref-2)

**Experimental connection: test the integer rule.** Measure the incident angle
and the centres of resolved outgoing order peaks, using the angle convention
of Notebook 2. With known $d$ and $\lambda$, the preceding equation predicts

$$\frac{d}{\lambda}(\sin\theta_m-\sin\theta_i)=m.$$

For a finite grating, compare the peak centres with these integer values
within the angular measurement uncertainty. The phase-cycle diagrams explain
this observable angular pattern; a camera does not directly count the
unwrapped field-phase cycles across one grating period.


**A simpler case: normal incidence.** Here $k_{ix}=0$, so

$$\Delta\phi_m=k_{mx}d=2\pi m.$$

An outgoing component returns to the same phase after one grating period: its phase has changed by a whole number of cycles. This connects to the periodic-wave picture in the EMPossible explanation [3](#ref-3).

For tilted incidence, use the **difference** between outgoing and incident phase changes: $(k_{mx}-k_{ix})d=2\pi m$.

This selects candidate orders. The next view shows the candidates with real outgoing directions so that we can draw their plane waves.

### Small calculation 2: build the selected order's wave

Reproduce the section-5 starting selection: $\lambda=d=2\,\mu\mathrm m$, $\theta_i=30^\circ$, and $m=-1$. First compute $k_{mx}$ from the phase-repeat condition. Then choose the negative $k_{mz}$ branch, so the wave travels into $z<0$.

`remaining` is $k^2-k_{mx}^2$. Its sign tells us whether the selected component has a real outgoing direction. `np.sqrt` supplies the vertical component, and `np.arctan2` recovers the signed angle from $-z$. The cosine at the end is the same plane-wave expression used for the outgoing part of the interactive image.

**Predict, then try:** change `m` to `0`, then to `1`, keeping the other code settings fixed. Which selection continues along the incident direction, and which cannot be drawn as a real outgoing wave?


```python
wavelength, d, theta_i_deg = 2.0, 2.0, 30.0
m = -1  # integer order; independent of the dropdown
k = 2*np.pi / wavelength
k_ix = k * np.sin(np.deg2rad(theta_i_deg))
k_mx = k_ix + m*2*np.pi/d
remaining = k**2 - k_mx**2
if remaining < -1e-12*k**2:
    print(f'm = {m:+d}: no real outgoing direction')
else:
    k_mz = -np.sqrt(max(0.0, remaining))
    theta_m_deg = np.rad2deg(np.arctan2(k_mx, -k_mz))
    incident_cycles = k_ix*d / (2*np.pi)
    outgoing_cycles = k_mx*d / (2*np.pi)
    x, z = 0.0, -1.0  # one pixel below the grating, in micrometres
    field_at_pixel = np.cos(k_mx*x + k_mz*z)  # t = 0, unit amplitude
    print('Outgoing order (k_mx, k_mz), rad/um:', np.round([k_mx, k_mz], 3))
    print(f'Angle from -z: {theta_m_deg:+.2f} degrees')
    print('Incident, outgoing, difference (cycles):',
          np.round([incident_cycles, outgoing_cycles, outgoing_cycles-incident_cycles], 3))
    print(f'Signed field at (x,z)=({x:g},{z:g}): {field_at_pixel:+.3f}')
```

For $m=-1$, $(k_{mx},k_{mz})\approx(-1.571,-2.721)\,\mathrm{rad}/\mu\mathrm m$ and $\theta_m=-30^\circ$. The incident phase change is $+0.5$ cycles; the outgoing change is $-0.5$ cycles. Their difference is $-1$ full cycle. At the chosen pixel the signed field is approximately $-0.913$.

For $m=0$, the outgoing angle is $+30^\circ$ and both phase changes are $+0.5$ cycles, giving zero difference. For $m=+1$, $k_{mx}=1.5k$: this selection has no real outgoing direction. These answers refer to the fixed parameters written in this cell.

</details>


### Small calculation 3: test the seam in phase cycles

Use the comparison view's example: $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, normal incidence, and $m_{\mathrm{test}}=0.5$. The end of one cell gains $k_{\mathrm{out},x}d$ relative to A. The next copy starts with the incident offset $k_{ix}d$ relative to the same A.

The complex phase factor `np.exp(1j*phase)` has magnitude one. Adding $2\pi$ to its phase leaves its value unchanged. Dividing the next-copy factor by the end-of-cell factor tests the seam: a ratio of 1 means agreement. `np.isclose` allows for tiny numerical rounding errors.

**Predict, then try:** change `m_test` from `0.5` to `1.0`. Will the printed phase changes become numerically equal, or will their complex phase factors agree?


```python
wavelength, d, theta_i_deg = 2.0, 4.0, 0.0
m_test = 0.5  # independent of the comparison slider
k = 2*np.pi / wavelength
k_ix = k * np.sin(np.deg2rad(theta_i_deg))
k_out_x = k_ix + m_test*2*np.pi/d
if abs(k_out_x) > k*(1+1e-12):
    raise ValueError('Choose a comparison value with a real outgoing direction.')
end_phase = k_out_x*d
next_phase = k_ix*d
end_factor = np.exp(1j*end_phase)
next_factor = np.exp(1j*next_phase)
seam_ratio = next_factor / end_factor
print('End, next start, end-minus-next (cycles):',
      np.round(np.array([end_phase, next_phase, end_phase-next_phase])/(2*np.pi), 3))
print('End factor; next-copy factor:', np.round(end_factor, 3), np.round(next_factor, 3))
print('Next/end ratio:', np.round(seam_ratio, 3))
print('Copies join:', bool(np.isclose(seam_ratio, 1.0, rtol=0, atol=1e-10)))
```

<details><summary>Answer for the stated starting values and changes</summary>

For $m_{\mathrm{test}}=0.5$, the orange end phase is $0.5$ cycles and the green next-start phase is zero. Their factors are $-1$ and $+1$, and the seam ratio is $-1$: a half-cycle mismatch.

For $m_{\mathrm{test}}=1$, the phase changes are still different numbers—one cycle and zero cycles—but both factors are $+1$. Their ratio is 1 and the copies join. The comparison is between **phase changes relative to A**, not between absolute field colours at two unrelated points.

</details>


## 6. Connect phase cycles to path length

One wavelength of extra travel corresponds to one full phase cycle:

$$\text{extra cycles}=\frac{\Delta L}{\lambda},\qquad
\text{extra phase}=k\Delta L,\qquad k=\frac{2\pi}{\lambda}.$$

**Start at normal incidence:** set $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, and click **Normal incidence** in the view below. Its controls are shared with section 5: changing either view updates both. Turn on **Compare non-integer values** and set the comparison value to $0.5$.

- **Left:** the same continuous outgoing plane wave, with $\mathbf{k}_{\mathrm{out}}$ and its projection $k_{\mathrm{out},x}$. Here A and B are at the grating surface, one period apart. Their phase difference is still $k_{\mathrm{out},x}d$.
- **Right:** paths through those same two openings. The green incoming paths begin on a common incident wavefront. The blue outgoing paths end on a common plane perpendicular to their direction. The grey dashed lines join those starting and ending points. This is the parallel-ray geometry of a far-field direction. The path-projection construction follows the grating-equation explanation in [4, §2.1, Fig. 2-2](#ref-4).

At normal incidence the incoming lengths are equal. The orange segment is the extra outgoing distance for one route. The orange dashed line and right-angle mark show how that distance is obtained by projecting $d$ along the outgoing direction.

We define the **signed** path difference as

$$\Delta L=(\text{total route through A})-(\text{total route through B}).$$

For normal incidence, $\Delta L=d\sin\theta_{\mathrm{out}}$. Since $k_{\mathrm{out},x}=k\sin\theta_{\mathrm{out}}$,

$$\boxed{k\Delta L=k\,d\sin\theta_{\mathrm{out}}=k_{\mathrm{out},x}d.}$$

The left picture compares the phase of one plane wave at A and B. The right compares the travel phases of two contributions on the outgoing reference plane. With the stated sign convention, they give the same cycle count at normal incidence.

**Change just the comparison value:**

| Value | Extra path $\Delta L$ | Phase $k\Delta L$ | Cycles $\Delta L/\lambda$ |
| --- | --- | --- | --- |
| $0.5$ | $1\,\mu\mathrm m=\lambda/2$ | $\pi$ | $0.5$: mismatch |
| $1$ | $2\,\mu\mathrm m=\lambda$ | $2\pi$ | $1$: agreement |
| $-1$ | $-2\,\mu\mathrm m=-\lambda$ | $-2\pi$ | $-1$: agreement |

For the negative direction, the route through B is longer. The highlighted segment has a positive geometric length; the readout carries the sign of A minus B. **An integer number of extra wavelengths is the same condition as an integer number of extra phase cycles.**

### Then tilt the illumination

Keep $d=4\,\mu\mathrm m$ and $\lambda=2\,\mu\mathrm m$. Set $\theta_i=30^\circ$, choose $m=0$ in **Show order**, and click **Use selected order**.

The incoming lengths now differ too. The diagram marks the extra incoming segment in green and the extra outgoing segment in orange. Define their signed projections as

$$\ell_i=d\sin\theta_i,\qquad \ell_{\mathrm{out}}=d\sin\theta_{\mathrm{out}}.$$

For **total route A minus total route B**, the incoming contribution is $-\ell_i$ and the outgoing contribution is $+\ell_{\mathrm{out}}$. Thus

$$\Delta L=\ell_{\mathrm{out}}-\ell_i
=d(\sin\theta_{\mathrm{out}}-\sin\theta_i).$$

Multiplying by $k$ recovers our existing phase condition:

$$\boxed{k\Delta L=(k_{\mathrm{out},x}-k_{ix})d=\Delta\phi_{\mathrm{out}}-\Delta\phi_i.}$$

In this zero-order example, $\ell_i=\ell_{\mathrm{out}}=2\,\mu\mathrm m$: the two contributions cancel in the total path difference, so $\Delta L=0$. Next select $m=-1$: the outgoing direction is normal, $\ell_{\mathrm{out}}=0$, and $\Delta L=-2\,\mu\mathrm m=-\lambda$. The paths still agree in phase, now with one negative full cycle of difference.

We can therefore express the same order condition in three ways:

$$\boxed{\Delta L=m\lambda
\quad\Longleftrightarrow\quad
\Delta\phi_{\mathrm{out}}-\Delta\phi_i=2\pi m
\quad\Longleftrightarrow\quad
k_{mx}=k_{ix}+m\frac{2\pi}{d}.}$$

This gives the conventional grating equation in our signed-angle convention:

$$d(\sin\theta_m-\sin\theta_i)=m\lambda.$$

The Newport path-length explanation [4, §2.1](#ref-4) uses the same integer-wavelength principle, with its own angle convention. Harvey and Pfisterer also explain how angle and order-sign conventions affect the written grating equation [5, §2, Eqs. (1)–(3)](#ref-5). Here the wavelength is in the common medium on both sides. These diagrams explain phase agreement and order directions; they do not calculate order strengths.


### Small calculation 4: convert extra path into phase

Start with $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, normal incidence, and an outgoing angle of $30^\circ$. These are the path view's $m=+1$ settings. Compute both signed projections, subtract them in the **route A minus route B** convention, and compare $k\Delta L$ with $(k_{\mathrm{out},x}-k_{ix})d$.

This cell begins with angles rather than an order label, so the path lengths come directly from the geometry. **Predict, then try:** keep the outgoing angle at $30^\circ$ and change only `theta_i_deg` to `30.0`. Then keep that incidence and set `theta_out_deg = 0.0`. Compare with the two tilted-incidence examples above.


```python
wavelength, d = 2.0, 4.0  # micrometres
theta_i_deg, theta_out_deg = 0.0, 30.0
k = 2*np.pi / wavelength
sin_i, sin_out = np.sin(np.deg2rad([theta_i_deg, theta_out_deg]))
ell_i = d*sin_i
ell_out = d*sin_out
delta_L = ell_out - ell_i  # total route A minus total route B
phase_from_path = k*delta_L
k_ix, k_out_x = k*sin_i, k*sin_out
phase_from_components = (k_out_x-k_ix)*d
print('Incoming projection, outgoing projection, total difference (um):',
      np.round([ell_i, ell_out, delta_L], 3))
print('Phase/pi from path and from k_out_x - k_ix:',
      np.round(np.array([phase_from_path, phase_from_components])/np.pi, 3))
print(f'Extra wavelengths = extra phase cycles: {delta_L/wavelength:+.3f}')
```

<details><summary>Answer for the stated starting values and changes</summary>

With normal incidence and $\theta_{\mathrm{out}}=30^\circ$, the projections are $\ell_i=0$ and $\ell_{\mathrm{out}}=2\,\mu\mathrm m$. Thus $\Delta L=\lambda$ and both phase calculations give $2\pi$: one cycle.

With both angles at $30^\circ$, the two projections are equal and $\Delta L=0$, corresponding to $m=0$. Keeping incidence at $30^\circ$ and setting the outgoing angle to zero gives $\Delta L=-2\,\mu\mathrm m=-\lambda$ and a phase difference of $-2\pi$, corresponding to $m=-1$.

</details>


The summed spreading-wave example and its two calculations are now in [Spreading waves and finite arrays](../supplement/Spreading_Waves_and_Finite_Arrays.ipynb). That separate notebook introduces a source model and relative strengths.

## Final checkpoint

1. **Can the outgoing phase change be half a cycle and still belong to an order?**

   Yes—if its difference from the incident phase change is an integer number of cycles.

2. **Does adding more openings move the order directions?**

   No—with wavelength, spacing and incidence fixed, the order directions stay fixed; the finite-array peaks narrow. [5, §1, Figs. 1–2](#ref-5)

3. **Does a white point in the signed-field snapshot necessarily mean zero intensity?**

   No—it may simply be an instantaneous zero. The distinction between the instantaneous field and time-averaged intensity is discussed in [1, §3.4](#ref-1).

**Looking ahead to [Notebook 4: Evanescent diffraction orders](../4_Evanescent_Diffraction_Orders.ipynb)**

The plots in this notebook show only orders with real outgoing directions, including grazing ones. The integer phase-repeat condition also permits candidates with $|k_{mx}|>k$, which have no real outgoing direction. [2, §2.4](#ref-2)

**“What happens to the integer orders that our plots leave out?”**

Notebook 4 will explore **evanescent orders**: what their fields look like near the grating, how their amplitudes decay with distance from it, and how changing wavelength or period takes an order between propagating and evanescent behaviour through the grazing condition.
