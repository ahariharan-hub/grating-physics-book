### Two ideal binary transmission gratings

**Binary** means two transmission levels or regions; it does not require two different amplitudes. **Transmission** describes the output side of the device. Let $R(x)$ be one in the centred interval $-D_{\mathrm{ridge}}d/2\leq x<D_{\mathrm{ridge}}d/2$, zero elsewhere in a period, and repeated with period $d$.

An ideal **binary amplitude grating** alternates between transparent and opaque regions:

$$t_A(x)=a_g+(a_r-a_g)R(x),\qquad a_r=1,\quad a_g=0.$$

Here $R=1$ labels an opening. These real levels are defined at an ideal mask plane; zero transmission does not specify whether blocked light is reflected or absorbed. This is a different device from our lossless dielectric grating.

An ideal **binary dielectric phase grating** has two phase levels:

$$t_P(x)=e^{i\phi_g}+\left(e^{i\phi_r}-e^{i\phi_g}\right)R(x),\qquad
\phi_r=k_0n_rh,\quad\phi_g=k_0n_gh.$$

Its amplitude is everywhere one, while its phase changes between regions. The phase delays use $k_0=2\pi/\lambda_0$, transmission from $z=0$ to $z=-h$, and time dependence $e^{-i\omega t}$. Thus the positive propagation phase and the absolute exit-plane reference match the earlier dielectric model.

If a relative transmission $e^{ik_0[n(x)-n_{\mathrm{out}}]h}$ is used instead, restore $e^{ik_0n_{\mathrm{out}}h}$ before comparing with the BVP. An overall phase rotates all coefficients without changing their magnitudes.

### One coefficient formula, two physical prescriptions

For unit illumination and either pair of levels $(t_r,t_g)$,

$$\boxed{C_m=\frac{1}{d}\int_{-d/2}^{d/2}t(x)e^{-imGx}\,dx
=t_g\delta_{m0}+(t_r-t_g)D_{\mathrm{ridge}}\,\operatorname{sinc}(mD_{\mathrm{ridge}}),\qquad G=\frac{2\pi}{d}.}$$

Here $\operatorname{sinc}(v)=\sin(\pi v)/(\pi v)$ with $\operatorname{sinc}(0)=1$. Use $(1,0)$ for the amplitude mask and $(e^{i\phi_r},e^{i\phi_g})$ for the dielectric phase mask. The order label $m$, magnitude $\lvert C_m\rvert$ and phase $\arg C_m$ describe the corresponding complex Fourier coefficient.

**Predict:** can a grating with $\lvert t(x)\rvert=1$ everywhere produce orders other than $m=0$?

**Explore:** compare both ideal masks, then select $m=0$, $1$ and $2$ in the complex plane. Enable **Show finite Fourier sum** and reduce $M$. The solid transmission curves remain the exact local functions; only their finite reconstruction changes. This control does not alter the BVP solution used later.

<details><summary>Optional mathematics: coefficients and normalization</summary>

The constant groove term contributes $t_g\delta_{m0}$. The centred rectangular region contributes

$$\frac{t_r-t_g}{d}\int_{-D_{\mathrm{ridge}}d/2}^{D_{\mathrm{ridge}}d/2}e^{-imGx}\,dx
=(t_r-t_g)D_{\mathrm{ridge}}\operatorname{sinc}(mD_{\mathrm{ridge}}).$$

The $1/d$ normalization makes $C_0$ the mean complex transmission. Parseval's identity gives

$$\sum_{m=-\infty}^{\infty}\lvert C_m\rvert^2
=\frac{1}{d}\int_{-d/2}^{d/2}\lvert t(x)\rvert^2\,dx
=D_{\mathrm{ridge}}\lvert t_r\rvert^2+(1-D_{\mathrm{ridge}})\lvert t_g\rvert^2.$$

This mean square is $D_{\mathrm{ridge}}$ for the amplitude mask and one for the phase mask. A truncated sum omits the remaining harmonics. These are Fourier norms, not the longitudinal-flux efficiencies discussed below.

</details>

The amplitude grating changes transmitted amplitude directly. The phase grating changes the **complex field** through phase alone, so it can produce multiple nonzero orders even with unit transmission amplitude. If its two phases coincide modulo $2\pi$, it becomes uniform and only $m=0$ remains.

At zero amplitude there is no complex-plane direction, so phase is undefined. The plots omit phases below $10^{-6}$ of unit illumination, including opaque regions and negligible coefficients. A blank phase is not a phase of zero.

The finite Fourier sum can ring near a discontinuity and leak into an ideal opaque region. That is a truncation effect, not transmission through the exact opaque mask. Increasing $M$ improves the reconstruction away from the edges; at a jump the Fourier series tends to the two-sided mean.

**The same period permits the same transverse wavevectors, while the chosen transmission levels determine their complex coefficients.** A finite-thickness physical grating can generate both amplitude and phase modulation, even when a simple local approximation changes only phase.

### Return to the dielectric grating: compare at the same output plane

For this comparison, use **only the dielectric phase mask and the HBVP of the same lossless structure**. The opaque amplitude mask is not a surrogate for that material.

The phase-mask approximation prescribes

$$U_{\mathrm{thin}}(x,-h)=t_P(x)U_i(x,0),\qquad U_i(x,0)=1.$$

It assumes local transmission with no reflection, internal round trips, lateral coupling or vector polarization effects. The present exterior indices are equal and the material indices are real. The scalar BVP retains reflection and finite-thickness coupling.

Use the lossless BVP coefficients calculated in the setup cell:

$$C_m^{\mathrm{HBVP}}(z_{\mathrm{ref}})=A_m(z_{\mathrm{ref}}).$$

For either model, an output plane $z_{\mathrm{ref}}=-h-\Delta$ with $\Delta\geq0$ gives

$$C_m(z_{\mathrm{ref}})=C_m(-h)e^{i\beta_m\Delta},\qquad
U(x,z_{\mathrm{ref}})=\sum_{m=-M}^{M}C_m(z_{\mathrm{ref}})e^{imGx}.$$

### Read the complex differences

The coefficient view reports

$$\Delta C_m=C_m^{\mathrm{HBVP}}-C_m^{\mathrm{thin}},\qquad
\Delta\phi_m=\arg\!\left[C_m^{\mathrm{HBVP}}(C_m^{\mathrm{thin}})^*\right].$$

In the complex plane, the cyan arrow runs from the thin-mask coefficient to the HBVP coefficient. Its displacement is $\Delta C_m$. A dash in the phase table means that the amplitude is below $10^{-6}$ of the incident amplitude; $\Delta\phi_m$ requires **both** coefficients to be significant.

At half duty the thin mask has no nonzero even orders, but the finite-thickness scalar problem can populate them through coupled scattering. The models also predict different amplitudes and phases for the other orders. Similar intensity curves therefore do not establish equal complex fields: intensity alone loses the local field phase, while relative order phases affect the reconstructed pattern.

Absolute coefficient phases depend on the reference plane. Here both models acquire the same factor for a given order, so their $\Delta\phi_m$ stays fixed while it remains meaningful. Relative phases **between different orders** generally change because their $\beta_m$ differ.

Both plots retain the same 41 harmonics. Close to mask discontinuities, truncation produces Gibbs oscillations; the finite sum does not reproduce an ideal step pointwise. All plots share unit incident normalization and fixed scales; no phase alignment or power renormalization is applied.

### Compare flux, not just squared amplitudes

For propagating transmitted orders,

$$\eta_m^{(T)}=\frac{\operatorname{Re}\beta_{m,\mathrm{out}}}{k_0n_{\mathrm{in}}}\lvert C_m\rvert^2.$$

Evanescent and grazing orders carry zero outgoing normal power in this model. The table computes efficiencies and all-order totals directly from the two predictions.

A unit-modulus local phase mask does **not** guarantee unit transmitted normal flux when its spectrum includes appreciable diffraction angles and evanescent harmonics. Although reflection is omitted, its power deficit is not calculated reflection or absorption. This is a limitation of applying the local approximation outside the paraxial regime. The lossless HBVP instead balances reflected and transmitted scalar flux.

A useful exact check is a completely uniform, index-matched medium: both models give the same plane wave. A uniform slab with a different index can still reflect, so it need not match the reflection-free thin mask. See the [numerical acceptance report](../wave_optics/thin_grating.md) for these checks and measured errors.

### Diffraction Scattering Coefficients: From Scalar to Vector Optics

Our $C_m$ is one complex amplitude per order for a specified incident plane wave. A vector description resolves polarization. Schematically, for a selected outgoing side,

$$\begin{pmatrix}C_{\mathrm{TE},m}^{\mathrm{out}}\\C_{\mathrm{TM},m}^{\mathrm{out}}\end{pmatrix}
=\mathbf S_m(\mathbf k_{\mathrm{in}})
\begin{pmatrix}A_{\mathrm{TE}}^{\mathrm{in}}\\A_{\mathrm{TM}}^{\mathrm{in}}\end{pmatrix}.$$

The matrix contains complex scattering amplitudes with specified reference planes and polarization bases. Its entries can depend on incidence angle and polarization; the allowed orders obey $k_{x,m}=k_{x,\mathrm{in}}+mG$. General vector geometries can mix TE and TM. For a one-dimensional isotropic grating at normal incidence, these polarizations decouple, so cross-polarization coupling is not implied by our drawing.

Maxwell matching concerns tangential electric and magnetic fields. The present scalar equation and derivative matching coincide with the TE $E_y$ problem for this nonmagnetic, $y$-invariant geometry, but they do not also give the TM response or a general vector solution. **This notebook does not calculate $\mathbf S_m$**; a later vector diffraction lesson will do that.

For a lithography connection, see Thomas Vincent Pistor, *Electromagnetic Simulation and Modeling with Applications in Lithography*, UC Berkeley (2001), [official thesis](https://www2.eecs.berkeley.edu/Pubs/TechRpts/2001/Archive/ERL-01-19.pdf) ([catalogue record](https://www.semanticscholar.org/paper/Electromagnetic-Simulation-and-Modeling-With-In-Pistor/4391a014ea787c9baab7db639eb24da20a650b0b)). Section 4.3 represents object scattering using plane waves; §4.7 discusses mask models, from the thin scalar transmission model (§4.7.2.1) to thick vector models (§§4.7.2.2–4.7.2.3). Figures 4-11, 4-14 and 4-15 discuss thin-mask limitations and observation/reference planes. Pistor is a technical reference for this connection, not the originator of scattering coefficients. We stop at the scattered field; aerial-image formation is outside this lesson.

**Takeaway:** periodicity constrains the order wavevectors; the physical model determines their complex scattering amplitudes.
