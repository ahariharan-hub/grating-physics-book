## 1. Meet the diffraction orders

In Notebook 2.1, one isolated slit had a continuous range of candidate outgoing directions. Repeating the slit indefinitely adds a constraint: only discrete directions fit the periodic structure and its illumination.

Beyond the grating, the transmitted field can be described using outgoing plane-wave components called **diffraction orders**. For an indefinitely repeating grating illuminated by one plane wave, periodicity selects discrete possible values of the wavevector component across the grooves. We will determine which of those values give real outgoing directions. [1, §2.2.1](#ref-1) In a far-field pattern from a finite illuminated grating, orders that are present have finite angular width around the ideal directions. [7, §1, Figs. 1–4](#ref-7)

The integer $m$ labels an order: $m=0$ is the zero order, $m=\pm1$ the first orders, and $m=\pm2$ the second orders. It does not count openings. With the same medium on both sides, the transmitted zero-order direction is the incident direction. [7, §3, Fig. 8](#ref-7)

### Describe an order with its wavevector

Notebook one used a wavevector to describe a plane wave's direction and spatial phase. Here the subscripts identify which wave we mean:

| Symbol | Meaning |
| --- | --- |
| $\mathbf k_i=(k_{x,i},k_{y,i},k_{z,i})$ | Incident wavevector, at normal or tilted incidence |
| $\mathbf k_m=(k_{x,m},k_{y,m},k_{z,m})$ | Outgoing wavevector of diffraction order $m$ |
| $k=2\pi/\lambda$ | Common scalar magnitude for these real wavevectors in the same medium |

Thus $|\mathbf k_i|=|\mathbf k_m|=k$: their directions and components can differ while their lengths agree. For the transmitted zero order in this geometry, $\mathbf k_0=\mathbf k_i$. In $k_{x,m}$, the $m$ identifies the order and the $x$ identifies its component across the grooves. Each component is a signed scalar.
- The single-slit notation $\mathbf k(\theta)$ in Notebook 2.1 labels a continuous candidate direction; an integer order label becomes appropriate only after introducing the periodic grating.

The grating equation can be written directly in these components:

$$\boxed{k_{x,m}=k_{x,i}+mG,\qquad k_{y,m}=k_{y,i}},\qquad G=\frac{2\pi}{d}.$$

Across the grooves, the candidate values of $k_x$ are separated by $G$. Like $k$, $G$ has units of radians per unit length: a step of $G$ adds $Gd=2\pi$ to the phase advance over one period. Along the grooves, the structure is uniform, so $k_y$ is preserved. In our current $x$-$z$ examples, $k_{y,i}=k_{y,m}=0$. We will explain why the steps must be integer multiples in Notebook 3. [3, §2.3, Eq. (2.50), and §2.4](#ref-3)

The same rule has a compact **vector form at the surface**:

$$\boxed{\mathbf k_{m,\parallel}=\mathbf k_{i,\parallel}+m\mathbf G},\qquad \mathbf G=\frac{2\pi}{d}\,\hat{\mathbf x}.$$

Here $\parallel$ means **parallel to the grating surface**: it collects the $x$ and $y$ components, excluding $z$. The symbol $\hat{\mathbf x}$ is a unit vector along $+x$, so $\mathbf G$ points across the grooves.

### Complete the outgoing wavevector

The surface rule supplies $k_{x,m}$ and $k_{y,m}$. In the same surrounding medium, an outgoing propagating plane wave still has magnitude $k=2\pi/\lambda$, so

$$k_{x,m}^2+k_{y,m}^2+k_{z,m}^2=k^2,\qquad
\boxed{k_{z,m}=-\sqrt{k^2-k_{x,m}^2-k_{y,m}^2}}.$$

For a positive quantity under the square root, the **negative** root gives $k_{z,m}<0$ and selects transmission into $z<0$. Zero gives $k_{z,m}=0$: a **grazing** direction parallel to the grating. A negative value gives no real outgoing direction; the corresponding evanescent field will be discussed later. The normal component is determined by this magnitude condition and the outgoing side, separately from the vector rule at the surface. [3, Eqs. (2.54)–(2.58) and §2.4](#ref-3)

In the plots below, **allowed directions** means real directions, including grazing ones. The order explorer reports both the total and the split between orders travelling into $z<0$ and grazing orders. This is a count of possible directions, not a count of bright or nonzero-amplitude beams.

### Recover the angle equation

For our $x$-$z$ examples, $k_{x,i}=k\sin\theta_i$ and $k_{x,m}=k\sin\theta_m$. Substituting these into the $x$-component rule and dividing by $k$ gives

$$k\sin\theta_m=k\sin\theta_i+m\frac{2\pi}{d}
\quad\Longrightarrow\quad
\boxed{\sin\theta_m=\sin\theta_i+m\frac{\lambda_0}{nd}=\sin\theta_i+m\frac{\lambda}{d}}.$$

Here $n$ is the common exterior index and $\lambda=\lambda_0/n$. With different incident and output indices, the corresponding law is

$$\boxed{n_t\sin\theta_m-p_i\sin\theta_i=m\frac{\lambda_0}{d}.}$$

- Both angles are measured from $-z$ toward $+x$, with $-90^\circ\leq\theta_m\leq90^\circ$ for transmitted or grazing directions.
- Positive $m$ increases $k_{x,m}$ and negative $m$ decreases it; at tilted incidence the sign of $m$ need not be the sign of $\theta_m$.
- Requiring the normalized expression for $\sin\theta_m$ to lie between $-1$ and $+1$ is the same check for a real direction as the square-root condition above. [3, §2.4](#ref-3)

The **fixed preview** below uses **$\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, and $\theta_i=0^\circ$**. For these values, the real orders are $m=-2,-1,0,+1,+2$, all with $k_{z,m}<0$. This preview does not change when you use the later interactive controls. Arrows share an origin only to compare directions; each represents the direction of an extended plane-wave component.

**Direction is not strength.** The relation does not give the amplitude of an order, and an allowed direction can have zero amplitude. In this fixed preview, the outgoing arrows have equal length, colour, and line width; they carry no information about power. Determining strengths requires an additional calculation.
- [1, §2.8](#ref-1); [2, §1](#ref-2)

### Small calculation 1: build one outgoing wavevector

Use the preview's settings: $\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, and normal incidence. Construct order $m=+1$ by adding $mG$ to the incident $x$ component, retaining $k_y=0$, and choosing the negative $k_z$ branch. In the code, `k_i` and `k_m` are three-component vectors, while `k` is their common scalar magnitude. All printed components are in radians per micrometre.

`np.sqrt` takes a square root and `np.linalg.norm` gives a vector's length. The `if` statement checks for a real direction before taking the square root. A small tolerance treats numerical rounding at grazing as zero. The final line distinguishes transmission into $z<0$ from grazing. This calculation uses the values written in its code cell, independently of the interactive controls.

**Predict, then try:** keep wavelength, period, and incidence angle fixed, and change `m` to `-1`, then `0`, then `3`. Which components change, and which choice has no propagating direction?

```python
wavelength, d, theta_deg = 1.0, 2.5, 0.0  # lengths in micrometres
m = 1  # integer diffraction-order label
k, G = 2*np.pi / wavelength, 2*np.pi / d
theta = np.deg2rad(theta_deg)
k_i = k * np.array([np.sin(theta), 0.0, -np.cos(theta)])
k_x_m, k_y_m = k_i[0] + m*G, k_i[1]
remaining = k**2 - k_x_m**2 - k_y_m**2
tolerance = 1e-12 * k**2  # rounding tolerance at grazing
if remaining < -tolerance:
    print(f'm = {m:+d}: no real propagating direction')
else:
    k_z_m = 0.0 if abs(remaining) <= tolerance else -np.sqrt(remaining)
    k_m = np.array([k_x_m, k_y_m, k_z_m])
    print('Incident k_i:', np.round(k_i, 3))
    print(f'Order {m:+d} k_m:', np.round(k_m, 3))
    print('|k_i|, |k_m| (rad/um):', np.round([np.linalg.norm(k_i), np.linalg.norm(k_m)], 3))
    print('Direction:', 'grazing along z=0' if k_z_m == 0 else 'into z<0')
```

<details><summary>Answer for the stated starting values and changes</summary>

With $\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, and normal incidence as written in the code, the $m=+1$ vector is approximately $(2.513,\,0,\,-5.759)\,\mathrm{rad}/\mu\mathrm m$. Changing to $m=-1$ reverses its $x$ component; at normal incidence these two orders have the same negative $z$ component. Both wavevector lengths are $k\approx6.283\,\mathrm{rad}/\mu\mathrm m$.

For $m=0$, $\mathbf k_m=\mathbf k_i$: the transmitted zero order continues downward along the incident direction. For $m=3$ with these same parameters, $|k_{x,m}|>k$, leaving no real value of $k_{z,m}$. The code therefore reports no propagating direction. These are checks of allowed directions, not calculations of order amplitudes.

</details>

### The same orders on two surfaces

The number of real order directions depends on wavelength, period, and incidence angle. Represent each direction by a ray from the origin. Every such ray meets a **reference sphere centred at the grating origin**. Rays with $k_{z,m}<0$ also meet a **flat observation plane** at $z=-L$; grazing rays run parallel to that plane and never meet it. [4, Figs. 1 and 5](#ref-4); [7, §3, Fig. 9](#ref-7)

These markers represent directions of extended plane-wave components. The common ray origin is a drawing convention, and the sphere is a reference surface rather than a spherical wavefront. Neither surface predicts focused spots, spot widths, or order power.

The **order explorer** starts with $\lambda=1\,\mu\mathrm m$, $d=2.5\,\mu\mathrm m$, and normal incidence, with $R=2\,\mu\mathrm m$ and $L=3\,\mu\mathrm m$. These starting values give five real orders, $m=-2,-1,0,+1,+2$, all travelling into $z<0$. The live readout gives the count for the **current controls**, including a separate grazing count.

The main plot is an $x$-$z$ section at $y=0$: the sphere appears as a semicircle and the plane as a horizontal line. Coordinates are lengths in micrometres. Below it, a horizontal ruler projects the blue sphere markers onto their $x$ coordinates; the ruler is a coordinate aid, not another observation surface.

The controls have two roles:

- **Wavelength, period, and incidence angle** determine the allowed directions and order count.
- **Sphere radius $R$ and plane distance $L$** determine where the reference surfaces intersect those directions. Changing either length leaves the order directions and count unchanged.

**Reset example** restores this explorer's starting values and the component-view selection to $m=+1$. Notebook 2.4 initializes its own copy of these controls for the three-panel component view. The incident-wave explorer in Notebook 2.2 has separate controls.

Change one parameter at a time:

- Click **Reset example**, then change $L$ from $3$ to $6\,\mu\mathrm m$. Which markers move? Reset, then change $R$ from $2$ to $4\,\mu\mathrm m$. Which markers move? Reset, then set $\theta_i=30^\circ$. Does the zero-order direction remain on the $z$ axis?
- Reset, then vary wavelength or period. Watch the order count and the approach to grazing.
- Reset, then set $d=4\,\mu\mathrm m$ and $R=4\,\mu\mathrm m$, keeping $\lambda=1\,\mu\mathrm m$ and normal incidence. Compare $m=+3$ and $+4$ on the semicircle and on the ruler. Is their horizontal gap larger than the other gaps?

The plot limits stay fixed. A finite plane intersection can fall outside the displayed $x$ range near grazing; the readout lists those orders, and arrows at the plot edges show the rays continuing. Such an off-screen intersection still exists. An exactly grazing ray has no plane intersection at any finite distance.

**Our next question:** why does equal horizontal spacing not necessarily mean equal spacing in angle or along the semicircle?

These rays are geometrical representatives of plane-wave directions, drawn from a common reference origin. They are not the boundaries or separated paths of finite order beams; the ideal component extends throughout the homogeneous output region. Sphere and plane markers locate that construction, not calculated illuminated spots. [3, §2.3](#ref-3)

### If the media on the two sides differ

The explorer below retains equal incident and output indices. More generally, at fixed vacuum wavelength $\lambda_0$, let the incident and transmitted lossless media have real indices $n_i$ and $n_t$. Periodicity still sets

$$k_{x,m}=k_{x,i}+m\frac{2\pi}{d},\qquad k_{y,m}=k_{y,i}.$$

The output dispersion relation instead uses $k_t=n_tk_0$:

$$k_{z,m}=-\sqrt{(n_tk_0)^2-k_{x,m}^2-k_{y,m}^2},\qquad k_0=2\pi/\lambda_0.$$

For strictly propagating transmission the radicand is positive; zero is grazing, and a negative value selects the outgoing evanescent decay branch. In the $xz$ plane, angles measured from $-z$ toward $+x$ give

$$\boxed{n_t\sin\theta_m=n_i\sin\theta_i+m\frac{\lambda_0}{d}.}$$

For $m=0$ this is Snell's law. For equal exterior indices it reduces to the medium-wavelength grating equation already used here. Thus periodicity fixes the transverse increment, while the output index fixes the available propagation disk and angles. This does not determine coefficient magnitudes or phases; refraction and interface reflection belong to the scattering problem.

**Worked comparison.** Keep normal incidence, $\lambda_0=0.600$ micrometre and $d=1.000$ micrometre. The $+1$ order has the same $k_x=2\pi/d$ for output air and output index 1.5. Its angle changes from $36.87$ degrees to $23.58$ degrees. Air supports only $m=0,\pm1$ as strictly propagating orders; index 1.5 supports $0,\pm1,\pm2$. The unchanged controls below do not simulate this unequal-medium interface.

**Predict:** at fixed $\lambda_0$ and incidence, does increasing $n_t$ change the transverse order spacing, or only the longitudinal components, angles and propagation eligibility?

<details><summary>Answer</summary>

The transverse step remains $2\pi/d$. Increasing $n_t$ enlarges the output propagation disk; the longitudinal components and angles adjust. This kinematic statement does not predict how the redistributed power is shared among orders.

</details>
