# 9.1 Defining the Complex Exit Field

A thin-mask calculation begins with the incident field and a prescribed transmission. Their product gives the exit field. Comparing normal and tilted illumination shows why the mask's periodicity and the field's spatial variation must be distinguished.

We follow one binary grating across a fixed 10 µm field. Start with its transmission and one incident plane wave, multiply them, then compare normal and tilted illumination. The same interactive view is used throughout. Sampling and FFT recovery follow in [9.2](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb).

## 1. The exit field records amplitude and phase

Choose the plane immediately after the mask, $z=0^-$. Its complex field is

$$\boxed{U_{\mathrm{exit}}(x)=U(x,0^-)
=\underbrace{A_{\mathrm{exit}}(x)}_{\text{amplitude at }x}
\underbrace{e^{i\phi_{\mathrm{exit}}(x)}}_{\text{phase at }x}.}$$

Each position has one complex value. Its magnitude gives the amplitude; its angle gives the phase. Where the field is zero, phase is undefined. This describes the field before choosing how to obtain it.

## 2. Build that field from a mask and illumination

Our scalar thin-mask model prescribes ([Kirchauer, Eq. 4.37](#ref-9-1-kirchauer-mask) [\[1\]](#ref-9-1-kirchauer-mask))

$$\boxed{U_{\mathrm{exit}}(x)
=\underbrace{t(x)}_{\text{mask transmission}}
\underbrace{U_i(x)}_{\text{incident field}}.}$$

In general, both factors may have amplitude and phase:

$$t(x)=a_t(x)e^{i\phi_t(x)},\qquad U_i(x)=A_i(x)e^{i\phi_i(x)},$$
$$U_{\mathrm{exit}}(x)=a_t(x)A_i(x)e^{i[\phi_t(x)+\phi_i(x)]}.$$

**Amplitudes multiply; phases add.** For our binary mask, transmission is 1 in an opening and 0 behind an opaque strip. It adds no phase in the openings. This ideal prescription neglects thickness-dependent scattering and reflection.

| Setting | Baseline |
| --- | --- |
| Displayed field | $L=10$ µm, from $-5$ to $+5$ µm |
| Period control | $d=2000$ nm $=2$ µm |
| Duty control | $D=0.5$, defined as the **open fraction** |
| Derived widths | $W_{\mathrm{open}}=Dd=1$ µm; $W_{\mathrm{close}}=(1-D)d=1$ µm |
| Illumination | One plane wave, $A_i=1$, wavelength in the incident medium $\lambda_i=2$ µm; starts at $\theta_i=30^\circ$ |

The readout also gives

$$\frac{L}{d}=5\ \text{periods},\qquad
f_g=\frac1d=0.5\ \text{cycles/}\mu\mathrm{m},\qquad
G=\frac{2\pi}{d}=\pi\approx3.142\ \text{rad/}\mu\mathrm{m}.$$

Changing $D$ changes the opening fraction; changing $d$ changes the order spacing. The plotted window is a portion of the periodic grating and does not itself impose PBC.

**Define the two factors.** Openings are centred at $x=jd$, for integer $j$:

$$t(x)=\begin{cases}
1,& |x-jd|<Dd/2\quad\text{for some integer }j,\\
0,&\text{otherwise}.
\end{cases}$$

The incident wave travels from $z>0$ toward the mask. Its phase at the surface is

$$U_i(x)=e^{ik_{x,i}x},\qquad k_{x,i}=\frac{2\pi}{\lambda_i}\sin\theta_i.$$

**Multiply at each position:**

$$\boxed{U_{\mathrm{exit}}(x)=t(x)e^{ik_{x,i}x}
=\begin{cases}
e^{ik_{x,i}x},&\text{inside an opening},\\
0,&\text{behind an opaque strip}.
\end{cases}}$$

Use consistent length units in the equations. The controls convert nm to µm. The view extends [2.2](2_2_Illumination_and_the_Surface_Field.ipynb): the upper panels show the illumination; the lower panels compare the actual exit field (solid) with an ordinary-PBC cell copy (dashed). Their amplitudes coincide; their complex components and phases can differ. Blue bands mark openings.

This is one angular component of an [ideal Köhler illuminator](#ref-9-1-kirchauer-mask) [\[1\]](#ref-9-1-kirchauer-mask): one source-pupil point supplies a plane wave across the field.

## 3. What repeats between neighbouring cells?

Keep $d=2$ µm, $D=0.5$ and $\lambda_i=2$ µm. The view opens at $30^\circ$ so the comparison is visible. Try normal incidence, then return to $30^\circ$:

1. **Normal incidence, $\theta_i=0^\circ$.** Here $k_{x,i}=0$, so $U_{\mathrm{exit}}=t$. Every opening passes the same complex value, and the actual and PBC-copy curves coincide.
2. **Tilted incidence, $\theta_i=30^\circ$.** Here $k_{x,i}=\pi/2$ rad/µm. Moving one period advances the phase by $\Delta\phi=k_{x,i}d=\pi=180^\circ$.

| Value at an opening centre | $x=0$ | $x=d$ | $x=2d$ |
| --- | --- | --- | --- |
| Actual field, normal incidence | $1$ | $1$ | $1$ |
| Actual field, $30^\circ$ | $1$ | $-1$ | $1$ |
| Ordinary-PBC copy, $30^\circ$ | $1$ | $1$ | $1$ |

The dashed curves repeat the central cell $-d/2\le x<d/2$ with identical complex values in every cell. The actual tilted field instead changes sign between neighbouring cells. Compare corresponding points:

$$\underbrace{t(x+d)=t(x)}_{\text{mask repeats}},\qquad
\underbrace{|U_{\mathrm{exit}}(x+d)|=|U_{\mathrm{exit}}(x)|}_{\text{amplitude repeats}},$$
$$\boxed{U_{\mathrm{exit}}(x+d)=e^{i\Delta\phi}U_{\mathrm{exit}}(x).}$$

The boxed relation is **Bloch periodicity**: the same local profile acquires a known phase factor between cells. **Ordinary PBC** is its special case $e^{i\Delta\phi}=1$. Copying identical complex values into every cell is appropriate only in that case. See the [ordinary and Bloch boundary definitions in Meep](#ref-9-1-meep) [\[2\]](#ref-9-1-meep).

At the tilted baseline, two cells give a full $2\pi$ phase advance, so the actual field repeats after $2d=4$ µm. Across the 10 µm window the advance is $5\pi$: the mask fits five periods, but the full field acquires a factor of $-1$.

**Phase wrapping is a display choice.** The upper incident-phase plot is unwrapped; both lower exit-phase curves are wrapped. For example, $225^\circ$ and $-135^\circ$ describe the same complex value. The difference between solid and dashed curves comes from the cell-to-cell phase factor, not from wrapping. Exit-field phase is undefined behind the opaque strips.

## 4. Two routes to the same exit spectrum

For our unit-amplitude plane wave, both routes describe the same exit field and its Bloch relation. The mask $t(x)$ is periodic in both. The product and shifted-order forms follow [Kirchauer, Eqs. 4.37–4.42](#ref-9-1-kirchauer-mask) [\[1\]](#ref-9-1-kirchauer-mask).

**1. Illuminate first: Bloch periodicity appears in the position-space field.**

$$\underbrace{t(x)}_{\text{periodic mask}}
\xrightarrow{\;\times e^{ik_{x,i}x}\;}
\underbrace{U_{\mathrm{exit}}(x)}_{\text{Bloch-periodic field}}
\xrightarrow{\;\mathrm{FT}\;}
\boxed{\widetilde U_{\mathrm{exit}}(k_x)}.$$

**2. Transform the mask first: the shifted order locations encode Bloch periodicity.**

$$\underbrace{t(x)}_{\text{periodic mask}}
\xrightarrow{\;\mathrm{FT}\;}
\underbrace{\widetilde t(k_x)}_{\text{lines at }mG}
\xrightarrow{\;\text{shift by }+k_{x,i}\;}
\boxed{\widetilde U_{\mathrm{exit}}(k_x)}.$$

For this periodic mask, the spectra consist of discrete lines. Their complex coefficients remain $C_m=t_m$; their locations become $k_{x,m}=k_{x,i}+mG$. Every shifted order carries the same phase factor across one cell:

$$\underbrace{e^{ik_{x,m}d}=e^{ik_{x,i}d}}_{\text{Bloch phase in every order}},
\qquad mGd=2\pi m.$$

**One-cell FFT:** in route 1, first remove the known carrier,
$U_{\mathrm{exit}}(x)e^{-ik_{x,i}x}=t(x)$. In route 2, $t(x)$ is already periodic. A [normalized FFT](#ref-9-1-numpy-fft) [\[3\]](#ref-9-1-numpy-fft) of one period of this periodic factor estimates $t_m=C_m$; the illumination shift then specifies the physical order locations.

[9.2](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb) develops the sample grid and FFT normalization; it writes the period as $p$. The derivation below gives continuous-transform and discrete-order representations, including an exact three-order check.

### Continuous representation

A simple component $e^{ik_xx}$ has a phase that changes uniformly across $x$. Larger $|k_x|$ means faster spatial variation; the sign gives the direction of that phase change. Fourier analysis finds how much of each component builds the field:

$$\boxed{\widetilde U_{\mathrm{exit}}(k_x)
=\int_{-\infty}^{\infty}U_{\mathrm{exit}}(x)e^{-ik_xx}\,dx.}$$

Adding those components back gives

$$U_{\mathrm{exit}}(x)=\frac1{2\pi}\int_{-\infty}^{\infty}
\widetilde U_{\mathrm{exit}}(k_x)e^{ik_xx}\,dk_x.$$

Here $k_x$ is in rad/nm. The FFT appendix also uses $f_x=k_x/(2\pi)$ in cycles/nm; the matching inverse integral then has $e^{i2\pi f_xx}$ and no $1/(2\pi)$ factor.

| Expression | Category | What it describes |
| --- | --- | --- |
| $U_{\mathrm{exit}}(x)$ | Position-space field | Amplitude and phase at a position |
| $\widetilde U_{\mathrm{exit}}(k_x)$ | Continuous spectrum | Component weights indexed by transverse wavevector |
| $\{C_m\}$ | Discrete order-space data, when applicable | A list of component amplitudes and phases |

Fourier analysis changes the **representation**. A physical model is still needed to obtain the field in the first place.

### A periodic mask

Now also require $t(x+p)=t(x)$. The physical period $p$ introduces $G=2\pi/p$. Its Fourier series is

$$t(x)=\sum_m
\underbrace{t_m}_{\text{mask coefficient}}
\underbrace{e^{imGx}}_{\text{mask harmonic}}.$$

$t(x)$ is the transmission at a position. Each complex $t_m$ describes one component of the whole repeating pattern. Periodicity can be entirely in amplitude: even when $\phi_t(x)=0$, a repeating amplitude pattern introduces $G$.

Insert this series into the thin-mask product $U_{\mathrm{exit}}=tU_i$ with $U_i=A_i e^{ik_{x,i}x}$:

$$U_{\mathrm{exit}}(x)
=\underbrace{\left[\sum_m t_m e^{imGx}\right]}_{\text{periodic mask}}
 \underbrace{\left[A_i e^{ik_{x,i}x}\right]}_{\text{incident plane wave}}.$$

Multiply the incident field into every term:

$$\boxed{U_{\mathrm{exit}}(x)=\sum_m
\underbrace{A_i t_m}_{C_m}
e^{i\underbrace{(k_{x,i}+mG)}_{k_{x,m}}x}.}$$

| Expression | Category | Meaning |
| --- | --- | --- |
| $t_m$ | Mask coefficient | Dimensionless complex weight of a transmission harmonic |
| $C_m=A_i t_m$ | Exit-field coefficient | Field amplitude and phase of order $m$ |
| $k_{x,i}$ | Illumination property | Incident transverse phase rate |
| $mG$ | Grating contribution | Integer harmonic shift set by the mask period |
| $k_{x,m}=k_{x,i}+mG$ | Order location | Total transverse wavevector of that component |

Thus $U_{\mathrm{exit}}(x)=\sum_m C_m e^{ik_{x,m}x}$ is a **position-space reconstruction from order-space data**. It still gives the field as a function of $x$. A Fourier integral extracts components; this sum adds known components back together.

The relation $C_m=A_i t_m$ belongs to this thin-mask example. FMM generally supplies different $C_m$ from its scattering solution.

For a **centred binary mask**, the opening fraction is

$$D=\frac{W_{\mathrm{open}}}{W_{\mathrm{open}}+W_{\mathrm{close}}},\qquad
t_m=D\,\operatorname{sinc}(mD),\qquad
\operatorname{sinc}(u)=\frac{\sin(\pi u)}{\pi u},\quad\operatorname{sinc}(0)=1.$$

Therefore the same baseline field can be written as

$$\boxed{U_{\mathrm{exit}}(x)=t(x)e^{ik_{x,i}x}
=\sum_{m\in\mathbb Z}\underbrace{D\,\operatorname{sinc}(mD)}_{C_m\text{ for }A_i=1}
e^{i\underbrace{(k_{x,i}+mG)}_{k_{x,m}}x}.}$$

For the example choice $D=0.4$: $C_0=0.4$, $C_{\pm1}\approx0.30273$, $C_{\pm2}\approx0.09355$, and $C_{\pm3}\approx-0.06237$.

- Changing $d$ at fixed $D$ changes $G$ and the order spacing; these thin-mask coefficients stay the same.
- Changing $D$ at fixed $d$ changes the widths and coefficients; the order locations stay the same.
- Changing only the illumination angle shifts the physical order locations through $k_{x,i}$; these coefficients stay the same **in this prescribed thin-mask model**.

The binary mask has infinitely many nonzero harmonics. The series equals the mask product away from its sharp edges; at an ideal edge it converges to the midpoint value. A finite order sum only approximates the sharp-edged field.


### An exact three-order check

The binary baseline needs infinitely many harmonics. For this final algebra check only, replace it with a different, smooth transmission that has exactly three nonzero coefficients:

$$t(x)=0.6+0.4\cos(Gx)
=0.2e^{-iGx}+0.6+0.2e^{iGx}.$$

With $A_i=1$, $C_{-1}=0.2$, $C_0=0.6$, and $C_{+1}=0.2$. The local product and the order sum are

$$U_{\mathrm{exit}}(x)=[0.6+0.4\cos(Gx)]e^{ik_{x,i}x},$$

$$U_{\mathrm{exit}}(x)
=0.2e^{i(k_{x,i}-G)x}+0.6e^{ik_{x,i}x}+0.2e^{i(k_{x,i}+G)x}.$$

Both calculate the **same complex number at every position**. Direct substitution checks that equality, the phase step between cells, and periodicity after removing the incident phase. This example defines the field analytically; no FFT or propagation has been used.
