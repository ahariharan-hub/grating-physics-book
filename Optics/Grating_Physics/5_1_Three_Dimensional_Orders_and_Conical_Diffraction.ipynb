# 5.1 Three-Dimensional Orders and Conical Diffraction

When an incident wave travels partly along a grating's grooves, the outgoing orders acquire a three-dimensional geometry. Their directions can no longer be described by a single drawing in the plane across the grooves. The additional component also changes the condition that separates propagating, grazing and evanescent orders.

[Chapter 3](3_0_Why_Diffraction_Orders_Appear.ipynb) established the order integer, and [Chapter 4](4_0_Evanescent_Diffraction_Orders.ipynb) followed grazing into evanescence. We now extend those arguments to incidence with a component along the grooves. The grating's two translation symmetries determine the surface components of each outgoing wavevector; the wave equation then supplies its normal component. Together, these relations explain the cone of outgoing directions.

The incident and output media have the same index in these constructions; refraction between unequal exterior media is not simulated. The lesson determines direction geometry, while scattering amplitudes and efficiencies require further information. Its two interactive views share controls, and the small calculations have separate editable parameters. Run the cells in order within this lesson; no earlier notebook needs to be running.

## 1. Let the incident wave travel along the grooves

[Chapter 4](4_0_Evanescent_Diffraction_Orders.ipynb) completed each order using the two-dimensional relation
$$k_{x,m}^2+k_{z,m}^2=k^2.$$
Allowing phase variation along a groove introduces $k_{y,m}$, so the same wave equation gives
$$k_{x,m}^2+k_{y,m}^2+k_{z,m}^2=k^2.$$
We will determine the two surface components first, then use this relation to complete the normal component.

We retain the earlier geometry. The grating lies at $z=0$, repeats along $x$ with period $d$, and has uniform grooves along $y$. Incidence is from $z>0$; transmission is into $z<0$. Both exterior regions have the same homogeneous, isotropic, lossless medium. $\lambda$ is the wavelength **in that medium**, $k=2\pi/\lambda$, and lengths are in micrometres.

Use $\mathbf k_i=(k_{x,i},k_{y,i},k_{z,i})$ for the incident wavevector and $\mathbf k_m=(k_{x,m},k_{y,m},k_{z,m})$ for diffraction order $m$. Plain $k$ is the positive scalar wavenumber of the medium. For the real incident vector, $|\mathbf k_i|=k$.

The incident complex scalar field and its real, time-dependent field are
$$U_i(x,y,z)=A_i e^{i(k_{x,i}x+k_{y,i}y+k_{z,i}z)},\qquad
u_i=\operatorname{Re}\{U_i e^{-i\omega t}\}.$$

Previously $k_{y,i}=0$: the phase did not vary along a groove. With $k_{y,i}\ne0$, the surface field is
$$U_i(x,y,0)=A_i e^{i(k_{x,i}x+k_{y,i}y)}.$$

The controls use the component fractions
$$a_i=\frac{k_{x,i}}k,\qquad b_i=\frac{k_{y,i}}k,\qquad
\frac{k_{z,i}}k=-\sqrt{1-a_i^2-b_i^2}.$$
We require $a_i^2+b_i^2<1$ for incidence onto the surface. These fractions will later become the direction-cosine coordinates $\alpha_i,\beta_i$.

The next view extends the earlier $xz$ wave picture into three dimensions. A wavefront is a plane of constant phase, perpendicular to $\mathbf k_i$. At $t=0$, the orange crest planes satisfy
$$\mathbf k_i\cdot\mathbf r=2\pi n,\qquad n\in\mathbb Z.$$
Here $n$ labels successive **incident wave crests**, not diffraction orders. Adjacent planes are separated by $\lambda$ measured along their normal. Orange lines mark their intersections with $z=0$; the same lines appear in the surface map. At normal incidence a crest coincides with the whole grating plane, so there is no distinct intersection line.

Begin with **Normal incidence**, then **Reset: in-plane**, then **Conical: β = 0.4**. The last two keep $a_i=0.3$ fixed, isolating the new along-groove variation. All three presets use $\lambda=2$ and $d=4$.

At **Normal incidence**, the wavefronts are parallel to the grating. Every point on its surface receives the same incident phase.
With **Reset: in-plane**, the planes tilt across the openings. Their surface traces run along $y$, so phase stays constant along each opening.
With **Conical: β = 0.4**, the planes also tilt along the openings. The surface traces become oblique; following one opening along $y$ now crosses successive phases.


The surface colour is $\cos(k_{x,i}x+k_{y,i}y)$ at $t=0$, on a fixed scale $[-1,1]$: signed incident field, not intensity. As in the earlier notebooks, **red is $+1$, white is $0$, and blue is $-1$**. The colour bar is a field scale; unwrapped phase in radians is available on hover. The orange crest planes follow [Chapter 1](1_0_Plane_Waves_and_Wavevectors.ipynb)'s warm-colour convention for $u_i/A=+1$.

The right panel shows that incident field over the entire grating plane; the left shows it in the openings of a 50%-open schematic screen ($W_{\mathrm{open}}=d/2$), extending the earlier slit picture. No scattering amplitudes are being calculated. All crest planes intersecting the displayed volume are included; the edges of the drawn planes and grating are display limits, not beam or grating edges.

Changing $d$ alone separates the geometry of the openings from that of the illumination. The openings move, but the incident wavefronts do not. Changing $\lambda$ alone at fixed direction changes the wavefront spacing. Rotate the view until you can see that $\mathbf k_i$ remains perpendicular to the wavefronts.

The controls keep incidence physical. If increasing $|k_{x,i}/k|$ requires a smaller $|k_{y,i}/k|$, the latter is adjusted with a notice. Presets restore $\lambda,d$, set the stated incidence, and select $m=+1$ for the later order views. The grazing and beyond presets appear with those outgoing orders.

### Small calculation 1: complete the incident vector

Construct the incident vector and check its length. Polar angle $\theta_i$ is measured from $-z$; azimuth $\varphi_i$ is measured in the $xy$-plane from $+x$ toward $+y$:
$$\frac{\mathbf k_i}{k}=
(\sin\theta_i\cos\varphi_i,\ \sin\theta_i\sin\varphi_i,\ -\cos\theta_i).$$

Unlike the signed $xz$-plane angle in the preceding notebooks, this **polar angle is nonnegative**; the azimuth identifies the transverse direction. For example, the earlier planar angle $-30^\circ$ is now $\theta_i=30^\circ$, $\varphi_i=180^\circ$. Both descriptions give $\mathbf k_i/k=(-1/2,0,-\sqrt3/2)$. The Cartesian axes and outgoing $-z$ convention are unchanged.

**Predict and rerun:** change only $b_i$ from $0.4$ to $0$. Which Cartesian component remains fixed? Does the polar angle remain fixed?

```python
wavelength = 2.0  # micrometres
a_i, b_i = 0.3, 0.4
k = 2*np.pi/wavelength

if a_i**2+b_i**2 >= 1:
    print("Choose a_i**2 + b_i**2 < 1 for incidence onto the surface.")
else:
    s_i = np.array([a_i, b_i, -np.sqrt(1-a_i**2-b_i**2)])
    theta_i = np.rad2deg(np.arccos(-s_i[2]))
    phi_i = (np.rad2deg(np.arctan2(b_i, a_i))
             if np.hypot(a_i, b_i) > 0 else None)
    print("k_i/k =", np.round(s_i, 6))
    print("k_i (rad/um) =", np.round(k*s_i, 6))
    print("|k_i|/k =", np.linalg.norm(s_i))
    print("theta_i from -z (degrees) =", round(theta_i, 6))
    print("azimuth phi_i (degrees) =", phi_i)
```

:::{dropdown} Answer

For $(a_i,b_i)=(0.3,0.4)$, $\mathbf k_i/k=(0.3,0.4,-0.866025)$, $\theta_i=30^\circ$, and $\varphi_i\approx53.130^\circ$.

With $b_i=0$, $k_{x,i}/k$ remains ${0.3}$, but $\theta_i\approx17.458^\circ$. At fixed polar angle, changing azimuth generally changes **both** tangential components. At normal incidence azimuth is undefined.

:::

## 2. What does the grating change, and what does it preserve?

Across the periodic direction, [Chapter 3](3_0_Why_Diffraction_Orders_Appear.ipynb)'s argument still applies:
$$e^{i(k_{x,m}-k_{x,i})d}=1
\quad\Longrightarrow\quad
\boxed{k_{x,m}=k_{x,i}+mG,\qquad G=\frac{2\pi}{d},\quad m\in\mathbb Z.}$$

Along $y$, the grating is unchanged by **any** displacement $\Delta y$, not just one period. Under plane-wave illumination its field has the incident translation factor $e^{ik_{y,i}\Delta y}$. Each outgoing component must satisfy
$$e^{i(k_{y,m}-k_{y,i})\Delta y}=1\quad\text{for every }\Delta y,
\quad\Longrightarrow\quad
\boxed{k_{y,m}=k_{y,i}.}$$

If $k_{y,m}-k_{y,i}$ were nonzero, a sufficiently small displacement would give a nonzero phase change smaller than one full cycle, so the factor could not remain 1 for every displacement. Thus the difference must be zero.

The key distinction is **discrete periodicity across the grooves** and **continuous uniformity along the grooves**. This is still a linear grating with **one order integer**; allowing three-dimensional propagation has not introduced a second grating period.

Maystre's pseudo-periodicity and Rayleigh-expansion derivation provides the foundation for the discrete components [\[1\]](#ref-5-1-maystre).

### Small calculation 2: check both translation factors

Compare incident and outgoing surface phases across one period along $x$, then across an arbitrary displacement along $y$.

**Predict and rerun:** change the integer $m$ and the along-groove displacement. Should the two complex translation factors still agree?

```python
wavelength, d = 2.0, 4.0
a_i, b_i, m = 0.3, 0.4, 1
delta_y = 0.7  # any displacement in micrometres
k = 2*np.pi/wavelength
k_x_i, k_y_i = k*a_i, k*b_i
k_x_m, k_y_m = k_x_i+m*2*np.pi/d, k_y_i

incident_x, outgoing_x = np.exp(1j*k_x_i*d), np.exp(1j*k_x_m*d)
incident_y, outgoing_y = np.exp(1j*k_y_i*delta_y), np.exp(1j*k_y_m*delta_y)
print("Across one period: incident =", incident_x, "; outgoing =", outgoing_x)
print("Along a groove:    incident =", incident_y, "; outgoing =", outgoing_y)
print("Across-groove phase difference / (2*pi) =", (k_x_m-k_x_i)*d/(2*np.pi))
print("Along-groove phase difference =", (k_y_m-k_y_i)*delta_y)
```

:::{dropdown} Answer

The across-groove phase difference is $2\pi m$; the along-groove difference is zero. Both translation factors match. A trial $k_{y,m}=k_{y,i}+2\pi/L$ would match translations by $L$, but generally fail other displacements. Uniform grooves require the condition for **every** displacement.

:::

## 3. Complete the third component

The homogeneous medium requires
$$k_{x,m}^2+k_{y,m}^2+k_{z,m}^2=k^2.$$

Insert the tangential components:
$$D_m=k^2-(k_{x,i}+mG)^2-k_{y,i}^2.$$

As in [Chapter 4](4_0_Evanescent_Diffraction_Orders.ipynb), choose the outgoing or decaying branch for $z<0$:
$$\boxed{k_{z,m}=
\begin{cases}
-\sqrt{D_m},&D_m>0\quad\text{(propagating)},\\
0,&D_m=0\quad\text{(grazing)},\\
-i\sqrt{-D_m},&D_m<0\quad\text{(evanescent)}.
\end{cases}}$$

The new term is $k_{y,i}^2$. A real outgoing direction requires
$$\left(a_i+m\frac{\lambda}{d}\right)^2+b_i^2\le1.$$
Equality is grazing. The condition $|k_{x,m}|\le k$ alone is no longer sufficient.

For an evanescent component, write $\kappa_m=\sqrt{-D_m}$ and $s=-z\ge0$:
$$U_m(x,y,-s)=C_m e^{i(k_{x,m}x+k_{y,i}y)}e^{-\kappa_m s}.$$
The phase varies along the surface; the amplitude decays normally with length $\ell_m=1/\kappa_m$. Use algebraic squares in the dispersion relation when $k_{z,m}$ is complex: $\mathbf k_m\cdot\mathbf k_m=k^2$, without complex conjugation. The ordinary length statement $|\mathbf k_m|=k$ applies only to real propagating or grazing vectors; $k$ remains the medium wavenumber for evanescent orders too.

TU Delft derives the 3D square-root relation [\[2, Eq. (375)\]](#ref-5-1-delft-diffraction). Its outgoing side is $+z$; our branch signs are reversed.

### Small calculation 3: move an order through the 3D boundary

Keep $a_i=0.3$ and $\lambda/d=0.5$. For $m=+1$, $k_{x,m}/k=0.8$, so grazing occurs at
$$b_i^2=1-0.8^2=0.36.$$

**Predict:** what happens to $m=+1$ at $b_i=0.4,0.6,0.8$? At the last value, $k_{x,m}$ is still smaller than $k$. Is that enough for propagation?

```python
wavelength, d = 2.0, 4.0
a_i = 0.3
orders = np.arange(-3, 4)

for b_i in [0.4, 0.6, 0.8]:
    a_m = a_i+orders*wavelength/d
    r = 1-a_m**2-b_i**2  # D_m / k**2
    r[np.abs(r) <= 1e-12] = 0.0
    gamma = -np.sqrt(r+0j)
    states = np.where(r == 0, "grazing", np.where(r > 0, "propagating", "evanescent"))
    print(f"\nb_i = {b_i:.1f}")
    print(" m   k_mx/k  k_my/k       k_mz/k            state")
    for m, a, g, state in zip(orders, a_m, gamma, states):
        print(f"{m:+2d}  {a:+6.2f}  {b_i:+6.2f}  {g.real:+.4f}{g.imag:+.4f}j   {state}")
```

:::{dropdown} Answer

At $b_i=0.4$, $m=-2,-1,0,+1$ propagate. At $b_i=0.6$, $m=+1$ is grazing; $m=-2,-1,0$ propagate. At $b_i=0.8$, only $m=-1,0$ propagate.

For $m=+1$ at $b_i=0.8$, $k_{x,m}^2+k_{y,m}^2=1.28k^2$, although each tangential component is smaller than $k$. Here $k_{z,m}/k=-i\sqrt{0.28}$ and, for $\lambda=2\,\mu\mathrm m$, $\ell\approx0.602\,\mu\mathrm m$.

:::

## 5. Why is the diffraction called conical?

For every real outgoing order,
$$\frac{\mathbf k_m}{k}\cdot\hat{\mathbf y}
=\frac{k_{y,m}}k=b_i.$$
Its angle $\chi_m$ to the positive groove axis satisfies $\cos\chi_m=b_i$. All orders make the **same angle to the groove axis**; their rays lie on a cone with its vertex at the grating origin.

Press **Conical: β = 0.4**, then enable **Show cone / planar limit** in the spatial view. The translucent surface connects candidate directions; the discrete marked rays are the propagating orders. Only the $z\le0$ portion is drawn.

The cone describes **rays from a common origin**. Their **endpoints at a common radius** form a circle in a plane of constant $y$. These are two views of the same direction construction. In wavevector space, each real order obeys
  $$k_{y,m}=k_{y,i},\qquad k_{x,m}^2+k_{z,m}^2=k^2-k_{y,i}^2.$$

The cone axis is along the grooves, not the surface normal. At $b_i=0$, the ray locus becomes the $xz$-plane. For negative $b_i$, it opens toward $-y$; the angle to $+y$ is obtuse, while its semi-angle about $-y$ is $\arccos|b_i|$.

Any nonzero along-groove component gives conical geometry under our assumptions; large angles merely make it easier to see. See [\[3\]](#ref-5-1-harvey-vernold), [\[4, Fig. 6\]](#ref-5-1-harvey-pfisterer), and the handbook's conical equation [\[5\]](#ref-5-1-palmer).

### Small calculation 4: verify the common cone angle

Compute the real outgoing unit vectors and measure their angles to $+y$. Their endpoint circle on a unit sphere has radius
$$\rho=\sqrt{1-b_i^2}.$$

**Predict and rerun:** replace $b_i=0.4$ with $-0.4$, then $0$. Which results change?

```python
wavelength, d = 2.0, 4.0
a_i, b_i = 0.3, 0.4
orders = np.arange(-3, 4)
a_m = a_i+orders*wavelength/d
r = 1-a_m**2-b_i**2
real = r >= -1e-12
directions = np.column_stack([a_m[real], np.full(np.count_nonzero(real), b_i),
                              -np.sqrt(np.maximum(r[real], 0))])
chi = np.rad2deg(np.arccos(np.clip(directions[:, 1], -1, 1)))
circle_radii = np.hypot(directions[:, 0], directions[:, 2])
print("  m    angle to +y (degrees)   circle radius")
print(np.column_stack([orders[real], chi, circle_radii]))
print("Expected radius =", np.sqrt(1-b_i**2))
```

:::{dropdown} Answer

At $b_i=0.4$, all four real orders make an angle $\chi\approx66.422^\circ$ to $+y$, with circle radius $\rho\approx0.916515$. Their individual directions differ despite this common angle.

At $b_i=-0.4$, the real order labels and circle radius are unchanged; the angle to $+y$ is $113.578^\circ$. At $b_i=0$, $\chi=90^\circ$ and the circle is a great-circle section: the rays lie in the $xz$-plane.

:::
