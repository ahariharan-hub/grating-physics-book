**How do we connect a physical optical structure to a field and then to a prediction?**

**Medium and wavelength convention.** Most geometry lessons use a homogeneous, isotropic, lossless surrounding medium and take $\lambda$ as its in-medium wavelength; the material-scattering lessons instead take vacuum wavelength $\lambda_0$ and explicit indices. These inputs must not be interchanged. The course convention below distinguishes them.

## Motivation and prerequisites

A field equation tells us which waves are possible. An optical experiment also requires illumination, a structure and boundary conditions to select the field that actually occurs. This opening chapter separates those tasks before later chapters examine their geometry and mathematics in detail.

Begin with basic algebra, trigonometry and the idea of a wave. Complex numbers and derivatives appear in the lessons with physical explanations; detailed derivations can wait until a second reading.

## The central distinction

- **Known reference-plane field:** propagate it through a homogeneous region.
- **Known structure and illumination:** solve the scattering problem to find the field first.
- A prescribed aperture mask approximates the second task; it does not solve material scattering.

The scalar model represents one optical field component. Its usefulness depends on geometry, polarization and material assumptions; the examples state those assumptions locally. A numerical result must also be checked for convergence before being interpreted physically.

## Refractive index: which wavelength are we using?

At fixed optical frequency $\omega$, a real surrounding-medium index $n=n(\omega)$ fixes the phase wavelength and wavenumber:

$$\boxed{\lambda_0=\frac{2\pi c}{\omega},\quad k_0=\frac{\omega}{c},\qquad
\lambda=\frac{\lambda_0}{n},\quad k=nk_0=\frac{2\pi}{\lambda}.}$$

Frequency stays the same across a stationary interface; phase speed and wavelength can change. In a dispersive material, evaluate $n$ at that frequency. The simple time-domain wave-speed examples additionally assume a nondispersive medium.

| Model input | Meaning | How index enters |
| --- | --- | --- |
| Wave, order-geometry and Gaussian explorers: $\lambda$ | Wavelength in the homogeneous surrounding medium | Index is already incorporated in the entered wavelength; there is no independent index slider. |
| Material-scattering configuration: `wavelength` | Vacuum wavelength $\lambda_0$ | Incident, output and layer indices enter separately. |
| Lithography example: 193 nm | Vacuum wavelength $\lambda_0$ | Refer medium wavelength, NA and sigma coordinates to the specified optical side. |

**Holding $\lambda$ fixed is not holding the laser frequency fixed when changing $n$.** To represent a fixed-frequency source in another medium, keep $\lambda_0$ fixed and recompute $\lambda=\lambda_0/n$. For example, $\lambda_0=0.600$ micrometre gives $\lambda=0.400$ micrometre at $n=1.5$.

The surrounding index specifies propagation outside the structure. The grating's material index may instead vary with position and be complex. For $\widetilde n=n'+i\kappa$ with $\kappa>0$ under $e^{-i\omega t}$, an outgoing plane wave along positive path distance has phase factor $e^{ik_0n'\ell}$ and attenuation $e^{-k_0\kappa\ell}$. The real-medium wavelength rule is not a complete description of that absorbing field. Our scalar material solver uses the stated permittivity and boundary conditions; it does not replace them by a single effective exterior index.

**One exit field, consistent notation**

[9.1: Defining the Complex Exit Field](9_1_Defining_the_Complex_Exit_Field.ipynb) explains these expressions step by step. [9.2](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb) then examines sampling.

The exit field is the complex scalar field restricted to the outgoing side of the structure:

$$\boxed{U_{\mathrm{exit}}(x,y)=U(x,y,z_e).}$$

$z_e$ names the exit plane. It is $0^+$ for the positive-$z$ slit example, $0^-$ for the negative-$z$ ideal mask, and $-h$ on the transmitted side of our finite layer. These choices describe different geometries, with the same field notation.

| Symbol | Use throughout the course |
| --- | --- |
| $U_i$ | Incident complex field, evaluated at the stated plane |
| $t(x,y)$; $t_m,t_{mn}$ | Prescribed thin-mask transmission; its dimensionless Fourier coefficients |
| $U_{\mathrm{exit}}$ | Complex field at $z=z_e$, obtained from the chosen physical model |
| $C_m=C_m(z_e)$; $C_{mn}=C_{mn}(z_e)$ | Exit-field order coefficients; show the plane explicitly when it changes |
| $U_{\mathrm{ref}}(x,y)=U(x,y,z_{\mathrm{ref}})$ | Field at a general reference plane; it equals $U_{\mathrm{exit}}$ when $z_{\mathrm{ref}}=z_e$ |
| $F_m(z)$ | Solver's depth-dependent harmonic function, with $C_m(z_{\mathrm{ref}})=F_m(z_{\mathrm{ref}})$ in the outgoing region |
| $\widetilde U_{\mathrm{exit}}$; $\widetilde U_{\mathrm{ref}}$ | Continuous Fourier spectra of the named plane fields |

**Obtain the field, then represent it.** A thin mask prescribes $U_{\mathrm{exit}}=tU_i$. Scalar FMM determines $U_{\mathrm{exit}}$ by solving the scattering problem. Either field can then be represented by its amplitude and phase, spatial samples, or Fourier components. A finite numerical representation must resolve the required content.

For one-dimensional periodic structure under one incident plane wave,

$$U_{\mathrm{exit}}(x)=\sum_m C_m e^{i(k_{x,i}+mG)x},\qquad G=2\pi/p.$$

The normal-incidence case has $k_{x,i}=0$. Under the thin-mask prescription, $C_m=A_i t_m$ for $U_i(x)=A_i e^{ik_{x,i}x}$ with phase zero at the origin of the same exit plane. A different incident reference phase is included in $U_i$ and therefore in $C_m$. A Gaussian envelope generally requires a continuous spectrum. Earlier geometry lessons call the period $d$; Chapter 9 calls it $p$.

Always state the plane and incident phase reference. For $U_i(z)=e^{-ik_0z}$ with phase zero at $z=0$, the ideal mask placed at $z_e=-h$ has $U_{\mathrm{exit}}=e^{ik_0h}t$. Removing a common carrier for a comparison plot is stated next to that plot.

Backend names remain compatible: `solution.transmission` stores the exit coefficients $C_m(z_e)$. A derivation mapping internal arrays to this vector identifies that mapping explicitly. Generic Fourier examples may use $c_m$ for an arbitrary signal; use $t_m$ for a mask and $C_m$ for an optical exit field.
