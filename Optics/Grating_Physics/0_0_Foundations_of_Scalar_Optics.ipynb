# 0.0 Foundations of scalar optics

Consider light incident on a diffraction grating. The grating's period helps determine the directions in which diffraction orders can propagate, but it does not by itself determine how much light enters each order. To predict the outgoing field, we must also describe the illumination and the grating's response to it. We can then propagate that field to a detector and calculate the intensity there.

This sequence connects three parts of an optical calculation: the physical structure, the field it produces and the field's subsequent propagation. Sometimes the outgoing field is already specified, so propagation is the starting point. In other problems, finding that field is the main task. The distinction runs through this book and helps us understand what a calculation can predict from the information supplied.

## From a structure to an optical field

When a complex field is known at a reference plane, propagation through a homogeneous region starts from specified field data. The propagation calculation describes how that field evolves away from the plane. Its result therefore depends on the field supplied at the starting plane: two calculations that use the same propagation method need not predict the same output if their starting fields differ.

When the structure and illumination are specified instead, the field must first be found by solving a scattering problem. The structure's geometry and material properties enter that problem through the governing equation and its boundary conditions. A prescribed aperture mask supplies an approximation to this stage by assigning a transmission function; it does not, by itself, solve the material-scattering problem. Once a field has been obtained from either description, its propagation can be studied.

Scalar optics represents one component of the optical field. The usefulness of that representation depends on the geometry, polarisation and material assumptions, which will be stated in the individual lessons. Numerical convergence must also be checked before a computed result is given a physical interpretation. Choosing a physical model and checking its numerical representation are therefore both part of constructing an optical prediction.

## Wavelength and refractive index

The distinction between a propagation model and a material model also determines how a wavelength input must be read. Most geometry lessons describe propagation in a homogeneous, isotropic, lossless surrounding medium and use its in-medium wavelength $\lambda$. The material-scattering lessons specify the vacuum wavelength $\lambda_0$ and supply the material indices separately. At fixed optical frequency $\omega$, a real surrounding-medium index $n=n(\omega)$ relates these quantities through

$$\boxed{\lambda_0=\frac{2\pi c}{\omega},\quad k_0=\frac{\omega}{c},\qquad
\lambda=\frac{\lambda_0}{n},\quad k=nk_0=\frac{2\pi}{\lambda}.}$$

At a stationary interface the frequency is unchanged, although the phase speed and wavelength can change. In a dispersive material, the index in these relations is evaluated at that frequency. The simpler time-domain wave-speed examples additionally assume a nondispersive medium. The conventions used by the different examples are collected below so that an input wavelength can always be interpreted in the appropriate model.

| Model input | Meaning | How index enters |
| --- | --- | --- |
| Wave, order-geometry and Gaussian explorers: $\lambda$ | Wavelength in the homogeneous surrounding medium | Index is already incorporated in the entered wavelength; there is no independent index slider. |
| Material-scattering configuration: `wavelength` | Vacuum wavelength $\lambda_0$ | Incident, output and layer indices enter separately. |

For a fixed-frequency source, $\lambda_0$ remains fixed while the wavelength in a new medium is recomputed as $\lambda=\lambda_0/n$. Thus $\lambda_0=0.600$ micrometre gives $\lambda=0.400$ micrometre at $n=1.5$. Holding the medium wavelength $\lambda$ fixed while changing $n$ would describe a change of source frequency as well.

These wavelength relations describe the real surrounding index that specifies propagation outside the structure. The grating's material index may instead vary with position and be complex. For $\widetilde n=n'+i\kappa$ with $\kappa>0$ under $e^{-i\omega t}$, an outgoing plane wave along positive path distance has phase factor $e^{ik_0n'\ell}$ and attenuation $e^{-k_0\kappa\ell}$. The real-medium wavelength rule is not a complete description of that absorbing field. Our scalar material solver uses the stated permittivity and boundary conditions; it does not replace them by a single effective exterior index.

**One exit field, consistent notation**

[9.1: Defining the Complex Exit Field](9_1_Defining_the_Complex_Exit_Field.ipynb) explains these expressions step by step. [9.2](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb) then examines sampling.

The exit field is the complex scalar field restricted to the outgoing side of the structure:

$$\boxed{U_{\mathrm{exit}}(x,y)=U(x,y,z_e).}$$

$z_e$ names the exit plane. It is $0^+$ for the positive-$z$ slit example, $0^-$ for the negative-$z$ ideal mask, and $-h$ on the transmitted side of our finite layer. These choices describe different geometries, with the same field notation.

| Symbol | Use throughout the course |
| --- | --- |
| $U_i$ | Incident complex field, evaluated at the stated plane |
| $t(x,y)$; $t_m,t_{mp}$ | Prescribed thin-mask transmission; its dimensionless Fourier coefficients |
| $U_{\mathrm{exit}}$ | Complex field at $z=z_e$, obtained from the chosen physical model |
| $C_m=C_m(z_e)$; $C_{mp}=C_{mp}(z_e)$ | Exit-field order coefficients; show the plane explicitly when it changes |
| $U_{\mathrm{ref}}(x,y)=U(x,y,z_{\mathrm{ref}})$ | Field at a general reference plane; it equals $U_{\mathrm{exit}}$ when $z_{\mathrm{ref}}=z_e$ |
| $F_m(z)$ | Solver's depth-dependent harmonic function, with $C_m(z_{\mathrm{ref}})=F_m(z_{\mathrm{ref}})$ in the outgoing region |
| $\widetilde U_{\mathrm{exit}}$; $\widetilde U_{\mathrm{ref}}$ | Continuous Fourier spectra of the named plane fields |

**Obtain the field, then represent it.** A thin mask prescribes $U_{\mathrm{exit}}=tU_i$. Scalar FMM determines $U_{\mathrm{exit}}$ by solving the scattering problem. Either field can then be represented by its amplitude and phase, spatial samples, or Fourier components. A finite numerical representation must resolve the required content.

For one-dimensional periodic structure under one incident plane wave,

$$U_{\mathrm{exit}}(x)=\sum_m C_m e^{i(k_{x,i}+mG)x},\qquad G=2\pi/p.$$

The normal-incidence case has $k_{x,i}=0$. Under the thin-mask prescription, $C_m=A_i t_m$ for $U_i(x)=A_i e^{ik_{x,i}x}$ with phase zero at the origin of the same exit plane. A different incident reference phase is included in $U_i$ and therefore in $C_m$. A Gaussian envelope generally requires a continuous spectrum. Earlier geometry lessons call the period $d$; Chapter 9 calls it $p$.

Always state the plane and incident phase reference. For $U_i(z)=e^{-ik_0z}$ with phase zero at $z=0$, the ideal mask placed at $z_e=-h$ has $U_{\mathrm{exit}}=e^{ik_0h}t$. Removing a common carrier for a comparison plot is stated next to that plot.

Backend names remain compatible: `solution.transmission` stores the exit coefficients $C_m(z_e)$. A derivation mapping internal arrays to this vector identifies that mapping explicitly. Generic Fourier examples may use $c_m$ for an arbitrary signal; use $t_m$ for a mask and $C_m$ for an optical exit field.

## The introductory sequence

The seven lessons in this introductory series form a conceptual overview of scalar diffraction. They begin with the quantity used to represent monochromatic light and the field shapes that can satisfy its equation. Boundary conditions then connect those possible fields to a physical problem. With that distinction established, the discussion turns to propagation of a known field, scattering by a physical grating, and the representation of a scattered field by diffraction orders. The final comparison examines when a prescribed binary mask is a useful approximation to a physical absorbing grating.

Each lesson develops one part of this connection through an experiment or visualisation. The sequence is summarised in the following table.

| Lesson | Central question | Experiment or visualization |

## Reading and computation

Basic algebra, trigonometry and the idea of a wave are sufficient starting points. Complex numbers and derivatives are introduced with physical explanations as they become necessary. On a first reading, the visualisations establish the physical phenomenon and the equations provide a way to describe and predict it. The optional derivations can be revisited once that connection is familiar.

The introductory series provides orientation before the more systematic treatment that begins with [Chapter 1: Plane waves and wavevectors](1_0_Plane_Waves_and_Wavevectors.ipynb) and [1.1: Oscillation and wavelength](1_1_Oscillation_and_Wavelength.ipynb). Deriving or implementing the Fourier modal method is not a prerequisite for continuing. Readers who want that mathematical development can consult [Appendix_1: Mathematics of the scalar grating solver](Appendix_1_Fourier_Modal_Method_for_Scalar_Grating_Scattering.ipynb), which follows Fourier coupling, layer modes and interface matching through executable checks.

The sequence is intended to be read continuously, although each notebook constructs its own examples and can run in a fresh kernel. Within a notebook, the code cells should be run in order. The opening of each lesson identifies the prerequisites and distinguishes the essential discussion from optional background.
