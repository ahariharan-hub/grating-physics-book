## 1. Kernels and multipliers: the preliminary picture

An **integral kernel is a continuous counterpart of a matrix**. A matrix entry weights one input's contribution to one output; $K(P,Q)$ weights the connection from input point $Q$ to output point $P$. The idea also appears in heat diffusion and imaging: the physical problem determines the particular kernel.

For propagation, take a boundary patch at $Q=(\xi,\eta,0)$ and an observation point $P=(x,y,s)$. Its contribution is

$$\boxed{dU(P)=K(P,Q)\,U(Q)\,dA_Q.}$$

The kernel's **magnitude** sets the strength of this connection; its **phase** sets the direction of the contribution's complex arrow. Multiplying by the boundary value and patch area gives a field contribution.

Define the transverse separation and full distance:

$$\rho^2=(x-\xi)^2+(y-\eta)^2,\qquad R=\sqrt{s^2+\rho^2}.$$

$Q$ is a boundary sample, not a newly inserted physical point source. No particular boundary field is prescribed here.

The same geometry is used throughout: Q is on the reference plane and P on the observation plane. The selected points lie in the $y=\eta=0$ section. The geometry controls in Sections 1-5 are linked; changing one updates the others.


**A multiplier updates one component independently.** For homogeneous plane-to-plane propagation, transverse translation invariance makes the spatial kernel depend on coordinate differences. Writing $U_{\mathrm{ref}}(\xi,\eta)=U(\xi,\eta,0)$, adding the patch contributions gives a convolution:

$$U(x,y,s)=\iint K_{\rm RS1}(x-\xi,y-\eta;s)\,U_{\mathrm{ref}}(\xi,\eta)\,d\xi\,d\eta.$$

Fourier transformation turns that integral into multiplication. For full angular-spectrum propagation:

$$\boxed{\widetilde U_s(k_x,k_y)=H_{\rm ASM}(k_x,k_y;s)\,\widetilde U_{\mathrm{ref}}(k_x,k_y)},
\qquad H_{\rm ASM}=\mathcal F_\perp\{K_{\rm RS1}\}.$$

Here $\mathcal F_\perp$ is the **two-dimensional Fourier transform over the transverse separations** $(\Delta x,\Delta y)=(x-\xi,y-\eta)$, giving $(k_x,k_y)$. **Distance $s$ is held fixed**; there is no Fourier transform along the propagation coordinate.

| Representation | What it weights | Units in this problem |
| --- | --- | --- |
| Spatial kernel $K_{\rm RS1}$ | Boundary point $Q$ to observation point $P$ | Inverse area |
| Fourier multiplier $H_{\rm ASM}$ | One plane-wave component $(k_x,k_y)$ | Dimensionless |

Both carry amplitude and phase. In matrix language, the Fourier representation is **diagonal**: each spatial frequency evolves independently. This relies on translation invariance; a system that mixes spatial frequencies requires a kernel connecting different components in Fourier space.

**Grating example.** A patterned thin mask uses a position-space multiplier, $U_{\mathrm{exit}}(x)=t(x)U_i(x)$, which couples spatial frequencies. FMM/RCWA describes a finite-thickness grating through an order-coupling matrix: $C_m(z_e)=\sum_p T_{mp}C_p^{\rm in}$, where $T_{mp}$ connects incident order $p$ to transmitted order $m$. Subsequent homogeneous propagation updates each outgoing component independently.

The general input plane is $z_{\mathrm{ref}}=0$ in this local propagation coordinate. It may coincide with the physical exit plane, in which case $U_{\mathrm{ref}}=U_{\mathrm{exit}}$. A different reference plane has different field values obtained by propagation.

**Notation:** use $U_{\mathrm{exit}}$ at the exit plane, $U_{\mathrm{ref}}$ at a general reference plane, and $C_m$ (or $C_{mp}$) for field coefficients. [Shared convention](0_0_Foundations_of_Scalar_Optics.ipynb#exit-field-notation).

## 2. Full propagation: spatial kernel and ASM multiplier


From 8.2, the full RS I kernel is

$$\boxed{K_{\rm RS1}(P,Q)=-2\partial_s g
=\frac{s\,e^{ikR}(1-ikR)}{2\pi R^3}},\qquad
g=\frac{e^{ikR}}{4\pi R}.$$

Here $g$ is the point-source Green response; its derivative supplies the **planar propagation kernel**. Keep the complete $(1-ikR)$ factor.

The transverse Fourier transform of this same kernel gives full ASM:

$$\boxed{\widetilde K_{\rm RS1}(k_x,k_y;s)=H_{\rm ASM}(k_x,k_y;s)=e^{i\kappa s}},\qquad
\kappa=\sqrt{k^2-k_x^2-k_y^2}.$$

Choose the outgoing branch: real $\kappa\geq0$ for propagating components and positive imaginary $\kappa$ for evanescent decay. **RS I and full ASM are equivalent representations of the same propagation operator.** The Fourier transform uses inverse measure $dk_x\,dk_y/(2\pi)^2$. [Weyl connection](#ref-8-3-handbook) [\[1\]](#ref-8-3-handbook)

**Explore.** Move Q or P, or change the separation. The full kernel retains the true path and directional weighting; blue fronts mark its phase contours.

**Kernel magnitude.**

$$|K_{\rm RS1}|=\frac{s\sqrt{1+(kR)^2}}{2\pi R^3}=\underbrace{\frac{1}{\lambda R}}_{\text{spreading}}\,\underbrace{\cos\theta}_{\text{directional weighting}}\,\underbrace{\sqrt{1+\frac{1}{(kR)^2}}}_{\text{near-field correction}},\qquad\cos\theta=\frac{s}{R}.$$


## 3. Fresnel: keep the quadratic path variation


**First separate the factors.** Using $k=2\pi/\lambda$, rewrite the full kernel exactly as

$$K_{\rm RS1}(P,Q)=\frac{e^{ikR}}{i\lambda s}
\underbrace{\left(\frac{s}{R}\right)^2
\left(1+\frac{i}{kR}\right)}_{\text{directional and near-field factor}}.$$

**Then approximate the path and the prefactor separately.** For $\rho/s\ll1$,

$$R=s\sqrt{1+\frac{\rho^2}{s^2}}
=s+\frac{\rho^2}{2s}-\frac{\rho^4}{8s^3}+\cdots.$$

Keep the quadratic path term in the exponential; use $R\approx s$ and $kR\gg1$ in the remaining factor:

$$\begin{aligned}
e^{ikR}&\approx e^{iks}\,e^{ik\rho^2/(2s)},\\
\left(\frac{s}{R}\right)^2\left(1+\frac{i}{kR}\right)&\approx1.
\end{aligned}$$

Substitution now gives the Fresnel kernel:

$$\boxed{K_{\rm F}(P,Q)=\frac{e^{iks}}{i\lambda s}
\exp\!\left(\frac{ik\rho^2}{2s}\right)}.$$

The quadratic phase remains; the transverse magnitude weighting becomes uniform. Check small angles, $ks\gg1$, and the neglected phase $k\rho^4/(8s^3)$ over the connections of interest.

**In Fourier space, expand the longitudinal wavevector instead:**

$$\begin{aligned}
\kappa&=k\sqrt{1-\frac{k_x^2+k_y^2}{k^2}}
\approx k-\frac{k_x^2+k_y^2}{2k},\\
H_{\rm ASM}=e^{i\kappa s}&\approx e^{i[k-(k_x^2+k_y^2)/(2k)]s}\\
&=\underbrace{e^{iks}e^{-is(k_x^2+k_y^2)/(2k)}}_{H_{\rm F}}.
\end{aligned}$$

This is the same paraxial propagation: $\widetilde K_{\rm F}=H_{\rm F}$. It does not retain evanescent decay. [Approximation reference](#ref-8-3-mit-approximations) [\[2\]](#ref-8-3-mit-approximations)

**Explore.** The geometry is shared with the other panels. The **upper-right plot** compares exact and quadratic extra paths; the **lower-right plot** compares full and Fresnel kernel magnitudes on the same fixed scale.

**Kernel magnitude.**

$$|K_{\rm RS1}|=\underbrace{\frac{1}{\lambda R}}_{\text{spreading}\ \to\ 1/(\lambda s)}\,\underbrace{\cos\theta}_{\text{directional weighting}\ \to\ 1}\,\underbrace{\sqrt{1+\frac{1}{(kR)^2}}}_{\text{near-field correction}\ \to\ 1}\quad\Longrightarrow\quad\boxed{|K_{\rm F}|=\frac{1}{\lambda s}}.$$

These simplify the magnitude using $R\approx s$, small angles and $kR\gg1$; the quadratic path variation is still retained in the phase.


## 4. Fraunhofer: keep the tilt, neglect the curvature

**Fresnel keeps the wavefront's curvature. Fraunhofer keeps its local tilt and neglects the curvature.** A small part of a distant spherical wavefront looks approximately planar. Here we examine the kernel's phase across boundary points Q for a fixed observation point P.


**Expand the square inside the Fresnel phase:**

$$\rho^2=(x-\xi)^2+(y-\eta)^2
=(x^2+y^2)-2(x\xi+y\eta)+(\xi^2+\eta^2).$$

The exponential therefore separates into three factors:

$$\begin{aligned}
K_{\rm F}(P,Q)
&=\frac{e^{iks}}{i\lambda s}
\underbrace{e^{ik(x^2+y^2)/(2s)}}_{\text{observation phase}}\\
&\quad\times
\underbrace{e^{-ik(x\xi+y\eta)/s}}_{\text{tilt: Fourier-transform factor}}\\
&\quad\times
\underbrace{e^{ik(\xi^2+\eta^2)/(2s)}}_{\text{curvature: boundary phase}}.
\end{aligned}$$

Together, these still depend only on the transverse coordinate differences. That is why Fresnel remains a convolution and has the multiplier $H_{\rm F}$.

**Flatten the response locally.** The cross term gives a linear phase ramp across the boundary: its tilt carries the observation direction. The quadratic boundary term supplies curvature. Fraunhofer discards this curvature by using

$$e^{ik(\xi^2+\eta^2)/(2s)}\approx1
\quad\text{when}\quad
\frac{k(\xi^2+\eta^2)}{2s}\ll1$$

across the relevant boundary region, **in addition to** the Fresnel conditions. Omitting just that factor gives

$$\boxed{K_{\rm FF}(P,Q)=\frac{e^{iks}}{i\lambda s}
e^{ik(x^2+y^2)/(2s)}e^{-ik(x\xi+y\eta)/s}}.$$

**Recognize the remaining Fourier transform.** Leave the boundary field $U_{\mathrm{ref}}$ unspecified and insert this kernel into the boundary integral:

$$\begin{aligned}
U_{\rm FF}(x,y,s)
&=\frac{e^{iks}}{i\lambda s}e^{ik(x^2+y^2)/(2s)}\\
&\quad\times\underbrace{\iint U_{\mathrm{ref}}(\xi,\eta)
 e^{-i[(k\,x/s)\xi+(k\,y/s)\eta]}\,d\xi\,d\eta}
 _{\displaystyle\widetilde U_{\mathrm{ref}}(k\,x/s,\,k\,y/s)}.
\end{aligned}$$

Each observation point selects a phase tilt, hence the Fourier weight at $(k\,x/s,k\,y/s)$. The underbraced integral is the **input spectrum**; the left side is the **spatial output field**. The carrier, observation phase and absolute amplitude factor remain.

This explains the missing $H_{\rm FF}$: after dropping the boundary phase, the approximate kernel depends on $P$ and $Q$ separately. Fraunhofer gives a **scaled spectrum-to-position mapping**, rather than another same-frequency multiplier. The underlying homogeneous propagation still preserves spatial frequencies; full ASM can also calculate the far field.

Notice the useful special case: **at $\xi=\eta=0$, Fresnel and Fraunhofer kernels coincide**, even when they both differ from the full kernel. One central sample cannot establish agreement across a plane.

**Explore.** Set P to $x=0$: the green phase line is horizontal. Move P sideways to tilt it; the green origin-to-P connector marks the observation direction. Move Q to sample the curves. The green phase line is tangent to the Fresnel parabola at $\xi=0$; the shaded gap is the phase error from neglecting curvature.

## 5. What does the kernel contribute at P?

Let the three kernels predict the contribution of the **same tiny boundary patch at Q**. Take $U(Q)$ real and positive (unit amplitude and zero phase in the explorer). Writing the kernel in magnitude and phase form gives

$$\begin{aligned}
K(P,Q)&=|K|e^{i\phi},\\
dU_{\rm real}(P,t)&=|K|\,U(Q)\,dA_Q\cos(\phi-\omega t).
\end{aligned}$$

**Magnitude sets the oscillation's height; phase shifts it along the time axis.** Changing the geometry changes both. Here $\phi$ is the phase of the complete kernel, including its prefactor.

The default **Patch response** plots $\operatorname{Re}[\lambda^2 K e^{-i\omega t}]$ over two periods: the contribution per unit patch strength $U(Q)dA_Q/\lambda^2$. All three curves use the same geometry and a fixed vertical scale.

Blue fronts mark equal phase of the full kernel in the $y=\eta=0$ section; $\theta$ measures the angle from the plane normal. The lower plots compare magnitude and unwrapped phase error relative to full RS I, including whole turns.

Each curve represents one patch's contribution. Adding contributions across the boundary gives the total field. The displayed boundary interval is a comparison window, not an aperture.

## 6. In Fourier space: follow one plane-wave component

Real-space kernels ask **how Q contributes to P**. Fourier multipliers ask **what happens to one plane-wave component**. Keep the same $k_x,k_y$ notation and positive downstream distance $s$.

**First resolve the wavevector.** For a propagating component, its magnitude is $k=2\pi/\lambda$. Its component along global $z$ is $k_z$; the transverse projection has magnitude $k_{xy}$, with components $k_x,k_y$:

$$k^2=k_z^2+k_{xy}^2,\qquad k_{xy}^2=k_x^2+k_y^2.$$

Our downstream coordinate is $s=-z$, so $k_z=-\kappa$. The exact longitudinal component and its Fresnel approximation are

$$\kappa=\sqrt{k^2-k_x^2-k_y^2}
\quad\longrightarrow\quad
\kappa_{\rm F}=k-\frac{k_x^2+k_y^2}{2k}.$$

The first is a section of the wavevector sphere; Fresnel replaces it with a paraboloid near the axis. The resulting multipliers are

$$\widetilde U_s=H\widetilde U_{\mathrm{ref}},\qquad
\boxed{H_{\rm ASM}=e^{i\kappa s}},\qquad
\boxed{H_{\rm F}=e^{i\kappa_{\rm F}s}}.$$

**Multiplier magnitude.** With transverse components $(k_x,k_y)$ and longitudinal component $\kappa$, the outgoing branch gives

$$|H_{\rm ASM}|=\begin{cases}
1,&k_x^2+k_y^2\leq k^2,\\
e^{-s\sqrt{k_x^2+k_y^2-k^2}},&k_x^2+k_y^2>k^2,
\end{cases}
\qquad |H_{\rm F}|=1.$$

For propagating components, distance changes phase without weakening an individual plane wave. Above the cutoff, ASM gives evanescent decay; Fresnel's formal unit-magnitude continuation does not describe that physics. Select $k_x/k>1$ to see the distinction. Near the axis the two arrows agree; a small longitudinal-wavevector error can accumulate into a large phase error over a long distance.

**Where does Fraunhofer fit?** Section 4's retained tilt is the Fourier weight

$$e^{-i(k_x^{\rm obs}\xi+k_y^{\rm obs}\eta)},\qquad

In this paraxial mapping, **moving P selects a coordinate of the input spectrum**. The usual Fraunhofer expression reads that spectrum as a spatial far-field pattern, with its prefactor retained. It supplies a different representation of the calculation; it does not supply a third universal same-frequency multiplier. ## 7. The two calculation chains

The input is the complex field $U_{\mathrm{ref}}(\xi,\eta)$ on the reference plane. We can reconstruct its propagated field by adding **boundary-patch contributions**, or work with its **plane-wave spectrum**.

**Real space: weight the patches, then add.** With $Q=(\xi,\eta,0)$ and $P=(x,y,s)$,

$$U_{\mathrm{ref}}(Q)
\xrightarrow{\times\,K_j(P,Q)}
K_j(P,Q)U_{\mathrm{ref}}(Q)
\xrightarrow{\displaystyle\iint dA_Q}
U_j(P).$$

Use $K_{\rm RS1}$ for full propagation (RS I, equivalent to ASM), $K_{\rm F}$ for Fresnel, or $K_{\rm FF}$ for Fraunhofer. The sum is coherent: it adds complex contributions, including their phases.

**Fourier space: ASM and Fresnel update the spectrum, then transform back.**

$$U_{\mathrm{ref}}
\xrightarrow{\mathcal F_\perp}
\widetilde U_{\mathrm{ref}}
\xrightarrow{\times\,H_j}
\widetilde U_j
\xrightarrow{\mathcal F_\perp^{-1}}
U_j,\qquad j=\mathrm{ASM},\mathrm F,$$

$$H_{\rm ASM}=e^{is\sqrt{k^2-k_x^2-k_y^2}},\qquad
H_{\rm F}=e^{iks}e^{-is(k_x^2+k_y^2)/(2k)}.$$

The full multiplier uses the outgoing/decaying square-root branch; Fresnel uses its paraxial approximation.

**Fraunhofer reads the input spectrum at mapped observation positions.**

$$U_{\mathrm{ref}}
\xrightarrow{\mathcal F_\perp}
\widetilde U_{\mathrm{ref}}(k_x,k_y)
\xrightarrow{\substack{(k_x,k_y)=(k\,x/s,\,k\,y/s)\\
\text{then multiply by }A_s(x,y)}}
U_{\rm FF}(x,y;s),$$

$$A_s(x,y)=\frac{e^{iks}}{i\lambda s}\,e^{ik(x^2+y^2)/(2s)}.$$

The usual Fraunhofer approximation makes the boundary integral itself a scaled input Fourier transform. Its chain therefore has no final inverse transform; the prefactor retains the carrier, spreading and observation phase. The validity conditions from Sections 3 and 4 still apply.
