# 9.0 From a Physical Model to a Numerical Experiment

**How do we turn a diffraction model into a numerical calculation whose results we can assess?**

## Motivation and prerequisites

Chapter 8 established how a known complex field propagates. Chapter 9 addresses the computational choices: how to obtain that starting field, represent it with finite data, and calculate its evolution accurately.

## The four-stage workflow

| Stage | Choices |
| --- | --- |
| **1. Define the mask and illumination** | Geometry, materials, wavelength, and incident field |
| **2. Obtain the exit field** | Thin-mask prescription $U_{\mathrm{exit}}=tU_i$, or a scalar FMM scattering calculation |
| **3. Represent the exit field** | Spatial samples $U_{\mathrm{exit}}(x,y)$, or spectral coefficients $C_{mp}$ |
| **4. Propagate the field** | Full ASM, Fresnel, or Fraunhofer within the method's validity range |

These decisions answer different questions. A material model determines the exit field; a representation stores it; a propagation operator evolves it. Changing representation need not change the physical model. Introducing an approximation can.

We will follow a two-dimensional periodic grating, first under plane-wave illumination and then under Gaussian illumination. The first case introduces discrete diffraction orders. The second adds finite illumination, broadened spectra, and computational-window choices.

## What makes a calculation trustworthy?

Keep **physical inputs** separate from **numerical controls**. Changing wavelength, opening width, or material index changes the optical problem. Increasing the retained harmonic count or refining a sampling grid should improve the representation of the same problem.

A smooth image or a balanced power budget is useful evidence, but neither establishes accuracy by itself. We will check convergence for the complex coefficients, fields, or intensity measurements we actually need, and distinguish numerical error from physical approximation.

## Before you begin

- Can more plotted pixels recover harmonics omitted from a scattering calculation?
- Does FFT agreement verify the physical model, or a change of representation?
- Which parameters would you vary to test accuracy without changing the grating?
