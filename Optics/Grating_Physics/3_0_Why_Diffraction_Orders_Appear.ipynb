# 3. Phase change and diffraction orders

[Download numerical explorers or standard Python scripts](Downloads.md).

**Why do diffraction directions carry integer labels?**

 For these direction constructions the incident and output media have the same index; refraction between unequal exterior media is not simulated.

## Motivation and prerequisites

Chapter 2 identified diffraction orders. Here we explain their discreteness by following phase between equivalent positions in neighboring periods. This gives the grating equation a physical meaning beyond a rule to memorize.

Review [2.2: illumination and the surface field](2_2_Illumination_and_the_Surface_Field.ipynb) and [2.3: allowed orders](2_3_Allowed_Diffraction_Orders.ipynb). You need relative phase, wavevector components and the fact that adding $2\pi$ leaves a complex phase factor unchanged.

## The central idea

Under plane-wave illumination, translating by one grating period need not leave the field itself unchanged: tilted incidence contributes a definite phase step. An outgoing order must reproduce that same translation factor. The incident and outgoing transverse phase steps can therefore differ only by an integer number of complete cycles.

This is a phase-matching condition. It fixes transverse wavevector differences, not the strength of an order. The path-difference picture describes the same condition geometrically; signed projections matter when the illumination is tilted. The examples assume a perfectly periodic structure and monochromatic illumination. Finite arrays broaden the ideal discrete-direction picture and are outside this chapter's derivation.
