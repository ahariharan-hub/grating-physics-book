# 3.2 Selected orders and phase continuity

**What does phase agreement look like for an integer order and for a non-integer trial?**

 For these direction constructions the incident and output media have the same index; refraction between unequal exterior media is not simulated.

Prerequisite: [3.1: the order rule](3_1_Phase_Change_and_the_Order_Rule.ipynb). Recall $(k_{x,m}-k_{x,i})d=2\pi m$, $k=2\pi/\lambda$, and $k_{x,i}=k\sin\theta_i$. The grating lies at $z=0$, repeats along $x$, and transmits into $z<0$ with $k_y=0$.

Run cells in order; no earlier notebook needs to be executed. Code parameters are independent of slider values. Lengths are in micrometres; wavelength is in the common medium. We use $e^{-i\omega t}$ and angles measured from $-z$ toward $+x$.

**Same-medium angle law, with index explicit:**
$$n\sin\theta_m=n\sin\theta_i+m\frac{\lambda_0}{d},\qquad \lambda=\frac{\lambda_0}{n}.$$
The factor $n$ is absorbed only when the equation is written using medium wavelength $\lambda$.

## 1. Follow the wave into a selected order

Extend the incident-wave representation from Notebook [3.1](3_1_Phase_Change_and_the_Order_Rule.ipynb). Set the **wavelength**, **period**, and **incidence angle**, then choose an order from **Show order**.

- **Left:** the incident wave above the grating and the selected outgoing plane-wave component below it, using the same signed-field colour scale as Notebook [3.1](3_1_Phase_Change_and_the_Order_Rule.ipynb). The arrows follow $\mathbf k_i$ and $\mathbf k_m$, respectively.
- **Right:** the possible order directions and their angles. Orange marks the selected order in both the drawing and the table.

The readout connects the drawing to the phase condition:

$$\Delta\phi_m-\Delta\phi_i=(k_{x,m}-k_{x,i})d=2\pi m.$$

To draw a real outgoing direction, we keep the orders with $|k_{x,m}|\leq k$, where $k=2\pi/\lambda$. Equality is a **grazing** order: its wave travels parallel to the grating. Both sides use the same wavelength in the medium. The real-direction test follows from the wavevector-length constraint [\[1, §§3.5–3.6\]](#ref-3-2-delft-wavefield); see also the discussion of allowed orders in [\[2, §2.2.1\]](#ref-3-2-palmer), [\[3, §2.2.1\]](#ref-3-2-newport-equation).

**Start with $\lambda=d=2\,\mu\mathrm m$ and $\theta_i=30^\circ$.** The dropdown contains $m=-1$ and $m=0$.

1. Select $m=-1$: the outgoing wave travels at $-30^\circ$. Its phase step is $-\pi$, while the incident step is $+\pi$; their difference is $-2\pi$.
2. Select $m=0$: the outgoing wave has the same direction and phase step as the incident wave. Its stripes continue along the same direction.
3. Increase $d$ to $4\,\mu\mathrm m$. More orders become available. Select them in turn and compare the stripe orientations, arrows, and phase steps.

**Reset example** restores the starting parameters and selects $m=-1$. The parameter controls here are independent of Notebook [3.1](3_1_Phase_Change_and_the_Order_Rule.ipynb).

Each displayed wave has unit amplitude and zero phase at the origin. The lower half shows **one outgoing component**; its actual strength and the total field at the grating require a scattering model. The picture illustrates direction and phase, with no order efficiencies assigned.

- It is an extended plane-wave component, not a finite beam bounded by its arrow or the image frame.
- The copied cells test exact periodicity; their displayed number does not set a physical grating length.
- A finite illuminated array would require an additional envelope model, which this continuity test does not supply. [\[4, §2.3\]](#ref-3-2-maystre) 

### Connect the seam to $k_{\mathrm{out},x}d$

Turn on **Compare non-integer values**. These plots share the wavelength, period, incidence, and physical coordinate ranges of the outgoing-wave region above. The ruler spans one period $d$; dotted lines mark cell boundaries.

The slider chooses a direction through

$$k_{\mathrm{out},x}=k_{x,i}+\frac{2\pi}{d}m_{\mathrm{test}}.$$

It includes directions between the integer orders. We call this trial vector $\mathbf k_{\mathrm{out}}$; it becomes $\mathbf k_m$ when $m_{\mathrm{test}}$ is an integer $m$. **Left:** a continuous plane wave at that direction. **Right:** its pattern in one cell, copied into neighbouring cells with the phase change of the illumination.

**Which phase are we following?** It is the phase of the outgoing wave,
$\phi_{\mathrm{out}}=k_{\mathrm{out},x}x+k_{\mathrm{out},z}z-\omega t$. The colours show the field $\cos\phi_{\mathrm{out}}$, not the phase itself.

1. **Look at the blue $\mathbf{k}_{\mathrm{out}}$ arrow.** It points along propagation, perpendicular to the stripes. Its orange horizontal projection is $k_{\mathrm{out},x}$: the phase change per micrometre along $x$. The arrows are scaled to fit the drawing; the readout gives their values.
2. **Walk from A to B.** These points are separated by $d$ at the same height and instant. Only $x$ changes, so
   $$\phi_{\mathrm{out}}(B)-\phi_{\mathrm{out}}(A)=k_{\mathrm{out},x}d.$$
   The orange graph starts at zero because it measures **phase change from A**, not absolute phase. Its orange dot is the phase reached at the end of this cell.
3. **Look at the green circle at B.** This is the starting phase of the next copy, also measured relative to A. Identical grating cells see the same illumination with a phase shift $k_{x,i}d$, so their responses must repeat with that shift. This is the **incident phase offset**. At normal incidence, it is zero.

- The seam compares those two values **at the same location B**: the phase just before the cell ends and the phase as the next copy begins.
- They join when the values differ by a whole number of $2\pi$ cycles.
- The arrow on the right describes the wave *inside* one copied cell.

**Set up the comparison:** $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, normal incidence. Here $k=\pi\,\mathrm{rad}/\mu\mathrm m$ and $k_{x,i}=0$.

| Comparison value | Orange: phase gained from A to B | Green: next copy starts at | Seam |
| --- | --- | --- | --- |
| $m_{\mathrm{test}}=0.5$ | $k_{\mathrm{out},x}d=\pi$ | $0$ | Half-cycle mismatch: broken |
| $m_{\mathrm{test}}=1$ | $k_{\mathrm{out},x}d=2\pi$ | $0$ | One full cycle: joins |

For $0.5$, the outgoing $k_{\mathrm{out},x}=\pi/4$ gives the half-cycle change over $d=4$. Moving to $1$ doubles $k_{\mathrm{out},x}$: the wave now gains a full cycle across the cell. **The two plotted phase values need not be numerically equal; they must represent the same position in a cycle.**

The broken seam belongs to this cell-copy test, not to a physical discontinuous outgoing field. This illustration applies the phase-repeat condition from [\[4, §2.3\]](#ref-3-2-maystre), with the normal-incidence continuity picture in [\[5\]](#ref-3-2-em-discrete). An integer value lets a continuous plane wave satisfy the grating's phase-repeat condition. Selecting an order or clicking **Use selected order** restores an exact order.
- The comparison slider includes only real outgoing directions.

### Small calculation 1: build the selected order's wave

Reproduce the explorer’s starting selection: $\lambda=d=2\,\mu\mathrm m$, $\theta_i=30^\circ$, and $m=-1$. First compute $k_{x,m}$ from the phase-repeat condition. Then choose the negative $k_{z,m}$ branch, so the wave travels into $z<0$.

`remaining` is $k^2-k_{x,m}^2$. Its sign tells us whether the selected component has a real outgoing direction. `np.sqrt` supplies the vertical component, and `np.arctan2` recovers the signed angle from $-z$. The cosine at the end is the same plane-wave expression used for the outgoing part of the interactive image.

**Predict, then try:** change `m` to `0`, then to `1`, keeping the other code settings fixed. Which selection continues along the incident direction, and which cannot be drawn as a real outgoing wave?


```python
wavelength, d, theta_i_deg = 2.0, 2.0, 30.0
m = -1  # integer order; independent of the dropdown
k = 2*np.pi / wavelength
k_x_i = k * np.sin(np.deg2rad(theta_i_deg))
k_x_m = k_x_i + m*2*np.pi/d
remaining = k**2 - k_x_m**2
if remaining < -1e-12*k**2:
    print(f'm = {m:+d}: no real outgoing direction')
else:
    k_z_m = -np.sqrt(max(0.0, remaining))
    theta_m_deg = np.rad2deg(np.arctan2(k_x_m, -k_z_m))
    incident_cycles = k_x_i*d / (2*np.pi)
    outgoing_cycles = k_x_m*d / (2*np.pi)
    x, z = 0.0, -1.0  # one pixel below the grating, in micrometres
    field_at_pixel = np.cos(k_x_m*x + k_z_m*z)  # t = 0, unit amplitude
    print('Outgoing order (k_x_m, k_z_m), rad/um:', np.round([k_x_m, k_z_m], 3))
    print(f'Angle from -z: {theta_m_deg:+.2f} degrees')
    print('Incident, outgoing, difference (cycles):',
          np.round([incident_cycles, outgoing_cycles, outgoing_cycles-incident_cycles], 3))
    print(f'Signed field at (x,z)=({x:g},{z:g}): {field_at_pixel:+.3f}')
```

:::{dropdown} Answer for the stated starting values and changes

For $m=-1$, $(k_{x,m},k_{z,m})\approx(-1.571,-2.721)\,\mathrm{rad}/\mu\mathrm m$ and $\theta_m=-30^\circ$. The incident phase change is ${+0.5}$ cycles; the outgoing change is ${-0.5}$ cycles. Their difference is ${-1}$ full cycle. At the chosen pixel the signed field is approximately ${-0.913}$.

For $m=0$, the outgoing angle is $+30^\circ$ and both phase changes are ${+0.5}$ cycles, giving zero difference. For $m=+1$, $k_{x,m}=1.5k$: this selection has no real outgoing direction. These answers refer to the fixed parameters written in this cell.

:::


### Small calculation 2: test the seam in phase cycles

Use the comparison view's example: $d=4\,\mu\mathrm m$, $\lambda=2\,\mu\mathrm m$, normal incidence, and $m_{\mathrm{test}}=0.5$. The end of one cell gains $k_{\mathrm{out},x}d$ relative to A. The next copy starts with the incident offset $k_{x,i}d$ relative to the same A.

The complex phase factor `np.exp(1j*phase)` has magnitude one. Adding $2\pi$ to its phase leaves its value unchanged. Dividing the next-copy factor by the end-of-cell factor tests the seam: a ratio of 1 means agreement. `np.isclose` allows for tiny numerical rounding errors.

**Predict, then try:** change `m_test` from `0.5` to `1.0`. Will the printed phase changes become numerically equal, or will their complex phase factors agree?


```python
wavelength, d, theta_i_deg = 2.0, 4.0, 0.0
m_test = 0.5  # independent of the comparison slider
k = 2*np.pi / wavelength
k_x_i = k * np.sin(np.deg2rad(theta_i_deg))
k_out_x = k_x_i + m_test*2*np.pi/d
if abs(k_out_x) > k*(1+1e-12):
    raise ValueError('Choose a comparison value with a real outgoing direction.')
end_phase = k_out_x*d
next_phase = k_x_i*d
end_factor = np.exp(1j*end_phase)
next_factor = np.exp(1j*next_phase)
seam_ratio = next_factor / end_factor
print('End, next start, end-minus-next (cycles):',
      np.round(np.array([end_phase, next_phase, end_phase-next_phase])/(2*np.pi), 3))
print('End factor; next-copy factor:', np.round(end_factor, 3), np.round(next_factor, 3))
print('Next/end ratio:', np.round(seam_ratio, 3))
print('Copies join:', bool(np.isclose(seam_ratio, 1.0, rtol=0, atol=1e-10)))
```

:::{dropdown} Answer for the stated starting values and changes

For $m_{\mathrm{test}}=0.5$, the orange end phase is ${0.5}$ cycles and the green next-start phase is zero. Their factors are ${-1}$ and ${+1}$, and the seam ratio is ${-1}$: a half-cycle mismatch.

For $m_{\mathrm{test}}=1$, the phase changes are still different numbers—one cycle and zero cycles—but both factors are ${+1}$. Their ratio is 1 and the copies join. The comparison is between **phase changes relative to A**, not between absolute field colours at two unrelated points.

:::
