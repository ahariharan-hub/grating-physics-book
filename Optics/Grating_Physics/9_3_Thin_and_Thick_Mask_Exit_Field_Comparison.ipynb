# 9.3 Thin and Thick Mask Exit-Field Comparison

[9.1](9_1_Defining_the_Complex_Exit_Field.ipynb) introduced the thin-mask rule at $z=0$. [9.2](9_2_Sampling_Complex_Fields_and_Their_Spectra.ipynb) showed how to sample a known field. We now compare thin and thick masks, starting with their geometry and the planes where their fields are defined.

## 1. Geometry and field definitions

Consider an absorbing layer of thickness $h$ with an opening of width $W_{\mathrm{open}}$, as shown below. The layer stands in air, with no substrate. Air also fills the opening. Light arrives from above and travels toward negative $z$. We put the top surface at $z=0$, so the layer lies between $z=0$ and $z_e=-h$. The drawing shows one opening.

![One opening in an absorbing layer in air. The chosen thin-mask plane is at the top, and the physical exit plane is at the bottom.](assets/9_3_single_opening_reference_planes.png)

*Figure 9.3.1. The orange dashed line marks the chosen thin-mask plane, $z_{\mathrm{obj}}=0$. The blue dotted line marks the thick mask's exit plane, $z_e=-h$.*

As in 9.1, we place the mathematical thin mask at $z_{\mathrm{obj}}=0$. This is our chosen object plane. The thin-mask rule multiplies the incident field by the transmission function ([1](#ref-9-3-thin-mask), Eq. 4.37). For the centred opening,

$$
\begin{gathered}
U_{\mathrm{thin}}(x,0^-)=t(x)U_i(x,0),\\[0.4em]
t(x)=\begin{cases}
1,& |x|<W_{\mathrm{open}}/2,\\
0,& |x|>W_{\mathrm{open}}/2.
\end{cases}
\end{gathered}
$$

Here $0^-$ means just below the thin-mask plane. This rule replaces the thick layer with a transmission function at one plane. It does not give the actual field inside the thick opening. The dark blocks in the drawing show the thick geometry we want to compare with this thin-mask model.

For the thick mask, we define the exit field just below its bottom surface:

$$
U_{\mathrm{exit}}^{\mathrm{thick}}(x)
=U_{\mathrm{total}}(x,z_e^-)
=U_{\mathrm{total}}(x,-h^-).
$$

Here $-h^-$ means just below $z=-h$, in the air. The exit field covers the whole plane, including the regions below the absorber. We must solve how the light interacts with the material to find this field. We cannot assume it is $tU_i$.

We choose the top surface, $z_{\mathrm{obj}}=0$, as the object plane to match 9.1. We must test how well the thin-mask model at this plane agrees with the thick-mask result. Applying $tU_i$ at the bottom surface is a different choice: it is not the same as applying it at the top and then propagating that field through a distance $h$ in air.

We observe the fields at $z_{\mathrm{ref}}=z_e-\ell=-h-\ell$, where $\ell\geq0$ is the distance below the thick mask's exit. For a fair field comparison, both predictions must eventually be evaluated at a **common observation plane**.

## 2. Bring the thick-mask response to the object plane

Start with unit-amplitude normal incidence, with $U_i(x,0)=1$. Set $\ell=0$, so the observation plane lies at the thick mask's exit, $z_{\mathrm{ref}}=z_e=-h$. We first solve how the light interacts with the material and find the field at this plane. We then back-propagate its outgoing components through a distance $h$ in air to the chosen object plane, $z_{\mathrm{obj}}=0$:

$$
\begin{gathered}
\{\text{thick geometry},\,U_i\}
\xrightarrow{\text{solve scattering}}
U_{\mathrm{thick}}(x,z_{\mathrm{ref}}=z_e=-h)
\\[0.7em]
\xrightarrow{\text{back-propagate outgoing components by }h}
U_{\mathrm{eq}}^{\mathrm{thick}}(x,z_{\mathrm{obj}}=0)
\\[0.7em]
\xrightarrow{\text{compare amplitude and phase}}
U_{\mathrm{thin}}(x,0^-)=t(x)U_i(x,0).
\end{gathered}
$$

The two fields in the comparison are now defined at the same object plane and use the same incident amplitude and phase. For this normal-incidence starting case, the ideal thin-mask field is simply $t(x)$: it has amplitude 1 and phase 0 in the opening, and is zero outside. Where the field is zero, its phase is undefined.

The back-propagated field $U_{\mathrm{eq}}^{\mathrm{thick}}$ is an equivalent outgoing field at the object plane. It is not the actual total field at the top of the thick geometry. Back-propagation here is a mathematical step applied to the outgoing field in air; it does not mean sending light backward through the material. Comparing this equivalent field with $t(x)$ shows how well the simple thin-mask rule represents the thick mask's response. For the equivalent-mask construction, see [3](#ref-9-3-pistor), Figures 4-14 and 4-15.

### Calculate and compare

The following calculation uses the absorbing screen from 0.7: period $d=1.5\,\mu\mathrm{m}$, opening width $W_{\mathrm{open}}=0.75\,\mu\mathrm{m}$, vacuum wavelength $\lambda_0=0.633\,\mu\mathrm{m}$ and illustrative absorber index $1.5+0.8i$. The numerical model repeats the opening periodically; the earlier single-opening drawing establishes the geometry, not an isolated-slit simulation. We begin at normal incidence with thickness $h=0.3\,\mu\mathrm{m}$.

The solver gives the outgoing coefficients $C_m(z_e)$ at the physical exit. Each propagating coefficient is brought back to the object plane using its own longitudinal wavevector:

$$C_m^{\mathrm{eq}}(0)=C_m(z_e)e^{-i\beta_mh},\qquad
U_{\mathrm{eq}}^{\mathrm{thick}}(x,0)=\sum_{m\in\mathcal P}C_m^{\mathrm{eq}}(0)e^{ik_{x,m}x},$$

where $k_{x,m}=k_{x,i}+mG$ and $\beta_m=\sqrt{k_0^2-k_{x,m}^2}$ on the outgoing branch. Each factor follows angular-spectrum propagation in air ([2](#ref-9-3-propagation), Angular Spectrum Method). The ideal thin-mask curve uses the same propagating orders $\mathcal P$, with coefficients $t_m=(W_{\mathrm{open}}/d)\operatorname{sinc}(mW_{\mathrm{open}}/d)$ and unit incident amplitude at $z=0$.

The top row shows one opening, the chosen planes, and all propagating order directions. Use **Show order m** to highlight one direction and its row in the power table. The wavefront sketch shows the incident wave above the layer and the selected outgoing wave below it, each with unit amplitude to show direction. It is not the total scattered field. The green curves below show that order's calculated contribution; the blue and orange curves still sum all considered orders.

Start by changing the thickness. The middle row shows the physical exit field at $z_{\mathrm{ref}}=z_e$; the lower row compares the back-propagated response with the ideal thin mask at $z_{\mathrm{obj}}=0$. Then tilt the illumination and compare positive and negative angles. Use **Spatial phase** to switch between wrapped and unwrapped phase. Unwrapping follows the field along $x$ within each region where its amplitude is nonzero; it does not connect phase across a zero-field gap. Neither curve is scaled or phase-shifted to match the other.

The lower row uses all solved propagating orders. The ideal curve uses those same orders, so it is a smooth Fourier projection rather than the exact binary boundary. Small negative lobes outside the opening have phase $\pi$; the exact ideal field is zero there. The optional controls show the exact ideal boundary as a separate dotted curve and let you change the scattering cutoff $M$. The material solve and the middle exit-field plots still include all solved orders; only the backward comparison is restricted to propagating orders.

## 3. Compare the diffraction orders

Open **Compare orders** in the same view above. The thickness, angle and selected order are shared by both tabs. The spatial fields are sums of these complex coefficients, so differences between individual orders help explain differences between the blue and orange field curves. Both models are compared at $z_{\mathrm{obj}}=0$:

$$C_m^{\mathrm{thin}}(0)=t_m,\qquad
C_m^{\mathrm{eq}}(0)=C_m^{\mathrm{thick}}(z_e)e^{-i\beta_mh}.$$

For the ideal mask, $t_m=(W_{\mathrm{open}}/d)\operatorname{sinc}(mW_{\mathrm{open}}/d)$ is known analytically. We could also recover it by taking the FFT of the complex ideal field after removing the incident carrier, as in 9.2. The thick coefficients come from the scattering solve and the backward propagation; they do not come from an FFT of the ideal mask.

The first plot compares magnitudes. The second shows the wrapped phase difference, $\arg[C_m^{\mathrm{eq}}(0)C_m^{\mathrm{thin}}(0)^*]$, where both coefficients are large enough to define a phase. The table gives the selected order's real and imaginary parts, magnitude and phase, followed by the complex difference. These coefficients are field amplitudes, not power fractions.

Start with normal incidence and vary the thickness. Select an order whose ideal coefficient is nonzero, then compare its magnitude and phase. Next change the angle and check which orders differ most. Switching tabs or selecting an order reuses the existing solution.

## 4. Compare the fields at an observation plane

Open **Observation plane** in the same view. Both object-plane predictions are now propagated to $z_{\mathrm{ref}}=-h-\ell$. Since they start at $z_{\mathrm{obj}}=0$, the forward distance is $h+\ell$:

$$\begin{aligned}
C_m^{\mathrm{thin}}(0)&\xrightarrow{\text{propagate }h+\ell}
C_m^{\mathrm{thin}}(z_{\mathrm{ref}})=t_m e^{i\beta_m(h+\ell)},\\[0.4em]
C_m^{\mathrm{eq}}(0)&\xrightarrow{\text{propagate }h+\ell}
C_m^{\mathrm{replay}}(z_{\mathrm{ref}})=C_m^{\mathrm{eq}}(0)e^{i\beta_m(h+\ell)}.
\end{aligned}$$

We also query the thick-mask solution directly at $z_{\mathrm{ref}}$. This is equivalent to propagating its exit coefficients through the shorter distance $\ell$. The direct and replay fields must agree when reconstructed with the same propagating orders. The grey markers provide this check, and the readout gives their relative complex-field difference. This checks the reference-plane calculation; it does not replace a check of the scattering cutoff.

Start at $\ell=0$, where the observation plane is the physical exit. Move **Distance below exit** to compare amplitude, phase and intensity farther below the layer. The thin and thick curves may change as their components interfere, while the replay and direct thick results should stay together. Use **Spatial phase** to switch between wrapped and unwrapped phase. The distance control moves the observation plane; it does not solve a new scattering problem.

All three curves use the same propagating orders. Even at $\ell=0$, they show the propagating part of the field rather than the complete near field, which also contains evanescent components.

## What have we learned?

The thin mask prescribes a field at $z_{\mathrm{obj}}=0$. The thick mask requires a scattering calculation, which gives its outgoing field at $z_e=-h$. These are different planes. To compare the models, we bring their predictions to the same plane and retain the same incident amplitude and phase.

Back-propagation gives an equivalent outgoing thick-mask field at the object plane. Its amplitude and phase can differ from the thin-mask prediction because the physical layer reflects, absorbs and scatters light. This equivalent field is not the actual total field at the top surface. The ideal mask adds no phase in its openings at the object plane, but propagation to another plane changes its field phase.

Our backward comparison retains all solved propagating orders. The ideal curve uses those same orders, so it is a smooth projection of the binary boundary. It does not reproduce the complete near field. Forward propagation brings both predictions to $z_{\mathrm{ref}}=-h-\ell$; agreement between the replay and direct thick fields checks the plane changes. Changing the scattering cutoff $M$ is a separate accuracy check.
